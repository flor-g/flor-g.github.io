# <a id="tc1" name="tc1"></a>Text cell 1: Scalar inference as predictive coding

**Goal.** Build a predictive-coding / free-energy model of scalar inference with:

- a latent **world state** $s$: the proportion of a set that satisfies the predicate, as in the fraction of a jar of marbles that sank

- an observed **utterance** $y \in \{\textit{no},\textit{some},\textit{all}\}$

- a base world prior over $s$ whose sharpness is under the modeller's control, and is swept in Text cell 6

Text cell 3 fixes each of these and states the notation the rest of the notebook keeps to: $s$ lies
in the open interval $(0,1)$ and is carried in logit coordinates on a quadrature grid; $y$ clamps a
lexical field through the exclusion set of its entry; the base world prior is sampled on the same grid. The
letters here already follow that convention, so the observation is written $y$, since Text cell 3's
commitment 5 reserves $u$ for the utility level.

**Predictive-coding motivation.** Bogacz (2017) defines a log-joint quantity $F$ and performs gradient ascent on it to infer latent states (his Eq. (6), **p. 200**):

$$
F(\phi;y) = \ln p(\phi) + \ln p(y\mid \phi),
\qquad
\dot\phi \propto \frac{\partial F}{\partial \phi}.
$$

## <a id="tc2" name="tc2"></a>Text cell 2: High-level predictive-coding pseudocode (reference implementation plan)

This cell is **pseudocode**. It sketches the classical predictive-coding pattern:

1. Maintain **state units** (latent causes) and **error units** (prediction errors).

2. Run **inner-loop inference dynamics** until errors settle.

3. Run **outer-loop learning** (Hebbian-style) on synapses generating predictions.

**Connection to Bogacz (2017).**
- In the hierarchical predictive-coding network, state and error dynamics take the form (Bogacz Eqs. (53)–(54), **p. 206**):

$$
\dot\phi_i = -\varepsilon_i + h'(\phi_i)\,\Theta_{i-1}^\top\varepsilon_{i-1}
$$

$$
\dot\varepsilon_i = \phi_i - \Theta_i h(\phi_{i+1}) - \Sigma_i\varepsilon_i.
$$

- Learning the generative weights is Hebbian in the sense that updates are proportional to a product of an error term and an activity term (e.g., Bogacz Eq. (50), **p. 206**):

$$
\Delta\Theta \propto \varepsilon\,h(\phi)^\top.
$$

- For learning **noise/uncertainty** parameters with local plasticity, Bogacz introduces an auxiliary inhibitory interneuron $e_i$ and dynamics (Eqs. (59)–(60), **p. 207**) so that error nodes converge to precision-weighted errors (Bogacz Eq. (61), **p. 207**):

$$
\dot\varepsilon_i = \phi_i - g_i(\phi_{i+1}) - e_i,\qquad
\dot e_i = \Sigma_i\varepsilon_i - e_i.
$$

### <a id="tc2-bogacz" name="tc2-bogacz"></a>Bogacz's construction in one dimension, step by step

The accompanying paper (§2.2) keeps what its later sections use of Bogacz (2017). This section sets the construction out step by step, in his scalar notation, with his section and equation numbers.

1. **Two constraints** (his §1). Bogacz states two constraints that a model would need to satisfy to be considered biologically plausible: local computation, in which a neuron computes only from the activity of its inputs and the weights of those inputs, and local plasticity, in which a synapse changes only on the basis of its pre-synaptic and post-synaptic activity.
2. **The problem** (§2, Eqs. 1–3). An organism infers the size $v$ of a food item from a light intensity $u$. The likelihood is $p(u\mid v)=f(u;g(v),\Sigma_u)$ and the prior $p(v)=f(v;v_p,\Sigma_p)$.
3. **Exact inference** (§2.1, Eqs. 4–5). The exact posterior requires the normalizing integral $p(u)$, which is what makes exact inference hard. For a discrete distribution, circuits in the basal ganglia have been proposed to compute the normalization (Bogacz & Gurney, 2007, as reported by Bogacz, 2017, §2.1).
4. **The most likely value** (§2.2, Eqs. 6–9). The estimate $\phi$ is found by gradient ascent on $F=\ln p(\phi)+\ln p(u\mid\phi)$, which moves it towards a maximum of $F$; with a nonlinear $g$, the ascent can stop at a local maximum. The gradient (Eq. 8) is a sum of two terms, weighted by the reliabilities of the prior over $v$ and of the input.
5. **Prediction errors** (§2.3, Eqs. 10–14). The two terms are named prediction errors, $\varepsilon_p$ and $\varepsilon_u$ (Eqs. 10–11), and a network of nodes computes them with local operations. In that network the prior mean $v_p$ is the weight of a connection from a tonically active neuron (§2.3).
6. **Learning the parameters** (§2.4, Eqs. 15–21). $v_p$, $\Sigma_p$ and $\Sigma_u$ are learned by ascending $F$, with rules of Hebbian form. They are updated after each observation and, the observations differing, never converge; they stay near the values at which the expected change is zero (§2.4). The variances are where precision learning enters (§2.4).
7. **Learning a relationship between levels** (§2.5, Eqs. 22–29), of the form $g(v)=\theta h(v)$. In the linear case, $g(v)=\theta v$, the rule for $\theta$ is the product of an error and a state activity, $\partial F/\partial\theta=\varepsilon_u\phi$ (Eq. 25).
8. **Free energy** (§3, Eqs. 30–35). The posterior is approximated by a distribution $q(v)$ chosen to minimize the Kullback–Leibler divergence from the true posterior. Since the divergence is non-negative, the negative free energy of a suitable $q$ is a lower bound on $\ln p(u)$, so maximizing it both fits $q$ and learns the parameters: one objective serves inference and learning. Bogacz then takes $q$ to be a delta at $\phi$ and drops the term $-\int q\ln q$, which does not depend on $\phi$ (his Eq. 34). For a delta the dropped term is not finite, so what remains, the $F$ of step 4, is not itself the bound: the reduction yields the objective of maximum a posteriori inference, whose gradient in $\phi$ is the limit of the bound's gradient in the centre of a Gaussian $q$ whose width tends to zero.
9. **A hierarchy** (§4.2, Eqs. 51–56), each level predicting the one below; it is stated here with one scalar per level. In the vector form the weights between two levels are a matrix, each entry learned by its own product of an error and a state activity (Eq. 56), and the same matrix carries the prediction down and, transposed, the error up (Eqs. 53–54).
10. **Local plasticity for the variances** (§5, Eqs. 57–66). In the vector form the rules that learn the variances contain a matrix inverse, so the change of one weight would depend on the weights of other synapses. Bogacz removes the inverse by adding to each prediction-error node an interneuron whose activity holds the quantity the rule needs (§5.1, Eqs. 59–63). The error nodes are assumed to settle much faster than the rest of the network, and under that separation they converge to stable fixed points (§§5.1, 6).

**The two standard decompositions of the free energy.** The second is Bogacz's Eq. (33) rearranged:

$$
\mathcal F(q)=D_{KL}[q(v)\|p(v)]-\mathbb E_q[\log p(u\mid v)],
$$
$$
\mathcal F(q)=D_{KL}[q(v)\|p(v\mid u)]-\log p(u).
$$

Free energy therefore upper-bounds surprisal. The quantity above is minimized; Bogacz's $F$ is its negative and is maximized (his Eqs. 32–35), and this notebook follows his sign (Text cell 3, preamble item 4).

**What the construction assumes to be Gaussian, and why it breaks without it.**

- What is assumed Gaussian is the generative model's densities: the prior over the hidden variable and the likelihood of the input (Eqs. 1–3). The posterior need not be Gaussian. Bogacz's own $g(v)=v^2$ makes it non-Gaussian, which is why he approximates it; the posterior "may have a complicated shape" (§3, recalling §2.1).
- A Gaussian log-density is quadratic in the difference between an activity and its prediction, so the gradient (Eq. 8) is a sum of such differences, each divided by a variance, the likelihood's term multiplied by $g'(\phi)$. Each difference is what an error node computes by subtraction (Eqs. 10–11), and it is why each update reads only its own afferents (Eqs. 7–11). Under another family the score, and with it the rules that learn the variances, generally changes, so Bogacz's displayed equations would have to be re-derived.
- For the Gaussian, the normalizers collect into terms in $\ln\Sigma$ and a constant, so they drop out of the gradient in $\phi$ (Eq. 7).
- The rules that learn the variances are derived from the Gaussian's dependence on $\Sigma$ (§2.4, Eqs. 15–21; §5.1, Eqs. 57–66).
- The delta approximation keeps only a mode of the posterior (§3, Eqs. 30–34).

<a id="pseudocode" name="pseudocode"></a>

```python
# PSEUDOCODE
# shapes: batch B, latent dim D2, sensory dim D1
#batch size is the number of independent scenarios/participants we are simlutating at time t
#dimension is the number of features in a given layer

phi2 = torch.zeros(B, D2)          # state units (layer 2)
eps1 = torch.zeros(B, D1)          # error at sensory level
eps2 = torch.zeros(B, D2)          # error at latent level (optional, if adding a prior)
e1   = torch.zeros(B, D1)          # interneuron for local Sigma(variance) learning (optional, sets how much we care about errors form different sources)

for t in range(T): #time iteration
    pred1 = (h(phi2) @ Theta1.T)   # generate(phi2) = Theta1 h(phi2)

    if local_sigma:               # cf. Bogacz Eq. 59–60
        eps1 = eps1 + dt * (u - pred1 - e1) # error = (input - prediction) - sigma_eps
        e1   = e1   + dt * (eps1 * Sigma1 - e1)
    else:                         # cf. Bogacz Eq. 54
        eps1 = eps1 + dt * (u - pred1 - eps1 * Sigma1)

    # phi2 update (Eq. 53): bottom-up drive from eps1
    bu = (eps1 @ Theta1)          # take the sensory error eps1 and multiply by the weight matrix to feed the higher layer
    phi2 = phi2 + dt * (hprime(phi2) * bu - eps2)  # take derivative (dh/dphi) and multiply it by bottom-up error, subtract with top-down pressure (eps2)

# After inference settles: Hebbian weight update (Bogacz Eq. 56 / 50)
Theta1 += lr_theta * (eps1.T @ h(phi2)) / B

# If learning Sigma locally: Bogacz Eq. 71 (diagonal case -> elementwise)
if local_sigma:
    Sigma1 += lr_sigma * ((eps1 * e1).mean(dim=0) - 1.0)
    Sigma1 = Sigma1.clamp_min(1e-3)   # keep positive
```

## <a id="tc3" name="tc3"></a>Text Cell 3: three-level predictive-coding network over a continuous world state

The model makes several commitments that the reader should note while reading:

1. The chain of prediction runs utterance ($y$) $\leftarrow$ lexical ($L$) $\leftarrow$ situation ($S$) $\leftarrow$ utility ($u$).

2. The world scale is the closed interval $[0,1]$ (§3, item 2), continuous by stipulation (§1). The state is tracked in logit coordinates, which cover the open interval $(0,1)$, on a fixed quadrature grid. The grid's node count is a numerical substrate and not a state space; its **half-width** is a different matter, and §1 says why.

3. At this phase, the model does not implement any precision inference. Hence, every variance ($\sigma$) is fixed at $1$. The value is the multiplicative identity because $\sigma$ weights a synaptic connection, so fixing every $\sigma$ at $1$ assigns equal weight to every connection. The **exposure ensemble** is stipulated in the same way and for the same reason: this phase does not model how often an utterance is heard, so $p(y)$ is uniform over the three, which weights every utterance alike and makes the ensemble vanish from Eq. (B2) (Appendix B).

4. **Sign and language convention.** $\mathcal F$ below is equivalent to the $F$ defined by Bogacz (2017), henceforth "Bogacz". $F$ is *negative* free energy, with dynamics given by gradient *ascent*. Every gradient sign in this document follows Bogacz. The word "construction" refers to the general constraints that the free-energy principle imposes on a
predictive-coding network, as explicated in Bogacz.

5. **Subscript convention.** Four letters index four levels, capitalized exactly when the level is field-valued: $L$ and $S$ carry $\mathbb R^K$ fields, $y$ and $u$ do not. Each letter names one thing only. $y$ denotes the observed utterance ($E_y$, $\chi_y$, and the read-out $q_y$ of Text cell 4). $L$ denotes the lexical level ($\varphi_L$, $\varepsilon_L$, $\sigma_L$, $g_L$). $S$ denotes the situation level ($\varphi_S$, $\varepsilon_S$, $\sigma_S$, $g_S$); $S$ is capitalized because the level is field-valued, and the lowercase $s$ is reserved for the world state itself, the proportion of Eq. (1). $u$ denotes the utility level ($\theta_u$, $\varphi_u$, $\varepsilon_u$, $\sigma_u$, $\mu_u$) and does not denote utterance at any point.

6. **$\theta_u$ is learned, and belongs to the configuration.** The network's one modifiable parameter carries $\theta_u^\ast$, the maximizer of $\tilde{\mathcal F}$ given in closed form by Eq. (B2). It is a function of everything a configuration fixes: $\ell_0$, $\Lambda$, $\mu_u$, $\theta_L$ (through $n$), $K$ and the grid half-width, the mask sharpness, $B$ (through $m$), every $\sigma$, and the exposure ensemble of Appendix B, uniform at this phase. Two configurations therefore carry two different $\theta_u^\ast$, and every evaluation below re-learns it wherever it respawns the network. A pinned $\theta_u$ appears only as a control, and is labelled as one where it appears. The flow of Eq. (20) starts at $\theta_u(0)=0$, one start shared by every configuration.

7. **A timescale commitment, not merely a timescale ordering.** The error units are held to $\tau_\varepsilon\le\tau_\varphi/\!\left(4\lambda_{\max}(H)\right)$, where $H$ is the Hessian of §8.1, taken in the metric §8 states, at the current $\theta_u$. §8.3 derives it. Since $\lambda_{\max}(H)$ grows as $\theta_u^2$, the separation the model demands tightens as the square of the learned parameter. What the commitment secures is a precondition of the *monotone* rise of $\mathcal F$ (§8.2(iii)), which also uses the silent start of the error units, and not convergence, which holds at every separation (§8.2(ii)). The accompanying paper calls this bound the critical-damping bound (§3.4).

---

### <a id="tc3-1" name="tc3-1"></a>1. State space and numerical substrate

The world state is a proportion $s\in(0,1)$, carried in logit coordinates

$$\zeta=\operatorname{logit}(s)=\log\frac{s}{1-s}\in\mathbb R,
\qquad
s=\operatorname{logit}^{-1}(\zeta)=\varsigma(\zeta)=\frac{1}{1+e^{-\zeta}},
\tag{1}$$

where $\varsigma$ is the standard logistic function. It is written $\varsigma$ rather than the usual
$\sigma$ throughout, because $\sigma$ is reserved here for the variances
$\sigma_y,\sigma_L,\sigma_S,\sigma_u$ of §3. Since $\operatorname{logit}$ is a bijection
$(0,1)\to\mathbb R$, Eq. (1) loses nothing of the state. The construction's Gaussian densities are over the network's
state variables, $\varphi_S$ and $\varphi_u$ (§§4–5), which are therefore real-valued; they do not
constrain the domain that labels a field's components. That the world scale is continuous is
therefore a stipulation of this implementation, motivated by quantifiers and gradable adjectives
both organizing meaning over scales, so that one continuous representation serves both. That its
coordinate is logit is a further stipulation: the construction does not require an unbounded
coordinate, and the grid truncates this one in any case (below). Results depend on the choice: the grid is uniform in
$\zeta$, the mode is read over $\zeta$ (Text cell 4, Part D), and $\theta_L$ is a log-gain in it
(Appendix A). It is chosen because translating by $\theta_L$ in $\zeta$ multiplies the odds, so the
threshold is a log-gain, and because over $\zeta$ every $\mathrm{Beta}(\alpha,\beta)$ base world prior has its mode inside the scale, which the mode criteria need (Text cell 4, Part D). The quadrature, the grid
and the default base world prior live in $\zeta$.

Integrals over $\zeta$ are evaluated on a **fixed** grid ($\zeta_1,\dots,\zeta_K$) with fixed
positive quadrature weights ($w_1,\dots,w_K$), where $w_k,\zeta_k\in\mathbb R$. For $f,h\in\mathbb R^K$ we write

$$\langle f,h\rangle=\sum_{k=1}^K w_k\,f_k h_k,
\qquad
\|f\|^2=\langle f,f\rangle,
\qquad
w_k,\,f_k,\,h_k\in\mathbb R,\quad w_k>0 .
\tag{2}$$

We call such $\langle\cdot,\cdot\rangle$ the **quadrature inner product**: the inner product induced on
$\mathbb R^K$ by the quadrature rule. Equivalently, $\langle\cdot,\cdot\rangle$ is the inner product
of the weighted space $\ell^2(w)$. With $w_k>0$, Eq. (2) is a genuine inner product and $\|\cdot\|$ a norm. §8.1's proof will use this fact.

Where matrix notation is more convenient we write $W=\operatorname{diag}(w_1,\dots,w_K)$ for the
same measure, so that

$$\langle f,h\rangle=f^{\mathsf T}Wh,\qquad
B^{\mathsf T}Wf=\big(\langle b_1,f\rangle,\dots,\langle b_m,f\rangle\big)^{\mathsf T},\qquad
B^{\mathsf T}WB=\big(\langle b_i,b_j\rangle\big)_{ij}.
\tag{3}$$

The two notations are the same object throughout and are used interchangeably: $\langle\cdot,\cdot\rangle$
where a single pairing is meant, $W$ where several projections are taken at once. $W$ is diagonal and
positive, and never a free parameter: it is the grid's own measure, fixed with the grid.

The scalar $\langle f,h\rangle$ is a discretization of the $L^2$ inner product over the grid's own
interval, $\int_{-Z}^{Z} f(\zeta)h(\zeta)\,d\zeta$, which it approaches as $K\to\infty$ at a fixed
half-width $Z$.

**The grid carries two parameters and they are not of the same kind.** $K$ is an accuracy
parameter: refining it converges the settled fields and the expectations read from them, and no
quantity below is defined in terms of a state indexed by $k$. A criterion of Text cell 4 is a
threshold on those fields, so its status can still change at a boundary cell as $K$ moves;
Appendix A prints the four criteria along a ladder in $K$. The half-width $Z$ is not. Eq. (1) carries the state onto the whole of $\mathbb R$, so cutting
it at $\pm Z$ **bounds the state space**, and there is no $Z\to\infty$ limit for a quantity to
converge to: a different $Z$ is a different truncation, not a better approximation to the same one.
Eq. (1) is a stipulated choice of coordinates; **the truncation that follows it
is a further stipulation**, one of this implementation rather than a consequence of the semantics, and
Appendix A fixes only a lower bound on it, $\theta_L<Z$, so that $E_{\textit{no}}$ keeps an
unexcluded node. Nothing fixes $Z$ beyond that.

One consequence is carried here rather than left implicit, and it is **not** that the truncation
clips a tail. The basis $B$ of §3 is built on the grid: its columns are orthonormalized under
$\langle\cdot,\cdot\rangle$, which $Z$ fixes. So **$B$ is rebuilt by $Z$**, and the settled fields
of §6 depend on the half-width through it, in shape and not merely in scale. What that costs the
statistics Text cell 4 defines is said there, where those statistics are.

**Why the situation level carries a field over $\zeta$.** In Bogacz's examples a state node
estimates a world quantity, and its settled activity is that quantity's most likely value. The
situation level here is built differently: its $K$ units carry the log-weight of the world state at
each grid node, so the hidden variables are a field over $\zeta$, and $s$ appears only as the label
of a unit. A single node for $\zeta$ is the alternative, and Bogacz's framework admits it. The
entry would enter through a nonlinear generative map, as the size of a food item enters his first
example through $g(v)=v^2$ (Bogacz §2); the update would be local; and the settled activity would be
the most likely $\zeta$, with nothing left to read off. It is not taken, for three reasons.

1. *The entry is a set.* $E_y$ is a region of the scale (Eq. 5), and a region enters a single node
   only through a step-shaped log-likelihood. Added to the log-density of the base world prior, it leaves $\mathcal F$
   non-concave in that node, with two maxima wherever the excluded region holds the mode of the base world prior (one at that mode, one at the edge of the region left open), so the uniqueness of §8.1, the closed forms of §6, and with them the exact
   test of the implementation would be lost. Carried as a field, the entry is a vector that the
   level above predicts affinely (Eq. 9), and $\mathcal F$ stays a sum of squares in it.
2. *A single value carries no mass.* The mass criteria of Text cell 4 are statements about mass, as
   are the accounts they are compared with.
3. *The utility level acts on the belief's shape.* It tilts and widens it (Text cell 5 Part B); a
   point has no width.

The price is stated with the choice. The network carries a value at every node, which is an
unnormalized distribution over $s$, and a statement about $s$ has to be read off that field; Text
cell 4 gives the two ways it is read.

### <a id="tc3-2" name="tc3-2"></a>2. Inventory of quantities

| Symbol | Description | Space | Status |
|---|---|---|---|
| $\varphi_L$ | lexical exclusion field | $\mathbb R^K$ | **fixed**  |
| $\varphi_S$ | situation log-density field | $\mathbb R^K$ | **variational** (fast) |
| $\varphi_u$ | utility state | $\mathbb R^m$ | **variational** (fast) |
| $\varepsilon_L;\varepsilon_S;\varepsilon_u$ | error units | $\mathbb R^K,\mathbb R^K,\mathbb R^m$ | **variational** (fast) |
| $\theta_u$ | the scalar of the generative function $g_S(\varphi_u)$ | $\mathbb R$ | **variational** (slow); learned, $\theta_u^\ast$ of Eq. (B2), per configuration |
| $\varepsilon_y; \theta_L$ | utterance error; a scalar of $g_y(\varphi_L)$ | $\mathbb R^{\lvert\mathcal Y\rvert}$, $\mathbb R_{>0}$ | fixed ($\varepsilon_y\equiv0$ by Eq. 8) |
| $\mathcal Y$ | utterance inventory, one word-form unit per member | $\{\textit{no},\textit{some},\textit{all}\}$, $\lvert\mathcal Y\rvert=3$ | fixed |
| $E_y$ | emission of observed utterance | $\subseteq[0,1]$ | fixed |
| $K$ | grid nodes | $\mathbb N$ | fixed |
| $\langle\zeta_k,w_k\rangle$; $W$ | quadrature, where k indexes (1, ..., K); $W=\operatorname{diag}(w_k)$ is its matrix form (Eq. 3) | $\langle \mathbb R, \mathbb R \rangle$; $\mathbb R^{K\times K}$ | fixed |
| $\chi_y$ | quadratured indicator of emission  | $\{0,1\}^K$ | fixed |
| $\Lambda$ | lexical strength, a scalar of $g_y(\varphi_L)$ | $\mathbb R_{>0}$ | fixed |
| $m$ | utility dimension | $\mathbb N$ | fixed |
| $B$ | utility basis | $\mathbb R^{K\times m}$ | fixed |
| $\ell_0$ | log of the base world prior | $\mathbb R^K$ | fixed (supplied; default $\mathcal N(0,1)$ in $\zeta$) |
| $\mu_u$ | mean of the terminating prior on $\varphi_u$ | $\mathbb R^m\setminus\{0\}$ | fixed|
| $\sigma_y;\sigma_L;\sigma_S;\sigma_u$ | variances | $\mathbb R_{>0}$ | fixed |
| $\tau_\varepsilon;\tau_\varphi;\tau_\theta$ | time constants | $\mathbb R_{>0}$ | fixed; $\tau_\varepsilon\le\tau_\varphi/\!\left(4\lambda_{\max}(H)\right)\ll\tau_\varphi\ll\tau_\theta$ (§8.3) |

**Where $\varepsilon_y$ lives.** The $y$ level is lowercase under commitment 5 for the same reason $u$
is: it carries one unit per word form, not a field over the grid. $g_y$ *contracts* the $K$-node
lexical field onto those $\lvert\mathcal Y\rvert$ units, by $\hat\varphi_y=\theta_L AW\varphi_L$ with
$A\in\mathbb R^{\lvert\mathcal Y\rvert\times K}$ (Appendix A, Eq. A2), whereas $g_S$ expands $\mathbb R^m$ into a
field. Hence

$$\varepsilon_y=\sigma_y^{-1}\big(\varphi_y-\theta_L AW\varphi_L\big)\in\mathbb R^{\lvert\mathcal Y\rvert},
\tag{4}$$

the mirror of $\varepsilon_u\in\mathbb R^m$ at the other end of the chain, and the space that makes
Eq. (A4) a scalar as $\theta_L$ requires. Eq. (8) pins $\varepsilon_y$ at $0$ because $\varphi_L$ is
clamped, not because the space is trivial: at $\lvert\mathcal Y\rvert=3$ the three clamped fields
$\Lambda\chi_y$ have Gram $X^{\mathsf T}WX$ of rank $3$ (condition number $21.0$), so the $A$ of Eq. (A2) exists, with $\theta_L AWX=I_3$ to
$3.3\times10^{-15}$ (Code Cell A), and unclamping reopens a genuinely three-dimensional error channel.

§§4–7 are written with $m=1$, writing $b$ for the single column of $B$, purely for legibility;
§8.4 gives the general form, and the implementation uses it.

### <a id="tc3-3" name="tc3-3"></a>3. Fixed quantities and their algebraic dependencies

1. **Variances $\sigma$** \
   $\sigma_y=\sigma_L=\sigma_S=\sigma_u=1$. This is the minimum that Friston (2005)
   imposes on the estimated variance, as Bogacz reports it (p. 202; Friston writes
   $\Sigma^{1/2}=1+\lambda$, p. 822), to keep the error-unit dynamics from diverging; it is also the whole of our
   commitment on precision, since this model performs no precision inference. Where $\sigma$ appears below
   it is retained symbolically so that the equations remain readable if a later phase unfixes it.

   Note that fixing $\sigma$ removes precision-learning (Bogacz Eqs. 20–21, 48–49, 55), but the absence of
   precision learning does not remove Hebbian learning from this implementation. In particular, the outer Hebbian
   loop performs $\Theta$-learning (see §8, $\theta_u$ remains plastic on the slow timescale (Eq. 20)). Precision learning is what
   requires the interneuron construction (Bogacz §5), which this model therefore does not carry. For further notes see Appendix B.

2. **Emission $E_y$** \
   Each utterance $y$ carries an exclusion set $E_y\subseteq[0,1]$, the set of states
   its lexical entry rules out. On the scale $\{\textit{no},\textit{some},\textit{all}\}$,

   $$E_{\textit{no}}=(0,1],\qquad E_{\textit{some}}=\{0\},\qquad E_{\textit{all}}=[0,1).
\tag{5}$$

   $E$ stands for emission and its denotation is not related to the error quantity $\varepsilon$. Note the latter would not be
   true if $E_y$ defines the inclusion of a word. Note that, at this phase, $E_y$ is fixed by the utterance entry
   alone and makes no reference to which other utterances might exist in some alternatives space.

   $E_y$'s indicator on the grid is $\chi_y\in\{0,1\}^K$, with $\chi_{y,k}=1$ exactly where $\zeta_k$
   lies in the region that Eq. (A1) of Appendix A assigns to $E_y$ through the threshold $\theta_L$,
   since the endpoints at which Eq. (5) anchors *no* and *all* have no $\zeta$ coordinate. The
   lexical field is $\chi_{y}$ scaled by $\Lambda\in\mathbb R_{>0}$, which denotes lexical strength
   (see entry "Lexical strength"):

   $$\varphi_L=\Lambda\,\chi_y.
\tag{6}$$

   For further notes, see Appendix A on boundary resolution, and Appendix D on why the entry is fixed by exclusion
   rather than by the truth set.

3. **Utility basis $B$** \
   The $u$ level carries $\varphi_u\in\mathbb R^m$ and the $S$ level carries a
   field, so the $u$ level can address the $S$ level only through a fixed profile matrix
   $B\in\mathbb R^{K\times m}$, as in Eq. (10). Every result in §8 is stated for arbitrary fixed $B$ and
   arbitrary $m$, and no proof below in this cell constrains the choice. Mind that, even though this cell is
   indifferent to the column size of $B$, the model's expressive power is not (see Text cell 5, Part A).
   The implementation takes $m=2$. Note that $B$ is the spatial profile of a single projection, which is what makes Eq. (20) an instance of
   the local plasticity rule. For further notes see Appendix B.

4. **Base world prior $\ell_0$**\
   $\ell_{0,k}=\log p_0(\zeta_k)$ samples a base world prior $p_0$ over the world state at the grid nodes.
   $p_0$ may be supplied by a distribution elicited for the predicate, and need not be. Where none is
   supplied, the default is the normal distribution in $\zeta$ with mean $0$ and variance $1$,
   $p_0=\mathcal N(0,1)$, one of the base world priors Text cell 4 Part D and Text cell 4b run. The default is a stipulation, motivated by three properties: it is
   symmetric about the scale's midpoint, $s=1/2$, so it favours neither end; it is unimodal; and it is
   positive at every node, so $\ell_0$ is finite on the grid, which every $p_0$ must be. $\ell_0$ is not a
   state unit, and the construction's Gaussian assumption, which concerns the latent fields, neither
   requires nor motivates a Gaussian $p_0$.
   It enters through $g_L$ (Eq. 9) as the reference point of the lexical complement, and is then renormalized
   so that $\sum_k w_k e^{\ell_{0,k}}=1$, since an analytically normalized density is not grid-normalized once the grid is truncated.

5. **Utility prior mean $\mu_u$**\
   $\mu_u\in\mathbb R^m$ is the vector of weights on the terminating tonically
   active node, whose activity is $1$ by construction (Bogacz, Fig. 3). $\mu_u\neq0$ is a standing requirement of this implementation, a
   stipulation and not a requirement of the construction.
   The terminating contribution to $g_S$ is $\theta_u B\mu_u$ (see §4 and §8.5), which at $\mu_u=0$ vanishes for every $\theta_u$; the
   chain then stops rather than terminates, and the learning objective has no maximizer, cf. Bogacz's linear example, whose optimum
   $\theta^\ast=u/v_p$ does not exist at $v_p=0$ while his rule, Eq. 25, remains defined. However, different from Bogacz Eq. 25, $\mu_u\neq0$ is necessary but not by itself sufficient; the exact requirement
   is stated in Appendix B. The implementation takes $\mu_u=\mathbf 1\in\mathbb R^m$.

6. **Lexical strength $\Lambda$**\
   $\Lambda>0$ mediates between the lexicon and the world belief. $\Lambda\to\infty$ recovers a hard truth-conditional constraint;
   finite $\Lambda$ makes it soft and differentiable. The implementation takes $\Lambda=8$ if not said otherwise. See also Text cell 6 for a
   sweep on $\Lambda$.

---

### <a id="tc3-4" name="tc3-4"></a>4. Generative model

The chain runs $E_y \leftarrow \varphi_L \leftarrow \varphi_S \leftarrow \varphi_u \leftarrow 1$, each
level predicting the one below:

$$\hat\varphi_y=g_y(\varphi_L),\qquad
\hat\varphi_L=g_L(\varphi_S),\qquad
\hat\varphi_S=g_S(\varphi_u),\qquad
\hat\varphi_u=\mu_u\cdot 1 .
\tag{7}$$

1. **$g_y$: lexical content to word form.** \
   $g_y$ maps the lexical field onto the word-form units of the $y$ level. A word-recognition model,
   mapping an auditory input to word forms, is out of scope: it would be stacked beneath the $y$ level,
   and stacking it would unclamp $y$, not $\varphi_L$. In the present phase $\varphi_L$ is clamped by
   $y$ through the entry (Eqs. 5–6), so that

   $$\varepsilon_y\equiv 0
\tag{8}$$

   identically. Eq. (8) is therefore a *consequence* of the clamp on $\varphi_L$. Because $\varphi_L$ is clamped, the $y$ level takes no part in the variational dynamics at this
   phase. It is retained rather than deleted so that unclamping $\varphi_L$ later reopens the channel
   with no change to anything below $\varphi_L$; the definition of $g_y$, and what follows from it,
   are left to Appendix A.

2. **$g_L$: situation belief to lexical exclusion.** \
   $\varphi_S$ is an unnormalized log-density; a state is
   excluded to the degree its log-weight is low. Hence

   $$g_L(\varphi_S)=\ell_0-\varphi_S.
\tag{9}$$

   This function is a transported set-complement map. On indicator vectors, the complement is $x\mapsto 1-x$,
   which is the order-reversing involution of $[0,1]$; on log-weights, the corresponding order-reversing involution of
   $\mathbb R$ is $x\mapsto \ell_0-x$, reflection through the base world prior. To paraphrase: by placing the base world prior in $g_L$,
   we take $\ell_0$ as the reference instead of zero; that is, the complement of lexical exclusion remains when the base world prior is subtracted. As both are involutions and are affine, Eq. (9) therefore preserves the complement structure.

   Note $g_L'=-I$, that is, the lexical error ($\varepsilon_L$) reaches the situation belief ($\varphi_S$) *inhibitorily*.
   Appendix D derives this sign, and $\ell_0$'s place here, from the exclusion convention of §3.

3. **$g_S$: utility to situation belief.**

   $$g_S(\varphi_u)=\theta_u\varphi_u\,b.
\tag{10}$$

   This function is an instance of $\Theta h(\varphi)$ (cf. Bogacz Eq. 42) with $\Theta=\theta_u b$ and
   $h(\varphi)=\mathrm{id}(\varphi)$. Eq. (10) is written at $m=1$, as are §§8.1–8.3; the general form is
   $g_S(\varphi_u)=\theta_u B\varphi_u$, with $b$ the single column of $B$ at $m=1$, and §8.4 gives the
   lift. The scalar $\theta_u$ is the whole of the parameter here: $B$ is fixed (§3).

   Note that $g_S$ is **homogeneous**, that is, it depends on $\theta_u$ and $\varphi_u$ only through their product.
   Since $\varphi_u$ is anchored only by $\mu_u$ (Eq. 11), the map $(\theta_u,\mu_u)\mapsto(-\theta_u,-\mu_u)$ leaves
   every belief unchanged: the sign of $\theta_u$ carries no content on its own, and is fixed only relative to $\mu_u$.
   What $\mu_u$ does and does not move between utterances is covered in Text cell 5 Part B.

4. **Residuals.** \
   With every $\sigma=1$ the error units and the raw residuals coincide at equilibrium,
   so write

   $$\varepsilon_L=\sigma_L^{-1}(\varphi_L-g_L(\varphi_S))=1(\varphi_L-\ell_0+\varphi_S)=r_L,\qquad
   \varepsilon_S=\sigma_S^{-1}(\varphi_S-g_S(\varphi_u))=1(\varphi_S-\theta_u\varphi_u b)=r_S,\qquad
   \varepsilon_u=\sigma_u^{-1}(\varphi_u-\mu_u)=r_u.
\tag{11}$$

5. **Read-out.** \
   The construction fixes a posterior, and not a read-out of it. Under Bogacz's delta approximation
   (his §3, Eq. 34), the posterior is a point mass at the settled state
   $(\varphi_S^\ast,\varphi_u^\ast)$, and Part D reports $\varphi_S^\ast\in\mathbb R^K$ as it stands.
   For comparison with the Rational Speech Act literature (Frank & Goodman, 2012; Goodman & Frank,
   2016), whose foundational assumptions differ from those of this framework, we also read
   $\varphi_S$ as unnormalized log-weights over the scale, giving the normalized density

   $$q(\zeta_k)=\frac{e^{\varphi_{S,k}}}{\sum_j w_j e^{\varphi_{S,j}}},\qquad
   \sum_k w_k\,q(\zeta_k)=1.
\tag{12}$$

   The statistics reported from Eq. (12) in the evaluation, namely $\mathbb E[s]$, $\mathrm{sd}(s)$, the region
   $q$-masses and the leak, are summaries of the read-out $q$. They are not model quantities, and are defined in
   Text cell 4 at Eqs. (25)–(27). Nothing in the architecture dictates Eq. (12): it is kept for the
   comparison, and it takes no part in the dynamics or in Eq. (20).

---

### <a id="tc3-5" name="tc3-5"></a>5. Free energy

Following Bogacz Eq. (7) generalized to a hierarchy (his Eqs. 51–54), with the quadrature inner
product of Eq. (2) supplying the measure on the two field levels:

$$\boxed{\;
\mathcal F=-\frac{\|r_L\|^2}{2\sigma_L}-\frac{\|r_S\|^2}{2\sigma_S}-\frac{\|r_u\|^2}{2\sigma_u}
\;-\;\tfrac12\big(K\log\sigma_L+K\log\sigma_S+m\log\sigma_u\big)+C\;}
\tag{13}$$

Each level contributes one $\log\sigma$ per component, hence $K$ at the two field levels and $m$ at
the utility level; $\|r_u\|^2$ is the ordinary squared norm on $\mathbb R^m$, the two field norms
being those of Eq. (2).

The $\log\sigma$ terms vanish at $\sigma=1$ and are constant in every variational quantity; they are
carried only to mark where a later precision-bearing version would reintroduce them. $r_y\equiv0$
contributes nothing. $\mathcal F$ is to be **maximized**.

---

### <a id="tc3-6" name="tc3-6"></a>6. Targets of the variational quantities

The target of each quantity is its value at the unique stationary point of $\mathcal F$ (uniqueness is
established in §8).

**Error units.** Each error unit's target is its residual,

$$\varepsilon_L^\ast=\frac{r_L}{\sigma_L},\qquad
\varepsilon_S^\ast=\frac{r_S}{\sigma_S},\qquad
\varepsilon_u^\ast=\frac{r_u}{\sigma_u},
\tag{14}$$

which is Bogacz Eq. (57).

**World belief.** Setting $\partial\mathcal F/\partial\varphi_S=0$ gives
$\varepsilon_S=-\varepsilon_L$, i.e. $r_S/\sigma_S=-r_L/\sigma_L$, and the residual form
$r_S=-r_L$ holds only at $\sigma_L=\sigma_S$; hence

$$\varphi_S^\ast=\frac{\sigma_L\,\theta_u\varphi_u b+\sigma_S\big(\ell_0-\varphi_L\big)}
{\sigma_L+\sigma_S}.
\tag{15}$$

Two remarks follow, both stated at $\sigma=1$ for concreteness. First, the excluded region receives
$-\Lambda/2$, so its $q$-mass
vanishes as $\Lambda\to\infty$. Second, the factor $\tfrac12$ is a genuine temperature, which is defined
as the precision-weighted combination of two unit-variance channels, and, because the read-out (12) is exponential,
it flattens the reported belief $q$. That the factor equals $\tfrac12$ is a consequence of $\sigma_L=\sigma_S=1$
and is the natural place for a later phase to intervene.

**Utility state.** Setting $\partial\mathcal F/\partial\varphi_u=0$ gives
$\varepsilon_u=\theta_u\langle\varepsilon_S,b\rangle$; substituting Eq. (15),

$$\varphi_u^\ast=\frac{\mu_u+\dfrac{\sigma_u\theta_u}{S}\,\langle\ell_0-\varphi_L,\;b\rangle}
{1+\dfrac{\sigma_u\theta_u^{2}}{S}\,\|b\|^2},
\qquad S=\sigma_L+\sigma_S.
\tag{16}$$

**Two fixed-point identities**:

$$\varepsilon_S=-\varepsilon_L,\qquad\qquad \varepsilon_u=\theta_u\langle\varepsilon_S,b\rangle.
\tag{17}$$

Eqs. (15)–(17) are closed-form and are the correctness test for the implementation: the network's
numerical fixed point must reproduce them.

---

### <a id="tc3-7" name="tc3-7"></a>7. Dynamics

Error units converge toward their residuals (Bogacz Eqs. 53–54 at $\Sigma=1$):

$$\tau_\varepsilon\dot\varepsilon_L=\frac{r_L}{\sigma_L}-\varepsilon_L,\qquad
\tau_\varepsilon\dot\varepsilon_S=\frac{r_S}{\sigma_S}-\varepsilon_S,\qquad
\tau_\varepsilon\dot\varepsilon_u=\frac{r_u}{\sigma_u}-\varepsilon_u.
\tag{18}$$

State units ascend $\mathcal F$ and converge at its maximum. Differentiating Eq. (13) in the metric
§8 states, and using $g_L'=-I$ and $\partial g_S/\partial\varphi_u=\theta_u b$:

$$\tau_\varphi\dot\varphi_S=\nabla_{\varphi_S}\mathcal F=-\varepsilon_S-\varepsilon_L,
\qquad
\tau_\varphi\dot\varphi_u=\nabla_{\varphi_u}\mathcal F
=-\varepsilon_u+\theta_u\langle\varepsilon_S,b\rangle.
\tag{19}$$

The middle expressions equal the right-hand ones when each error unit sits at its residual,
$\varepsilon=r/\sigma$, the target of Eq. (14). During a transient the error units lag their
residuals and the state units follow the right-hand sides, which then differ from the gradient
(§8.2).

Both lines are instances of Bogacz Eq. (53), $\dot\varphi_i=-\varepsilon_i+h'(\varphi_i).\!*\,
\Theta_{i-1}^{\mathsf T}\varepsilon_{i-1}$: the first has $\Theta_{L}=-I$, the second $\Theta_S=b$
contracted under Eq. (2).

The slow parameter follows Bogacz's own gradient. His Eq. (25) gives
$\partial F/\partial\theta=\varepsilon_u\phi$, the error *below* times the state *above*, and his
Eq. (29) gives the same for a nonlinear $h$. Both are stated for his single-node example (§2.5), and both are
exactly our quantity:

$$\tau_\theta\,\dot\theta_u=\frac{\partial\mathcal F}{\partial\theta_u}
=\varphi_u\langle\varepsilon_S,b\rangle,
\qquad
\tau_\varepsilon\le\frac{\tau_\varphi}{4\lambda_{\max}(H)}\ll\tau_\varphi\ll\tau_\theta.
\tag{20}$$

*The learning rate is a ratio of time constants.* Bogacz supplies the gradient and says that
parameters move proportionally to it; here the constant of proportionality is a **time constant**, of
exactly the kind Eqs. (18) and (19) already carry for the error and state units, so the factor that
plays a learning rate's part is $\tau_\varphi/\tau_\theta$, the one §8.5's stopping rule reads. What
the model does not carry is a step size set apart from its ordering of timescales. So $\theta_u$ takes a
third, $\tau_\theta$, and the model carries one ordering of timescales. Its slow end, $\tau_\varphi\ll\tau_\theta$, is what §8.5 needs, exactly as its
fast end, the bound of commitment 7, is what §8.3 needs. $\tau_\theta$ sets the **rate only**:
the fixed point of Eq. (20) is the stationary point of $\tilde{\mathcal F}$ and does not depend on it.

**Where the flow starts, and what the network is there.** Eq. (20) is integrated from $\theta_u(0)=0$, the one start every configuration shares. At $\theta_u=0$ the utility level is disconnected from the situation level, and the network sits at the *tempered* version of the control that Text cell 4 reports against: $\varphi_S=\frac{\sigma_S}{\sigma_L+\sigma_S}(\ell_0-\varphi_L)$, which is $\tfrac12(\ell_0-\varphi_L)$ at the variances of commitment 3. It is not the literal listener $q_{\mathrm{lit}}$, which requires $\theta_u=0$ **and** $\sigma_S\to\infty$; the two are distinct objects and are never equated below. What *literal* denotes here is distinct from what it denotes in the Rational Speech Act literature (Goodman & Frank, 2016, p. 820) and in the Gricean literature (Grice, 1975, p. 44). The flow leaves zero immediately, because the gradient there is $\langle\mu_u,\sum_y c_y\rangle/\lvert\mathcal Y\rvert\neq0$, which is where §3's requirement $\mu_u\neq0$ is felt, and it rises monotonically toward $\theta_u^\ast$: the two roots of Eq. (B2) have opposite signs, so zero lies between the minimum and the maximizer, on the maximizer's rising side. Appendix B gives the argument, and says why a start at $\theta_u(0)=1$ would not do.

Note: Eqs. (19) and (20) are written at $m=1$, and §8.4 lifts the convergence proof to $m>1$. **However, at $m>1$, the
locality of Eq. (20) no longer holds.** At $m>1$ this update requires a quantity that no single
synapse holds, so the rule is Hebbian as written here and not in the case the implementation runs.
The implementation runs at $m=2$, so the rule as implemented departs from the
locality of plasticity that the construction imposes. This departure is stated, not repaired:
Appendix B states it in exact terms, says why a relay does not repair it, and names what would.

---

### <a id="tc3-8" name="tc3-8"></a>8. Convergence

Throughout §§8.1–8.3, $\theta_u$ is held fixed; the two-timescale separation that licenses this is
§8.5. Write $x=(\varphi_S,\varphi_u)\in\mathbb R^{K}\times\mathbb R$.

**The metric.** The field level carries the quadrature inner product of Eq. (2) and the utility
level the ordinary one, so $x$ lives in $\mathbb R^K\times\mathbb R^m$ with the metric
$G=\operatorname{diag}(W,I_m)$. Gradients and Hessians of $\mathcal F$ in this section are taken
in that metric: $\nabla_x\mathcal F=G^{-1}\partial\mathcal F/\partial x$, whose $\varphi_S$ block is
$-\varepsilon_S-\varepsilon_L$ where the ordinary partial derivative is
$-W(\varepsilon_S+\varepsilon_L)$, and $H=-G^{-1}\partial^2\mathcal F/\partial x^2$, whose
eigenvalues are real because $G^{1/2}HG^{-1/2}$ is symmetric. Eq. (19) is ascent in this metric,
and in the coordinates $\sqrt{w_k}\,\varphi_{S,k}$, where the metric is the ordinary one, it is
Bogacz's Eq. (53) as written. Eqs. (21) and (22) below are the quadratic form of the coordinate
Hessian $\partial^2\mathcal F/\partial x^2$, with the field norms of Eq. (2). That form equals the
$G$-pairing $\langle y,\nabla^2\mathcal F\,y\rangle_G$ of the metric Hessian, and not its ordinary
pairing $y^{\mathsf T}\nabla^2\mathcal F\,y$, since $\nabla^2\mathcal F=G^{-1}\partial^2\mathcal
F/\partial x^2$ is self-adjoint in $G$ and not symmetric. So $H$ is positive definite **in $G$**:
its eigenvalues are positive, while the ordinary symmetric part of $H$ need not be.
$\lambda_{\max}(H)$ is likewise a quantity in $G$, and every value this notebook quotes is the one
in $G$.

#### <a id="tc3-8-1" name="tc3-8-1"></a>8.1 $\mathcal F$ is strictly concave in $x$

Each of $r_L,r_S,r_u$ is **affine** in $x$. This is where the linearity of $g_L$ (Eq. 9) and of $g_S$
in $\varphi_u$ (Eq. 10) is used, and it is the only place either is needed. Hence $\mathcal F$ is a
negative sum of squares of affine functions, so concave. For strictness, take a direction
$(v,t)\in\mathbb R^K\times\mathbb R$, $(v,t)\neq 0$, and evaluate the quadratic form of the Hessian by
reading off the directional derivative of each residual: $\partial_{(v,t)}r_L=v$, $\partial_{(v,t)}r_S
=v-t\theta_u b$, $\partial_{(v,t)}r_u=t$. Therefore

$$(v,t)^{\mathsf T}\frac{\partial^2\mathcal F}{\partial x^2}(v,t)
=-\frac{\|v\|^2}{\sigma_L}-\frac{\|v-t\theta_u b\|^2}{\sigma_S}-\frac{t^2}{\sigma_u}.
\tag{21}$$

Every term is $\le 0$; the first vanishes only if $v=0$ and the third only if $t=0$. So the form is
strictly negative for $(v,t)\neq0$: $\partial^2\mathcal F/\partial x^2$ is negative definite, and
so, in $G$, is $\nabla^2\mathcal F=-H$.

Three consequences are worth recording. **(i)** $\mathcal F$ has a unique stationary point, which is its
global maximum; Eqs. (15)–(16) are therefore *the* solution, not *a* solution. **(ii)** Eq. (21) holds
for every $\theta_u\in\mathbb R$ and every $\Lambda>0$, so neither the lexical strength nor the
generative weight can destabilize the fast subsystem. **(iii)** $\varphi_L$ does not appear in Eq. (21)
at all: it enters $r_L$ additively and so shifts the maximizer without touching the Hessian. **The
result is independent of the shape of the lexical field.** §9.2 and §9.3 rest on this.

#### <a id="tc3-8-2" name="tc3-8-2"></a>8.2 The fast subsystem converges

§8.1 gives a unique optimum. Three further results concern the dynamics, and they rest on
different assumptions, so they are stated apart.

**(i) Ascent with instantaneous errors.** If every error unit sits at its residual at every moment,
the limit $\tau_\varepsilon\to0$ of Eq. (18), then Eq. (19) is gradient ascent in the metric $G$,
and along it $\dot{\mathcal F}=\langle\nabla_x\mathcal F,\dot x\rangle_G=
\tau_\varphi^{-1}\|\nabla_x\mathcal F\|_G^2\ge0$, with equality only at the stationary point.
$\mathcal F$ is bounded above (it is a negative sum of squares) and strictly concave, so this flow
converges globally to the unique maximizer of §8.1, from any initialization, with $\mathcal F$
rising throughout.

*Standard result invoked without proof, following the omission convention of Bogacz's paper:*
convergence of gradient flow on a strictly concave, bounded-above, coercive objective. Coercivity is
immediate: $\mathcal F\to-\infty$ as $\|x\|\to\infty$, since $\|r_L\|^2+\|r_u\|^2$ already dominates
$\|x\|^2$.

The identity for $\dot{\mathcal F}$ holds in that limit only. In the dynamics as implemented the
error units have states of their own, they differ from their residuals during the transient, and
$\dot x$ is then not the gradient.

**(ii) Convergence of the coupled network.** Eqs. (18)–(19) together are linear, and §8.3 reduces
their stability to one quadratic per eigenvalue of $H$, all of whose coefficients are positive.
Every mode therefore decays at every positive $\tau_\varepsilon$ and $\tau_\varphi$: the network
with its error units converges to the maximizer of §8.1 from any initialization, and convergence
needs no separation of timescales.

**(iii) Monotone $\mathcal F$ under the implemented start.** A monotone rise needs more than
convergence. Commitment 7 (§8.3) makes every mode non-oscillatory, which is necessary and not
sufficient: an overdamped mode started at the maximizer with an error unit away from its residual
first moves off the maximizer, and $\mathcal F$ falls meanwhile. The implementation starts every
error unit silent, the belief at the base world prior $\ell_0$ and the utility state at $\mu_u$ (`infer`,
Code cell 1). Under that start, with
commitment 7 and $\mathrm dt=\tau_\varepsilon/2$, $\mathcal F$ does not decrease at any step of the
three inferences Part A integrates (Text cell 4). That is a measured property of this start, not a
theorem for every initial state.

#### <a id="tc3-8-3" name="tc3-8-3"></a>8.3 The error units do not disturb this

Eqs. (18)–(19) are jointly linear, so their combined fixed point is the one computed above and its
stability is a matter of the eigenvalues of a constant matrix. Bogacz gives an analysis of
the same kind for a different subsystem, an error node and its interneuron with $\varphi$ held
constant (his §5.1, Eq. 66), and states that $\varepsilon$ converges to the residual when
$\varphi$ evolves on a slower timescale. We adopt his method, not his result, and derive the
requirement for the coupled system here, in the exact form the implementation enforces.

Eliminating the error units from Eqs. (18)–(19) leaves, for each eigenvalue $\lambda$ of
$H$ (§8.1, in the metric of §8), the scalar characteristic equation

$$\tau_\varepsilon\tau_\varphi\,s^{2}+\tau_\varphi\,s+\lambda=0 ,$$

whose roots are real, so that the mode relaxes without oscillating, exactly when
$\tau_\varepsilon\le\tau_\varphi/(4\lambda)$. Imposing it on every mode gives commitment 7,

$$\tau_\varepsilon\le\frac{\tau_\varphi}{4\,\lambda_{\max}(H)} ,
\qquad\text{with the Euler step then taken at }\mathrm{d}t=\tau_\varepsilon/2 .$$

(The eliminated system's remaining root is $s=-1/\tau_\varepsilon$, of multiplicity $K$, real and
fast for any positive $\tau_\varepsilon$.) This is stronger than the loose
$\tau_\varepsilon\ll\tau_\varphi$ it replaces, and it is not a fixed ratio: $\lambda_{\max}(H)$
grows as $\theta_u^{2}$, so a separation that suffices at one $\theta_u$ fails at a larger one.
Eq. (28) of Text cell 4 is the $\varphi_u$ diagonal entry of the same $H$ and sits one below
$\lambda_{\max}(H)$ at the Gaussian $\theta_u^\ast$, $809.69$ against $810.69$ (Code Cell 2).

**What the commitment secures, and what it asserts.** The quadratic's three coefficients are
positive, since $\lambda>0$ by §8.1, so both roots have negative real part at every positive
$\tau_\varepsilon$; that is §8.2(ii). What the commitment adds is real roots, the precondition of
§8.2(iii)'s *monotone* rise of $\mathcal F$, and not convergence: at $\tau_\varepsilon=0.1$ the
system still reaches the same fixed point, with $\mathcal F$ dipping during the transient. What it asserts is a separation of
$4\lambda_{\max}(H)$ between the error and the state units, which at a learned $\theta_u^\ast$ is
of order $10^{3}$ under the Gaussian base world prior and $10^{8}$ under the flat one. No circuit has
separations of that order, so the honest statement is a conditional: either the model asserts them
at large $\theta_u^\ast$, or §8.2(iii) weakens there to convergence with a non-monotone
transient. Text cell 4 reports what the commitment costs in integration, and how little of
$\theta_u^\ast$ the evaluation's criteria turn out to need.

#### <a id="tc3-8-4" name="tc3-8-4"></a>8.4 The proof does not depend on dimension or basis

§§8.1–8.3 are written at $m=1$; the implementation takes $m=2$, so the lift is stated here. Let
$\varphi_u\in\mathbb R^m$ and $B\in\mathbb R^{K\times m}$, with $\theta_u$ the scalar of Eq. (10), so
$g_S=\theta_u B\varphi_u$. For a direction $(v,t)\in\mathbb R^K\times\mathbb R^m$ the residual
derivatives are $\partial_{(v,t)}r_L=v$, $\partial_{(v,t)}r_S=v-\theta_u Bt$ and
$\partial_{(v,t)}r_u=t$, which is Eq. (21) with $t\theta_u b$ replaced by $\theta_u Bt$, so

$$(v,t)^{\mathsf T}\frac{\partial^2\mathcal F}{\partial x^2}(v,t)
=-\frac{\|v\|^2}{\sigma_L}-\frac{\|v-\theta_u Bt\|^2}{\sigma_S}-\frac{\|t\|^2}{\sigma_u},
\tag{22}$$

the field norms being those of Eq. (2) and $\|t\|^2$ the ordinary norm on $\mathbb R^m$.

**Why $B$ cannot matter.** $B$ appears only in the middle term, and that term does no work: it is
$\le0$ throughout, while strictness is carried by the outer two: the third excludes $t\neq0$, and
the first and second together exclude $t=0,\,v\neq0$. Eq. (22) is therefore negative definite for
**every** $B\in\mathbb R^{K\times m}$ (rank-deficient, non-orthonormal, $m>K$, any profile shape in
$\zeta$, and $B=0$ included) and for **every** $\theta_u\in\mathbb R$, zero and negative alike. Nothing
in §§8.1–8.2 used $m=1$, the shape of $b$ as a function of $\zeta$, or the dimensionality of $\zeta$.
(The linearity that *is* used is linearity of $g_S$ in $\varphi_u$, which makes $r_S$ affine in $x$;
that is a different property, named in §8.1.)

**The matrix case.** Eq. (22) is negative definite with any $M\in\mathbb R^{K\times m}$ in place of $\theta_u B$, so
concavity would also survive promoting $\theta_u$ to a matrix. Mind that is a statement about the reach of the
proof, not a licence to adopt it: with $\theta_u$ a matrix the $u\to S$ weights become $K\times m$
independently modifiable synapses, which is a different circuit from the one specified. For further notes
see Appendix B. This implementation still takes $\theta_u\in\mathbb R$.

#### <a id="tc3-8-5" name="tc3-8-5"></a>8.5 The slow parameter

$\mathcal F$ is concave in $\theta_u$ for fixed $x$, but **not jointly concave** in
$(\theta_u,\varphi_u)$: the term $\theta_u\varphi_u b$ is bilinear. Under
$\tau_\varphi\ll\tau_\theta$ the fast subsystem equilibrates between $\theta_u$ updates, so Eq. (20)
is ascent on $\tilde{\mathcal F}(\theta_u)=\max_x\mathcal F(x,\theta_u)$, whose gradient coincides
with $\partial\mathcal F/\partial\theta_u$ at the equilibrium $x^\ast(\theta_u)$ (Danskin, 1967, p. 24;
the theorem is invoked without proof. It requires a unique maximizer, which §8.1 gives, and a compact
set for $x$, which the coercivity of §8.2 supplies, its bound being free of $\theta_u$). $\tilde{\mathcal F}$
is bounded above by $0$, being a maximum of negative sums of squares, so the ascent converges.

Nevertheless, that $\tilde{\mathcal F}$ converges does not by itself supply a maximizer to converge to.
Appendix B treats the premise this rests on, the sign and location of $\theta_u^\ast$, the locality of
Eq. (20), and what a learned $\theta_u$ implies about the alternative space of an observed utterance.

**How the flow stops.** It starts at $\theta_u(0)=0$, the tempered control described under
Eq. (20), and rises monotonically from there. It stops when its own update falls below a
**tolerance**,
$|\Delta\theta_u|=(\tau_\varphi/\tau_\theta)\,|\partial\tilde{\mathcal F}/\partial\theta_u|<\mathrm{tol}$.
That is the only stopping rule, and it reads the size of the step the unit has just taken and
nothing else: no closed form, and nothing about the shape of the trajectory, to which the unit has
no access. The fast subsystem already stops the same way at its own timescale, where the tolerance
is set above the roundoff floor of the arithmetic that forms the derivative and therefore scales
with $\lambda_{\max}(H)$.

**The value is ad hoc, and what is reported does not use it.** Nothing here fixes what tolerance a
system has, and this cell takes no position on whether a living one needs a tolerance for the
reason a floating-point computation does. So what the notebook reports is $\theta_u^\ast$, the
stationary point of $\tilde{\mathcal F}$, from the closed form of Eq. (B2) and cross-checked
against an independent bisection on the gradient: it is the **asymptote** of this flow and carries
no tolerance at all. A $\theta_u$ taken *from* the flow is labelled **realizable**, and carries on
its face the tolerance at which it halted.

**Two consequences follow from the rule and are stated as such.** $\tilde{\mathcal F}$ is shallow
near its maximum, so any tolerance coarse enough to be plausible halts well short of
$\theta_u^\ast$ while the gradient is still positive: the integration shows the direction of the
flow, the monotone rise of $\tilde{\mathcal F}$ and a shrinking gradient, not arrival. And where
the flow *starts* slowly, the same rule halts it at once and the belief stays at the tempered
control. No claim below rests on the flow arriving; Text cell 4 shows that the mass criteria it
reports are met long before $\theta_u^\ast$ would be, which is a fact about the shape of the
update rather than about where the flow stops.

#### <a id="tc3-8-6" name="tc3-8-6"></a>8.6 A note on what the convergence in this section does and does not show

With $\varphi_L$ clamped and both $g_L,g_S$ linear, the model is linear-Gaussian at this phase and
Eqs. (15)–(16) solve it in closed form. The recurrent dynamics are therefore a claim about *neural
implementation*, the same status they have in Bogacz's own linear examples, and the closed forms make
§6 an exact test of that implementation.

That exactness is what lets the evaluation read a configuration it cannot afford to integrate. Under
commitment 7 an inference costs about $180\lambda_{\max}(H)$ Euler steps, and $\lambda_{\max}(H)$
grows as $\theta_u^2$, so at a large learned $\theta_u^\ast$ the closed form is the only route and
Text cell 4 says at every table which route it took. What the closed form does not show is that the
dynamics *reach* those fixed points at such $\theta_u$, and that is a real gap rather than a formal
one; it is narrowed by the fact that the mass criteria Text cell 4 reports are met at a
$\lvert\theta_u\rvert$ small enough to integrate, where the two agree to about $10^{-9}$, the
accuracy the stopping rule of §8.3 settles for.
Nonlinearity would be present if $h$ is made nonlinear in $\varphi_u$ (Eq. 10, currently the identity).
However, the shape of $b$ as a function of $\zeta$ is not such a route: $b$ is a fixed vector on the grid,
so $g_S$ stays linear in $\varphi_u$ whatever profile it carries (§8.4).

To conclude, the proofs in this section depend on three assumptions: that $g_L$ and $g_S$ are affine
in $x$ (§8.1); that $\varphi_L$ is clamped, so that $\varphi_L$ is not a variable (§8.6); and that
$\sigma_u$ is finite with at least one of $\sigma_L,\sigma_S$ finite as well. Finite $\sigma_u$ carries
strictness in $t$, and the finite member of $\sigma_L,\sigma_S$ carries it in $v$. Should a later phase
unfix $\sigma$ (§3), that last assumption is the one Eq. (22) needs.

### <a id="tc3-9" name="tc3-9"></a>9. Foreseen extensions

#### <a id="tc3-9-1" name="tc3-9-1"></a>9.1 The flat direction: resolved, with a standing requirement

**Claim: compatible; the concern does not arise.** An unnormalized log-density is ordinarily defined
only up to an additive constant, so one expects $\mathcal F$ to be flat along $\mathbf 1$ and
$\varphi_S$ to drift without converging. Here it does not: §8.1 shows $\nabla^2\mathcal F$ is negative
definite, so there is no flat direction in any direction, $\mathbf 1$ included. The reason is structural: $\varphi_S$ is a *state* with its own Gaussian errors at both the $L$ and $S$ levels, and
those pin it in absolute terms. It is not merely a parameterization of the distribution (12).

**Caveat.** Because the absolute scale of $\varphi_S$ is pinned,
that scale does work: it is the temperature noted under Eq. (15), and the read-out (12) is sensitive to
it. The consequence is that the factor $\tfrac12$ in Eq. (15) is a real flattening of the reported
belief, so any future change to $\sigma_L$ or $\sigma_S$ changes the reported distribution's sharpness
even when it leaves the support untouched.

#### <a id="tc3-9-2" name="tc3-9-2"></a>9.2 One-dimensional dense-scale predicates

**Claim: compatible.** By §8.1(iii) the
convergence result is independent of the shape of $\varphi_L$: the lexical field enters only additively,
and the Hessian does not see it. Moving from a sharp quantifier to a dense-scale predicate replaces
$\chi_y\in\{0,1\}^K$ with a smooth profile $\chi_y\in[0,1]^K$ and changes nothing else: not the level structure, not the messages, not $\theta_u$,
and not any proof in §8. Eqs. (15)–(17) hold verbatim.

The state space is already continuous (Eq. 1), so no discretization is being relaxed; $K$ is a
quadrature parameter throughout.

**Caveats.**
1. *Quadrature accuracy runs opposite to intuition.* A smooth $\chi_y$ is **easier** to integrate on a
   fixed grid than a step. The sharp quantifier boundaries of the present phase are the numerically
   demanding case, incurring $O(1/K)$ error near the discontinuity; dense-scale predicates relax this.
   The grid is thus the representation that serves both phases.

2. *$\Lambda$ is bounded in practice.* Large $\Lambda$ makes $\varphi_S$ badly scaled, and
   $\Lambda\to\infty$ is a limit, not a setting. The read-out (12) should be computed in log-space.

3. *A parametric read-out is closed off.* If a later phase wants $q$ to be a two-parameter family
   (logit-normal, say) rather than a grid-borne field, sharp exclusions
   become unrepresentable: an exponential family with fixed support cannot move its support. This is a restriction on *parametric* $q$, not on the
   present formulation, and it is the reason the grid is retained.

4. *Unclamping $\varphi_L$ is a separate step.* A predicate whose lexical content is itself inferred
   requires $\varphi_L$ to become variational, which reopens $\varepsilon_y$ and therefore requires
   $g_y$ (Appendix A). The architecture accommodates this without modification; the
   *content* of $g_y$ is not supplied here.

#### <a id="tc3-9-3" name="tc3-9-3"></a>9.3 Multidimensional predicates

**Claim: theoretically compatible.** §8.4 establishes negative definiteness for arbitrary $M$ and
arbitrary $m$, and nothing in §§4–7 assumed $\zeta$ to be scalar. A predicate over $d$ jointly inferred
degree scales requires only that $\zeta\in\mathbb R^d$, that $b$ become a matrix, and that $\varphi_u$
become a vector. Every equation and every proof above carries over unchanged.

**Caveats, and this is where a real ceiling sits.**
1. *The grid is exponential in $d$.* A tensor-product quadrature costs $K^d$ nodes. At $d=1$, $K\approx101$
   against two variational states is negligible; at $d=3$ it is already $10^6$ nodes per iteration. The
   limit is **dimensional, not dense**: refining a one-dimensional scale is cheap, adding a scale is not.

2. *The escape routes conflict with §9.2's caveat 3.* Factored, parametric, or sampled representations
   are the standard remedies for grid blow-up, but a parametric $q$ cannot represent sharp exclusions.
   **Multidimensional semantics with sharp lexical boundaries is therefore the one corner this
   formulation does not reach**, and it is declared as a non-compatibility rather than a caveat. A
   factored grid over dimensions that are semantically independent is the most promising route, but it
   assumes an independence that scalar predicates may not license, and no claim is made here.

3. *Identifiability is untested.* With $m>1$ utility parameters addressing a $d$-dimensional field,
   $\theta_u$ acquires rotational degeneracies that §8.5's local convergence claim does not address.
   Nothing above rules this out; nothing above establishes it either.

<a id="code1" name="code1"></a>

````python
# === code cell 1: three-level predictive-coding network over a continuous
#     world state. ARCHITECTURE ===
#
# Implements the specification in the preceding markdown cell. Equation numbers below
# refer to that cell throughout.
#
# SIGN CONVENTION: F is Bogacz's negative free energy and is MAXIMIZED; every state
# equation is gradient ASCENT (spec preamble item 4, Eq. 13).

import math
import torch

DTYPE = torch.float64

UTTERANCES = ["no", "some", "all"]

# `infer`'s stopping tolerance, PER UNIT of lambda_max(H) (decision I3, revised).
#
# The stopping test is max|derivative| < tolerance. A derivative cannot be resolved below
# the roundoff floor of the arithmetic that forms it, and that floor is NOT a constant:
# measured, it is 4.547e-13 * lambda_max(H), flat to four significant figures from
# theta_u = 13.4 to 61.3 (agent/audits/2026-09-21-tolerance-halting/). The mechanism is
# arithmetic rather than fitted: a derivative is a rate times a state error, the state
# error sits at machine precision relative to the state, and the stiffest rate is
# lambda_max(H), so the floor is about 2.4 * eps * |phi_S| * lambda_max(H).
#
# I3 originally set a FIXED 1e-9, chosen to sit above the floor measured at the theta_u of
# the day (F34). The PRINCIPLE was right and the CONSTANT does not travel: 1e-9 is 12.2
# times the floor at lambda = 181, 1.8 times at lambda = 1206, where the stopping step
# already drifts 31 steps under a change in theta_u of one part in 1e15, which is exactly
# the non-reproducibility F34 set the tolerance to avoid, and BELOW the floor above
# lambda = 2177, where the test can never fire at all and the run ends at its step cap.
#
# So the tolerance is keyed to the quantity the floor tracks. The value below is 12.1
# floors, which is 1e-9's own margin at the lambda I3 was calibrated at, made scale-free:
# the old constant becomes a consequence rather than a stipulation. lambda_max(H) is not a
# quantity smuggled in from outside the system: `fast_time_constant` already computes it
# to set tau_error, and the architecture's timescale commitment is stated in terms of it
# (tau_error <= tau_state / (4 lambda_max(H))).
#
# What it buys, measured: convergence at every theta_u tested, including two that ran to
# the cap under 1e-9; 181-224 Euler steps per unit lambda against 216-220 before; and a
# stopping step that does not move at all under a 1e-15 perturbation of theta_u.
DERIVATIVE_TOLERANCE_PER_RATE = 5.5e-12


# 1. NUMERICAL SUBSTRATE (spec Sec. 1)

def build_grid(num_nodes=101, half_width=6.0, dtype=DTYPE, device=None):
    """Fixed quadrature grid over zeta, with trapezoidal weights (Eqs. 1-2).

    The grid discretizes the integral, not the state: `num_nodes` is an accuracy
    parameter, and no quantity in the model is defined in terms of a state indexed
    by a node.
    """
    if num_nodes < 3:
        raise ValueError("num_nodes must be at least 3")
    if half_width <= 0:
        raise ValueError("half_width must be positive")

    zeta = torch.linspace(-half_width, half_width, num_nodes, dtype=dtype, device=device)
    spacing = (zeta[1] - zeta[0]).item()

    weights = torch.full_like(zeta, spacing)
    weights[0] *= 0.5
    weights[-1] *= 0.5
    return zeta, weights


def logistic(x):
    """s = sigma(zeta), recovering the proportion from its logit (Eq. 1)."""
    return torch.sigmoid(x)


# 2. LEXICON (spec Sec. 3, Eqs. 5-6)

def exclusion_indicator(utterance, zeta, theta_L, sharpness=None):
    """chi_y on the grid: 1 where the entry for `utterance` excludes the state (Eq. A1).

        E_no    ->  excluded where zeta >  -theta_L   (outside the cell of s=0)
        E_some  ->  excluded where zeta <= -theta_L   (inside  the cell of s=0)
        E_all   ->  excluded where zeta <   theta_L   (outside the cell of s=1)

    `theta_L` is semantic, not a parameter of the discretization: Appendix A gives its
    denotation (Eq. A5), the Voronoi reading of Eq. (A1), and the grid cap. The
    step is also the demanding case of Sec. 9.2 caveat 1.

    `sharpness=None` gives the step of the current phase (chi in {0, 1}); a positive
    float gives a logistic profile (chi in [0, 1]) for dense-scale predicates. Spec
    Sec. 8.1(iii) guarantees every convergence result is independent of this choice.
    """
    if utterance not in UTTERANCES:
        raise ValueError(f"utterance must be one of {UTTERANCES}")
    if theta_L <= 0:
        raise ValueError("theta_L must be positive")
    if sharpness is not None and sharpness <= 0:
        raise ValueError("sharpness must be positive when given")

    # Signed distance into the excluded region, positive where excluded. "some" has
    # no margin of its own: E_some is the complement of E_no (Eq. 5), so chi_some is
    # built as 1 - chi_no. That holds chi_no + chi_some = 1 at every node, the
    # boundary zeta = -theta_L included, as Eq. (A1) states it and Eq. (C1) requires;
    # a margin of its own would need a non-strict test the shared step cannot express.
    complement = utterance == "some"
    if utterance == "all":
        margin = theta_L - zeta
    else:  # "no", and "some" through its complement
        margin = zeta + theta_L

    if sharpness is None:
        chi = (margin > 0).to(zeta.dtype)
    else:
        chi = torch.sigmoid(margin / sharpness)
    return 1.0 - chi if complement else chi


# 3. BASE WORLD PRIORS (spec Sec. 3, the fixed quantity ell_0)

# ell_0 is a log-density in zeta. Because the state is carried in logit coordinates,
# a base world prior specified over the proportion s must be PUSHED FORWARD through the change of
# variables, picking up the Jacobian ds/dzeta = s(1-s). The distinction is not cosmetic:
# a distribution that is flat in s is NOT flat in zeta.

def gaussian_world_prior(mean=0.0, precision=1.0):
    """Gaussian in zeta: the default base world prior ell_0 of Sec. 3, used when no elicited
    distribution is supplied. Sec. 3 gives the reason for a Gaussian default."""
    if precision <= 0:
        raise ValueError("precision must be positive")

    def log_density(zeta):
        return (
            0.5 * math.log(precision / (2.0 * math.pi))
            - 0.5 * precision * (zeta - mean).square()
        )

    log_density.description = f"Gaussian in zeta (mean {mean}, precision {precision})"
    return log_density


def beta_world_prior(alpha, beta):
    """Beta(alpha, beta) in s, pushed forward to zeta.

    With s = sigma(zeta) and ds/dzeta = s(1 - s), the Jacobian absorbs one factor from
    each side of the Beta density and the log-density collapses to

        log p(zeta) = alpha * log s + beta * log (1 - s) - log B(alpha, beta),

    evaluated through softplus for numerical stability. Note the exponents are alpha
    and beta rather than the usual alpha - 1 and beta - 1: that shift IS the Jacobian.

    Beta(1, 1) is the uniform base world prior over the proportion; alpha != beta is skewed.
    """
    if alpha <= 0 or beta <= 0:
        raise ValueError("alpha and beta must be positive")

    log_normalizer = (
        math.lgamma(alpha) + math.lgamma(beta) - math.lgamma(alpha + beta)
    )

    def log_density(zeta):
        log_s = -torch.nn.functional.softplus(-zeta)          # log sigma(zeta)
        log_one_minus_s = -torch.nn.functional.softplus(zeta)  # log (1 - sigma(zeta))
        return alpha * log_s + beta * log_one_minus_s - log_normalizer

    log_density.description = f"Beta({alpha}, {beta}) in s"
    return log_density


def uniform_world_prior():
    """Literal Uniform(0, 1) over the proportion.

    Implemented as its own case, NOT as the low-precision limit of the Gaussian:
    sending precision -> 0 inflates the variance in
    zeta and the sigmoid squashes the result into bimodal mass of the base world prior at the endpoints, which
    is the opposite of uniform. Uniform in s is Beta(1, 1), whose push-forward to zeta
    is the logistic density, unimodal about zeta = 0 and not flat there.
    """
    density = beta_world_prior(1.0, 1.0)
    density.description = "Uniform(0, 1) in s"
    return density


BASE_WORLD_PRIORS = {
    "gaussian": gaussian_world_prior(0.0, 1.0),
    "flat": uniform_world_prior(),
    "skewed low": beta_world_prior(1.0, 3.0),
    "skewed high": beta_world_prior(3.0, 1.0),
}


# 4. THE NETWORK

class LexicalPredictiveCodingNetwork:
    """y <- phi_L <- phi_S <- phi_u <- 1, with explicit error units.

    Clamped:      y, phi_L                       (Eqs. 6, 8)
    Variational:  phi_S, phi_u, and the error units    (fast, Eqs. 18-19)
                  theta_u                              (slow, Eq. 20)
    Fixed:        every sigma, Lambda, b, ell_0, mu_u, and the grid (Sec. 3)

    theta_u is LEARNED by default: `theta_u="learned"` sets it to theta_u*, the
    maximizer of F~ (Appendix B, Eq. B2), which is specific to every input and
    hyperparameter the network is built with, so `respawn` re-learns it on every
    override. A number instead of "learned" holds theta_u fixed, which the evaluation
    uses only as an explicitly stated control. The flow of Eq. (20) itself starts at
    `theta_u_initial` = 0, where the network is the TEMPERED control (theta_u = 0 at
    the model's sigmas), not the literal listener, which also needs sigma_S -> infinity.
    """

    def __init__(
        self,
        *,
        num_nodes=101,
        grid_half_width=6.0,
        base_prior=None,
        prior_mean=0.0,
        prior_precision=1.0,
        theta_L=None,
        num_atoms=10,
        lexical_strength=8.0,
        mu_u=1.0,
        theta_u="learned",
        theta_u_initial=0.0,
        basis_degree=2,
        sigma_lexical=1.0,
        sigma_state=1.0,
        sigma_utility=1.0,
        exclusion_sharpness=None,
        device=None,
    ):
        if prior_precision <= 0:
            raise ValueError("prior_precision must be positive")
        if lexical_strength <= 0:
            raise ValueError("lexical_strength (Lambda) must be positive")
        if theta_L is None and (int(num_atoms) != num_atoms or num_atoms < 2):
            raise ValueError(
                "num_atoms (n) must be an integer >= 2; below 2 the entries for "
                "\"some\" and \"all\" coincide"
            )
        if min(sigma_lexical, sigma_state, sigma_utility) <= 0:
            raise ValueError("variances must be positive")
        if min(sigma_lexical, sigma_state, sigma_utility) < 1.0:
            raise ValueError(
                "variances below 1 are outside the floor adopted in Sec. 3 "
                "(Friston 2005; Bogacz footnote 3)"
            )

        self.dtype = DTYPE
        self.device = device

        self.zeta, self.weights = build_grid(
            num_nodes=num_nodes,
            half_width=grid_half_width,
            dtype=self.dtype,
            device=device,
        )
        self.num_nodes = num_nodes

        # Fixed quantities (Sec. 3). Every sigma is carried symbolically although the
        # present phase fixes all three at 1, so that a later phase can unfix them
        # without the equations below changing shape.
        self.sigma_lexical = float(sigma_lexical)
        self.sigma_state = float(sigma_state)
        self.sigma_utility = float(sigma_utility)

        self.lexical_strength = float(lexical_strength)

        # LEXICAL THRESHOLD theta_L.
        #
        # logit is asymptotic at s = 0 and s = 1, so it is a bijection only on the OPEN
        # interval and the endpoints where Eq. (5) anchors "no" and "all" have no zeta
        # coordinate. Realizing E_y on the grid therefore requires a threshold
        # theta_L > 0, with E_y mapping to {zeta > -theta_L}, {zeta <= -theta_L} and
        # {zeta < theta_L} for "no", "some" and "all".
        #
        # theta_L is a generative weight of g_y, the lexical-content-to-word-form map:
        # recovering y from phi_L requires the threshold that produced it. Its gradient
        # is dF/dtheta_L = eps_y * dg_y/dtheta_L, which vanishes identically while
        # eps_y == 0 (Eq. 8), so theta_L is fixed BECAUSE phi_L is clamped rather than
        # by stipulation; unclamping phi_L makes it learnable.
        #
        # n is theta_L's denotation: sigma(-theta_L) = 1/(2n) (Eq. A5). A count where the
        # predicate has atoms; where it has none, n is the number of distinguishable steps
        # the scale affords and need not be an integer. One name, one formula: the two
        # cases differ in where n comes from, not in what Eq. (A5) says (Appendix A, O1).
        if theta_L is None:
            self.theta_L = math.log(2 * int(num_atoms) - 1)
        else:
            self.theta_L = float(theta_L)
        if self.theta_L <= 0:
            raise ValueError("theta_L must be positive")
        # derived from theta_L (Eq. A5), not an independent quantity
        self.num_atoms = (math.exp(self.theta_L) + 1.0) / 2.0
        if self.theta_L >= grid_half_width:
            raise ValueError(
                f"theta_L = {self.theta_L:.4f} exceeds the grid half-width "
                f"{grid_half_width}: E_no leaves no grid node unexcluded. Widen the "
                f"grid, or lower n (n = {self.num_atoms:.4g} requires half_width > "
                f"{self.theta_L:.4f})."
            )
        self.exclusion_sharpness = exclusion_sharpness
        # theta_u is resolved at the end of the constructor, once ell_0, B and mu_u exist:
        # "learned" needs all three (Eq. B2).
        if not (theta_u == "learned" or isinstance(theta_u, (int, float))):
            raise ValueError('theta_u must be "learned" or a number (a fixed control)')
        self.theta_u_initial = float(theta_u_initial)

        # UTILITY BASIS B (K x m). The default m = 2 takes zeta and zeta^2, orthonormalized under the
        # quadrature inner product against the constant function and each other. The
        # linear component separates "no" from {"some", "all"} and the quadratic
        # separates "some" from {"no", "all"}, so all three utterances receive
        # distinct projections.
        if basis_degree < 1:
            raise ValueError("basis_degree must be at least 1")
        self.basis_degree = int(basis_degree)

        raw = [torch.ones_like(self.zeta)] + [
            self.zeta ** d for d in range(1, self.basis_degree + 1)
        ]
        orthonormal = []
        for vector in raw:
            for previous in orthonormal:
                vector = vector - self.inner(vector, previous) * previous
            norm = torch.sqrt(self.inner(vector, vector))
            if float(norm) <= 1e-12:
                raise ValueError("basis degenerated during orthonormalization")
            orthonormal.append(vector / norm)
        # discard the constant, keep the m read-out-relevant directions
        self.basis = torch.stack(orthonormal[1:], dim=1)          # (K, m)
        self.num_basis = self.basis.shape[1]

        # mu_u: the weight on the terminating tonic node, one per basis direction.
        # Scalars broadcast.
        #
        # mu_u MUST BE NONZERO. The terminating contribution to g_S is theta_u * B * mu_u,
        # so at mu_u = 0 it is identically zero for every theta_u: the chain stops rather
        # than terminates, phi_u loses its anchor, and theta_u becomes unidentifiable --
        # it diverges by shifting magnitude out of the priced factor phi_u into the
        # unpriced theta_u along the exact degeneracy (theta_u, phi_u) -> (c theta_u,
        # phi_u / c), which leaves g_S invariant. This is not a departure from Bogacz:
        # his Eq. (25) model diverges identically at v_p = 0 (argmax theta = u / v_p).
        # v_p != 0 is an unstated premise of his construction, whose worked example
        # takes v_p = 3. The value 1.0 is a stipulation; a terminating prior is a fixed
        # boundary condition and has no target to be derived from.
        mu_u_tensor = torch.as_tensor(mu_u, dtype=self.dtype, device=device)
        if mu_u_tensor.ndim == 0:
            mu_u_tensor = mu_u_tensor.expand(self.num_basis).clone()
        if mu_u_tensor.shape != (self.num_basis,):
            raise ValueError(f"mu_u must be a scalar or have shape ({self.num_basis},)")
        self.mu_u = mu_u_tensor

        # Log of the base world prior, ell_0 (Sec. 3). `base_prior` is any callable
        # zeta -> log-density; when omitted, `gaussian_world_prior` at the given
        # mean and precision.
        #
        # The result is renormalized against the quadrature measure so that
        # sum_k w_k exp(ell_0) = 1 exactly on the truncated grid. An analytically
        # normalized density is not grid-normalized once the grid is cut off, and the
        # spec (Sec. 9.1) notes that phi_S is pinned in absolute terms, so the constant
        # is worth getting right even though it happens to be invisible here: B is
        # orthogonal to the constant function by construction, so <1, b_j> = 0 and an
        # additive shift in ell_0 reaches neither phi_u* (Eq. 16) nor the read-out.
        if base_prior is None:
            base_prior = gaussian_world_prior(prior_mean, prior_precision)
        if not callable(base_prior):
            raise TypeError("base_prior must be a callable zeta -> log-density")

        self.prior_mean = float(prior_mean)
        self.prior_precision = float(prior_precision)
        self.base_prior = base_prior
        self.base_prior_description = getattr(base_prior, "description", "custom")

        raw_log_prior = base_prior(self.zeta)
        if raw_log_prior.shape != self.zeta.shape:
            raise ValueError("base_prior must return one log-density per grid node")
        shift = raw_log_prior.max()
        prior_grid_mass = torch.sum(self.weights * torch.exp(raw_log_prior - shift))
        self.base_log_prior = raw_log_prior - shift - torch.log(prior_grid_mass)

        # retained so that probe networks can be respawned with the same configuration
        self._configuration = dict(
            num_nodes=num_nodes,
            grid_half_width=grid_half_width,
            base_prior=base_prior,
            prior_mean=prior_mean,
            prior_precision=prior_precision,
            theta_L=theta_L,
            num_atoms=num_atoms,
            lexical_strength=lexical_strength,
            mu_u=self.mu_u.clone(),
            theta_u=theta_u,
            theta_u_initial=theta_u_initial,
            basis_degree=basis_degree,
            sigma_lexical=sigma_lexical,
            sigma_state=sigma_state,
            sigma_utility=sigma_utility,
            exclusion_sharpness=exclusion_sharpness,
            device=device,
        )

        # theta_u: learned (theta_u*, Eq. B2) unless a fixed control value is given.
        # "learned" is what `_configuration` carries, so every respawn re-learns it for
        # its own inputs and hyperparameters (a variational theta_u is a commitment of
        # the model, not a setting carried over between configurations).
        self.theta_u_learned = theta_u == "learned"
        self.theta_u = (
            self.learned_theta_u() if self.theta_u_learned else float(theta_u)
        )

    def respawn(self, **overrides):
        """A new network with this one's configuration, plus any overrides.

        theta_u is re-learned for the new configuration unless the override, or the
        configuration being copied, fixes it to a number."""
        return LexicalPredictiveCodingNetwork(**{**self._configuration, **overrides})

    # === quadrature (Eq. 2) ===

    def inner(self, f, h):
        """<f, h> under the fixed quadrature measure (Eq. 2)."""
        return torch.sum(self.weights * f * h)

    def project(self, f):
        """B^T W f: contract a field onto the m basis directions (Eq. 3)."""
        return torch.mv(self.basis.T, self.weights * f)

    def squared_norm(self, f):
        return self.inner(f, f)

    @staticmethod
    def dot(a, b):
        """Ordinary inner product, for the utility vector (no quadrature measure)."""
        return torch.sum(a * b)

    # === clamped lexical field (Eqs. 5-6) ===

    def lexical_field(self, utterance):
        """phi_L = Lambda * chi_y, clamped by the observed utterance."""
        chi = exclusion_indicator(
            utterance,
            self.zeta,
            self.theta_L,
            sharpness=self.exclusion_sharpness,
        )
        return self.lexical_strength * chi

    # === generative pass (Sec. 4) ===

    def predict_lexical(self, phi_S):
        """g_L(phi_S) = ell_0 - phi_S (Eq. 9).

        The complement, transported to log-weights and reflected through the base world
        base world prior rather than through zero. Still an affine involution --
        g_L(g_L(x)) = x, so the structure of Eq. (9) is unchanged; ell_0 supplies the
        reference point the complement is taken about."""
        return self.base_log_prior - phi_S

    def predict_state(self, phi_u, theta_u=None):
        """g_S(phi_u) = theta_u * B phi_u (Eq. 10, with M = B theta_u of Sec. 8.4).

        ell_0 has moved to g_L, so the utility level's prediction of the situation is
        now purely its own contribution, with no tonic offset. `theta_u=None` uses the
        network's own theta_u, which is theta_u* unless it was fixed as a control."""
        theta = self.theta_u if theta_u is None else float(theta_u)
        return theta * torch.mv(self.basis, phi_u)

    def residuals(self, phi_S, phi_u, phi_L, theta_u=None):
        """r_L, r_S, r_u of Eq. (11). r_y == 0 by Eq. (8) and is omitted; Appendix A
        gives its space (Eq. 4) and why omitting it is safe (Eq. A3)."""
        r_lexical = phi_L - self.predict_lexical(phi_S)
        r_state = phi_S - self.predict_state(phi_u, theta_u=theta_u)
        r_utility = phi_u - self.mu_u
        return r_lexical, r_state, r_utility

    def free_energy(self, phi_S, phi_u, phi_L, theta_u=None):
        """Bogacz's F, to be MAXIMIZED (Eq. 13). The log-sigma terms vanish at
        sigma = 1 and are constant in every variational quantity; they are carried
        only to mark where a precision-bearing version would reintroduce them."""
        r_lexical, r_state, r_utility = self.residuals(
            phi_S, phi_u, phi_L, theta_u=theta_u
        )
        quadratic = (
            -0.5 * self.squared_norm(r_lexical) / self.sigma_lexical
            - 0.5 * self.squared_norm(r_state) / self.sigma_state
            - 0.5 * self.dot(r_utility, r_utility) / self.sigma_utility
        )
        normalizer = -0.5 * (
            self.num_nodes * math.log(self.sigma_lexical)
            + self.num_nodes * math.log(self.sigma_state)
            + self.num_basis * math.log(self.sigma_utility)
        )
        return quadratic + normalizer

    # === read-out (Eq. 12) ===

    def read_out(self, phi_S):
        """Normalized belief q(zeta_k), computed in log-space (Sec. 9.2 caveat 2)."""
        shifted = phi_S - phi_S.max()
        unnormalized = torch.exp(shifted)
        return unnormalized / torch.sum(self.weights * unnormalized)

    # === closed-form targets (Sec. 6) ===

    def closed_form_fixed_point(self, utterance, theta_u=None):
        """Eqs. (15)-(16), generalized to arbitrary fixed sigma and to B of Sec. 8.4.

        Writing S = sigma_L + sigma_S, G = B^T W B and c = B^T W (ell_0 - phi_L),

            (I + sigma_u theta^2 G / S) phi_u* = mu_u + sigma_u theta c / S
            phi_S* = [sigma_L theta B phi_u* + sigma_S (ell_0 - phi_L)] / S

        The signs follow ell_0 having moved into g_L (Eq. 9): the utility level couples
        to the NET ell_0 - phi_L where it coupled to the sum ell_0 + phi_L before.

        At sigma = 1 and m = 1 these reduce to Eqs. (15)-(16) as written. G = I for the
        orthonormal basis built above, but the m x m system is solved in general so the
        result stays correct if a later phase supplies a non-orthonormal B.

        Because the model is linear-Gaussian while phi_L is clamped (Sec. 8.6), these
        are exact, and they are the correctness test for `infer`.
        """
        theta = self.theta_u if theta_u is None else float(theta_u)
        phi_L = self.lexical_field(utterance)
        total_sigma = self.sigma_lexical + self.sigma_state

        gram = self.basis.T @ (self.weights[:, None] * self.basis)
        # ell_0 now enters through g_L, so the utility level couples to the NET of ell_0
        # and lexicon, ell_0 - phi_L, where before it coupled to their sum.
        coupling = self.project(self.base_log_prior - phi_L)

        matrix = (
            torch.eye(self.num_basis, dtype=self.dtype, device=self.device)
            + self.sigma_utility * theta ** 2 * gram / total_sigma
        )
        rhs = self.mu_u + self.sigma_utility * theta * coupling / total_sigma
        phi_u_star = torch.linalg.solve(matrix, rhs)

        phi_S_star = (
            self.sigma_lexical * self.predict_state(phi_u_star, theta_u=theta)
            + self.sigma_state * (self.base_log_prior - phi_L)
        ) / total_sigma

        return phi_S_star, phi_u_star

    # === the literal listener (Part C, Part D) ===

    def literal_fixed_point(self, utterance):
        """phi_S of the LITERAL listener: the base world prior restricted by the entry.

        phi_S = ell_0 - phi_L, whose read-out is q_lit proportional to
        p_0 * exp(-Lambda chi_y): the base world prior times the soft semantic mask, and nothing
        else. This is the baseline the mass shift criterion of Text cell 4 Part C
        is measured against.

        It is a fixed point of THIS network, not an external construction. Eq. (15)
        reads phi_S* = [sigma_L theta_u B phi_u + sigma_S (ell_0 - phi_L)] / S with
        S = sigma_L + sigma_S, so sigma_S -> infinity sends it to ell_0 - phi_L: the
        situation level keeps no prior of its own and the lexical channel alone fixes
        the belief. That limit is what "literal" means here.

        NOT the same as theta_u = 0. At theta_u = 0 the r_S term still penalizes phi_S
        against a prediction of zero, which leaves phi_S* = (ell_0 - phi_L)/2: the
        literal posterior TEMPERED by one half, since the read-out (Eq. 12) is
        exponential. The half is a genuine temperature, so the theta_u = 0 belief is
        flatter than the literal one and holds systematically more mass at the ends of
        the scale. Both are reported below: this is the baseline, and theta_u = 0 is
        kept beside it as the temperature-matched control. The tempered control is
        also where the flow of Eq. (20) starts (`theta_u_initial`); the literal
        listener is not.
        """
        return self.base_log_prior - self.lexical_field(utterance)

    # === the learned theta_u (Sec. 8.5, Appendix B) ===

    def theta_u_stationary_points(self, utterances=None):
        """Both roots of Eq. (B2): (theta_u*, theta_min), the maximizer and the minimum.

        Eliminating phi_S and phi_u (exact by Sec. 8.6) reduces F, averaged over the
        exposure ensemble `utterances` (default: the three utterances at uniform
        p(y), Appendix B), to

            F~(theta) = const - sum_y ||theta mu_u - c_y||^2 / (2 (sigma_u theta^2 + S)),
                S = sigma_L + sigma_S,        c_y = B^T W (ell_0 - phi_L,y),

        whose stationarity equation is the quadratic

            sigma_u <mu_u, sum c> theta^2 + (n S ||mu_u||^2 - sigma_u sum ||c_y||^2) theta
                                                          - S <mu_u, sum c> = 0,

        n the number of utterances. The roots have product -S / sigma_u < 0: exactly one
        of each sign, the maximizer being the one with sign <mu_u, sum c>. The reduction
        uses G = B^T W B = I, which the basis built above satisfies.
        """
        utterances = UTTERANCES if utterances is None else list(utterances)
        gram = self.basis.T @ (self.weights[:, None] * self.basis)
        identity = torch.eye(self.num_basis, dtype=self.dtype, device=self.device)
        if float((gram - identity).abs().max()) > 1e-10:
            raise NotImplementedError(
                "the closed form of Eq. (B2) assumes B^T W B = I (the basis built here)"
            )
        couplings = [self.project(self.base_log_prior - self.lexical_field(y))
                     for y in utterances]
        total = torch.stack(couplings).sum(dim=0)
        S = self.sigma_lexical + self.sigma_state
        a = self.sigma_utility * float(self.dot(self.mu_u, total))
        if a == 0.0:
            raise ValueError(
                "<mu_u, sum_y c_y> = 0: F~ has no interior maximizer and theta_u "
                "diverges (Appendix B's degenerate ray)"
            )
        b = (len(utterances) * S * float(self.dot(self.mu_u, self.mu_u))
             - self.sigma_utility * sum(float(self.dot(c, c)) for c in couplings))
        c0 = -S * a / self.sigma_utility
        root = math.sqrt(b * b - 4.0 * a * c0)       # discriminant > 0: a * c0 < 0
        # numerically stable pair: q = -(b + sign(b) root) / 2
        q = -0.5 * (b + math.copysign(root, b))
        roots = (q / a, c0 / q)
        maximizer = roots[0] if (roots[0] > 0) == (a > 0) else roots[1]
        minimum = roots[1] if maximizer == roots[0] else roots[0]
        return maximizer, minimum

    def learned_theta_u(self, utterances=None):
        """theta_u*, the maximizer of F~ (Eq. B2): the value Eq. (20) ascends to."""
        return self.theta_u_stationary_points(utterances)[0]

    # === Hessian check (Eq. 21) ===

    def hessian_quadratic_form(self, v, t, theta_u=None):
        """(v, t)^T grad^2 F (v, t) of Eq. (21), in the sigma-weighted form

            -||v||^2/sigma_L - ||v - theta_u B t||^2/sigma_S - ||t||^2/sigma_u,

        obtained by reading off the directional derivative of each residual. Strictly
        negative for (v, t) != 0, for every theta_u and every Lambda (Sec. 8.1), and
        for arbitrary B and m (Sec. 8.4).
        """
        theta = self.theta_u if theta_u is None else float(theta_u)
        return (
            -self.squared_norm(v) / self.sigma_lexical
            - self.squared_norm(v - theta * torch.mv(self.basis, t)) / self.sigma_state
            - self.dot(t, t) / self.sigma_utility
        )

    # === recurrent inference (Sec. 7) ===

    def stiffest_state_rate(self, theta_u=None):
        """lambda_max(H), H = -grad^2 F in the state (Eqs. 21-22), quadrature metric.

        The rate of the stiffest state mode under Eq. (19) with instantaneous errors.
        Eq. (28) is H's diagonal phi_u entry, 1/sigma_u + theta_u^2 lambda_max(G)/sigma_S;
        the coupling to phi_S puts the top eigenvalue slightly above it (810.69 against
        809.69 at the Gaussian theta_u*).

        Computed blockwise, without decomposing H (Appendix G Sec. 1). In the symmetrized
        coordinates diag(W^1/2, I) H diag(W^-1/2, I), which have H's eigenvalues, H couples
        phi_S to phi_u only through W^1/2 B, whose singular values are the square roots
        of the eigenvalues gamma_j of G = B^T W B. So H splits into one 2 x 2 block per
        gamma_j,

            [[a,                            -theta sqrt(gamma_j) / sigma_S       ],
             [-theta sqrt(gamma_j) / sigma_S, 1/sigma_u + theta^2 gamma_j / sigma_S]],

        with a = 1/sigma_L + 1/sigma_S, and acts as a on the directions of phi_S that B
        does not reach. Every block's larger root is at least a, so lambda_max(H) is the
        largest of those roots. Exact for every sigma and every B, orthonormal or not, and
        Theta(K m^2) where a dense decomposition of H is Theta(K^3). At sigma = 1 and
        G = I it is theta_u^2 + 2, Eq. (G1).
        """
        theta = self.theta_u if theta_u is None else float(theta_u)
        a = 1.0 / self.sigma_lexical + 1.0 / self.sigma_state
        gram = self.basis.T @ (self.weights[:, None] * self.basis)
        gammas = torch.linalg.eigvalsh(gram).clamp(min=0.0)       # roundoff below zero
        d = 1.0 / self.sigma_utility + theta ** 2 * gammas / self.sigma_state
        coupling = theta ** 2 * gammas / self.sigma_state ** 2
        roots = 0.5 * ((a + d) + torch.sqrt((a - d) ** 2 + 4.0 * coupling))
        return float(roots.max())
    def fast_time_constant(self, theta_u=None, tau_state=1.00):
        """The D4 commitment: the largest tau_error that keeps every mode non-oscillatory.

        Eliminating the error units from Eqs. (18)-(19) gives, for each eigenvalue
        lambda of H, the pair of rates s solving

            tau_error tau_state s^2 + tau_state s + lambda = 0,

        plus s = -1/tau_error for the error directions the states do not read. Both
        roots are real iff tau_error <= tau_state / (4 lambda), so the bound is set by
        lambda_max(H) and is a critical damping, not a rule of thumb. It scales as
        theta_u^-2: the error units must be fast relative to the stiffest state mode,
        not only relative to tau_state (Sec. 8.3's tau_error << tau_state is the
        theta_u -> 0 end of it). Above the bound the stiff mode rings and F is not
        monotone along the flow, which Sec. 8.2 asserts.
        """
        return tau_state / (4.0 * self.stiffest_state_rate(theta_u))

    @torch.no_grad()
    def infer(
        self,
        utterance,
        *,
        dt=None,
        tau_error=None,
        tau_state=1.00,
        max_steps=None,
        max_time=1000.0,
        derivative_tolerance=None,
        patience=10,
        record_history=True,
        theta_u=None,
    ):
        """Integrate Eqs. (18)-(19) to the fixed point.

        The time constants follow the D4 commitment (`fast_time_constant`): by default
        tau_error sits at the critical-damping bound tau_state / (4 lambda_max(H)),
        which depends on theta_u, and dt = tau_error / 2. `max_steps` defaults to
        `max_time` / dt, so the time horizon, not the step count, is what is fixed.
        F is monitored but never differentiated: the messages are computed locally and
        F only records that they are ascending it.

        derivative_tolerance: the stopping test is max|derivative| < this for
        `patience` consecutive steps. It is set ABOVE the floating-point floor on
        purpose, because a tolerance inside that band is crossed by roundoff rather
        than by the dynamics: the step count then moves by hundreds under a change in
        theta_u of one part in 1e15, and two implementations of the same gradient
        disagree on it.

        THE FLOOR IS NOT A CONSTANT. It is 4.547e-13 * lambda_max(H) (see
        `DERIVATIVE_TOLERANCE_PER_RATE`), so the tolerance is not one either: by default
        it is DERIVATIVE_TOLERANCE_PER_RATE * lambda_max(H), which holds the margin above
        the floor fixed at 12.1 at every theta_u. The old fixed 1e-9 is what this
        returns at lambda = 182, and it is reported in the result so that anything
        checked against it can be checked in units of it rather than against a constant
        of its own.

        The fixed point is correspondingly accurate to about the tolerance: the
        residual displacement lies in H's slowest direction, where the rate is 1, so
        |phi - phi*| is the tolerance rather than the tolerance over lambda. That is
        1.6e-10 at theta_u = 5.3 and 2.1e-8 at theta_u = 61.3, four orders below
        anything reported in either case.
        """
        if utterance not in UTTERANCES:
            raise ValueError(f"utterance must be one of {UTTERANCES}")
        theta = self.theta_u if theta_u is None else float(theta_u)
        if derivative_tolerance is None:
            derivative_tolerance = (
                DERIVATIVE_TOLERANCE_PER_RATE * self.stiffest_state_rate(theta)
            )
        bound = self.fast_time_constant(theta, tau_state=tau_state)
        tau_error = bound if tau_error is None else tau_error
        dt = 0.5 * tau_error if dt is None else dt
        if min(dt, tau_error, tau_state) <= 0:
            raise ValueError("dt and both time constants must be positive")
        if dt > tau_error:
            raise ValueError("require dt <= tau_error (Euler, Sec. 8.3)")
        if tau_error >= tau_state:
            raise ValueError("require tau_error << tau_state (Sec. 8.3)")
        if tau_error > bound * (1.0 + 1e-12):
            raise ValueError(
                f"tau_error={tau_error:.3g} violates the timescale commitment "
                f"tau_error <= tau_state / (4 lambda_max(H)) = {bound:.3g} at "
                f"theta_u = {theta:.4g}: the stiffest mode would oscillate"
            )
        # The loop updates the error units first and the states from the updated errors,
        # so each mode of H with eigenvalue lambda steps by a 2 x 2 map of determinant
        # 1 - dt/tau_error and trace 2 - dt/tau_error - dt^2 lambda / (tau_error tau_state)
        # (Appendix G Sec. 3). Under the guards above both of its factors are real and lie
        # in [0, 1) for every lambda <= lambda_max(H), so the loop is stable and does not
        # oscillate without a separate check.
        if max_steps is None:
            max_steps = int(math.ceil(max_time / dt))

        phi_L = self.lexical_field(utterance)

        # initialization: the belief starts at the base world prior, the utility state at the
        # mean of its terminating prior, and every error unit silent.
        phi_S = self.base_log_prior.clone()
        phi_u = self.mu_u.clone()

        eps_lexical = torch.zeros_like(phi_S)
        eps_state = torch.zeros_like(phi_S)
        eps_utility = torch.zeros_like(phi_u)

        history = []
        settled = 0
        converged = False

        for step in range(max_steps):
            r_lexical, r_state, r_utility = self.residuals(
                phi_S, phi_u, phi_L, theta_u=theta
            )

            # Eq. (18): tau_e * eps_dot = r - eps
            eps_lexical_dot = (r_lexical / self.sigma_lexical - eps_lexical) / tau_error
            eps_state_dot = (r_state / self.sigma_state - eps_state) / tau_error
            eps_utility_dot = (
                r_utility / self.sigma_utility - eps_utility
            ) / tau_error

            eps_lexical = eps_lexical + dt * eps_lexical_dot
            eps_state = eps_state + dt * eps_state_dot
            eps_utility = eps_utility + dt * eps_utility_dot

            # Eq. (19): state units ascend F. The lexical error arrives inhibitorily
            # because g_L' = -I.
            phi_s_dot = (-eps_state - eps_lexical) / tau_state
            phi_u_dot = (
                -eps_utility + theta * self.project(eps_state)
            ) / tau_state

            max_derivative = max(
                float(eps_lexical_dot.abs().max()),
                float(eps_state_dot.abs().max()),
                float(eps_utility_dot.abs().max()),
                float(phi_s_dot.abs().max()),
                float(phi_u_dot.abs().max()),
            )

            if record_history:
                history.append(
                    {
                        "step": step,
                        "free_energy": float(
                            self.free_energy(phi_S, phi_u, phi_L, theta_u=theta)
                        ),
                        "max_derivative": max_derivative,
                        "phi_u": phi_u.tolist(),
                    }
                )

            settled = settled + 1 if max_derivative < derivative_tolerance else 0
            if settled >= patience:
                converged = True
                break

            phi_S = phi_S + dt * phi_s_dot
            phi_u = phi_u + dt * phi_u_dot

            if not (torch.isfinite(phi_S).all() and torch.isfinite(phi_u).all()):
                raise FloatingPointError(
                    "non-finite activity: reduce dt or lower the lexical strength"
                )

        r_lexical, r_state, r_utility = self.residuals(
            phi_S, phi_u, phi_L, theta_u=theta
        )

        return {
            "utterance": utterance,
            "converged": converged,
            "steps": step + 1,
            "phi_S": phi_S,
            "phi_u": phi_u,
            "phi_L": phi_L,
            "q": self.read_out(phi_S),
            "free_energy": self.free_energy(phi_S, phi_u, phi_L, theta_u=theta),
            "error_units": {
                "lexical": eps_lexical,
                "state": eps_state,
                "utility": eps_utility,
            },
            "residuals": {
                "lexical": r_lexical,
                "state": r_state,
                "utility": r_utility,
            },
            "theta_u": theta,
            "dt": dt,
            "tau_error": tau_error,
            "derivative_tolerance": derivative_tolerance,
            "history": history,
        }

    # === the slow parameter (Sec. 7 Eq. 20, Sec. 8.5) ===

    def theta_u_gradient(self, result):
        """dF/dtheta_u = phi_u <eps_S, b> (Eq. 20).

        This is Bogacz Eq. (25), dF/dtheta = eps_u * phi, the error BELOW times the
        state ABOVE, with phi_S being a field rather than a scalar, which is the only
        reason the contraction against b appears. Eq. (29) is the same quantity for a
        nonlinear h; ours is h = identity."""
        return self.dot(
            result["phi_u"], self.project(result["error_units"]["state"])
        )

    def learn_theta_u(
        self,
        utterances,
        *,
        tolerance,
        tau_theta=20.0,
        num_updates=50,
        tau_state=1.00,
        theta_u_initial=None,
        **infer_kwargs,
    ):
        """Integrate Eq. (20), tau_theta * d(theta_u)/dt = dF/d(theta_u), from theta_u(0).

        There is no learning rate here, and none is needed. Bogacz gives the GRADIENT
        (his Eqs. 25 and 29) and states that parameters move proportionally to it; the
        constant of proportionality is a time constant of exactly the kind Eqs. (18)
        and (19) already carry for the error and state units. So theta_u takes a third,
        tau_theta, and the architecture requires

            tau_error <= tau_state / (4 lambda_max(H)),    tau_state << tau_theta,

        the slow end of the same ordering whose fast end is `fast_time_constant`.
        tau_theta sets the RATE only: the fixed point of the flow is the stationary
        point of F~(theta_u) = max_x F(x, theta_u) and does not depend on it.

        The flow starts at `theta_u_initial` (default: the network's, 0), one start
        shared by every configuration. There the network is the TEMPERED control; the
        update is <mu_u, sum_y c_y>-driven and nonzero, so theta_u leaves zero at once,
        toward theta_u*: the roots of Eq. (B2) have opposite signs, so zero lies
        between the minimum and the maximizer, on the maximizer's rising side. The
        network's theta_u is left where the flow ends. Every update re-derives the fast
        time constants from the current theta_u, since H stiffens as theta_u^2.

        HALTING (A19). The flow stops when its own update falls below `tolerance`:
        |delta theta_u| = (tau_state / tau_theta) |gradient| < tolerance. That is the
        ONLY stopping rule. It reads the size of the step the unit has just taken and
        nothing else: no closed form, and nothing about the shape of the trajectory,
        which the unit has no access to. `infer` halts the fast subsystem by the same
        mechanism at its own timescale.

        `tolerance` HAS NO DEFAULT, so every call names it and no result inherits a
        value silently. Halting by a tolerance is offered as a DIRECTION and not as a
        commitment (A19, demoted): the fast loop's own tolerance is keyed to
        lambda_max(H), which is an unsettled locality problem of the same kind as the
        timescale bound's (D12). The value here is ad hoc in any case. Predictions are therefore reported
        at theta_u* (Eq. B2), the asymptote of this flow, which carries no tolerance;
        a theta_u taken FROM this flow is a REALIZABLE one and carries on its face the
        tolerance it halted at.

        `num_updates` is a CAP and not a stopping rule. Reaching it means the flow did
        not halt, which is a fact to report rather than a result to use, so it raises.

        Requires <mu_u, sum_y c_y> != 0 (Appendix B). mu_u != 0 (Sec. 3) is
        necessary but not sufficient; otherwise F~ has no stationary point and this
        flow diverges.
        """
        if tolerance <= 0:
            raise ValueError(
                "tolerance must be positive: it is the halting mechanism (A19), and "
                "a nonpositive one never fires"
            )
        if num_updates < 1:
            raise ValueError(
                "num_updates caps the flow and must be at least 1; it is a cap, not "
                "a stopping rule"
            )
        if tau_theta <= tau_state:
            raise ValueError(
                "require tau_state << tau_theta: the fast subsystem must equilibrate "
                "between theta_u updates for Eq. (20) to be ascent on F~"
            )
        # necessary part only; the full requirement is Appendix B's
        if float(self.mu_u.abs().max()) == 0.0:
            raise ValueError(
                "mu_u = 0 leaves theta_u unidentifiable (Sec. 3). This test is the "
                "necessary part only: Appendix B requires <mu_u, sum_y c_y> != 0, "
                "which a nonzero mu_u on a degenerate ray still violates "
                "(Text cell 5 Part B Sec. 6)."
            )
        if "dt" in infer_kwargs or "tau_error" in infer_kwargs:
            raise ValueError(
                "dt and tau_error are re-derived at every update from the current "
                "theta_u (`fast_time_constant`); a fixed value would break the "
                "commitment as |theta_u| grows"
            )

        self.theta_u = self.theta_u_initial if theta_u_initial is None else float(theta_u_initial)
        self.theta_u_learned = False       # theta_u now follows the flow, not theta_u*

        trace = []
        for update in range(num_updates):
            gradient = 0.0
            free_energy = 0.0
            for utterance in utterances:
                result = self.infer(
                    utterance, record_history=False, tau_state=tau_state, **infer_kwargs
                )
                if not result["converged"]:
                    raise RuntimeError(
                        f"fast subsystem did not equilibrate for {utterance!r}; "
                        "the two-timescale argument of Sec. 8.5 does not apply"
                    )
                gradient += float(self.theta_u_gradient(result))
                free_energy += float(result["free_energy"])

            gradient /= len(utterances)
            free_energy /= len(utterances)

            trace.append(
                {"update": update, "theta_u": self.theta_u,
                 "gradient": gradient, "free_energy": free_energy}
            )
            step = (tau_state / tau_theta) * gradient
            self.theta_u = self.theta_u + step
            if abs(step) < tolerance:
                return trace

        raise RuntimeError(
            f"theta_u did not halt within num_updates={num_updates}: the last update "
            f"was {abs(step):.3e}, at or above tolerance={tolerance:.3e}, with theta_u "
            f"at {self.theta_u:.6f}. num_updates is a cap and not a stopping rule "
            "(A19), so a run that reaches it HAS NOT HALTED: report the configuration "
            "as one that does not halt at this tolerance within this cap, rather than "
            "reading its theta_u as a realizable value."
        )
````

## <a id="tc4" name="tc4"></a>Text cell 4: evaluation

The evaluation has four parts.

### <a id="tc4-parta" name="tc4-parta"></a>Part A
**A** tests every testable claim of the preceding specification: the closed
forms of Eqs. (15)-(16), the fixed-point identities of Eq. (17), monotonicity of $\mathcal F$
(Sec. 8.2), negative definiteness of the Hessian (Eq. 21) across $\theta_u$, the independence
of the relaxation rate from $\varphi_L$ (Sec. 8.1(iii)), agreement of the analytic messages
with finite differences (Sec. 7), stability of the belief under grid refinement (Sec. 1), and
compatibility with a smooth lexical field (Sec. 9.2). Three of the fourteen concern the learned
parameter: that $\theta_u^\ast$ of Eq. (B2) agrees with an independent bisection on the reduced
gradient (relative gap $4.6\times10^{-14}$), that $\tilde{\mathcal F}$ has an interior maximum there
rather than an asymptote (margin $+0.2398$), and that Eq. (20) ascends toward it from
$\theta_u(0)=0$, which is the flow as Sec. 8.5 describes it and not a check that it arrives.

The relaxation rate is compared **per unit time**, as a least-squares slope of
$\log\lVert\varphi_S-\varphi_S^\ast\rVert$ over a window selected by value rather than by step
index ($1.0002312$, $1.0002304$, $1.0002312$ for the three utterances, relative spread
$7.7\times10^{-7}$). A per-step contraction factor would not test Sec. 8.1(iii) at all under the
timescale commitment: $\mathrm dt$ is re-derived from $\theta_u$ at every configuration, so per-step
factors differ between runs that relax at identical rates, and at the $\mathrm dt$ the commitment
forces they sit within roundoff of $1$.

Two of them are tested in ways that need comment. **Grid refinement** is a claim about *convergence* in $K$,
so it is tested on the tail of the sequence, $\lvert\mathbb E[s]_{801}-\mathbb E[s]_{401}\rvert$,
rather than on the gap between two coarse grids. The distinction matters here because the rate is
set by something Sec. 1 does not mention: $\chi_y$ is a step, and trapezoidal quadrature on a
discontinuous integrand converges at $O(h)$ with the step's position jittering against the nodes as
$K$ changes, which is §9.2 caveat 1 arriving as a measurement. The tail is
$1.7\times10^{-3}$ under the hard $\chi_y$ and $3.3\times10^{-6}$ under the smooth one, against a
coarse $51$–$401$ spread of $1.1\times10^{-2}$; the cell reports all three, and the smooth mask is
the control that identifies which of them is the model and which is the quadrature.

**Everything in this cell is reported at the learned $\theta_u^\ast$**, which under the default
Gaussian base world prior is $-28.4375$: negative, and nowhere near the $1$ a constructor default would supply. Every
network an evaluation respawns re-learns its own (commitment 6), so the value differs from row to row
and each table says which it carries. Where a $\theta_u$ is held fixed instead, the table says so and
the reason is stated with it; `theta_u_dependence` is the case in point, recomputing every headline
quantity at the $\theta_u=0$ control, at $\pm1$ and at $\theta_u^\ast$ so that the dependence on
$\theta_u$, and on its sign, is visible rather than assumed. Two things it shows are worth carrying:
the *all*$-$*some* $\varphi_S$ contrast *grows* with $\lvert\theta_u\rvert$
($9.6995\to10.9524\to13.9124$), so unlike $\mu_u$ and $\ell_0$ the utility level does move contrasts;
and the difference between $\theta_u$ and $-\theta_u$ is largest near $\lvert\theta_u\rvert=1$
($0.0483$ in $\mathbb E[s]$, $0.0902$ in $P(\text{all-region})$) and decays to $0.0041$ and $0.0083$
at $\lvert\theta_u^\ast\rvert$, because only the $\mu_u$-carried linear term changes sign while the
$\theta_u^2$ term does not. The sign of $\theta_u$ therefore matters least where the model actually
sits.

### <a id="tc4-partb" name="tc4-partb"></a>Part B
**B** runs the network and sweeps $\Lambda$, confirming that a hard lexical constraint is
recovered as the $\Lambda\to\infty$ limit. Each $\Lambda$ is a configuration of its own and carries
its own $\theta_u^\ast$, which runs $-18.730$, $-19.192$, $-20.897$, $-28.437$, $-76.638$,
$+1535.417$, $+342.254$ over $\Lambda=1\dots64$: the sweep is not a curve at fixed $\theta_u$, and
the sign turns over between $\Lambda=16$ and $\Lambda=32$. The leak falls monotonically to
$5.03\times10^{-20}$ at $\Lambda=64$ regardless, which is the limit the part is testing. From
$\Lambda=32$ upward $\lambda_{\max}(H)$ is far past what the timescale commitment lets the
evaluation integrate, so the sweep is read from the closed form of Eqs. (15)–(16), exact by §8.6.

### <a id="tc4-partc" name="tc4-partc"></a>Part C
**C** asks whether the network produces a scalar implicature. The question needs an operational
form, and Part C states two criteria for it rather than deriving one. Three quantities have to be
named first, because the criteria are relations between them.

**The pragmatic listener and the literal baseline.** Write $q_y$ for the settled read-out under
utterance $y$, the pragmatic listener, written $q_H$ where the contrast needs a name. The baseline
it is measured against, and the dashed line in every panel of Part D and Text cell 4b, is
$q_{\mathrm{lit}}$, the
**untempered literal listener**: the belief $\varphi_S=\ell_0-\varphi_L$ that the base world prior
and the lexical entry fix between them, whose read-out is $p_0e^{-\Lambda\chi_y}$ normalized, the
base world prior restricted by the entry and nothing else. It is a fixed point of this network and not a
construction from outside it, being what Eq. (15) returns as $\sigma_S\to\infty$, where the
situation level keeps no prior of its own and the lexical channel alone fixes the belief. Reading
implicature strength off the gap between a literal and a pragmatic listener follows the practice of
the probabilistic accounts (Goodman & Frank, 2016, pp. 819–820).

**The $\theta_u$ control is a third quantity, not the baseline.** Eq. (15) at $\theta_u=0$ leaves
$\varphi_S^\ast=(\ell_0-\varphi_L)/2$, which is $q_{\mathrm{lit}}$ tempered by one half, the
halving surviving into the belief because the read-out of Eq. (12) is exponential. The two differ
by that temperature and by nothing else. $q_{\mathrm{lit}}$ is the baseline because it is the
literal listener the mass shift criterion names; the control is kept in view because it holds the temperature
fixed, so what separates it from $q_y$ is the utility level alone. The shift against
$q_{\mathrm{lit}}$ therefore carries both the tempering and the utility level, and Part D reports
the two masses side by side, and the shift split into the two parts, so that they can be read off.

The control is also **where learning starts**: Eq. (20) is integrated from $\theta_u(0)=0$, so the
model begins at the tempered control and moves away from it as $\theta_u$ rises (Text cell 3, under
Eq. 20). That makes the tempering the shift's starting value rather than a fixed offset carried
throughout: under the default Gaussian base world prior the shift for *some* is $+0.0175$ at $\theta_u=0$ and $+0.0008$
at $\theta_u^\ast$. It does not make the control the baseline, and it does not make it
$q_{\mathrm{lit}}$: the literal listener needs $\sigma_S\to\infty$ as well, and no run of this model
is ever at it.

**The all-region is the continuous form of the top state.** $\theta_L$ is fixed by
$\varsigma(-\theta_L)=1/2n$ at $n=10$ (Eq. A5), so the region $\zeta\ge\theta_L$ is $s\ge0.95$, which on
a ten-atom scale contains the atom $s=1$ and no other. $P(\text{all-region})$ of Eq. (27) is
accordingly what $q(s{=}10)$ names on a discrete scale. The region has two ends and only the near
one is semantic: $\theta_L$ fixes it, while the far one is the grid's truncation, so the cell's
extent is $Z-\theta_L$ and $n$ and the half-width bear on it between them. The truncation reaches
further than the cell's extent, because $B$ is orthonormalized on the grid, so $\varphi_S^\ast$
itself is shaped by it (Text cell 3 §1). The threshold of Eq.
(A5) also makes the
all-region Appendix A's Voronoi **cell of *all***, the proportions closer to $s=1$ than to any
other count, so the two names carry one region: *all-region* is written where a $q$-mass is taken
over it, *cell of all* where a mode is placed against it.

**The criteria.** Two criteria are stated on the all-region mass under *some*. The **shift
criterion**, $q_H(\text{all}\mid\textit{some})<q_{\mathrm{lit}}(\text{all}\mid\textit{some})$, is
met when the hierarchy lowers the mass the entry and the base world prior already assign. The **position
criterion**, $q_H(\text{all}\mid\textit{some})<\tfrac12$, is met when the states other than *all*
together hold the majority. The position criterion can be met on literal conditioning by itself,
since the non-all states stand together against the single all state; the shift criterion can be
met while *all* remains the most probable single outcome. The reader, in accordance with their own judgement, may take satisfaction of the disjunction, the conjunction, either one of the two alone, or even neither criteria as scalar implicature. This notebook takes no position on how the criteria should be interpreted. Writing
$\Delta_{\textit{some}}=P(\text{all-region};q_{\textit{some}})-P(\text{all-region};q_{\mathrm{lit},\textit{some}})$
for the **shift**, the shift criterion is $\Delta_{\textit{some}}<0$. Both criteria are stipulations
about how such an effect could show up *in this model*, not a definition of scalar implicature, and
everything reported below is relative to them. Both are stated on the read-out $q$ of Eq. (12),
which is kept for comparison with the probabilistic accounts, and are called the **mass shift
criterion** and the **mass position criterion** where the read-out needs naming; the posterior of the
construction itself, the delta at the settled state, is reported beside them in Part D and Text
cell 4b, with a mode shift and a mode position criterion of its own. The two pairs are not assumed to agree, and they do not always: over the
plane of Text cell 6 the two position criteria agree in all $121$ cells and the two shift criteria
disagree in $35$. Eq. (37) is this shift, written out where Text cell 6 maps it.

**Under the default Gaussian base world prior the mass shift criterion is not met.** There
$\Delta_{\textit{some}}=+0.0008$ at the learned $\theta_u^\ast=-28.4375$: the settled belief holds
more all-region mass than the literal listener, so the mass shift criterion fails while the mass position
criterion holds comfortably ($q_H=0.0025$). The utility level's own contribution there is negative, $-0.0167$; what
carries the shift positive is the tempering, $+0.0175$, which is the shift's value at the
$\theta_u=0$ start. Ascending $\mathcal F$ in $\theta_u$ therefore moves the shift *toward* the
mass shift criterion without meeting it: $+0.0175$ at the start, $+0.0009$ at the endpoint of the learning
probe ($\theta_u=-7.567$ after $60$ updates, gradient $-0.627$ and still ascending), $+0.0008$ at
$\theta_u^\ast$, which the closed form supplies rather than the flow. **Across Part D's four base world priors,
all at $\Lambda=8$, the conjunction holds under none**: the mass position criterion holds under each and
the mass shift criterion under none. **With every base world prior at $\Lambda=512$ (Text cell 4b), it holds under three**:
the flat base world prior ($\Delta_{\textit{some}}=-0.0146$, $q_H=0.0357$), $\mathrm{Beta}(3,1)$ ($-0.0955$,
$0.0413$), and the delta-like base world prior concentrated on the all-region, a $\mathrm{Beta}(64,1)$
approaching a point mass at $s=1$, which needs that $\Lambda$
($-0.5217$, $0.4351$). The conjunction does not need those base world priors' whole $\theta_u^\ast$: it turns
on at $\lvert\theta_u\rvert=3.078$, $1.021$ and $2.126$ respectively and holds from there to
$\theta_u^\ast$, which is why the conjunction can be shown in the dynamics and not only in the closed
form (*Integration cost and conditioning* below). Under the delta-like base world prior the mode position
criterion and the mode conjunction of Part D turn on a little earlier, both at
$\lvert\theta_u\rvert=2.024$, and also hold from there to $\theta_u^\ast$; under Part D's four base world priors
at $\Lambda=8$ the mode position criterion already holds at $\theta_u=0$ and the mode conjunction
at no $\theta_u$ (Code Cells 2 and 2b print both). Text cell 6 sweeps the $\mathrm{Beta}(\alpha,1)$
family, which contains the flat and delta-like base world priors, against $\Lambda$ and finds a band of $33$
cells where both hold at once (at $K=101$; Appendix A prints the plane at other node counts); the flat and delta-like rows of Text cell 4b are two of them.

*Nor is the mass shift criterion, where it is met, specific to* some. Under the Gaussian base world prior at
$\Lambda=8$ it is met by *all* ($-0.4557$) and by neither *no* ($+0.0003$) nor *some*. The entry
for *all* does not hold there: $0.6260$ of the settled belief given *all* lies in the region that
entry excludes. At $\Lambda=512$, where every entry holds, *no* and *all* are saturated under all
five base world priors, holding $0.0000$ and $1.0000$ of the all-region at baseline, so their zero shifts are
saturation and not restraint; only *some* had room to move, and it moved under three of the five.
Whether the result for *all* at $\Lambda=8$ should be read together with that leak is a fact we
note, and our position on it is reserved.

Two things have to be stated about the rows where the mass shift criterion is met, because they are
commitments anyone reading that outcome as scalar implicature thereby takes on.

*Firstly, the shift is relative to $\ell_0$ and to the strength of the entry together.* At
Part D's $\Lambda=8$ the shift keeps one sign across the four base world priors and varies a hundredfold in
size, from $+0.0004$ to $+0.0421$. At $\Lambda=512$ its sign follows $\ell_0$: it is negative under
the three base world priors that put the most mass on the all-region, $0.0479$, $0.1367$ and $0.9568$ of it,
deepening with that mass from $-0.0146$ through $-0.0955$ to $-0.5217$, and positive under the two
that put the least, $+0.0071$ at $0.0016$ and $+0.0089$ at $0.0001$. Raising $\Lambda$, each
configuration at its own $\theta_u^\ast$, is what carries the flat and $\mathrm{Beta}(3,1)$ base world priors
across zero, from $+0.0295$ and $+0.0421$; it carries the Gaussian base world prior further from meeting the mass shift criterion,
from $+0.0008$ to $+0.0071$ at a learned $\theta_u^\ast=+1580.81$. So the effect is a property of
the entry for *some* at a given strength against a particular world belief, and among the five
base world priors, at a strength where every entry holds, it appears exactly where $\ell_0$ most favours
*all*. A reader who wants the shift to follow from the lexical scale alone cannot have
it here.

*Secondly, the mechanism is non-discriminative towards all utterances in the inventory.* Differencing the closed form
of Eq. (16) against its own $\theta_u=0$ case, which is the control defined above and not the
baseline, gives

$$\varphi_S^\ast(\theta_u)-\varphi_S^\ast(0)=\tfrac{\theta_u}{2}\,B\varphi_u^\ast(\theta_u),
\tag{23}$$

The identity is exact to $1.8\times10^{-15}$ at the default base world prior's $\theta_u^\ast$, and growing $\lvert\theta_u\rvert$ drives

$$\varphi_S^\ast\;\longrightarrow\;\tfrac12\big(I+BB^{\mathsf T}W\big)(\ell_0-\varphi_L),
\tag{24}$$

which is the $\theta_u=0$ field with its component in $\operatorname{span}B$ **doubled** and its
orthogonal complement left alone. The limit is not a loose bound: it predicts
$\Delta_{\textit{some}}=-0.5217$ under the delta-like base world prior and that base world prior's own
$\theta_u^\ast=+1407.77$ delivers $-0.5217$,
and $\Delta_{\textit{some}}<0$ holds there at every $\theta_u\neq0$ of either sign that Code Cell 2b sweeps,
from $10^{-8}$ to $10^{7}$ in magnitude: it tends to the tempering, $-0.0586$, as $\theta_u\to0$, and to the
limit as $\lvert\theta_u\rvert$ grows, overshooting it slightly on the negative side ($-0.5219$ at $\theta_u=-100$). So the whole of the effect is an amplification of the utterance's own projection onto a
two-dimensional utility basis. Under the delta-like base world prior those projections, $c_y=B^{\mathsf T}W(\ell_0-\varphi_L)$ in the notation
Appendix B carries, are $c_{\textit{some}}=[+961.66,-529.77]$ and $c_{\textit{all}}=[+961.66,+208.51]$.
The two columns of $B$ at $m=2$ (Text cell 3 §3, item 3) are an odd and an even profile in $\zeta$, so the two entries
are identical in the odd (tilt) coordinate and differ only in the even (width) one, which is the only
coordinate in which they differ at all and the entire source of the *some*/*all* asymmetry above. Doubling a negative width
coordinate lowers both tails, the all-region among them.

*What the mechanism is not.* Three properties about the architecture hold regardless of the sign of $\theta_u$.
$\varphi_L$ is a function of the observed utterance alone (Eqs. 5–6); $g_S$ reaches the world belief
only through $\varphi_u$, which is driven by $\varepsilon_S$, which is driven by $\varphi_L$
(Eqs. 10–11, 19); and no term in Eq. (13) involves any utterance other than the observed one. Nothing
in the chain of prediction represents that *all* was available and not chosen, and no quantity is compared against
an alternative at any point in an inference pass. Where the mass shift criterion is met it is therefore met *without* a
within-trial alternatives computation, by amplification of a single entry's own projection. (These three facts settle the *absence* of an alternatives computation and nothing
more: they are indifferent to the sign of the shift, and they do not say what the shift is made of.
That it is amplification of a low-rank component, and not shrinkage toward $g_S$, is the separate
measurement above.)

*Note that the above does not suggest the model predicts that alternative competition does not contribute to scalar implicature.* A phenomenon that holds under
one name (i.e. scalar implicature) is not thereby the consequence of one mechanism. The result here does not show that explicit within-trial competition among
alternatives is absent from human scalar implicature, and nothing in this implementation
could show that: the model has no such competition to begin with. What the implementation does establish
is narrower, and is about the mass shift criterion alone: **the movement of $q$-mass out of the
all-region need not be the consequence of within-trial alternative inference**, since an
architecture containing none produces it. At $\Lambda=512$ it produces it under the delta-like
base world prior, which gives the *all* reading the most support and is where standard RSA has the most to
overcome, and under the flat and $\mathrm{Beta}(3,1)$ base world priors; at Part D's $\Lambda=8$ it produces it
under none of the four. The mass position criterion is met alongside it in all three, $q_H$ holding
$0.0357$, $0.0413$ and $0.4351$ of the all-region against the half the criterion allows, and in all
three both criteria are also met off an integrated fixed point. Over the $121$ cells Text cell 6
sweeps, the conjunction holds at $33$, a band that contains the flat and delta-like rows. So the
conjunction is reached in this architecture without explicit competition among alternatives, under three of five
base world priors at a lexical strength where every entry holds and over a band of the plane, and not under
Part D's four base world priors at $\Lambda=8$. Regarding whether direct alternative competition also contributes, and in what proportion, the implementation at this phase makes no prediction about either way.

Appendix B's *Alternative Spaces* section takes up the further question of what would be the nature of an implicature reaching
the belief through the learned $\theta_u^\ast$, namely conventionalized by exposure rather than
computed. This nature is a separate commitment from the one this part discussed.

### <a id="tc4-partd" name="tc4-partd"></a>Part D
**D** repeats the demonstration and the implicature probe under four base world priors, all at
$\Lambda=8$: a Gaussian in $\zeta$ (equivalently, $s$ logit-normal), a literal
$\mathrm{Uniform}(0,1)$ over the proportion, and two skewed $\mathrm{Beta}$ base world priors. Text cell 4b
runs the same four, and a delta-like $\mathrm{Beta}(64,1)$ concentrated on the all-region, at
$\Lambda=512$. The two beliefs being compared, the $\theta_u$ control kept beside them, the
all-region and the two criteria are all as Part C defines them.

A base world prior specified over $s$ must be pushed forward to $\zeta$, multiplying by
$\mathrm ds/\mathrm d\zeta=s(1-s)$. For a $\mathrm{Beta}$ base world prior the Jacobian absorbs one factor
from each side and the log-density collapses to
$\alpha\log s+\beta\log(1-s)-\log B(\alpha,\beta)$, writing $\beta$ for the second shape parameter. The
exponents are $\alpha,\beta$ rather than the usual $\alpha-1,\beta-1$, and that shift *is* the Jacobian. Each base world prior is then renormalized
against the quadrature measure, since an analytically normalized density is not grid-normalized
once the grid is truncated. Because a base world prior flat in $s$ is not flat in $\zeta$, the flat base world prior
is $\mathrm{Beta}(1,1)$ and not a low-precision Gaussian: sending the Gaussian's precision to zero
inflates the variance in $\zeta$, which the sigmoid squashes into bimodal mass at the endpoints, which is the opposite of flat.

This push-forward runs in the **opposite direction** to the transform the figures apply. The base world prior
enters multiplied by $\mathrm ds/\mathrm d\zeta$; a belief is plotted as a density in $s$ by
dividing by it, per the remark at Eq. (25). Both figures draw the base world prior as a grey line, and
that the flat panel plots as a horizontal line at $1.0$ exercises the whole chain of push-forward,
renormalization and plotting transform in a single check.

$\ell_0$ is fixed, and it sits in $g_L$ (Eq. 9) rather than in $g_S$: it
reaches the situation and utility levels only through the utterance-independent combination
$\ell_0-\varphi_L$ of Eqs. (15)–(16). $\varphi_S^\ast$ is affine in it with a coefficient that does
not depend on the utterance, so **at a shared $\theta_u$** the $\varphi_S$ contrasts are exactly
$\ell_0$-invariant: across the four base world priors at a common $\Lambda$ they agree to
$5.3\times10^{-15}$. Under learning that invariance no longer transfers across the rows of the
sweep, because $\ell_0$ also moves $\theta_u^\ast$ through $c_y$, and the coefficient does depend on
$\theta_u$; at each base world prior's own $\theta_u^\ast$ the same contrasts differ by up to $0.058$. What is
invariant is the map at fixed $\theta_u$, not the row of the table. The sweep is accordingly a
semantic test rather than a numerical one: Eq. (21) does not involve $\ell_0$ any more than it
involves $\varphi_L$, so every result in Sec. 8 holds unchanged and the closed forms stay exact
under each base world prior.

Two of the four rows are integrated, the Gaussian and $\mathrm{Beta}(1,3)$, and there the settled
state agrees with Eqs. (15)–(16) to $4.5\times10^{-9}$ and $1.2\times10^{-9}$. Those are the
tolerances those runs stop at, not limits of the closed form, and they differ between the rows
because the stopping tolerance is keyed to $\lambda_{\max}(H)$ rather than fixed (I3). The other two
carry $\theta_u^\ast$ of $5950.63$ and $55.01$, which puts $\lambda_{\max}(H)$ past what
the timescale commitment allows the evaluation to integrate, and are read from the closed form; the
table says which is which, and *Integration cost and conditioning* below says what that costs and
how much of it the criteria actually need.

**How far the baseline and the control separate.** They differ by temperature alone, and the sweep
shows how wide that is where the base world prior is diffuse: under *some* the all-region mass of
$q_{\mathrm{lit}}$ runs $0.0016$, $0.0504$, $0.0001$ and $0.1368$ down the four rows below, against
$0.0191$, $0.1357$, $0.0066$ and $0.2356$ for the tempered belief, a factor of twelve at the
Gaussian row. Tempering flattens, and a flatter belief holds more mass at the ends of the scale,
which is where the all-region sits.

What the sweep establishes is that $\Delta_{\textit{some}}$ depends on $\ell_0$ even where its sign does not: $+0.0008$ (Gaussian), $+0.0295$ (flat), $+0.0004$ ($\mathrm{Beta}(1,3)$) and $+0.0421$
($\mathrm{Beta}(3,1)$), each at that base world prior's own $\theta_u^\ast$, a hundredfold range of one sign.
Text cell 4b has the case in which the sign follows $\ell_0$ as well, and Part C records both as
the first of its two commitments. One might expect the shift to be the same under every base world prior, on
the ground that a quantity acting identically on every utterance can move no contrast between them. The premise is right and the inference does not
follow twice over: $\ell_0$-invariance is a property of the $\varphi_S$ contrasts at a fixed
$\theta_u$, which $\ell_0$ itself moves under learning, and $\Delta_y$ is not one of those contrasts
in the first place. It is a difference of $q$-masses (Eq. 27) taken after the softmax of
Eq. (12), and the *Reporting statistics* note below is the general form of the point: a statistic nonlinear in $\varphi_S$ need not inherit an invariance the $\varphi_S$ contrasts have. This is the
same failure mode Text cell 5 Part B §4 argues about under $\mu_u$, running in the opposite
direction: there an invariant contrast produces a drifting statistic, here a hundredfold spread,
and in Text cell 4b a change of sign.

**The spike near $s=1$ under the flat and $\mathrm{Beta}(3,1)$ base world priors.** Both $q_{\mathrm{lit}}$

**The delta read-out.** The figures plot $q$. After them, Code Cell 2 prints the posterior the
construction itself returns, the delta at the settled state (Text cell 3 §4, item 5): one row for

Two columns need care. The mode is the one column the two read-outs share: the exponential and

It is reported below in $s$, while $\theta_L$ is a threshold in $\zeta$; and since the mode belongs to $\varphi_S^\ast$ rather than to either read-out, whatever moves it moves both of them. and $-856.51$.

**Where the mode sits against the cell of *all*.** The cell of *all* is Part C's all-region,

**The delta read-out's two criteria.** Under *some*, Code Cell 2 also prints two criteria on the mode of $\varphi_S^\ast$, the node where it is largest. Writing $k^\ast$ for the grid node where
$\varphi_S^\ast$ is largest and $k_0$ for the node where $\ell_0$ is, the **mode shift criterion** is

$$k^\ast<k_0,$$

met when the mode lies below the mode of $\ell_0$ on the scale, and the **mode position criterion** is

$$\zeta_{k^\ast}<\theta_L,$$

met when the mode lies outside the cell of *all*. Both are statistics of the delta read-out, as the
mass criteria are of $q$, and neither is a model quantity.

**Why the mode is read over $\zeta$ and not over $s$.** A mode depends on the coordinate it is read
in, since a change of variable multiplies a density by its Jacobian (MacKay, 1998, pp. 77, 79). $\varphi_S^\ast$
is a log-density against the quadrature measure of Eq. (2), which is uniform in $\zeta$, so the node
where it is largest is the mode of $q$ over $\zeta$. Read as a density over $s$, the same
distribution is $q(\zeta)/s(1-s)$, and the factor $1/s(1-s)$ grows without bound at both ends of the
scale. For the base world priors of this notebook that is decisive. A $\mathrm{Beta}(\alpha,\beta)$
density over $s$ has its mode inside the scale only when $\alpha>1$ and $\beta>1$: the flat base world prior has no mode over $s$, and $\mathrm{Beta}(1,3)$, $\mathrm{Beta}(3,1)$ and every $\mathrm{Beta}(\alpha,1)$ of Text cells 4b and 6 have their mode at an endpoint, which the logit coordinates exclude (Eq. A1). On the grid such a mode falls on the outermost node, whose place is set by the half-width of §1, so a mode of $\ell_0$ read over $s$ would locate the edge of the grid rather than anything about the base world prior, and the mode
shift criterion would compare against it. Over $\zeta$ every $\mathrm{Beta}(\alpha,\beta)$ with $\alpha,\beta>0$ has its mode inside the scale, at $\zeta=\log(\alpha/\beta)$, where $s$ equals the mean
$\alpha/(\alpha+\beta)$: this is the two-outcome case of MacKay's (1998, p. 80) result that in the softmax
basis the most probable point of a Dirichlet is its mean. Appendix A reports, under
*some*, whether the node the criteria read can be found by comparing neighbours alone. **The mode position criterion is where the
grid's half-width enters most directly**: §1's truncation moves the mode of $\varphi_S^\ast$, which
is $\zeta_{k^\ast}$, which is the criterion's own left-hand side. Being comparisons of grid nodes, they resolve
a movement of the mode to one node; a tie would go to the first node, and the cell prints the gap
between the two largest values of $\varphi_S^\ast$, which is never zero in these rows. The cell
prints the mode shift in $s$ and in grid steps, $k^\ast-k_0$. The grid is uniform in $\zeta$ while
$s$ compresses toward the ends of the scale, so the steps, multiplied by the grid spacing of $0.12$,
give the shift in $\zeta$. The mode criteria are the counterparts on the delta read-out of the mass
shift and mass position criteria and are not assumed to agree with them, so the cell prints the four
side by side, with how often each pair agrees. As with the mass criteria, this notebook takes no
position on how they should be interpreted. At $\Lambda=8$ the mode lies $3$, $6$, $3$ and $5$ grid steps above the mode of $\ell_0$ under the Gaussian, flat, $\mathrm{Beta}(1,3)$ and
$\mathrm{Beta}(3,1)$ base world priors, so the mode shift criterion is met under none and the mode position
criterion under all four, and each agrees with its q counterpart in all four rows.

The mode also leaves the cell of the uttered entry itself in two rows: under *all* with the
$\mathrm{Beta}(1,3)$ base world prior it sits at $s=0.2535$, and under *no* with the $\mathrm{Beta}(3,1)$ base world prior
at $s=0.7465$. They are the two rows whose largest leak is $0.91$, and Text cell 4b reports the same modes at a $\Lambda$ where every entry holds. We note both and reserve our position on them.

### <a id="tc4-reporting-statistics" name="tc4-reporting-statistics"></a>Reporting statistics

Every number quoted above, apart from the delta read-out of Part D, is a functional of Eq. (12).
None is a model quantity. The posterior of the construction is the delta at the settled state
(Text cell 3 §4, item 5), and $q$ is the read-out kept for comparison; each statistic below is a
summary of $q$, defined here so that what is being reported is unambiguous.

We call

$$\mathbb E[s]
=\int_0^1 s\,q_s(s)\,\mathrm ds
=\int_{\mathbb R}\varsigma(\zeta)\,q(\zeta)\,\mathrm d\zeta
\;\approx\;\sum_k w_k\,\varsigma(\zeta_k)\,q(\zeta_k)
\tag{25}$$

the **expected proportion** under $q$. The two integrals agree because
$q_s(s)\,\mathrm ds=q(\zeta)\,\mathrm d\zeta$. Note where the change of variables lands: for an
*expectation* it acts on the integrand, so Eq. (25) is computed directly on the $\zeta$-grid with
**no Jacobian factor**. This is not true of the *density* $q_s(s)=q(\zeta)/\big(s(1-s)\big)$, which
the figures plot and which does carry one. The two are easily confused, and applying the density's
correction to the expectation, or omitting it from the density, is the natural error.

We call

$$\mathrm{sd}(s)=\Big(\sum_k w_k\,q(\zeta_k)\big(\varsigma(\zeta_k)-\mathbb E[s]\big)^2\Big)^{1/2}
\tag{26}$$

the **dispersion** of $q$, reported on the scale of $s$.

For a measurable region $R$ on the scale, we call

$$P(R)=\sum_k w_k\,q(\zeta_k)\,\mathbb 1[\zeta_k\in R]
\tag{27}$$

the **$q$-mass over $R$**. It is a number in $[0,1]$, since Eq. (12) integrates to $1$ over the whole
grid. It is not the field $\varphi_S$, which is unnormalized log-weights and carries no such
interpretation; the exponential and the normalizer of Eq. (12) separate the two.

The qualifier $q$ is not decorative, because two other masses appear in this notebook, and neither of
them is the $q$-mass. The *quadrature mass* of a region is the measure the grid itself carries there, the weights
$w_k$ of Eq. (2), defined with no density in play, and it is what the endpoints $s=0,1$ lack. The *base world prior
mass* of a region integrates the base world prior $p_0$ over it, and is what the grid renormalization of
$\ell_0$ divides out. Only $q$-mass is a statement about the read-out $q$.

Three regions are used. $P(\text{all-region})$ takes $R=\{\zeta\ge\theta_L\}$, where *all* is true, and
$P(\text{no-region})$ takes $R=\{\zeta\le-\theta_L\}$. We call the $q$-mass over the region the observed
entry excludes, $\langle q,\chi_y\rangle$, the **leak**; it coincides with $P(E_y)$ for a step
$\chi_y$, is its soft-weighted analogue for a smooth one, and vanishes as $\Lambda\to\infty$ by
Eq. (15).

**These statistics are nonlinear in $\varphi_S$, so differences between them are not differences
in $\varphi_S$.** Eq. (25) composes a softmax with $\varsigma$; adding one and the same field to
two different beliefs therefore does not move their $\mathbb E[s]$ equally. Any claim about a
*contrast* between utterances must be checked against $\varphi_S$ directly. Where a reported
contrast in $\mathbb E[s]$ drifts while the underlying $\varphi_S$ contrast is provably invariant,
the nonlinearity is sufficient to account for the drift on its own. That is the situation under $\mu_u$, and the reason Text cell 5 Part B §4 argues against reading its contrast columns as an
effect of that parameter.

### <a id="tc4-integration-cost" name="tc4-integration-cost"></a>Integration cost and conditioning

§8.3 fixes the error units' time constant at
$\tau_\varepsilon\le\tau_\varphi/(4\lambda_{\max}(H))$ and the Euler step at
$\mathrm dt=\tau_\varepsilon/2$. Both scale with the learned parameter, so what the evaluation can
afford to run is itself a function of $\theta_u^\ast$, and that is worth recording here.

Explicit Euler on a linear system is stable only while $\mathrm{d}t\,\lambda<2$ for every eigenvalue.
The $\varphi_S$ mode relaxes at $1/\sigma_L+1/\sigma_S$, a constant. The $\varphi_u$ mode relaxes at

$$\frac{1}{\sigma_u}+\frac{\theta_u^{2}\,\lambda_{\max}(B^{\mathsf T}WB)}{\sigma_S},
\tag{28}$$

which is **quadratic in $\theta_u$**: $r_S$ sees $\varphi_u$ only through $\theta_u B$, so curvature
in that direction scales as $\theta_u^2$, and the table below is accordingly written on
$\lvert\theta_u\rvert$; $B$ is orthonormal, giving
$\lambda_{\max}(B^{\mathsf T}WB)=1$ exactly. Learning $\theta_u$ therefore stiffens the fast
subsystem as it proceeds:

| $\lvert\theta_u\rvert$ | 1 | 10 | 28.44 | 50 |
|---|---|---|---|---|
| $\varphi_u$ rate | 2.0 | 101 | **809.7** | 2501 |
| stiffness ratio | 1.0 | 50.5 | **404.8** | 1250.5 |
| max $\mathrm{d}t$ at $\tau_\varphi=1$ | 1.0 | 0.0198 | **0.0025** | 0.0008 |

At the stationary point $\theta_u^\ast=-28.4375$ the system is $\sim\!405\times$ stiff, and under
the commitment an inference there takes some $170\lambda_{\max}(H)$ Euler steps: $138{,}390$ to
$138{,}398$ of them at $\mathrm dt=1.54\times10^{-4}$, which is seconds rather than milliseconds on
the machine that ran this notebook. Step counts are quoted below rather than seconds wherever the
comparison has to survive a change of machine.

Where the integration is stopped is a choice, and it is worth two paragraphs because it was got
wrong twice, in the same way. Integration stops when the largest derivative stays below a tolerance
for ten consecutive steps. The derivative does not decay indefinitely: it reaches a floating-point
floor, where the residuals are differences of nearly equal numbers. A tolerance *inside* that floor
is crossed by roundoff rather than by the dynamics, and then the step count is not a property of the
trajectory at all: a change in $\theta_u$ of one part in $10^{15}$ moves it by several hundred, and
two arrangements of the same arithmetic disagree about it. The first mistake was a tolerance of
$10^{-10}$, which sat inside the floor. It was replaced by $10^{-9}$, which sat above it.

**The second mistake was to think the floor was a number.** It is not: it is proportional to
$\lambda_{\max}(H)$, because a derivative is a rate times a state error, the state error sits at
machine precision relative to the state, and the stiffest rate is $\lambda_{\max}(H)$. Code Cell 2b
measures it at each $\theta_u$ it integrates and divides out the rate: the floor per unit of
$\lambda_{\max}(H)$ comes out the same at all three, to the figures printed. A *fixed* tolerance
therefore keeps its margin only near the $\theta_u$ where it was chosen, and the cell prints the
counterfactual beside the measurement: what the superseded $10^{-9}$ would have bought at each of
those $\theta_u$, and the $\lambda_{\max}(H)$ at which it would have equalled the floor exactly.
At the largest of the three it is **below** the floor, where the test can never fire however long
the run and the integration ends at its step cap.

The tolerance is therefore keyed to the quantity the floor tracks,
$\texttt{DERIVATIVE\_TOLERANCE\_PER\_RATE}\times\lambda_{\max}(H)$, which holds the margin at
$12.1$ floors at every $\theta_u$ (the cell prints that ratio three times and gets $12.1$ each time), and
which makes the old $10^{-9}$ a consequence rather than a stipulation, being what the rule
returns at the $\lambda_{\max}(H)$ it was calibrated at. Nothing is smuggled in from outside the
system: $\lambda_{\max}(H)$ is what §8.3's own commitment
$\tau_\varepsilon\le\tau_\varphi/(4\lambda_{\max}(H))$ is written in terms of, and `learn_theta_u`
already re-derives $\mathrm dt$ and $\tau_\varepsilon$ from the current $\theta_u$ at every update,
which is also why the cost of the slow flow rises as it proceeds.

The settled state is correspondingly accurate to about the tolerance, and so to a quantity that
scales with $\theta_u$ rather than a constant: four orders below anything this evaluation reports
at either end of the range it runs over. Part A therefore cannot check that agreement against a
constant of its own; a fixed threshold would hold only near the $\theta_u$ it was set at, which is
the mistake above a second time. It checks the gap in **multiples of the tolerance actually used**
and reports the multiple: $1.00$, $1.00$, $0.04$ and $1.00$ on the four such claims, against a limit
of $2$.

**What is integrated, and what is not.** The evaluation integrates wherever
$\lambda_{\max}(H)\le3\times10^{3}$ and reads Eqs. (15)–(16) elsewhere, which are exact by §8.6, and
every table says which it did. That bound is a **wall-clock budget on the machine that ran this**
and carries no commitment: it marks which runs are worth their seconds, not which are possible.
On Part D's four base world priors it admits the Gaussian and $\mathrm{Beta}(1,3)$ rows at $\theta_u^\ast$ and
not the flat and $\mathrm{Beta}(3,1)$; on Text cell 6's plane, $2$ of $121$ cells. At Text cell 4b's
$\Lambda=512$ no row is integrated *at $\theta_u^\ast$*, where $\lambda_{\max}(H)$ runs from
$1.98\times10^{6}$ to $2.53\times10^{6}$, but every row is integrated **at the $\theta_u$ its flow
halts at**, which is the $\theta_u$ a realization reaches, and three of the five are carried end to
end there. Wherever both are available they agree to the stopping tolerance, so correctness is not
what is at stake. What is at stake is whether the dynamics are shown to reach those fixed points,
and at $\theta_u^\ast$ under the delta-like base world prior they are not: that inference would be
$3.98\times10^{8}$ steps, about $5.0$ hours.

**How much of $\theta_u^\ast$ the criteria need.** Much less, and Code Cell 2b measures it rather
than assuming it. At $\Lambda=512$ the conjunction of Part C turns on at
$\lvert\theta_u\rvert=3.078$, $1.021$ and $2.126$ under the flat, $\mathrm{Beta}(3,1)$ and
delta-like base world priors, where $\lambda_{\max}(H)$ is $11.5$, $3.0$ and $6.5$, and under each it holds
from there all the way to $\theta_u^\ast$;
across the $33$ cells of Text cell 6 where both mass criteria are met the threshold runs from $0.100$ to $4.250$, with
$\lambda_{\max}(H)$ between $2.0$ and $20.1$, so **every one of those cells is integrable at the
$\theta_u$ the conjunction needs**, against $3.5\times10^{4}$ to $3.6\times10^{7}$ at their
$\theta_u^\ast$. Eq. (20) started from $\theta_u(0)=0$ passes the threshold within one or two
updates.

That the conjunction arrives so early is a fact about **the shape of the update**, not about where
the flow stops, and the two are reported separately because they are different claims. Where the
flow stops is its halt: at $\Lambda=512$ and a tolerance of $1$, $\theta_u=54.429$, $52.065$ and
$34.695$ after $3$, $3$ and $7$ updates, an order of magnitude beyond the thresholds just quoted and
still two orders short of $\theta_u^\ast$. Code Cell 2b runs each of those three rows end to end
with nothing in closed form: the updates of Eq. (20) integrated, then Eqs. (18)–(19) integrated at
the $\theta_u$ they halt at, in $574{,}120$, $527{,}295$ and $242{,}163$ steps. Off those integrated
fixed points $\Delta_{\textit{some}}$ is $-0.0145$, $-0.0953$ and $-0.5208$ and $q_H$ is $0.0359$,
$0.0415$ and $0.4361$, both criteria met in each. So what Part C reports is demonstrated in the
dynamics, at the $\theta_u$ a realization would actually reach rather than at one located by the
evaluator's own criterion. What the closed form carries is the result *at $\theta_u^\ast$*, and the
expensive object in this model is the maximizer of $\tilde{\mathcal F}$, not the implicature.

**What the cost means, and where the question belongs.** The seconds the cell reports are wall clock
and stand in for two quantities that are not: the number of Euler steps an inference takes, and the separation
$4\lambda_{\max}(H)$ that §8.3 demands between the error and the state units, which is $12$ to $46$ at the
three thresholds, $4.8\times10^{3}$ to $1.5\times10^{4}$ where the flow halts, and
$7.9\times10^{6}$ to $9.3\times10^{6}$ at those base world priors' $\theta_u^\ast$. No end of that range
is a cost the process being modelled could plausibly pay. That is a question about how much
machinery a scalar implicature should need, and it is the argument for a further representation
level rather than a larger $\theta_u$; since no such level is implemented here, it is left where it
is posed, in the accompanying paper (§6.2), and not pursued in this notebook.

The point generalizes past the numerics. §8.1 establishes that $\nabla^2\mathcal F$ is negative
definite for **every** $\theta_u$. Definiteness says nothing about *conditioning*: it secures
convergence, while the condition number, of order
$(1+\theta_u^2)/2$ here and $404.8$ at $\theta_u^\ast$, governs whether convergence is reachable in
practice, and now sets $\tau_\varepsilon$ as well as $\mathrm dt$. (At $\sigma_L=\sigma_S=\sigma_u=1$ the $\varphi_S$ rate is $1/\sigma_L+1/\sigma_S=2$, so
that expression and the stiffness ratio above coincide; they are different quantities and would
separate at unequal $\sigma$.) Slow-parameter learning and fast-subsystem conditioning are coupled,
and §8 does not say so.

<a id="code2" name="code2"></a>

````python
# === Code Cell 2: EVALUATION, covering specification checks, demonstration,
#     and the semantic question ===
#
# Part A tests every claim the specification makes that is testable.
# Part B shows what the network does.
# Part C reports the mass shift and mass position criteria of Text cell 4 Part C, on which a
# reader may judge whether it produces a scalar implicature; no position is taken on them.
#
# Every network here carries its own LEARNED theta_u* (Eq. B2), re-learned for every
# configuration it is respawned into. A fixed theta_u appears only as a stated control.

import math
import time
import torch


# ============================================================================
# SETTLING: integrated where feasible, closed form elsewhere
# ============================================================================

# Under the timescale commitment (`fast_time_constant`), tau_error and dt scale as
# 1 / lambda_max(H), which grows as theta_u^2, and an integration takes about
# 180 lambda_max(H) Euler steps. Where theta_u* is large that is out of reach: the flat
# base world prior's theta_u* = 5951 puts lambda_max(H) at 3.5e7. The fixed point is not in doubt
# there, since Eqs. (15)-(16) are exact while phi_L is clamped (Sec. 8.6) and Part A
# checks `infer` against them wherever it runs. So the evaluation integrates where
# lambda_max(H) is at most FEASIBLE_STIFFNESS, reads the closed form elsewhere, and says
# which it did.
#
# BE HONEST ABOUT WHAT THIS CONSTANT IS. Calling it a time budget ("about 7 s per
# inference") is not the whole truth. Measured on this machine at Lambda = 512 (agent/audits/2026-09-21-tolerance-halting/integrability_output.txt):
#
#     theta_u    lambda      steps   steps/lambda   seconds
#       13.37     180.8      39,007        215.8       1.73
#       34.70   1,206.1     264,724        219.5      11.64
#       52.07   2,713.3  21,706,280       8000.0     945.50
#       71.09   5,055.8  40,446,305       8000.0    1751.29
#
# `infer` takes 219 lambda steps at a flat 45 us/step up to lambda = 1206, and at 2713 and
# above it takes EXACTLY 8000 lambda, which is `max_steps` = max_time / dt with
# max_time = 1000 and dt = tau_state / (8 lambda). Those runs did not get slower; they hit
# the cap and returned converged = False.
#
# WHY, diagnosed (why_it_stops_converging.py). It is NOT a slow mode and NOT an Euler
# instability. The dynamics settle just as fast in both cases: max|derivative| decays
# exponentially at rate 1 per time unit and reaches its floor by t = 37 at either theta_u,
# then sits there unchanged out to t = 166. What differs is the HEIGHT OF THE ROUNDOFF
# FLOOR, which the stopping test is compared against:
#
#     theta_u    13.37   20.00   34.70   42.00   46.90   52.07   61.27
#     floor    8.2e-11 1.8e-10 5.5e-10 8.0e-10 1.0e-09 1.2e-09 1.7e-09
#     /theta^2 4.60e-13 4.57e-13 4.56e-13 4.55e-13 4.55e-13 4.55e-13 4.55e-13
#
# The floor is 4.55e-13 * theta_u^2, and equivalently 4.547e-13 * lambda_max(H), which is
# the flatter of the two (a spread of 1.0003 against 1.0107 over a 4.6-fold range in
# theta_u). theta_u enters the roundoff twice: the residuals carry it, and `phi_u_dot`
# multiplies the state error by it again, which is why it looks quadratic in theta_u;
# max|phi_S| is about 856 at BOTH theta_u above, so this is not the state growing.
#
# THIS CONSTANT DOES NOT MARK THAT BOUNDARY. `infer`'s tolerance is keyed to the same
# quantity the floor tracks (`DERIVATIVE_TOLERANCE_PER_RATE`), so the margin above the
# floor is held fixed at every theta_u and the test fires wherever the dynamics have
# settled. The rows above converge under the scaled tolerance: theta_u = 52.07 in
# 527,368 steps (23 s) and 61.27 in 720,281 (32 s), at 194 and 192 steps per unit
# lambda: the same linear cost as everywhere else. The table is here as the evidence for
# the scaled tolerance, not as a boundary the model has.
#
# So this constant is a WALL-CLOCK BUDGET on this machine, carrying NO THEORETICAL
# COMMITMENT. Eqs. (15)-(16) give the fixed point exactly at every theta_u, Part A
# checks `infer` against them wherever it runs, and the only question this gate settles
# is which rows are worth the seconds. It is set to admit the theta_u that the halting
# flow of Eq. (20) actually reaches at Lambda = 512 (lambda up to 2965 at the flat
# base world prior), and to refuse theta_u*, whose lambda of about 2e6 would cost hundreds of
# hours.
FEASIBLE_STIFFNESS = 3.0e3

# How far past its own stopping tolerance an integration may land and still be called
# agreement with Eqs. (15)-(16). Every quantity the stopping test bounds is checked in
# UNITS OF THAT TOLERANCE rather than against a constant of its own, because the
# tolerance is DERIVATIVE_TOLERANCE_PER_RATE * lambda_max(H) and so grows with theta_u:
# a fixed threshold holds only while theta_u stays where it was calibrated. The
# thresholds these replace were 1e-8, which had 2.2x headroom at the evaluation
# network's theta_u* and FAILED outright at theta_u = 52 (1.49e-8), while carrying 64x
# of unused slack at theta_u = 5.3: loose where it worked and broken where it did not.
# Measured, the residual displacement lies in H's slowest direction, where the rate is
# 1, so the gap IS the tolerance: the ratio runs 0.95 to 1.00. The margin is 2 by the
# user's cap of 2026-09-21, which this clears by very nearly a factor of two.
TOLERANCE_MARGIN = 2.0

# The tolerance each demonstration halts at (A19), BY THE CASE'S Lambda. Eq. (20)'s flow
# stops when its own update falls below it, and that is the only stopping rule the model
# has. Both VALUES are AD HOC, and the rule itself is offered as a DIRECTION rather than a
# commitment (A19, demoted 2026-09-22): the fast loop's tolerance is keyed to lambda_max(H),
# an unsettled locality problem of the same kind as the timescale bound's (D12). A living
# system's tolerance need not match a computer's. What the model PREDICTS is theta_u*
# (Eq. B2), the asymptote of the same flow, which carries no tolerance at all.
#
# WHY IT IS KEYED TO Lambda. A fixed absolute tolerance does not buy the same thing at
# different lexical strengths: the gradient of Eq. (20) grows with Lambda, so 1e-1 halts
# the Lambda = 8 flow after 15 updates and the Lambda = 512 flow only after 61 to 173. The
# hypothesis that a case's representative tolerance scales with its Lambda is stated in the
# paper and is NOT tested here. These are two named demonstration values, not a fitted
# function: nothing here claims the dependence has this or any particular form, which is
# why an unlisted Lambda raises instead of interpolating.
#
# This is NOT the relative tolerance |d theta| < tol * |theta| that A19 rejects as a guard.
# Lambda is a standing property of the configuration, fixed before the flow starts; |theta|
# is where the trajectory has got to. Keying to the first is a stipulation about the case;
# keying to the second would be the system reading its own trajectory, which the halting
# commitment forbids.
DEMONSTRATION_TOLERANCE = {8.0: 1.0e-1, 512.0: 1.0}
# `learn_theta_u` raises if it reaches this without halting, which is the point: a cap is
# not a stopping rule, and a flow that hits it has not halted and must be reported so.
HALTING_CAP = 100_000


def demonstration_tolerance(net):
    """The ad hoc tolerance this notebook demonstrates `net`'s case at (A19).

    Keyed to the network's Lambda, and strict: a Lambda with no named value raises rather
    than falling back on one, because a silent default is exactly the ad hoc quantity a
    result must not inherit.
    """
    strength = float(net.lexical_strength)
    if strength not in DEMONSTRATION_TOLERANCE:
        raise KeyError(
            f"no demonstration tolerance is named for Lambda = {strength:g}; the named "
            f"cases are {sorted(DEMONSTRATION_TOLERANCE)}. Name one deliberately rather "
            "than interpolating: the dependence on Lambda is a hypothesis, not a fit."
        )
    return DEMONSTRATION_TOLERANCE[strength]


def describe_theta(net):
    """How a network's theta_u was set, for table headers."""
    kind = "learned, theta_u*" if net.theta_u_learned else "fixed control"
    return f"theta_u = {net.theta_u:.4f} ({kind})"


def count_in_words(n):
    """A small count spelled out, for printed sentences."""
    words = ("none", "one", "two", "three", "four", "five", "six", "seven", "eight", "nine")
    return words[n] if 0 <= n < len(words) else str(n)


def settle(net, utterance, theta_u=None):
    """The settled state for one utterance: integrated if feasible, else closed form.

    Returns phi_S, phi_u, the read-out q, whether it was integrated, and the step
    count (None for the closed form)."""
    theta = net.theta_u if theta_u is None else float(theta_u)
    if net.stiffest_state_rate(theta) <= FEASIBLE_STIFFNESS:
        result = net.infer(utterance, theta_u=theta, record_history=False)
        if not result["converged"]:
            raise RuntimeError(f"{utterance!r} did not settle at theta_u = {theta:.4g}")
        return {"phi_S": result["phi_S"], "phi_u": result["phi_u"], "q": result["q"],
                "integrated": True, "steps": result["steps"]}
    phi_S, phi_u = net.closed_form_fixed_point(utterance, theta_u=theta)
    return {"phi_S": phi_S, "phi_u": phi_u, "q": net.read_out(phi_S),
            "integrated": False, "steps": None}


# ============================================================================
# A. SPECIFICATION CHECKS
# ============================================================================

def check_specification(net, *, seed=0, verbose=True):
    """Test each testable claim of the specification. Returns {claim: (ok, detail)}.

    Run at the network's own theta_u (theta_u* for the evaluation network), with the
    time constants of the timescale commitment (`fast_time_constant`)."""
    torch.manual_seed(seed)
    results = {}

    # Sec. 6 / Sec. 8.6: `infer` must reproduce the closed forms exactly, since the
    # model is linear-Gaussian while phi_L is clamped.
    # Ratios, in units of each run's own stopping tolerance (TOLERANCE_MARGIN); the
    # `_abs` companions are carried only so the detail lines can report both.
    worst_state = worst_utility = 0.0
    worst_identity_1 = worst_identity_2 = 0.0
    worst_state_abs = worst_utility_abs = 0.0
    worst_identity_1_abs = worst_identity_2_abs = 0.0
    worst_monotone = math.inf
    worst_q_mass = 0.0
    runs = {}
    for utterance in UTTERANCES:
        result = net.infer(utterance)
        runs[utterance] = result
        if not result["converged"]:
            results["convergence"] = (False, f"{utterance} did not converge")
        phi_S_star, phi_u_star = net.closed_form_fixed_point(utterance)
        tolerance = result["derivative_tolerance"]
        state_gap = float((result["phi_S"] - phi_S_star).abs().max())
        utility_gap = float((result["phi_u"] - phi_u_star).abs().max())
        worst_state_abs = max(worst_state_abs, state_gap)
        worst_utility_abs = max(worst_utility_abs, utility_gap)
        worst_state = max(worst_state, state_gap / tolerance)
        worst_utility = max(worst_utility, utility_gap / tolerance)

        # Eq. (17): eps_S = -eps_L and eps_u = theta_u B^T W eps_S
        units = result["error_units"]
        identity_1 = float((units["state"] + units["lexical"]).abs().max())
        identity_2 = float(
            (units["utility"] - net.theta_u * net.project(units["state"])).abs().max()
        )
        worst_identity_1_abs = max(worst_identity_1_abs, identity_1)
        worst_identity_2_abs = max(worst_identity_2_abs, identity_2)
        worst_identity_1 = max(worst_identity_1, identity_1 / tolerance)
        worst_identity_2 = max(worst_identity_2, identity_2 / tolerance)

        # Sec. 8.2: F is non-decreasing along the flow
        trace = [entry["free_energy"] for entry in result["history"]]
        worst_monotone = min(
            worst_monotone, min(trace[i + 1] - trace[i] for i in range(len(trace) - 1))
        )

        # Eq. (12): the read-out integrates to 1
        worst_q_mass = max(
            worst_q_mass, abs(float((net.weights * result["q"]).sum()) - 1.0)
        )

    results.setdefault("convergence", (True, "all three utterances settled"))
    stopping = runs[UTTERANCES[0]]["derivative_tolerance"]
    results["fixed point matches Eqs. (15)-(16)"] = (
        max(worst_state, worst_utility) < TOLERANCE_MARGIN,
        f"max |phi - phi*| = {max(worst_state_abs, worst_utility_abs):.2e} = "
        f"{max(worst_state, worst_utility):.2f} x the stopping tolerance "
        f"{stopping:.2e} (limit {TOLERANCE_MARGIN:g}x)",
    )
    results["Eq. (17) eps_S = -eps_L"] = (
        worst_identity_1 < TOLERANCE_MARGIN,
        f"max residual = {worst_identity_1_abs:.2e} = {worst_identity_1:.2f} x the "
        f"stopping tolerance",
    )
    results["Eq. (17) eps_u = theta_u B^T W eps_S"] = (
        worst_identity_2 < TOLERANCE_MARGIN,
        f"max residual = {worst_identity_2_abs:.2e} = {worst_identity_2:.2f} x the "
        f"stopping tolerance",
    )
    results["Sec. 8.2  F non-decreasing"] = (
        worst_monotone > -1e-9,
        f"min step = {worst_monotone:+.2e} (roundoff at the fixed point), at "
        f"tau_error = {runs['some']['tau_error']:.3e} = tau_state / (4 lambda_max(H))",
    )
    results["Eq. (12) read-out normalized"] = (
        worst_q_mass < 1e-10, f"max |1 - q-mass| = {worst_q_mass:.2e}"
    )

    # Sec. 8.1/8.4: the Hessian is negative definite, and Sec. 8.1(ii) says so
    # for every theta_u and Lambda. Run at m = num_basis, so the form tested is
    # Eq. (22) of Sec. 8.4, which reduces to Eq. (21) at m = 1. The list of theta_u is
    # a sweep, since the claim is about every theta_u; it includes the network's own.
    worst_form = -math.inf
    for theta in [-50.0, -1.0, 0.0, 1.0, 50.0, net.theta_u]:
        for _ in range(400):
            v = torch.randn(net.num_nodes, dtype=net.dtype)
            t = torch.randn(net.num_basis, dtype=net.dtype)
            scale = math.sqrt(float(v.pow(2).sum() + t.pow(2).sum()))
            worst_form = max(
                worst_form,
                float(net.hessian_quadratic_form(v / scale, t / scale, theta_u=theta)),
            )
    results["Eq. (22) Hessian negative definite"] = (
        worst_form < 0.0, f"max over unit directions = {worst_form:+.3e}"
    )

    # Sec. 8.1(iii): phi_L enters r_L additively and so shifts the maximizer without
    # touching the Hessian. The testable consequence is about the RELAXATION RATE, not
    # the step count: phi_L moves the fixed point, so different utterances start at
    # different distances from it and reach an absolute tolerance at different times.
    # What must agree is the asymptotic rate, which is set by the eigenvalues of the
    # Hessian alone. It is read per unit time, as the least-squares slope of
    # log(max derivative) against time, since dt is set by the timescale commitment and
    # a per-step factor at small dt sits so close to 1 that a tolerance on it tests
    # nothing. The window is chosen by VALUE: from 1e6 down to 1e3 times the settling
    # tolerance, late enough for the faster modes to have died out and early enough to
    # stay above the roundoff floor, which 1/tau_error amplifies at large theta_u.
    relaxation = {}
    for utterance in UTTERANCES:
        trace = [entry["max_derivative"] for entry in runs[utterance]["history"]]
        dt_used = runs[utterance]["dt"]
        window = [(k * dt_used, math.log(value)) for k, value in enumerate(trace)
                  if 1e3 * 1e-10 <= value <= 1e6 * 1e-10]
        mean_t = sum(time for time, _ in window) / len(window)
        mean_v = sum(value for _, value in window) / len(window)
        slope = (sum((time - mean_t) * (value - mean_v) for time, value in window)
                 / sum((time - mean_t) ** 2 for time, _ in window))
        relaxation[utterance] = -slope
    rates = list(relaxation.values())
    spread = (max(rates) - min(rates)) / max(rates)
    # tolerance reflects the precision of a rate estimated off a discrete trace near
    # the roundoff floor (about 1e-6 relative at theta_u*), not the sharpness of the
    # claim; at theta_u = 1 the same estimate agrees to 1e-10.
    results["Sec. 8.1(iii) relaxation rate independent of phi_L"] = (
        spread < 1e-5,
        "rate per unit time " + ", ".join(f"{y}={r:.7f}" for y, r in relaxation.items())
        + f" (relative spread {spread:.1e})",
    )

    # Sec. 7: the analytic messages are the gradients of Eq. (13). Checked against
    # central differences at an off-equilibrium point, which is where a sign error
    # would show.
    utterance = "some"
    phi_L = net.lexical_field(utterance)
    settled = runs[utterance]
    phi_S = settled["phi_S"] + 0.3 * torch.randn_like(settled["phi_S"])
    phi_u = settled["phi_u"] + 0.3 * torch.randn_like(settled["phi_u"])
    r_lexical, r_state, r_utility = net.residuals(phi_S, phi_u, phi_L)
    step = 1e-6

    analytic_s = -(r_state / net.sigma_state) - (r_lexical / net.sigma_lexical)
    worst_gradient = 0.0
    for k in range(0, net.num_nodes, 17):
        up, down = phi_S.clone(), phi_S.clone()
        up[k] += step
        down[k] -= step
        numeric = (
            float(net.free_energy(up, phi_u, phi_L))
            - float(net.free_energy(down, phi_u, phi_L))
        ) / (2 * step * float(net.weights[k]))
        worst_gradient = max(
            worst_gradient,
            abs(numeric - float(analytic_s[k])) / (abs(float(analytic_s[k])) + 1e-9),
        )

    analytic_u = -(r_utility / net.sigma_utility) + net.theta_u * net.project(
        r_state / net.sigma_state
    )
    for k in range(net.num_basis):
        up, down = phi_u.clone(), phi_u.clone()
        up[k] += step
        down[k] -= step
        numeric = (
            float(net.free_energy(phi_S, up, phi_L))
            - float(net.free_energy(phi_S, down, phi_L))
        ) / (2 * step)
        worst_gradient = max(
            worst_gradient,
            abs(numeric - float(analytic_u[k])) / (abs(float(analytic_u[k])) + 1e-9),
        )

    analytic_theta = float(net.dot(phi_u, net.project(r_state / net.sigma_state)))
    numeric_theta = (
        float(net.free_energy(phi_S, phi_u, phi_L, theta_u=net.theta_u + step))
        - float(net.free_energy(phi_S, phi_u, phi_L, theta_u=net.theta_u - step))
    ) / (2 * step)
    worst_gradient = max(
        worst_gradient,
        abs(numeric_theta - analytic_theta) / (abs(analytic_theta) + 1e-9),
    )
    results["Sec. 7 messages are gradients of Eq. (13)"] = (
        worst_gradient < 1e-4, f"max relative error = {worst_gradient:.2e}"
    )

    # Sec. 1: the grid discretizes the integral, not the state, so the reported
    # belief must be stable under refinement. What that claims is CONVERGENCE in K,
    # which is a statement about the tail of the sequence, not about how close two
    # coarse grids happen to sit. Testing it as a 51-vs-401 gap against a fixed
    # tolerance tests the convergence RATE instead, and the rate here is set by
    # something the claim does not mention: chi_y is a step, and trapezoidal
    # quadrature on a discontinuous integrand converges at O(h) with the step's
    # position jittering against the nodes as K changes. The smooth mask of Sec. 9.2
    # is the control, being the same model with a continuous integrand, and is
    # reported beside it.
    def expectations(num_nodes, sharpness):
        probe = net.respawn(num_nodes=num_nodes, exclusion_sharpness=sharpness)
        return [
            float((probe.weights
                   * probe.read_out(probe.closed_form_fixed_point(y)[0])
                   * logistic(probe.zeta)).sum())
            for y in UTTERANCES
        ]

    ladder = [51, 101, 201, 401, 801]
    hard = {k: expectations(k, None) for k in ladder}
    smooth = {k: expectations(k, 0.1) for k in ladder}
    spread = max(abs(hard[401][i] - hard[51][i]) for i in range(len(UTTERANCES)))
    tail = max(abs(hard[801][i] - hard[401][i]) for i in range(len(UTTERANCES)))
    smooth_tail = max(abs(smooth[801][i] - smooth[401][i]) for i in range(len(UTTERANCES)))
    results["Sec. 1  K is an accuracy parameter"] = (
        tail < 5e-3,
        f"E[s] converging: |401-801| = {tail:.1e} (hard chi_y), {smooth_tail:.1e} "
        f"(smooth chi_y); coarse 51-401 spread {spread:.1e} is the O(h) step artefact",
    )

    # Sec. 1, the other half of the same claim. Refining K converges, which the check
    # above reports. Widening the half-width at FIXED spacing has nothing to converge to:
    # Eq. (1) carries the state onto the whole of R, so cutting it at +-Z bounds the state
    # space rather than discretizing an integral over it, and the two ladders are what
    # Text cell 3 Sec. 1 means by K and Z not being parameters of the same kind. This
    # reports the contrast and nothing more; what the half-width does to the four criteria
    # is Appendix A's own ladder, beside its ladder in n.
    def expectations_at(half_width, spacing=0.12):
        probe = net.respawn(grid_half_width=half_width,
                            num_nodes=int(round(2 * half_width / spacing)) + 1)
        return [
            float((probe.weights
                   * probe.read_out(probe.closed_form_fixed_point(y)[0])
                   * logistic(probe.zeta)).sum())
            for y in UTTERANCES
        ]

    half_widths = [5.0, 6.0, 7.0, 8.0, 10.0]
    widened = {Z: expectations_at(Z) for Z in half_widths}
    z_spread = max(max(row[i] for row in widened.values())
                   - min(row[i] for row in widened.values())
                   for i in range(len(UTTERANCES)))
    z_tail = max(abs(widened[half_widths[-1]][i] - widened[half_widths[-2]][i])
                 for i in range(len(UTTERANCES)))
    results["Sec. 1  Z is not an accuracy parameter"] = (
        z_spread > 10 * tail,
        f"E[s] not settling: spread over Z in [{half_widths[0]:g}, {half_widths[-1]:g}] at "
        f"fixed spacing = {z_spread:.1e}, last step = {z_tail:.1e}, against the K tail "
        f"{tail:.1e} above",
    )

    # Sec. 8.5: F~(theta_u) = max_x F(x, theta_u) must have an interior stationary
    # point, which is exactly what mu_u = 0 destroys: there F~ climbs monotonically to
    # an asymptote and theta_u diverges. The network reads it from Eq. (B2)
    # (`learned_theta_u`); it is located here a second, independent way, by bisection
    # on the reduced gradient, and the two must agree. The bisection loses accuracy
    # at |theta_u*| above a few thousand, where the gradient flattens; at this
    # network's theta_u* it is exact to roundoff. The dynamics are checked below.
    #
    # THE SEARCH MUST COVER BOTH SIGNS. theta_u is an unconstrained real parameter:
    # Bogacz's Eqs. (25) and (56) give a gradient and no domain, and the optimum of his
    # linear example, theta* = u / v_p, takes whichever sign the data give. Eliminating phi_S and phi_u
    # analytically (Sec. 8.6) reduces the objective to
    #
    #     F~(theta) = const - ||theta mu_u - c||^2 / (2 (sigma_u theta^2 + S)),
    #         S = sigma_L + sigma_S,        c = B^T W (ell_0 - phi_L),
    #
    # whose stationarity equation is the quadratic
    #
    #     sigma_u <mu_u,c> theta^2 + (S ||mu_u||^2 - sigma_u ||c||^2) theta
    #                                                        - S <mu_u,c> = 0.
    #
    # Its roots have product -S / sigma_u < 0, so whenever <mu_u,c> != 0 there is
    # exactly ONE root of each sign; the maximizer is the one with sign(theta*) =
    # sign(<mu_u,c>) and the other root is a minimum. So the half-line to search is
    # read off dF~/dtheta at zero, which equals <mu_u,c> / S, a positive multiple of
    # the same inner product, rather than assumed. Bracketing on theta_u > 0 alone
    # is a restriction Bogacz does not impose, and under g_S = theta_u B phi_u (with
    # ell_0 in g_L, Eq. 9) the maximizer is negative, so such a bracket reports a
    # divergence that is an artefact of the bracket and not a property of the model.
    slow = net.respawn()
    lexical = {u: slow.lexical_field(u) for u in UTTERANCES}

    def reduced_energy(theta):
        return sum(
            float(slow.free_energy(
                *slow.closed_form_fixed_point(u, theta_u=theta), lexical[u], theta_u=theta))
            for u in UTTERANCES
        ) / len(UTTERANCES)

    def reduced_gradient(theta):
        total = 0.0
        for u in UTTERANCES:
            phi_S, phi_u = slow.closed_form_fixed_point(u, theta_u=theta)
            _, r_state, _ = slow.residuals(phi_S, phi_u, lexical[u], theta_u=theta)
            total += float(slow.dot(phi_u, slow.project(r_state / slow.sigma_state)))
        return total / len(UTTERANCES)

    ascent_at_zero = reduced_gradient(0.0)
    if ascent_at_zero == 0.0:
        raise RuntimeError(
            "dF~/dtheta_u = 0 at theta_u = 0, i.e. <mu_u, c> = 0: the quadratic above "
            "degenerates and no interior maximizer exists on either half-line"
        )
    sign = 1.0 if ascent_at_zero > 0 else -1.0

    low, high = 1e-3, 1e6                      # MAGNITUDES; the sign is fixed above
    for _ in range(120):                       # geometric bisection on the gradient
        mid = math.sqrt(low * high)
        if sign * reduced_gradient(sign * mid) > 0:
            low = mid
        else:
            high = mid
    theta_star = sign * math.sqrt(low * high)
    closed_form_star = slow.learned_theta_u()
    agreement = abs(theta_star - closed_form_star) / abs(closed_form_star)
    results["Sec. 8.5  theta_u* of Eq. (B2) matches an independent bisection"] = (
        agreement < 1e-8 and (not net.theta_u_learned or net.theta_u == closed_form_star),
        f"closed form {closed_form_star:.10f}, bisection {theta_star:.10f}, "
        f"relative gap {agreement:.1e}",
    )
    margin = reduced_energy(theta_star) - reduced_energy(sign * 1e8)
    interior = (
        reduced_energy(theta_star) > reduced_energy(0.5 * theta_star)
        and reduced_energy(theta_star) > reduced_energy(2.0 * theta_star)
        and margin > 0
    )
    results["Sec. 8.5  F~ has an interior maximum in theta_u"] = (
        interior and abs(reduced_gradient(theta_star)) < 1e-6,
        f"theta_u* = {theta_star:.4f}, |dF~/dtheta_u| = {abs(reduced_gradient(theta_star)):.1e}, "
        f"margin over the asymptote = {margin:+.4f}",
    )

    # and the flow of Eq. (20) must move toward it, with F~ monotone. It starts where
    # every configuration's flow starts, theta_u(0) = 0 (`theta_u_initial`), where the
    # network is the tempered control. The roots of the quadratic above have opposite
    # signs, so zero lies between the minimum and the maximizer, on the maximizer's
    # rising side: the gradient there already points at theta_u*, and no sign has to
    # be chosen. The flow halts when its own update falls below the tolerance (A19), and
    # it is slow near theta_u*, where F~ is shallow, so what the halted run can show is
    # direction, monotone F~ and a shrinking gradient, not arrival at theta_u*.
    # The shrinking-gradient clause needs at least two updates, so a tolerance coarse
    # enough to halt after one fails this check rather than weakening it silently.
    dynamic = net.respawn()
    tolerance = demonstration_tolerance(dynamic)
    trace = dynamic.learn_theta_u(UTTERANCES, tolerance=tolerance,
                                  tau_theta=20.0, num_updates=HALTING_CAP)
    energies = [entry["free_energy"] for entry in trace]
    worst_step = min(energies[i + 1] - energies[i] for i in range(len(energies) - 1))
    approaching = (
        trace[0]["theta_u"] == 0.0
        and abs(dynamic.theta_u - theta_star) < abs(trace[0]["theta_u"] - theta_star)
        and dynamic.theta_u * theta_star > 0
        and abs(trace[-1]["gradient"]) < abs(trace[0]["gradient"])
    )
    results["Sec. 8.5  Eq. (20) ascends toward it from theta_u = 0"] = (
        approaching and worst_step > -1e-9,
        f"theta_u {trace[0]['theta_u']:.3f} -> {dynamic.theta_u:.3f} (target {theta_star:.3f}) "
        f"in {len(trace)} updates, halted at tolerance {tolerance:g}, "
        f"|grad| {abs(trace[0]['gradient']):.3g} -> {abs(trace[-1]['gradient']):.3g}, "
        f"min dF~ step {worst_step:+.1e}",
    )

    # Sec. 9.2: a dense-scale (smooth) lexical field changes nothing structural.
    # Each probe has its OWN theta_u*, hence its own lambda_max(H) and its own stopping
    # tolerance, so the gap is accumulated in units of that tolerance and not in absolute
    # terms; otherwise the sharpest probe's tolerance would set the bar for all three.
    worst_dense = worst_dense_abs = 0.0
    for sharpness in [0.5, 1.0, 2.0]:
        probe = net.respawn(exclusion_sharpness=sharpness)       # its own theta_u*
        for y in UTTERANCES:
            result = probe.infer(y, record_history=False)
            phi_S_star, _ = probe.closed_form_fixed_point(y)
            gap = float((result["phi_S"] - phi_S_star).abs().max())
            worst_dense_abs = max(worst_dense_abs, gap)
            worst_dense = max(worst_dense, gap / result["derivative_tolerance"])
    results["Sec. 9.2 smooth lexical field is compatible"] = (
        worst_dense < TOLERANCE_MARGIN,
        f"max |phi_S - phi_S*| = {worst_dense_abs:.2e} = {worst_dense:.2f} x each "
        f"probe's own stopping tolerance (limit {TOLERANCE_MARGIN:g}x)",
    )

    if verbose:
        width = max(len(k) for k in results)
        print("SPECIFICATION CHECKS")
        print("=" * (width + 46))
        for claim, (ok, detail) in results.items():
            print(f"  {'PASS' if ok else 'FAIL'}  {claim:<{width}}  {detail}")
        print("=" * (width + 46))
        failures = sum(1 for ok, _ in results.values() if not ok)
        print(f"  {len(results) - failures}/{len(results)} passed")
        print()

    return results


# ============================================================================
# B. DEMONSTRATION
# ============================================================================

def demonstrate(net, *, verbose=True):
    """Run the network on each utterance and report the settled belief.

    Integrated where feasible (`settle`); the steps column reports the integration,
    and reads "closed" where the closed form was used instead."""
    proportion = logistic(net.zeta)
    # The all-region of Eq. (27). The threshold of Eq. (A5) makes it the same mask as
    # Appendix A's cell of "all", which `delta_readout_report` builds as `inside`; it is
    # kept as its own variable so that either definition can be varied on its own.
    upper_region = (net.zeta >= net.theta_L).to(net.dtype)
    lower_region = (net.zeta <= -net.theta_L).to(net.dtype)

    table = {}
    for utterance in UTTERANCES:
        result = settle(net, utterance)
        q = result["q"]
        excluded = exclusion_indicator(
            utterance, net.zeta, net.theta_L, sharpness=net.exclusion_sharpness
        )
        table[utterance] = {
            "q": q,
            "expected_s": float((net.weights * q * proportion).sum()),
            "leak": float((net.weights * q * excluded).sum()),
            "upper_q_mass": float((net.weights * q * upper_region).sum()),
            "lower_q_mass": float((net.weights * q * lower_region).sum()),
            "phi_u": result["phi_u"].tolist(),
            "steps": result["steps"],
        }

    if verbose:
        print("DEMONSTRATION")
        print(f"  Lambda = {net.lexical_strength}, {describe_theta(net)}, "
              f"mu_u = {net.mu_u.tolist()}, boundary = {net.theta_L}")
        print()
        print(f"  {'utterance':<10}{'E[s]':>9}{'P(s~0)':>10}{'P(s~1)':>10}"
              f"{'leak':>10}{'steps':>9}   phi_u (tilt, width)")
        for utterance, row in table.items():
            steps = "closed" if row["steps"] is None else str(row["steps"])
            print(f"  {utterance:<10}{row['expected_s']:>9.4f}{row['lower_q_mass']:>10.4f}"
                  f"{row['upper_q_mass']:>10.4f}{row['leak']:>10.4f}{steps:>9}"
                  f"   [{row['phi_u'][0]:+.3f}, {row['phi_u'][1]:+.3f}]")
        print()
        print("  'leak' is the q-mass over the region the entry excludes;")
        print("  it goes to zero as Lambda grows, which is the Eq. (15) limit of Sec. 6.")
        print("  'steps' are Euler steps at the time constants of the timescale commitment,")
        print(f"  dt = tau_error / 2 = {0.5 * net.fast_time_constant():.2e} here.")
        print()

    return table


def lexical_strength_sweep(net, strengths=(1, 2, 4, 8, 16, 32, 64)):
    """Sec. 6: a hard support constraint is the Lambda -> infinity
    limit of Eq. (15), not a separate assumption. Leak should fall monotonically to 0.

    Each Lambda is its own configuration and carries its own theta_u*, which runs to
    1535 at Lambda = 32, so the fixed points are read in closed form (exact by
    Sec. 8.6) at every row rather than integrated at some rows and not others."""
    rows = []
    for strength in strengths:
        probe = net.respawn(lexical_strength=float(strength))
        leaks = []
        for utterance in UTTERANCES:
            q = probe.read_out(probe.closed_form_fixed_point(utterance)[0])
            excluded = exclusion_indicator(
                utterance, probe.zeta, probe.theta_L,
                sharpness=probe.exclusion_sharpness,
            )
            leaks.append(float((probe.weights * q * excluded).sum()))
        rows.append((float(strength), max(leaks), probe.theta_u))

    print("LAMBDA -> INFINITY RECOVERS THE HARD MASK (Sec. 6)")
    print("  (closed form, each Lambda at its own learned theta_u*)")
    print(f"  {'Lambda':>8}{'theta_u*':>12}{'max leak':>12}")
    for strength, leak, star in rows:
        print(f"  {strength:>8.0f}{star:>12.3f}{leak:>12.6f}")
    monotone = all(rows[i + 1][1] <= rows[i][1] + 1e-12 for i in range(len(rows) - 1))
    print(f"  monotone decreasing: {monotone};  leak at Lambda={rows[-1][0]:.0f} "
          f"is {rows[-1][1]:.2e}")
    print()
    return rows


# ============================================================================
# C. THE SEMANTIC QUESTION: IS THERE A SCALAR IMPLICATURE?
# ============================================================================

def scalar_implicature_probe(net, *, verbose=True):
    """Compare the settled belief against the literal listener.

    The baseline is q_lit, the untempered literal posterior of
    `literal_fixed_point`: phi_S = ell_0 - phi_L, the base world prior restricted by the
    lexical entry and nothing else. The shift is measured against it.

    The theta_u = 0 belief is reported beside it as a secondary control. It is the
    same literal posterior tempered by one half (see `literal_fixed_point`), so it
    holds the temperature of the full network fixed while severing the utility level;
    q_lit holds the definition fixed instead. The two differ, and Text cell 4 Part D
    states by how much.

    The measured quantity is the change in P(s in all-region) between q_lit and the
    full network, the shift of Eq. (37). Text cell 4 Part C reports two criteria on it
    and takes no position on them (B10): the mass shift criterion, that the shift for
    "some" is negative, and the mass position criterion, that the full network's
    P(all-region | "some") is below 1/2.
    """
    upper_region = (net.zeta >= net.theta_L).to(net.dtype)
    proportion = logistic(net.zeta)

    rows = {}
    for utterance in UTTERANCES:
        baseline = net.read_out(net.literal_fixed_point(utterance))
        tempered = net.infer(utterance, theta_u=0.0, record_history=False)["q"]
        full = settle(net, utterance)["q"]
        rows[utterance] = {
            "tempered_upper": float((net.weights * tempered * upper_region).sum()),
            "baseline_upper": float((net.weights * baseline * upper_region).sum()),
            "full_upper": float((net.weights * full * upper_region).sum()),
            "baseline_expected_s": float((net.weights * baseline * proportion).sum()),
            "full_expected_s": float((net.weights * full * proportion).sum()),
            "baseline_q": baseline,
            "full_q": full,
        }
        rows[utterance]["shift"] = (
            rows[utterance]["full_upper"] - rows[utterance]["baseline_upper"]
        )

    if verbose:
        print("SCALAR IMPLICATURE PROBE")
        print(f"  full network: {describe_theta(net)}")
        print("  baseline: q_lit, the untempered literal posterior (phi_S = ell_0 - phi_L)")
        print("  also shown: the theta_u = 0 control, that posterior tempered by 1/2")
        print()
        print(f"  {'utterance':<10}{'E[s] base':>11}{'E[s] full':>11}"
              f"{'P(all) base':>13}{'P(all) full':>13}{'shift':>10}"
              f"{'P(all) th_u=0':>15}")
        for utterance, row in rows.items():
            print(f"  {utterance:<10}{row['baseline_expected_s']:>11.4f}"
                  f"{row['full_expected_s']:>11.4f}{row['baseline_upper']:>13.4f}"
                  f"{row['full_upper']:>13.4f}{row['shift']:>+10.4f}"
                  f"{row['tempered_upper']:>15.4f}")
        print()
        shift = rows["some"]["shift"]
        upper = rows["some"]["full_upper"]
        first, second = shift < 0, upper < 0.5
        print(f"  mass shift criterion, shift for \"some\" < 0:              "
              f"{'met' if first else 'NOT met'} ({shift:+.4f})")
        print(f"  mass position criterion, P(all-region | \"some\") < 1/2:    "
              f"{'met' if second else 'NOT met'} ({upper:.4f})")
        print(f"  both mass criteria: {'met' if first and second else 'NOT met'}; "
              f"under \"some\" the settled belief holds {abs(shift):.4f} "
              f"{'less' if shift < 0 else 'MORE'} all-region q-mass than q_lit.")
        print("  Whether the criteria are met is a measurement, and this notebook takes no")
        print("  position on how they should be interpreted (Text cell 4 Part C). Note that")
        print("  the mass shift criterion is met by other utterances too (see the shift column")
        print("  above), and that the shift's sign is ell_0-dependent (Code Cell 2b).")
        print()

    return rows


def theta_u_learning_probe(net, *, tau_theta=20.0, tolerance=None):
    """Sec. 8.5: integrate Eq. (20), tau_theta * d(theta_u)/dt = dF/d(theta_u).

    There is no learning rate: tau_theta is a time constant of the same kind as
    tau_error and tau_state, and the model's one timescale ordering runs
    tau_error <= tau_state / (4 lambda_max(H)) and tau_state << tau_theta. It sets the
    rate, not the fixed point.

    The flow starts at theta_u(0) = 0 (`theta_u_initial`), the one start every
    configuration shares. There the network is the TEMPERED control, not the literal
    listener, which would also need sigma_S -> infinity. The gradient at zero is
    <mu_u, sum_y c_y> / S, nonzero, and has the sign of theta_u* (Appendix B), so the
    flow leaves zero toward it without any sign being chosen.

    The flow halts when its own update falls below `tolerance` (A19), the only stopping
    rule the model has. The value is ad hoc and is printed with every number it fixes;
    theta_u*, which this block prints beside it, is the tolerance-free prediction.

    Requires <mu_u, sum_y c_y> != 0 (Appendix B); mu_u != 0 (Sec. 3) is necessary but
    not sufficient. Otherwise F~ has no stationary point and theta_u diverges, which
    is Bogacz's Eq. (25) at v_p = 0, not a departure from it.
    """
    net = net.respawn()
    tolerance = demonstration_tolerance(net) if tolerance is None else float(tolerance)
    star = net.learned_theta_u()
    trace = net.learn_theta_u(UTTERANCES, tau_theta=tau_theta, tolerance=tolerance,
                              num_updates=HALTING_CAP)
    energies = [entry["free_energy"] for entry in trace]
    monotone = all(
        energies[i + 1] >= energies[i] - 1e-9 for i in range(len(energies) - 1)
    )
    upper = (net.zeta >= net.theta_L).to(net.dtype)
    literal_upper = float(
        (net.weights * net.read_out(net.literal_fixed_point("some")) * upper).sum())

    def shift_for_some(theta):
        belief = net.read_out(net.closed_form_fixed_point("some", theta_u=theta)[0])
        return float((net.weights * belief * upper).sum()) - literal_upper

    at_start, at_end, at_star = (shift_for_some(theta)
                                 for theta in (trace[0]["theta_u"], net.theta_u, star))
    print("SLOW PARAMETER (Eq. 20, Sec. 8.5)")
    print(f"  mu_u = {net.mu_u.tolist()}  (must be nonzero)")
    print(f"  start: theta_u(0) = {trace[0]['theta_u']:.4f}, the tempered control; "
          f"dF~/dtheta_u there = {trace[0]['gradient']:+.4f}")
    print(f"  theta_u: {trace[0]['theta_u']:.4f} -> {net.theta_u:.6f}, halted after "
          f"{len(trace)} update(s) at tau_theta = {tau_theta:g};  "
          f"theta_u* = {star:.4f} (Eq. B2)")
    print(f"  Halting rule: the flow stops when its own update falls below the tolerance,"
          f" here {tolerance:g}, which is an AD HOC value.")
    print(f"  theta_u* = {star:.4f} above is the closed form of Eq. (B2), computed without"
          f" a tolerance.")
    print(f"  F~ increased monotonically: {monotone}")
    print(f"  final dF~/dtheta_u = {trace[-1]['gradient']:+.3e}; the update it produced is "
          f"what fell below the tolerance")
    print("  The flow is slow near the optimum because F~ is shallow there, so it halts far")
    print("  short of it; the target is the stationary point Part A locates in closed form,")
    print("  not this endpoint.")
    print()
    print(f"  shift for \"some\" (Eq. 37, against q_lit): {at_start:+.4f} at the start, "
          f"{at_end:+.4f} at the endpoint, {at_star:+.4f} at theta_u*")
    toward = at_star < at_start
    reached = at_star < 0
    print(f"  Ascending F~ moves the shift {'DOWN, toward' if toward else 'UP, away from'} "
          f"the mass shift criterion of Text cell 4 Part C, and "
          f"{'crosses zero' if reached else 'does not reach it'} at theta_u*.")
    print("  Eq. (13) contains no term rewarding informativity, so this is not F pursuing")
    print("  the mass shift criterion: raising |theta_u| amplifies the span(B) component of")
    print("  (ell_0 - phi_L), and the sign of the shift is set by where those directions")
    print("  put mass relative to the all-region. See `theta_u_dependence` below.")
    print()
    return net, trace


def theta_u_stationary_point(net):
    """theta_u* for this configuration, in closed form (Eq. B2).

    Delegates to the network (`learned_theta_u`), so it holds whether the network
    carries theta_u* or a fixed control value. Part A cross-checks it against an
    independent bisection on the reduced gradient."""
    return net.learned_theta_u()


def theta_u_dependence(net, *, verbose=True):
    """Every headline number in this cell, recomputed across theta_u: a CONTROL.

    The rest of the evaluation is reported at the learned theta_u*, which is NEGATIVE
    here (Appendix B). This probe holds theta_u fixed at chosen values so that the
    dependence of each number on theta_u, and on its sign, is visible rather than
    assumed. Four columns: the tempered control (theta_u = 0, which is also where the
    flow of Eq. (20) starts), +1 and -1, and the learned theta_u*.

    The shift row is Eq. (37), against q_lit, exactly as everywhere else in the
    notebook; the tempered-control column therefore shows what the theta_u = 0
    tempering does on its own rather than a column of zeros.

    Read it for two things. Which quantities depend on theta_u at all, and which of
    those depend on its SIGN: the utility level reaches phi_S through theta_u * B and
    the fixed point through theta_u^2 * G, so quantities dominated by the quadratic
    term are near-identical at +theta and -theta while the mu_u-carried linear term
    is what breaks the symmetry.
    """
    star = theta_u_stationary_point(net)
    settings = [("tempered (0)", 0.0), ("+1", 1.0),
                ("-1", -1.0), ("learned", star)]
    proportion = logistic(net.zeta)
    upper = (net.zeta >= net.theta_L).to(net.dtype)

    table = {}
    for label, theta in settings:
        column = {}
        for y in UTTERANCES:
            field = net.closed_form_fixed_point(y, theta_u=theta)[0]
            belief = net.read_out(field)
            base = net.read_out(net.literal_fixed_point(y))
            excluded = exclusion_indicator(
                y, net.zeta, net.theta_L, sharpness=net.exclusion_sharpness
            )
            column[y] = {
                "expected_s": float((net.weights * belief * proportion).sum()),
                "upper": float((net.weights * belief * upper).sum()),
                "leak": float((net.weights * belief * excluded).sum()),
                "shift": float((net.weights * belief * upper).sum())
                         - float((net.weights * base * upper).sum()),
            }
        contrast = (net.closed_form_fixed_point("all", theta_u=theta)[0]
                    - net.closed_form_fixed_point("some", theta_u=theta)[0])
        column["contrast"] = float(net.squared_norm(contrast)) ** 0.5
        table[label] = column

    if verbose:
        print("THETA_U DEPENDENCE OF THE REPORTED QUANTITIES")
        print(f"  theta_u* = {star:.4f} (closed form, Appendix B); the rest of this cell")
        print(f"  reports it. The other three columns hold theta_u fixed, as a control.")
        print()
        for name, key, fmt in [("E[s]", "expected_s", "{:>15.4f}"),
                               ("P(all-region)", "upper", "{:>15.4f}"),
                               ("shift vs q_lit (37)", "shift", "{:>+15.4f}"),
                               ("leak", "leak", "{:>15.2e}")]:
            print(f"  {name}")
            print(f"    {'utterance':<12}" + "".join(f"{lab:>15}" for lab, _ in settings))
            for y in UTTERANCES:
                print(f"    {y:<12}"
                      + "".join(fmt.format(table[lab][y][key]) for lab, _ in settings))
            print()
        print(f"    {'||contrast||':<12}"
              + "".join(f"{table[lab]['contrast']:>15.4f}" for lab, _ in settings))
        print("    (all - some, in the quadrature norm; it GROWS with |theta_u|, so the")
        print("     utility level does move contrasts, unlike mu_u and ell_0.)")
        print()
        symmetric = max(
            abs(table["+1"][y][key] - table["-1"][y][key])
            for y in UTTERANCES for key in ("expected_s", "upper")
        )
        print(f"  theta_u = +1 against theta_u = -1: largest difference {symmetric:.4f}.")
        print("  That is not small, and it is worth knowing WHERE it is large. phi_u*")
        print("  carries a term linear in theta_u (through mu_u) and one quadratic in it")
        print("  (through G); only the linear one changes under theta_u -> -theta_u, and")
        print("  it is relatively largest when |theta_u| is of order 1:")
        print()
        print(f"    {'|theta_u|':>12}{'max dE[s]':>12}{'max dP(all)':>14}")
        for magnitude in (0.5, 1.0, 2.0, 5.0, 10.0, abs(star), 100.0):
            gaps = [[], []]
            for y in UTTERANCES:
                plus = net.read_out(net.closed_form_fixed_point(y, theta_u=magnitude)[0])
                minus = net.read_out(net.closed_form_fixed_point(y, theta_u=-magnitude)[0])
                gaps[0].append(abs(float((net.weights * plus * proportion).sum())
                                   - float((net.weights * minus * proportion).sum())))
                gaps[1].append(abs(float((net.weights * plus * upper).sum())
                                   - float((net.weights * minus * upper).sum())))
            print(f"    {magnitude:>12.4f}{max(gaps[0]):>12.4f}{max(gaps[1]):>14.4f}")
        print()
        print("  The sign therefore matters most at |theta_u| of order 1 and progressively")
        print("  less toward |theta_u*|, where the quadratic term dominates and the two")
        print("  signs converge. The numbers this cell reports are read at theta_u*.")
        print()

    return table, star



# ============================================================================
# D. BASE WORLD PRIOR
# ============================================================================

def part_d_priors():
    """The rows Part D sweeps: the four base world priors, all at this network's Lambda.
    Shared with the realizability block below, so the two report the same configurations.
    The delta-like base world prior, which needs a Lambda of its own, is Code Cell 2b's."""
    return dict(BASE_WORLD_PRIORS)


def base_prior_sweep(net, priors=None, *, verbose=True):
    """Repeat the demonstration and the implicature probe under each base world prior.

    ell_0 is a fixed quantity (Sec. 3), so changing it moves the maximizer of Eq. (13)
    and, through c_y, the learned theta_u* each base world prior carries. It touches nothing in
    Sec. 8: by Eq. (22) the Hessian involves theta_u but not ell_0, and every
    convergence result holds at every theta_u (Sec. 8.1(ii)), so the closed forms stay
    exact. What theta_u* does change is the stiffness of the fast subsystem, so the
    dynamics are integrated only where that is feasible (`settle`) and the table says
    where. What the sweep tests is semantic, not numerical: whether the conclusion of
    Part C is an artefact of one base world prior.

    The flat base world prior is a literal Uniform(0, 1) over the proportion, not a
    low-precision Gaussian; see `uniform_world_prior` for why the two differ.

    A base world prior may be given either as a log-density or as a (log_density, overrides) pair,
    the overrides passed on to `respawn`. That is how Code Cell 2b gives every row a
    Lambda of its own.
    """
    priors = priors or part_d_priors()
    proportion = logistic(net.zeta)
    upper_region = (net.zeta >= net.theta_L).to(net.dtype)

    rows = {}
    for name, specification in priors.items():
        prior, overrides = (
            specification if isinstance(specification, tuple) else (specification, {})
        )
        probe = net.respawn(base_prior=prior, **overrides)
        prior_density = torch.exp(probe.base_log_prior)
        entry = {
            "description": probe.base_prior_description,
            "lexical_strength": probe.lexical_strength,
            "theta_u": probe.theta_u,
            "prior_expected_s": float((probe.weights * prior_density * proportion).sum()),
            "prior_upper_mass": float((probe.weights * prior_density * upper_region).sum()),
            "utterances": {},
            "network": probe,
        }
        for y in UTTERANCES:
            settled = settle(probe, y)
            baseline = probe.read_out(probe.literal_fixed_point(y))
            tempered = probe.infer(y, theta_u=0.0, record_history=False)["q"]
            full = settled["q"]
            excluded = exclusion_indicator(
                y, probe.zeta, probe.theta_L, sharpness=probe.exclusion_sharpness
            )
            entry["utterances"][y] = {
                "expected_s": float((probe.weights * full * proportion).sum()),
                "baseline_upper": float((probe.weights * baseline * upper_region).sum()),
                "tempered_upper": float((probe.weights * tempered * upper_region).sum()),
                "full_upper": float((probe.weights * full * upper_region).sum()),
                "leak": float((probe.weights * full * excluded).sum()),
                "integrated": settled["integrated"],
                "closed_form_error": (
                    float((settled["phi_S"] - probe.closed_form_fixed_point(y)[0])
                          .abs().max())
                    if settled["integrated"] else None
                ),
                "baseline_q": baseline,
                "full_q": full,
            }
            entry["utterances"][y]["shift"] = (
                entry["utterances"][y]["full_upper"]
                - entry["utterances"][y]["baseline_upper"]
            )
        rows[name] = entry

    # The spike at the right edge of the flat and Beta(3,1) panels is a property of the
    # PLOTTING coordinate: the figures draw a density in s, dividing by ds/dzeta, which is
    # 1/405 at the top node. The modes below are read in zeta, where the grid is uniform.
    spike = {}
    for name in ("flat", "skewed high"):
        if name not in rows:
            continue
        probe = rows[name]["network"]
        entry = rows[name]["utterances"]["some"]
        spike[name] = {}
        for label, field in (("q_lit", entry["baseline_q"]), ("full", entry["full_q"])):
            mode = int(torch.argmax(field))
            spike[name][label] = {
                "s": float(proportion[mode]),
                "density": float(field[mode]),
                "top_node": float(field[-1]),
            }

    if verbose:
        print("BASE WORLD PRIOR SWEEP")
        print()
        print(f"  {'p_0':<13}{'description':<43}{'Lambda':>8}{'theta_u*':>12}"
              f"{'E[s] p_0':>11}{'P0(all)':>9}{'  E[s] | no    some    all':<28}")
        for name, entry in rows.items():
            means = "  ".join(
                f"{entry['utterances'][y]['expected_s']:.3f}" for y in UTTERANCES
            )
            print(f"  {name:<13}{entry['description']:<43}{entry['lexical_strength']:>8.0f}"
                  f"{entry['theta_u']:>12.4f}"
                  f"{entry['prior_expected_s']:>11.4f}{entry['prior_upper_mass']:>9.4f}"
                  f"         {means}")
        print("  (each base world prior at its own learned theta_u*)")
        print()
        print("  implicature shift, P(all-region) full minus q_lit, the literal listener")
        print("  (the mass shift criterion for \"some\" is a NEGATIVE shift)")
        print()
        print(f"  {'p_0':<13}" + "".join(f"{y:>12}" for y in UTTERANCES)
              + f"{'max leak':>12}{'P(all|some)':>13}{'max |phi_S - phi_S*|':>24}")
        for name, entry in rows.items():
            shifts = "".join(
                f"{entry['utterances'][y]['shift']:>+12.4f}" for y in UTTERANCES
            )
            errors = [entry["utterances"][y]["closed_form_error"] for y in UTTERANCES]
            worst = ("closed form (stiff)" if None in errors
                     else f"{max(errors):.1e}")
            leak = max(entry["utterances"][y]["leak"] for y in UTTERANCES)
            print(f"  {name:<13}{shifts}{leak:>12.1e}"
                  f"{entry['utterances']['some']['full_upper']:>13.4f}{worst:>24}")
        print(f"  (the last column compares the integrated fixed point with Eqs. (15)-(16);")
        print(f"   rows marked closed form have lambda_max(H) above {FEASIBLE_STIFFNESS:.0e},")
        print(f"   where integration under the timescale commitment is not affordable)")
        print()
        first = [name for name, entry in rows.items()
                 if entry["utterances"]["some"]["shift"] < 0]
        both = [name for name in first
                if rows[name]["utterances"]["some"]["full_upper"] < 0.5]
        # the split Text cell 4 Part D reads: the theta_u = 0 control against q_lit is
        # the TEMPERING, and the rest of the shift is the utility level's own doing.
        print("  the \"some\" shift split into tempering (the theta_u = 0 control against")
        print("  q_lit) and the utility level's own contribution (the remainder), beside the")
        print("  all-region masses of q_lit, the tempered control and the full network:")
        print(f"  {'p_0':<13}{'q_lit':>8}{'tempered':>10}{'full':>8}"
              f"{'shift':>10}{'tempering':>12}{'utility':>10}")
        for name, entry in rows.items():
            row = entry["utterances"]["some"]
            tempering = row["tempered_upper"] - row["baseline_upper"]
            print(f"  {name:<13}{row['baseline_upper']:>8.4f}{row['tempered_upper']:>10.4f}"
                  f"{row['full_upper']:>8.4f}{row['shift']:>+10.4f}{tempering:>+12.4f}"
                  f"{row['shift'] - tempering:>+10.4f}")
        print()
        print(f"  mass shift criterion (shift for \"some\" < 0) met under {len(first)} of "
              f"{len(rows)} base world priors: " + (", ".join(first) if first else "none") + ".")
        print(f"  both mass criteria (also P(all-region | \"some\") < 1/2) met under {len(both)}: "
              + (", ".join(both) if both else "none") + ".")
        if 0 < len(first) < len(rows) or 0 < len(both) < len(rows):
            print("  Whether they are met is therefore NOT a property of the model alone; it")
            print("  depends on the base world prior the model is run against, which is a fact about the")
            print("  criteria as much as about the model. Text cell 4 Part C reports this.")
        print()

        # === what ell_0-invariance survives under learning ===
        # phi_S* is affine in ell_0 with an utterance-independent coefficient, so at a
        # SHARED theta_u the contrasts between utterances are exactly ell_0-invariant.
        # The coefficient depends on theta_u, and ell_0 moves theta_u* through c_y, so
        # under learning the invariance no longer transfers across the rows of a sweep.
        # Lambda is held at this network's own for every base world prior here, whatever Lambda the
        # rows carry: the comparison is across ell_0 alone, and phi_L must not move with it.
        contrasts = {}
        for key in ("fixed", "learned"):
            fields = []
            for specification in (priors or part_d_priors()).values():
                prior = (specification[0] if isinstance(specification, tuple)
                         else specification)
                probe = net.respawn(base_prior=prior)
                theta = 1.0 if key == "fixed" else None
                settled = {y: probe.closed_form_fixed_point(y, theta_u=theta)[0]
                           for y in UTTERANCES}
                fields.append((settled["all"] - settled["some"],
                               settled["no"] - settled["some"]))
            contrasts[key] = max(
                max(float((a - fields[0][0]).abs().max()),
                    float((b - fields[0][1]).abs().max()))
                for a, b in fields
            )
        print(f"  ell_0-invariance of the phi_S contrasts across these {count_in_words(len(fields))} base world priors, all at "
              f"Lambda = {net.lexical_strength:.0f}:")
        print(f"    at a shared theta_u = 1 (a control): {contrasts['fixed']:.1e}")
        print(f"    at each base world prior's own learned theta_u*: {contrasts['learned']:.4f}")
        print("  (exact at a shared theta_u; under learning ell_0 also moves theta_u*,")
        print("   so what is invariant is the map at fixed theta_u, not the row.)")
        print()

        # === the spike at the right edge of two panels ===
        if spike:
            print("  THE RIGHT-EDGE SPIKE in the \"some\" panels of the flat and Beta(3,1)")
            print("  rows, read in zeta, where the grid is uniform (the figures draw a")
            print("  density in s, dividing by ds/dzeta, which is 1/405 at the top node):")
            print(f"    {'p_0':<13}{'belief':<8}{'mode (s)':>10}{'density there':>15}"
                  f"{'density at top node':>21}")
            for name, entry in spike.items():
                for label, values in entry.items():
                    print(f"    {name:<13}{label:<8}{values['s']:>10.2f}"
                          f"{values['density']:>15.4f}{values['top_node']:>21.4f}")
            print()

    for name, entry in rows.items():
        entry["spike"] = spike.get(name)
    return rows


# ============================================================================
# REALIZABILITY: HOW MUCH OF theta_u* THE CRITERION NEEDS
# ============================================================================
#
# The criterion is read at each base world prior's learned theta_u*. Under the delta-like base world prior of
# Code Cell 2b that is theta_u* = 1408, where lambda_max(H) = 2.0e6 and one inference is hours:
# the row is exact by Sec. 8.6 but not integrated. That invites the wrong inference,
# that the implicature is the thing costing hours. It is not, and the two are separable:
#
#   the THRESHOLD, the least |theta_u| at which the conjunction of Text cell 4 Part C
#   holds, by bisection on the ray from 0 to theta_u*, with a check that it then holds
#   along the whole of that ray;
#
#   the ARRIVAL, how many updates of Eq. (20) from theta_u(0) = 0 carry the flow past
#   the threshold.
#
# The block ends with each configuration meeting both mass criteria run end to end with nothing in
# closed form, where its arrival is integrable: Eq. (20) integrated for those updates,
# with the fast subsystem integrated
# inside every one of them, then Eqs. (18)-(19) integrated at the theta_u they reach,
# and the conjunction read off that integrated fixed point.
#
# Seconds here are wall clock on the machine that ran the notebook, and stand in for two
# machine-independent quantities the table gives beside them: the Euler steps an
# inference takes, and the separation 4 lambda_max(H) that the timescale commitment
# demands between the error and the state units. Neither is a cost the process being
# modelled could plausibly pay, at the cheap end of the range or the expensive one. That
# is the plausibility cost the accompanying paper raises about a proposed alternatives
# level (its Sec. 6.2). That level is not implemented here, so this cell measures and
# does not argue; see the paper for the discussion.


def criterion_for_some(net, theta_u=None, *, belief=None):
    """Text cell 4 Part C's mass shift and mass position criteria for "some", as one record.

    shift = P(all-region | "some") - P_lit(all-region | "some")  (Eq. 37), and
    upper = P(all-region | "some"); the conjunction is shift < 0 AND upper < 1/2.
    Passing `belief` scores an INTEGRATED read-out with the same code that scores the
    closed form, so the run below is compared against the table on equal terms."""
    upper_region = (net.zeta >= net.theta_L).to(net.dtype)
    literal_upper = float(
        (net.weights * net.read_out(net.literal_fixed_point("some")) * upper_region).sum()
    )
    if belief is None:
        belief = net.read_out(net.closed_form_fixed_point("some", theta_u=theta_u)[0])
    upper = float((net.weights * belief * upper_region).sum())
    return {"shift": upper - literal_upper, "upper": upper,
            "both": (upper - literal_upper < 0.0) and (upper < 0.5)}


def mode_criteria_for_some(net, theta_u=None):
    """Text cell 4 Part D's mode shift and mode position criteria for "some", as one record:
    k* is the node where phi_S* is largest and k0 the node where ell_0 is; the mode shift
    criterion is k* < k0 and the mode position criterion zeta_{k*} < theta_L."""
    phi_S = net.closed_form_fixed_point("some", theta_u=theta_u)[0]
    k = int(torch.argmax(phi_S))
    shift = k < int(torch.argmax(net.base_log_prior))
    position = float(net.zeta[k]) < net.theta_L
    return {"shift": shift, "position": position, "both": shift and position}


# The criteria whose first arrival on the ray 0 -> theta_u* is located (Sec. 4.1.6).
CRITERION_TESTS = {
    "mass conjunction": lambda net, theta: criterion_for_some(net, theta)["both"],
    "mode position": lambda net, theta: mode_criteria_for_some(net, theta)["position"],
    "mode conjunction": lambda net, theta: mode_criteria_for_some(net, theta)["both"],
}


def criterion_threshold(net, *, which="mass conjunction", bisections=60, ray_checks=200):
    """The least |theta_u| on the ray 0 -> theta_u* at which criterion `which` is met.

    Nothing proved here makes a criterion monotone in theta_u, so the bisection is
    followed by a check at `ray_checks` points between the threshold and theta_u*: the
    threshold is reported together with the answer to "and does it hold from there on".
    Returns None where the criterion fails at theta_u* itself."""
    test = CRITERION_TESTS[which]
    star = net.learned_theta_u()
    if not test(net, star):
        return None
    if test(net, net.theta_u_initial):
        return {"theta_u": net.theta_u_initial, "holds_to_star": True,
                "rate": net.stiffest_state_rate(net.theta_u_initial)}
    low, high = 0.0, 1.0                               # fractions of theta_u*
    for _ in range(bisections):
        middle = 0.5 * (low + high)
        if test(net, middle * star):
            high = middle
        else:
            low = middle
    threshold = high * star
    holds = all(
        test(net, threshold + (star - threshold) * i / ray_checks)
        for i in range(ray_checks + 1)
    )
    return {"theta_u": threshold, "holds_to_star": holds,
            "rate": net.stiffest_state_rate(threshold)}


def theta_u_gradient_at_equilibrium(net, theta_u):
    """Eq. (20)'s gradient with the fast subsystem AT its fixed point, Eqs. (15)-(16).

    Eq. (20) is written for an equilibrated fast subsystem (Sec. 8.5), and the closed
    form is that equilibrium exactly (Sec. 8.6), so this is the quantity
    `theta_u_gradient` reads off an integration, without paying for one. It is what
    makes the survey affordable at theta_u where integration is not. The run at the end
    of the block integrates the same updates and checks that they agree."""
    total = 0.0
    for utterance in UTTERANCES:
        phi_S, phi_u = net.closed_form_fixed_point(utterance, theta_u=theta_u)
        _, r_state, _ = net.residuals(
            phi_S, phi_u, net.lexical_field(utterance), theta_u=theta_u
        )
        total += float(net.dot(phi_u, net.project(r_state / net.sigma_state)))
    return total / len(UTTERANCES)


def updates_to_criterion(net, *, which="mass conjunction", tau_theta=20.0, tau_state=1.00,
                         max_updates=60):
    """Updates of Eq. (20) from theta_u(0) = 0 that carry the flow past the threshold.

    The update is `learn_theta_u`'s, theta_u += (tau_state / tau_theta) * gradient. The
    count is a property of that step as much as of the model: what it reports is that
    the threshold is passed immediately, not that some integer is fundamental."""
    theta = net.theta_u_initial
    for update in range(1, max_updates + 1):
        theta = theta + (tau_state / tau_theta) * theta_u_gradient_at_equilibrium(
            net, theta
        )
        if CRITERION_TESTS[which](net, theta):
            return {"updates": update, "theta_u": theta,
                    "rate": net.stiffest_state_rate(theta)}
    return None


def halt_by_tolerance(net, *, tolerance, tau_theta=20.0, tau_state=1.00, cap=HALTING_CAP):
    """Where Eq. (20)'s flow halts: the REALIZABLE theta_u (A19).

    The same rule `learn_theta_u` applies (stop when |delta theta_u| < tolerance, which
    reads only the size of the step just taken), but with the gradient read off
    Eqs. (15)-(16) rather than an integration, so surveying every row is affordable. The
    end-to-end run below integrates the same flow and checks it lands on this theta_u.

    `cap` is a cap and NOT a stopping rule: a flow that reaches it has not halted, and
    that is a fact to report rather than a result to use, so it raises.
    """
    theta = net.theta_u_initial
    for update in range(1, cap + 1):
        step = (tau_state / tau_theta) * theta_u_gradient_at_equilibrium(net, theta)
        theta = theta + step
        if abs(step) < tolerance:
            return {"updates": update, "theta_u": theta,
                    "rate": net.stiffest_state_rate(theta)}
    raise RuntimeError(
        f"the flow did not halt within cap={cap} at tolerance={tolerance:g}; theta_u "
        f"reached {theta:.6f}. The cap is not a stopping rule (A19): report this "
        "configuration as one that does not halt at this tolerance."
    )


def integration_cost_report(net):
    """The table of Text cell 4's Integration cost and conditioning (Eq. 28), at this
    network's theta_u*."""
    top = float(torch.linalg.eigvalsh(net.basis.T @ (net.weights[:, None] * net.basis)).max())
    state_rate = 1.0 / net.sigma_lexical + 1.0 / net.sigma_state
    print("INTEGRATION COST: THE phi_u RATE OF Eq. (28) AGAINST |theta_u|")
    print(f"  {'|theta_u|':>10}{'phi_u rate':>12}{'stiffness ratio':>17}{'largest Euler dt':>18}")
    for magnitude in (1.0, 10.0, abs(net.theta_u), 50.0):
        rate = 1.0 / net.sigma_utility + magnitude ** 2 * top / net.sigma_state
        print(f"  {magnitude:>10.4f}{rate:>12.1f}{rate / state_rate:>17.1f}{2.0 / rate:>18.4f}")
    print(f"  (stiffness ratio: the phi_u rate over the phi_S rate, {state_rate:g}. Largest Euler dt:")
    print("   2 / rate, explicit Euler's own bound, before the timescale commitment tightens it.)")
    print(f"  at theta_u* = {net.theta_u:.4f}: Eq. (28) gives "
          f"{1.0 / net.sigma_utility + net.theta_u ** 2 * top / net.sigma_state:.2f}, "
          f"lambda_max(H) = {net.stiffest_state_rate():.2f}")
    print()


def realizability_report(net, priors=None, *, tau_theta=20.0, verbose=True):
    """What the criteria cost, against what theta_u* costs, on the given rows
    (Part D's by default)."""
    rows = {}
    for name, specification in (priors or part_d_priors()).items():
        prior, overrides = (
            specification if isinstance(specification, tuple) else (specification, {})
        )
        probe = net.respawn(base_prior=prior, **overrides)
        threshold = criterion_threshold(probe)
        rows[name] = {
            "network": probe,
            "theta_star": probe.theta_u,
            "star_rate": probe.stiffest_state_rate(probe.theta_u),
            "threshold": threshold,
            # where the flow HALTS: the realizable theta_u under A19, for every row,
            # whether or not the conjunction is ever met on the way
            "halt": halt_by_tolerance(probe, tolerance=demonstration_tolerance(probe),
                                      tau_theta=tau_theta),
            # where the conjunction is FIRST MET: not a halting rule, kept because it
            # reports that the conjunction arrives early in the flow
            "arrival": (updates_to_criterion(probe, tau_theta=tau_theta)
                        if threshold is not None else None),
            # the same for the mode criteria of Part D, on the same ray
            "mode": {which: {"threshold": criterion_threshold(probe, which=which)}
                     for which in ("mode position", "mode conjunction")},
        }
        for which, entry in rows[name]["mode"].items():
            t = entry["threshold"]
            entry["arrival"] = (None if t is None else
                                {"updates": 0} if t["theta_u"] == probe.theta_u_initial else
                                updates_to_criterion(probe, which=which, tau_theta=tau_theta))

    if not verbose:
        return rows

    tolerance = demonstration_tolerance(net)
    print("REALIZABILITY: WHERE THE FLOW HALTS, AGAINST WHAT theta_u* COSTS")
    print()
    print(f"  {'p_0':<13}{'theta_u*':>11}{'lambda(theta_u*)':>18}{'theta_halt':>12}"
          f"{'lambda(halt)':>14}{'4 lambda(halt)':>16}{'updates':>9}")
    for name, row in rows.items():
        halt = row["halt"]
        print(f"  {name:<13}{row['theta_star']:>11.2f}{row['star_rate']:>18.2e}"
              f"{halt['theta_u']:>12.3f}{halt['rate']:>14.1f}"
              f"{4 * halt['rate']:>16.0f}{halt['updates']:>9}")
    print(f"  (theta_halt: where Eq. (20)'s flow stops, its own update having fallen below")
    print(f"   the tolerance of {tolerance:g}. That is the only stopping rule the model has,")
    print("   and the value is AD HOC (A19). theta_halt is the REALIZABLE theta_u, what the")
    print("   dynamics reach; theta_u* is the asymptote they ascend toward, and it carries")
    print("   no tolerance. 4 lambda is the separation the timescale commitment demands")
    print("   there, tau_error <= tau_state / 4 lambda_max(H).)")
    print()
    print("  WHERE THE MASS CONJUNCTION IS FIRST MET: a fact about the SHAPE of the update,")
    print("  not about where the flow stops:")
    print(f"  {'p_0':<13}{'theta_crit':>12}{'lambda(crit)':>14}{'updates to it':>15}")
    for name, row in rows.items():
        if row["threshold"] is None:
            print(f"  {name:<13}{'not met at theta_u*, so no threshold':>41}")
            continue
        updates = row["arrival"]["updates"] if row["arrival"] else None
        print(f"  {name:<13}{row['threshold']['theta_u']:>12.3f}"
              f"{row['threshold']['rate']:>14.1f}"
              f"{(updates if updates is not None else '>60'):>15}")
    print("  (theta_crit: the least |theta_u| on the ray to theta_u* at which the shift")
    print("   for \"some\" is negative AND P(all-region | \"some\") < 1/2. NEITHER this nor")
    print("   the bisection that locates it is a halting rule: one is the evaluator's")
    print("   criterion, the other needs theta_u*. What they report is that the mass conjunction")
    print("   arrives EARLY in the flow, far below theta_halt above.)")
    met = [name for name, row in rows.items() if row["threshold"] is not None]
    holds = [name for name in met if rows[name]["threshold"]["holds_to_star"]]
    print(f"  A threshold exists under {len(met)} of {len(rows)} base world priors"
          + (f": {', '.join(met)}." if met else "."))
    if met:
        print(f"  The mass conjunction then holds all the way to theta_u* under {len(holds)} "
              f"of those {len(met)}.")
        print()
        print("  So the mass conjunction is not what makes a row unintegrable. Where it holds it")
        print("  holds at a |theta_u| of a few units, which the flow passes on its first")
        print("  updates, well below where it halts. theta_u* lies orders of magnitude")
        print("  beyond both, and it is theta_u* that this integrator cannot reach at all.")
        print()
    else:
        print()

    print("  WHERE THE MODE CRITERIA ARE FIRST MET, on the same ray (Text cell 4 Part D):")
    print(f"  {'p_0':<13}{'mode position':>15}{'lambda':>9}{'updates':>9}{'to star':>9}"
          f"{'mode conjunction':>18}{'lambda':>9}{'updates':>9}{'to star':>9}")
    for name, row in rows.items():
        line = f"  {name:<13}"
        for which, width in (("mode position", 15), ("mode conjunction", 18)):
            entry = row["mode"][which]
            t = entry["threshold"]
            if t is None:
                line += f"{'not met at theta_u*':>{width + 27}}"
                continue
            updates = entry["arrival"]["updates"] if entry["arrival"] else ">60"
            line += (f"{t['theta_u']:>{width}.3f}{t['rate']:>9.1f}{updates:>9}"
                     f"{('yes' if t['holds_to_star'] else 'no'):>9}")
        print(line)
    print("  (each column the least |theta_u| on the ray to theta_u* at which that criterion is met;")
    print("   0.000 where it is already met at theta_u = 0, the tempered control, whose mode is")
    print("   ell_0's; 'to star': whether it then holds at every checked point up to theta_u*.)")
    print()

    # ---- each configuration meeting both mass criteria, end to end, nothing in closed form ----
    # Every row with a threshold is run at the theta_u its flow HALTS at, where that is
    # integrable; a row that is not is named. The gate is FEASIBLE_STIFFNESS, which marks
    # where `infer` stops converging, not merely where it gets slow; see its comment.
    for name in met:
        halt = rows[name]["halt"]
        if halt["rate"] > FEASIBLE_STIFFNESS:
            print(f"  \"{name}\" is not run end to end: it halts at theta_u = "
                  f"{halt['theta_u']:.3f}, where lambda_max(H) = {halt['rate']:.1e} is above "
                  f"{FEASIBLE_STIFFNESS:.1e},")
            print("  the wall-clock budget of FEASIBLE_STIFFNESS. The fixed point there is "
                  "not in doubt;")
            print("  Eqs. (15)-(16) give it exactly, and this block reports only what it "
                  "integrated.")
            print()
    for name in met:
        row, halt = rows[name], rows[name]["halt"]
        if halt["rate"] > FEASIBLE_STIFFNESS:
            continue
        probe = row["network"].respawn()
        print(f"  RUN, NOTHING IN CLOSED FORM: the \"{name}\" row, "
              f"{halt['updates']} update(s) of Eq. (20) to its halt, then Eqs. (18)-(19)")
        print("  integrated at the theta_u they reach.")

        started = time.perf_counter()
        trace = probe.learn_theta_u(UTTERANCES, tau_theta=tau_theta,
                                    tolerance=tolerance, num_updates=HALTING_CAP)
        learning_seconds = time.perf_counter() - started
        started = time.perf_counter()
        settled = settle(probe, "some")
        settling_seconds = time.perf_counter() - started
        if not settled["integrated"]:
            raise RuntimeError("the halted theta_u was expected to be integrable")
        scored = criterion_for_some(probe, belief=settled["q"])
        closed_form_gap = float(
            (settled["phi_S"] - probe.closed_form_fixed_point("some")[0]).abs().max()
        )
        per_step = settling_seconds / settled["steps"]

        # Step counts are reproducible because `infer` stops at 1e-9, above the
        # floating-point floor of about 8e-11 to 1.6e-10 where the derivative stops
        # decaying. With the tolerance inside that band, as it was at 1e-10, the stopping
        # step was decided by which roundoff-level samples happened to land below it: a
        # change in theta_u of one part in 1e15 moved the count by hundreds, and forming the
        # same gradient in a different summation order gave a different number for the
        # same trajectory. See `infer` for the full argument.
        # Lines tagged "cost:" are wall clock and step counts, which belong to the machine
        # and to the particular integration path, not to the model: a second implementation
        # reaching the same fixed point along a slightly different path would spend a
        # different number of steps getting there, which is why they are tagged.
        print(f"    Eq. (20): theta_u {trace[0]['theta_u']:.4f} -> {probe.theta_u:.6f} in "
              f"{len(trace)} update(s), each one integrating all three utterances")
        print(f"      against the same flow halted at the closed-form equilibrium: "
              f"{halt['theta_u']:.6f}, agreeing to better than 1e-13")
        print(f"    Eqs. (18)-(19) at theta_u = {probe.theta_u:.4f}: "
              f"lambda_max(H) = {halt['rate']:.1f}, tau_error <= "
              f"{probe.fast_time_constant():.2e}")
        print(f"    mass conjunction off the INTEGRATED fixed point: shift for \"some\" = "
              f"{scored['shift']:+.4f} (< 0: {scored['shift'] < 0}), "
              f"P(all-region | \"some\") = {scored['upper']:.4f} (< 1/2: {scored['upper'] < 0.5})")
        print(f"    both mass criteria met by the integrated run: {scored['both']}; "
              f"max |phi_S - phi_S*| against Eqs. (15)-(16) = {closed_form_gap:.0e}")
        # Scaling this row's own steps-per-lambda out to theta_u* is legitimate, because
        # the ratio is linear in lambda: 181 to 224 steps per unit
        # across every theta_u measured, 13.4 to 61.3, because the tolerance is keyed to
        # lambda_max(H), so the extrapolation stays inside one regime. Under a FIXED
        # tolerance it did not: above lambda ~ 2e3 the run stopped converging at all, and
        # the same extrapolation crossed a boundary it was fitted below.
        steps_per_rate = settled["steps"] / halt["rate"]
        star_steps = steps_per_rate * row["star_rate"]
        star_seconds = star_steps * per_step
        print(f"    cost: {settled['steps']} Euler steps, {settling_seconds:.2f} s "
              f"({1e6 * per_step:.0f} us/step) on the machine that ran this, the {len(trace)} "
              f"update(s) of Eq. (20) a further {learning_seconds:.2f} s")
        print(f"    cost: the same inference at theta_u* = {row['theta_star']:.1f} would be "
              f"{star_steps:.2e} steps, about "
              + (f"{star_seconds:.0f} s" if star_seconds < 600 else f"{star_seconds / 3600:.1f} h")
              + f", at a separation of {4 * row['star_rate']:.1e}")
        print("    (so the mass conjunction is demonstrated in the dynamics here, and the closed")
        print("     form is what reports it at theta_u*, not what establishes it)")
        print()

        # === the floating-point floor `infer`'s stopping tolerance is set above ===
        # Run at the closed-form HALTED theta_u. The first 2000 steps past convergence are still the tail of
        # the decay, so the band is read over the 2000 after them.
        floor_network = row["network"]
        converged = floor_network.infer("some", theta_u=halt["theta_u"], record_history=False)
        # The tolerance stops the run at about 12 floors, and the decay is exponential at
        # rate 1 per simulated time unit, so the floor is another log(12) = 2.5 time units
        # away, and dt = tau_state / (8 lambda), so that is 20 lambda steps, not a
        # constant. Four time units of margin is taken.
        extra = int(4.0 * 8.0 * halt["rate"])
        past = floor_network.infer("some", theta_u=halt["theta_u"], derivative_tolerance=0.0,
                                   max_steps=converged["steps"] + extra, record_history=True)
        band = [entry["max_derivative"] for entry in past["history"][-2000:]]
        stopped_at = converged["derivative_tolerance"]
        floor = max(band)
        per_rate = floor / halt["rate"]
        print(f"  THE ROUNDOFF FLOOR. Integrated {extra} steps past convergence at theta_u = "
              f"{halt['theta_u']:.4f} (\"{name}\"); over the last 2000,")
        print(f"  the largest derivative stays between {min(band):.1e} and {max(band):.1e}, and "
              f"`infer` stopped at {stopped_at:.2e},")
        print(f"  which is {stopped_at / floor:.1f} times that floor. "
              f"DERIVATIVE_TOLERANCE_PER_RATE holds that ratio")
        print(f"  fixed at every theta_u; a constant tolerance would not, and below the floor "
              f"the test never fires.")
        # The floor per unit of lambda_max(H) is the constant the rule is keyed to, and the
        # counterfactual says what the superseded fixed 1e-9 would have bought HERE. Both are
        # printed because the prose quotes them (C6).
        print(f"    the floor per unit of lambda_max(H) = {floor:.3e} / {halt['rate']:.1f} = "
              f"{per_rate:.3e}, against DERIVATIVE_TOLERANCE_PER_RATE = "
              f"{DERIVATIVE_TOLERANCE_PER_RATE:.3e}")
        print(f"    a FIXED tolerance of 1e-9 would be {1e-9 / floor:.2f} floors here"
              + (", BELOW the floor, so the test could never fire"
                 if 1e-9 < floor else "")
              + f"; it equals the floor at lambda_max(H) = {1e-9 / per_rate:.0f}")
        print()
        rows[name]["run"] = {"theta_u": probe.theta_u, "steps": settled["steps"],
                             "seconds": settling_seconds, "scored": scored,
                             "closed_form_gap": closed_form_gap}
    return rows


# ============================================================================
# THE DELTA READ-OUT (Text cell 4 Part D, after the figures)
# ============================================================================

def delta_readout_report(net, realizability, *, tau_theta=20.0, finer_grids=(201, 401, 801)):
    """phi_S* itself, the centre of the delta posterior (Bogacz Eq. 34), by base world prior.

    The construction's posterior is a point mass at the settled state (phi_S*, phi_u*).
    q (Eq. 12) is built for comparison and appears in the last column only. The mode is
    the grid node where phi_S* is largest; the cell of "all" is zeta >= theta_L
    (Appendix A). Rows are the given base world priors at theta_u*, and again at the realizable
    theta_u of the block above, integrated, for every row that block ran end to end."""
    upper_region = (net.zeta >= net.theta_L).to(net.dtype)
    cell_start = float(logistic(torch.tensor(net.theta_L)))

    def peak(probe, field):
        k = int(torch.argmax(field))
        return k, float(logistic(probe.zeta[k]))

    def where(probe, k):
        return "inside" if float(probe.zeta[k]) >= probe.theta_L else "outside"

    print("THE DELTA READ-OUT: phi_S* AT THE SETTLED STATE, BY BASE WORLD PRIOR (Bogacz Eq. 34)")
    print()
    print(f"  {'p_0':<13}{'theta_u':<24}{'y':<6}{'phi_u*':>22}{'mode of phi_S* at s':>21}"
          f"{'max phi_S*':>12}{'min phi_S*':>12}{'q-mass, zeta >= theta_L':>24}")
    settled = {}
    for name, row in realizability.items():
        settings = [(f"theta_u* = {row['theta_star']:.4f}", row["network"], None)]
        if row.get("run") is not None:
            # The flow that reaches this theta_u was already integrated, in the
            # realizability block above, which recorded where it landed. Re-integrating it
            # here cost about as much again as everything else in this cell, and bought
            # nothing: what THIS block demonstrates is the settling AT that theta_u --
            # Eqs. (18)-(19) integrated, the "integrated" tag below, not the flow that
            # reached it. So the value is reused, not recomputed. Note this is not a
            # closed form standing in for the dynamics: row["run"]["theta_u"] is the
            # endpoint of the INTEGRATED flow (`halt_by_tolerance`'s closed-form value
            # agrees with it to better than 1e-13, and is deliberately not the one taken).
            learner = row["network"].respawn()
            learner.theta_u = row["run"]["theta_u"]
            learner.theta_u_learned = False      # the state `learn_theta_u` leaves
            settings.append((f"realizable {learner.theta_u:.4f}", learner, "integrated"))
        for label, probe, how in settings:
            for utterance in UTTERANCES:
                if how == "integrated":
                    run = settle(probe, utterance)
                    if not run["integrated"]:
                        raise RuntimeError("the halted theta_u was expected to be integrable")
                    phi_S, phi_u = run["phi_S"], run["phi_u"]
                else:
                    phi_S, phi_u = probe.closed_form_fixed_point(utterance)
                settled[(name, label, utterance)] = (probe, phi_S)
                k, s = peak(probe, phi_S)
                mass = float((probe.weights * probe.read_out(phi_S) * upper_region).sum())
                vector = "[" + ", ".join(f"{float(v):+.4f}" for v in phi_u) + "]"
                print(f"  {name:<13}{label:<24}{utterance:<6}{vector:>22}{s:>21.4f}"
                      f"{float(phi_S.max()):>12.2f}{float(phi_S.min()):>12.2f}{mass:>24.4f}")
    print("  (mode: the grid node where phi_S* is largest. The last column is Eq. (27) under the")
    print("   read-out q of Eq. (12), for comparison; the delta read-out does not use it.)")
    print()

    print(f"  THE PEAK OF ell_0 AGAINST THE CELL OF \"all\", zeta >= theta_L = {net.theta_L:.4f}, "
          f"s >= {cell_start:.4f} (Appendix A)")
    some_peak = 'phi_S*("some") mode (s)'
    print(f"    {'p_0':<13}{'ell_0 mode (s)':>22}{some_peak:>32}   at")
    same_node = True
    moved_out = []
    for name, row in realizability.items():
        probe = row["network"]
        k0, s0 = peak(probe, probe.base_log_prior)
        same_node = same_node and int(torch.argmax(probe.literal_fixed_point("some"))) == k0
        for (key, label, utterance), (holder, phi_S) in settled.items():
            if key != name or utterance != "some":
                continue
            k, s = peak(holder, phi_S)
            print(f"    {name:<13}{s0:>14.4f} {where(probe, k0):<7}{s:>24.4f} {where(holder, k):<7}"
                  f"   {label}")
            if where(probe, k0) == "inside" and where(holder, k) == "outside":
                moved_out.append((name, label, holder, phi_S, k))
    print(f"    under \"some\", ell_0 - phi_L has its mode at the node where ell_0 has its, under every base world prior: "
          f"{same_node}")
    print()

    for name, label, holder, phi_S, k in moved_out:
        inside = holder.zeta >= holder.theta_L
        margin = float(phi_S[inside].max() - phi_S[~inside].max())
        print(f"  AT THE BOUNDARY, \"{name}\" at {label}: phi_S*(\"some\") has its mode at zeta = "
              f"{float(holder.zeta[k]):.4f} (s = {float(logistic(holder.zeta[k])):.4f}); the node above, "
              f"zeta = {float(holder.zeta[k + 1]):.4f}, is inside the cell,")
        print(f"    and the largest phi_S* inside the cell is {margin:+.3f} against the largest outside it")
    if moved_out:
        name = moved_out[0][0]
        base = realizability[name]["network"]
        print(f"  the same at theta_u* on finer grids of the same half-width, \"{name}\":")
        for nodes in finer_grids:
            probe = base.respawn(num_nodes=nodes)
            phi_S = probe.closed_form_fixed_point("some")[0]
            k, s = peak(probe, phi_S)
            inside = probe.zeta >= probe.theta_L
            margin = float(phi_S[inside].max() - phi_S[~inside].max())
            print(f"    {nodes} nodes: theta_u* = {probe.theta_u:.2f}, mode s = {s:.4f} ({where(probe, k)}), "
                  f"inside against outside {margin:+.3f}")
    print()

    # === the mode shift and mode position criteria under "some" (B8), beside the mass criteria ===
    # Each is reported as a measurement, and no position is taken on how to read them (B10).
    # The mode is the grid node where the field is largest; torch.argmax takes the first such
    # node, so the gap between the two largest nodes of phi_S* is printed, and a tie shows as 0.
    # The mode shift is also counted in grid steps, k - k0: the grid is uniform in zeta while s
    # compresses toward the ends of the scale, so a shift that looks small in s can be many nodes.
    print("  THE MODE SHIFT AND MODE POSITION CRITERIA UNDER \"some\" (B8), BESIDE THE q SHIFT AND q POSITION")
    print("  CRITERIA (Part C)")
    print("    mode shift: mode(phi_S*) - mode(ell_0) < 0;  mode position: mode(phi_S*) outside the cell of")
    print("    \"all\", zeta < theta_L;  mass shift: shift for \"some\" < 0;  mass position: P(all-region | \"some\") < 1/2.")
    step = float(net.zeta[1] - net.zeta[0])
    print("    Modes in s. The mode shift is printed in s and in grid steps; the grid is uniform in zeta,")
    print(f"    so multiplying the steps by the grid spacing, {step:.2f}, recovers the shift in zeta.")
    print(f"    {'p_0':<13}{'theta_u':<24}{'ell_0':>8}{'ell_0-phi_L':>12}{'phi_S*':>8}{'shift':>9}{'steps':>7}"
          f"{'mode shift':>11}{'mode position':>14}{'both':>6}{'mass shift':>11}{'mass position':>14}"
          f"{'both':>6}{'top-2 gap':>11}")
    tally = {key: 0 for key in ("a", "b", "delta_both", "first", "second", "q_both",
                                "agree_first", "agree_second", "agree_both")}
    # The rows below are the base world priors handed in and, where the block above ran a flow end
    # to end, the same base world prior again at its realizable theta_u. A count over rows therefore
    # weights a repeated base world prior twice, so the four verdicts are kept per row and counted a
    # second time over distinct base world priors, at one row each (C6: the q blocks above report
    # their counts over base world priors, and a count over base world priors should be readable here too).
    by_row = {}
    counted = 0
    for (name, label, utterance), (holder, phi_S) in settled.items():
        if utterance != "some":
            continue
        k0 = int(torch.argmax(holder.base_log_prior))
        k_lit = int(torch.argmax(holder.literal_fixed_point("some")))
        k = int(torch.argmax(phi_S))
        largest = torch.topk(phi_S, 2).values
        at_s = lambda index: float(logistic(holder.zeta[index]))
        a, b = k < k0, float(holder.zeta[k]) < holder.theta_L
        upper = float((holder.weights * holder.read_out(phi_S) * upper_region).sum())
        literal = float((holder.weights * holder.read_out(holder.literal_fixed_point("some"))
                         * upper_region).sum())
        first, second = upper - literal < 0.0, upper < 0.5
        yes = lambda flag: "yes" if flag else "no"
        print(f"    {name:<13}{label:<24}{at_s(k0):>8.4f}{at_s(k_lit):>12.4f}{at_s(k):>8.4f}"
              f"{at_s(k) - at_s(k0):>+9.4f}{k - k0:>+7d}{yes(a):>11}{yes(b):>14}{yes(a and b):>6}{yes(first):>11}"
              f"{yes(second):>14}{yes(first and second):>6}{float(largest[0] - largest[1]):>11.1e}")
        counted += 1
        by_row[(name, label)] = (a, b, first, second)
        for key, flag in (("a", a), ("b", b), ("delta_both", a and b), ("first", first),
                          ("second", second), ("q_both", first and second),
                          ("agree_first", a == first), ("agree_second", b == second),
                          ("agree_both", (a and b) == (first and second))):
            tally[key] += int(flag)
    print(f"    over these {counted} rows: the mode shift criterion is met in {tally['a']}, the mode "
          f"position criterion in {tally['b']}, both in {tally['delta_both']}; the mass shift criterion in "
          f"{tally['first']}, the mass position criterion in {tally['second']}, both in {tally['q_both']}")
    print(f"    the mode and mass shift criteria agree in {tally['agree_first']} of {counted} rows, the mode "
          f"and mass position criteria in {tally['agree_second']}, the two conjunctions in "
          f"{tally['agree_both']}")
    at_star = {name: verdicts for (name, label), verdicts in by_row.items()
               if label.startswith("theta_u*")}
    repeats_agree = all(by_row[key] == at_star[key[0]] for key in by_row)
    over = lambda index: sum(int(v[index]) for v in at_star.values())
    both_of = lambda i, j: sum(int(v[i] and v[j]) for v in at_star.values())
    print(f"    over the {len(at_star)} distinct base world priors, one row each at its own theta_u*: the mode "
          f"shift criterion is met under {over(0)}, the mode position criterion under {over(1)}, "
          f"both under {both_of(0, 1)}; the mass shift criterion under {over(2)}, the mass position "
          f"criterion under {over(3)}, both under {both_of(2, 3)}")
    if counted == len(at_star):
        print("    (one row per base world prior here, so the two counts coincide)")
    else:
        print(f"    (the other {counted - len(at_star)} rows repeat a base world prior at a realizable theta_u; "
              f"each repeat carries the same four verdicts as its own theta_u* row: {repeats_agree})")
    print()


# ============================================================================
# RUN
# ============================================================================

evaluation_network = LexicalPredictiveCodingNetwork()

check_specification(evaluation_network)
demonstration = demonstrate(evaluation_network)
lexical_strength_sweep(evaluation_network)
implicature = scalar_implicature_probe(evaluation_network)
learned_network, theta_trace = theta_u_learning_probe(evaluation_network)
theta_table, theta_star_value = theta_u_dependence(evaluation_network)
prior_rows = base_prior_sweep(evaluation_network)
integration_cost_report(evaluation_network)
realizability = realizability_report(evaluation_network)


# ---- figure ----

try:
    import matplotlib.pyplot as plt

    proportion = logistic(evaluation_network.zeta).numpy()

    # Curves are plotted as densities in the PROPORTION s, so the belief, which is
    # normalized as a density in zeta, must be pushed through the change of variables
    #     p_s(s) = p_zeta(zeta) * |dzeta/ds| = q(zeta) / (s (1 - s)),
    # dividing by ds/dzeta rather than multiplying by it. The transform is checkable:
    # sum_k w_k * to_s_density(q)_k * (ds/dzeta)_k = 1 exactly, whereas multiplying by
    # the Jacobian gives ~0.035.
    ds_dzeta = proportion * (1.0 - proportion)

    def to_s_density(field):
        return field.numpy() / ds_dzeta

    all_region_start = float(logistic(torch.tensor(evaluation_network.theta_L)))

    figure, axes = plt.subplots(1, 3, figsize=(13, 3.6), sharey=True)
    for axis, utterance in zip(axes, UTTERANCES):
        row = implicature[utterance]
        axis.plot(proportion, to_s_density(torch.exp(evaluation_network.base_log_prior)),
                  color="0.6", linewidth=1.0, label="base world prior")
        axis.plot(proportion, to_s_density(row["baseline_q"]),
                  linestyle="--", label="literal listener $q_{\\mathrm{lit}}$")
        axis.plot(proportion, to_s_density(row["full_q"]), label="full network")
        axis.axvspan(all_region_start, 1.0, alpha=0.12, color="tab:red")
        axis.set_title(f'"{utterance}"')
        axis.set_xlabel("proportion $s$")
    axes[0].set_ylabel("read-out density $q(s)$")
    axes[0].legend(fontsize=8, loc="upper center")
    figure.suptitle(
        "Settled belief against the literal listener "
        "(shaded: the region where \"all\" is true)", fontsize=10)
    figure.tight_layout()
    plt.show()

    # second figure: "some" under each base world prior (Part D). Each panel carries its
    # own base world prior, so the gray line differs from panel to panel. Panels on a shared y-axis
    # have to be comparable, so only rows sharing this network's Lambda are drawn; Part
    # D's all do. The same base world priors at a strong Lambda are Code Cell 2b's.
    comparable = {
        name: entry for name, entry in prior_rows.items()
        if entry["lexical_strength"] == evaluation_network.lexical_strength
    }
    prior_figure, prior_axes = plt.subplots(
        1, len(comparable), figsize=(3.4 * len(comparable), 3.4), sharey=True
    )
    for axis, (name, entry) in zip(prior_axes, comparable.items()):
        row = entry["utterances"]["some"]
        axis.plot(proportion, to_s_density(torch.exp(entry["network"].base_log_prior)),
                  color="0.6", linewidth=1.0, label="base world prior")
        axis.plot(proportion, to_s_density(row["baseline_q"]),
                  linestyle="--", label="literal listener $q_{\\mathrm{lit}}$")
        axis.plot(proportion, to_s_density(row["full_q"]), label="full network")
        axis.axvspan(all_region_start, 1.0, alpha=0.12, color="tab:red")
        axis.set_title(f"{name}\n(shift {row['shift']:+.4f})", fontsize=9)
        axis.set_xlabel("proportion $s$")
    prior_axes[0].set_ylabel("read-out density $q(s)$")
    prior_axes[0].legend(fontsize=8)
    signs = {entry["utterances"]["some"]["shift"] > 0 for entry in comparable.values()}
    prior_figure.suptitle(
        '"some" under the four base world priors sharing Lambda = %g, each at its own '
        'learned $\\theta_u^\\ast$: the shift against the literal listener is %s across them'
        % (evaluation_network.lexical_strength,
           "of one sign" if len(signs) == 1 else "not of one sign"), fontsize=10)
    prior_figure.tight_layout()
    plt.show()

except ImportError:
    print("matplotlib unavailable; skipping the figure")


# ---- the delta read-out, after the q figures (Text cell 4 Part D) ----

delta_readout_report(evaluation_network, realizability)
````

## <a id="tc4b" name="tc4b"></a>Text cell 4b: Part D at a strong $\Lambda$

Code Cell 2b repeats the evaluations of Part D with $\Lambda=512$ for every base world prior and
every utterance, each configuration at its own learned $\theta_u^\ast$: Part D's four base world priors and a

**Why a strong $\Lambda$.** At Part D's $\Lambda=8$ the diffuse base world priors override the entries for
*no* and *all*: the largest leak in a row is $0.63$ under the Gaussian base world prior and $0.91$ under both
skewed base world priors. Part of the settled belief given one of those utterances then lies in the region its

criteria. The delta-like base world prior needs a stronger lexicon still. A literal $P_0(s{=}1)=1$ is not
representable, since $s$ lies in the *open* interval, $\zeta=\operatorname{logit}(1)$ is off the
grid, and $\ell_0$ would be $-\infty$ almost everywhere, making $r_S$ infinite. We therefore use the
limit family $\mathrm{Beta}(\alpha,1)$ that Text cell 6 sweeps. $\alpha=64$ puts the mode in $\zeta$
at $\log\alpha=4.16$, inside the all-region and clear of the grid ceiling Text cell 6 describes; it

$0.0003$ apart and the probe would be measuring the truncation. $\Lambda$ must rise with $\alpha$, or $\ell_0$ simply overrides the entry: Eq. (41) puts the threshold at
$\Lambda_{\mathrm{crit}}\approx\alpha\log2n=192$, and $\Lambda=8\alpha=512$ is about $2.7\times$
that. This cell holds all five base world priors at that $\Lambda$, so that its rows share one lexical
strength, and under it every entry holds under every base world prior, the largest leak in any row being

````python
# === Code Cell 2b: EVALUATION, Part D's base world priors at a strong Lambda ===
#
# Part D's four base world priors and the delta-like base world prior, all at Lambda = 512, each
# configuration at its own learned theta_u* (Text cell 4b). Code Cell 2's functions do
# the measuring; this cell supplies the rows, the blocks about the delta-like row, and
# the figures.

import math
import torch


# ============================================================================
# THE STRONG LAMBDA
# ============================================================================

# The delta-like row. Part D's four base world priors are all diffuse over the scale, so
# none of them puts the criterion under any pressure: hearing "some" under a base world prior
# that is already spread out, there is little all-region mass to move away from. The
# sharp case is the one worth testing, and it is the case Code Cell 4 is built around.
#
# A literal delta at s = 1 is not representable, since s lies in the OPEN interval, so
# zeta = logit(1) is off the grid and ell_0 = log p_0 would be -infinity almost
# everywhere. Beta(alpha, 1) is the limit family that approaches it (Text cell 6).
#
#   alpha = 64 puts the mode in zeta at log(alpha) = 4.16, well inside the all-region
#   (zeta > theta_L = 2.9444) and well clear of the grid ceiling, which Text cell 6
#   warns sets in above alpha ~ 300; it carries 95.7% of its mass in the
#   all-region and only 0.95% on the top node.
#
#   Lambda must then be raised to match, or ell_0 simply overrides the lexicon and
#   the sweep measures the override rather than the criterion. Eq. (41) puts the
#   override threshold at Lambda_crit ~ alpha * log(2n) = 64 * 2.9957 = 192. Taking
#   Lambda = 8 alpha = 512, about 2.7x that, drives every leak below 1e-70: the entry
#   holds against ell_0, which is the precondition for the probe to mean anything.
DELTA_ALL_ALPHA = 64.0
DELTA_ALL_LAMBDA = 8.0 * DELTA_ALL_ALPHA          # = 512, about 2.7 x Lambda_crit

# Every row of this cell carries that Lambda. At Part D's own Lambda the diffuse base world priors
# override the entries for "no" and "all" (Code Cell 2 prints the leak), so the same
# base world priors are read again here with every entry holding.
STRONG_LAMBDA = DELTA_ALL_LAMBDA


def delta_like_row():
    """The delta-like row, with the Lambda it needs (the derivation above)."""
    return {"delta (all)": (beta_world_prior(DELTA_ALL_ALPHA, 1.0),
                            {"lexical_strength": DELTA_ALL_LAMBDA})}


def strong_lambda_priors():
    """Part D's base world priors and the delta-like row, every one at STRONG_LAMBDA."""
    rows = {name: (prior, {"lexical_strength": STRONG_LAMBDA})
            for name, prior in part_d_priors().items()}
    return {**rows, **delta_like_row()}


def strong_lambda_comparison(base_rows, rows):
    """Each base world prior at Part D's Lambda beside the same base world prior at STRONG_LAMBDA, under "some"."""
    print(f"  THE SAME PRIORS AT PART D'S LAMBDA AND AT LAMBDA = {STRONG_LAMBDA:.0f}, under \"some\"")
    print(f"    {'p_0':<13}{'Lambda':>8}{'theta_u*':>12}{'shift':>10}{'P(all|some)':>13}"
          f"{'max leak':>11}{'mass shift':>12}{'mass position':>15}")
    for name, entry in rows.items():
        pairs = ([base_rows[name]] if name in base_rows else []) + [entry]
        for i, row in enumerate(pairs):
            some = row["utterances"]["some"]
            leak = max(row["utterances"][y]["leak"] for y in UTTERANCES)
            print(f"    {(name if i == 0 else ''):<13}{row['lexical_strength']:>8.0f}"
                  f"{row['theta_u']:>12.4f}{some['shift']:>+10.4f}{some['full_upper']:>13.4f}"
                  f"{leak:>11.1e}{('met' if some['shift'] < 0 else '-'):>12}"
                  f"{('met' if some['full_upper'] < 0.5 else '-'):>15}")
    print("    (mass shift criterion: shift for \"some\" < 0; mass position criterion: P(all-region | \"some\") < 1/2.")
    print("     The delta-like base world prior has no row at Part D's Lambda, since it needs this one.)")
    print()


def headroom_report(rows):
    """Room to fall and to rise in all-region mass, for every row and utterance."""
    print("  HEADROOM, EVERY ROW. A zero shift where q_lit holds none of the all-region, or all")
    print("  of it, is saturation and not restraint, so read these columns before the shifts.")
    print(f"    {'p_0':<13}{'utterance':<10}{'P(all) base':>13}{'P(all) full':>13}{'shift':>10}"
          f"{'room to fall':>14}{'room to rise':>14}")
    for name, entry in rows.items():
        for y in UTTERANCES:
            row = entry["utterances"][y]
            print(f"    {(name if y == UTTERANCES[0] else ''):<13}{y:<10}"
                  f"{row['baseline_upper']:>13.6f}{row['full_upper']:>13.6f}{row['shift']:>+10.4f}"
                  f"{row['baseline_upper']:>14.2e}{1.0 - row['baseline_upper']:>14.2e}")
    print()


def delta_like_row_report(net, rows):
    """The delta-like row's headroom and mechanism (Text cell 4 Part C, Eqs. 23-24).

    `net` is the default network: Eq. (23) is also checked at its theta_u*."""
    upper_region = (net.zeta >= net.theta_L).to(net.dtype)
    entry = rows["delta (all)"]
    print("  THE DELTA-LIKE ROW. This is the case the criterion is under the most")
    print("  pressure in: the base world prior already puts %.1f%% of its mass in the all-region,"
          % (100 * entry["prior_upper_mass"]))
    print("  so the literal listener, given \"some\", an entry that does NOT")
    print("  exclude the all-region, stays there. Leak stays below")
    print("  %.0e, so the lexicon is holding and the probe is measuring the utility"
          % max(entry["utterances"][y]["leak"] for y in UTTERANCES))
    print("  level rather than ell_0 overriding the entry.")
    print()
    print(f"    {'utterance':<10}{'P(all) base':>13}{'P(all) full':>13}{'shift':>10}"
          f"{'room to fall':>14}{'room to rise':>14}")
    for y in UTTERANCES:
        row = entry["utterances"][y]
        print(f"    {y:<10}{row['baseline_upper']:>13.6f}{row['full_upper']:>13.6f}"
              f"{row['shift']:>+10.4f}{row['baseline_upper']:>14.2e}"
              f"{1.0 - row['baseline_upper']:>14.2e}")
    print()
    some_row, all_row = entry["utterances"]["some"], entry["utterances"]["all"]
    shed_some = 1.0 - some_row["full_upper"] / some_row["baseline_upper"]
    print("  READ THE HEADROOM BEFORE THE SHIFTS. \"no\" has essentially no")
    print("  all-region mass to lose and \"all\" essentially none to gain, so their")
    print("  zeros are saturation, not restraint; only \"some\" is free to move in")
    print("  both directions, and it is the one that moves. What is NOT saturation")
    print(f"  is that \"all\" keeps its all-region mass (it moves by {all_row['shift']:+.1e})")
    print(f"  while \"some\" sheds {shed_some:.1%} of its own: both had room to fall, and only")
    print("  one fell. That difference is the even-parity (width) coordinate of their")
    print("  lexical projections, which is the one place the two entries differ.")
    print()

    # === the mechanism Text cell 4 Part C reads off this row (Eqs. 23-24) ===
    probe = entry["network"]
    couplings = {y: probe.project(probe.base_log_prior - probe.lexical_field(y))
                 for y in ("some", "all")}

    def eq23_gap(model):
        """max |phi_S*(theta) - phi_S*(0) - (sigma_L theta / S) B phi_u*| at theta_u*."""
        share = model.sigma_lexical / (model.sigma_lexical + model.sigma_state)
        worst = 0.0
        for y in UTTERANCES:
            phi_S, phi_u = model.closed_form_fixed_point(y)
            phi_S_zero = model.closed_form_fixed_point(y, theta_u=0.0)[0]
            worst = max(worst, float((phi_S - phi_S_zero - share * model.theta_u
                                      * torch.mv(model.basis, phi_u)).abs().max()))
        return worst

    field = probe.base_log_prior - probe.lexical_field("some")
    limit = 0.5 * field + 0.5 * torch.mv(probe.basis, probe.project(field))

    def shift_at(belief):
        return (float((probe.weights * belief * upper_region).sum())
                - some_row["baseline_upper"])

    print("  THE MECHANISM (Text cell 4 Part C, Eqs. 23-24), on this row")
    print(f"    ell_0's mode in zeta sits at log(alpha) = {math.log(DELTA_ALL_ALPHA):.2f}")
    print("    c_y = B^T W (ell_0 - phi_L), as (odd tilt, even width): "
          + "; ".join(f"{y} [{float(c[0]):+.2f}, {float(c[1]):+.2f}]"
                      for y, c in couplings.items()))
    print(f"    Eq. (23) holds to {eq23_gap(net):.1e} at the Gaussian theta_u* and to "
          f"{eq23_gap(probe):.1e} at this row's")
    print(f"    Eq. (24)'s limit gives a shift for \"some\" of "
          f"{shift_at(probe.read_out(limit)):+.4f}; this row's theta_u* = "
          f"{probe.theta_u:.2f} gives {some_row['shift']:+.4f}")
    sweep = (-1e7, -1e3, -1e2, -10.0, -1.0, -0.1, -1e-8,
             0.0, 1e-8, 0.1, 1.0, 10.0, 1e2, 1e3, 1e7)
    shifts = {theta: shift_at(probe.read_out(
        probe.closed_form_fixed_point("some", theta_u=theta)[0])) for theta in sweep}
    print("    shift for \"some\" at fixed theta_u, a control:")
    for part in (sweep[:7], sweep[7:]):
        print("      " + "  ".join(f"{theta:g}: {shifts[theta]:+.4f}" for theta in part))
    nonzero = [value for theta, value in shifts.items() if theta != 0.0]
    print(f"    negative at every theta_u != 0 above: {all(value < 0 for value in nonzero)}; "
          f"least negative {max(nonzero):+.4f}, most negative {min(nonzero):+.4f}")
    print()


def raised_lambda_control(net, rows, strong_rows):
    """The default Gaussian base world prior at the delta-like row's Lambda, against Part D's row, and the
    base world priors under which raising Lambda alone carries the shift for "some" below zero."""
    upper_region = (net.zeta >= net.theta_L).to(net.dtype)
    # === is it ell_0 or the raised Lambda? a control for Part C ===
    raised = net.respawn(lexical_strength=DELTA_ALL_LAMBDA)
    belief = raised.read_out(raised.closed_form_fixed_point("some")[0])
    literal = raised.read_out(raised.literal_fixed_point("some"))
    shifted = float((raised.weights * belief * upper_region).sum()
                    - (raised.weights * literal * upper_region).sum())
    print(f"  the default Gaussian base world prior at the delta row's Lambda = {DELTA_ALL_LAMBDA:.0f}: "
          f"theta_u* = {raised.theta_u:.2f}, shift for \"some\" = {shifted:+.4f},")
    print(f"  against {rows['gaussian']['utterances']['some']['shift']:+.4f} at "
          f"Lambda = {net.lexical_strength:.0f}. Raising Lambda alone moves this shift")
    print("  AWAY from the mass shift criterion. At the same Lambda the delta-like row reads "
          f"{strong_rows['delta (all)']['utterances']['some']['shift']:+.4f}, so between")
    print("  those two rows the difference is ell_0's (Text cell 4 Part C).")
    crossed = [name for name, entry in strong_rows.items() if name in rows
               and rows[name]["utterances"]["some"]["shift"] >= 0.0
               > entry["utterances"]["some"]["shift"]]
    print("  Raising Lambda alone carries the shift for \"some\" below zero under: "
          + (", ".join(crossed) if crossed else "no base world prior") + ".")
    print()


def strong_lambda_report(net, base_rows):
    """Part D's evaluations with every base world prior at STRONG_LAMBDA.

    `net` is the default network and `base_rows` Part D's rows from `base_prior_sweep`.
    Returns the strong network, its rows, and its realizability rows."""
    strong = net.respawn(lexical_strength=STRONG_LAMBDA)
    priors = strong_lambda_priors()
    print(f"PART D AT A STRONG LAMBDA: EVERY PRIOR AT LAMBDA = {STRONG_LAMBDA:.0f}")
    print()
    rows = base_prior_sweep(strong, priors)
    strong_lambda_comparison(base_rows, rows)
    headroom_report(rows)
    delta_like_row_report(net, rows)
    raised_lambda_control(net, base_rows, rows)
    realizability = realizability_report(strong, priors)
    return {"network": strong, "rows": rows, "realizability": realizability}


def cell_margin_report(realizability):
    """How far below the cell of "all" the mode of phi_S*("some") sits, every row at theta_u*."""
    print("  THE PEAK OF phi_S*(\"some\") AGAINST THE CELL OF \"all\", every row at theta_u*")
    print(f"    {'p_0':<13}{'mode (s)':>10}{'nodes below':>13}{'inside against outside':>24}")
    for name, row in realizability.items():
        probe = row["network"]
        phi_S = probe.closed_form_fixed_point("some")[0]
        k = int(torch.argmax(phi_S))
        inside = probe.zeta >= probe.theta_L
        first_inside = int(torch.nonzero(inside)[0])
        margin = float(phi_S[inside].max() - phi_S[~inside].max())
        print(f"    {name:<13}{float(logistic(probe.zeta[k])):>10.4f}"
              f"{max(0, first_inside - k):>13d}{margin:>+24.3f}")
    print("    (nodes below: grid steps from the mode up to the first node inside the cell, 0 if")
    print("     the mode is inside it; the last column is the largest phi_S* inside the cell")
    print("     less the largest outside it)")
    print()


# ============================================================================
# RUN
# ============================================================================

strong = strong_lambda_report(evaluation_network, prior_rows)


# ---- figures ----

try:
    import matplotlib.pyplot as plt

    strong_network, strong_rows = strong["network"], strong["rows"]
    proportion = logistic(strong_network.zeta).numpy()
    ds_dzeta = proportion * (1.0 - proportion)          # densities in s, as in Code Cell 2

    def to_s_density(field):
        return field.numpy() / ds_dzeta

    all_region_start = float(logistic(torch.tensor(strong_network.theta_L)))

    # All three figures use a LOGARITHMIC vertical axis. At this Lambda every entry holds,
    # so the settled densities pile against the ends of the scale and span several orders
    # of magnitude between panels; on a linear axis most panels would be a flat line under
    # one spike. The comparison to read is the full network against the literal listener
    # WITHIN each panel. The upper limit is set from the curves themselves.
    def draw(axis, prior, row, title):
        axis.plot(proportion, to_s_density(prior), color="0.6", linewidth=1.0,
                  label="base world prior")
        axis.plot(proportion, to_s_density(row["baseline_q"]), linestyle="--",
                  label="literal listener $q_{\\mathrm{lit}}$")
        axis.plot(proportion, to_s_density(row["full_q"]), label="full network")
        axis.axvspan(all_region_start, 1.0, alpha=0.12, color="tab:red")
        axis.set_yscale("log")
        axis.set_title(title, fontsize=9)
        axis.set_xlabel("proportion $s$")

    def ceiling(curves):
        return 3.0 * max(float(to_s_density(curve).max()) for curve in curves)

    # first figure: the three utterances under the default Gaussian base world prior
    gaussian = strong_rows["gaussian"]
    gaussian_prior = torch.exp(gaussian["network"].base_log_prior)
    figure, axes = plt.subplots(1, 3, figsize=(13, 3.6), sharey=True)
    for axis, utterance in zip(axes, UTTERANCES):
        row = gaussian["utterances"][utterance]
        draw(axis, gaussian_prior, row, f'"{utterance}"  (shift {row["shift"]:+.4f})')
    top = ceiling([gaussian_prior] + [gaussian["utterances"][y][key] for y in UTTERANCES
                                      for key in ("baseline_q", "full_q")])
    for axis in axes:
        axis.set_ylim(1e-4, top)
    axes[0].set_ylabel("read-out density $q(s)$, log scale")
    axes[0].legend(fontsize=8, loc="upper center")
    figure.suptitle(
        "Settled belief against the literal listener under the Gaussian base world prior at "
        "$\\Lambda=%g$, $\\theta_u^\\ast=%.2f$ (shaded: the region where \"all\" is true)"
        % (STRONG_LAMBDA, gaussian["theta_u"]), fontsize=10)
    figure.tight_layout()
    plt.show()

    # second figure: "some" under all five base world priors, which at this Lambda share their panels
    prior_figure, prior_axes = plt.subplots(
        1, len(strong_rows), figsize=(3.4 * len(strong_rows), 3.4), sharey=True
    )
    curves = []
    for axis, (name, entry) in zip(prior_axes, strong_rows.items()):
        row = entry["utterances"]["some"]
        prior = torch.exp(entry["network"].base_log_prior)
        draw(axis, prior, row, f"{name}\n(shift {row['shift']:+.4f})")
        curves += [prior, row["baseline_q"], row["full_q"]]
    top = ceiling(curves)
    for axis in prior_axes:
        axis.set_ylim(1e-4, top)
    prior_axes[0].set_ylabel("read-out density $q(s)$, log scale")
    prior_axes[0].legend(fontsize=8, loc="upper left")
    prior_figure.suptitle(
        '"some" under the five base world priors at $\\Lambda=%g$, each at its own learned '
        "$\\theta_u^\\ast$" % STRONG_LAMBDA, fontsize=10)
    prior_figure.tight_layout()
    plt.show()

    # third figure: the three utterances under the delta-like base world prior, the sharp base world prior the
    # criterion is under most pressure from, and the case Code Cell 4 is built around
    delta_entry = strong_rows["delta (all)"]
    delta_prior = torch.exp(delta_entry["network"].base_log_prior)
    delta_peak = max(
        float(to_s_density(curve).max())
        for y in UTTERANCES
        for curve in (delta_entry["utterances"][y]["baseline_q"],
                      delta_entry["utterances"][y]["full_q"], delta_prior)
    )
    delta_figure, delta_axes = plt.subplots(1, 3, figsize=(13, 3.6), sharey=True)
    for axis, utterance in zip(delta_axes, UTTERANCES):
        row = delta_entry["utterances"][utterance]
        draw(axis, delta_prior, row, f'"{utterance}"  (shift {row["shift"]:+.4f})')
        axis.set_ylim(1e-4, 3.0 * delta_peak)
    delta_axes[0].set_ylabel("read-out density $q(s)$, log scale")
    delta_axes[0].legend(fontsize=8, loc="upper left")
    delta_some = delta_entry["utterances"]["some"]
    delta_figure.suptitle(
        "The delta-like base world prior, $\\mathrm{Beta}(%g,1)$ at "
        "$\\Lambda=%g$ (shaded: the region where \"all\" is true).\nUnder "
        "\"some\" the literal listener holds %.1f%% of its mass inside that region "
        "and the full network %.1f%%"
        % (DELTA_ALL_ALPHA, DELTA_ALL_LAMBDA, 100 * delta_some["baseline_upper"],
           100 * delta_some["full_upper"]), fontsize=10)
    delta_figure.tight_layout()
    plt.show()

except ImportError:
    print("matplotlib unavailable; skipping the figures")


# ---- the delta read-out, after the q figures ----

delta_readout_report(strong["network"], strong["realizability"])
cell_margin_report(strong["realizability"])
````

## <a id="tc5" name="tc5"></a>Text cell 5: notes after the evaluation

Two matters that belong after the evaluation rather than inside the specification: why the model's
expressiveness depends on the utility basis dimension in a way its convergence proofs do not,
and a report on the settings of $\mu_u$ probed.

---

### <a id="tc5-parta" name="tc5-parta"></a>Part A. the utility basis dimension $m$

Text cell 3 refers to this part. Every result in Sec. 8 is stated for arbitrary fixed $B$ and arbitrary
$m$, and no proof there constrains the choice; what follows is about what the model can *express*.

*The model's expressive power is sensitive to the column size of $B$.* The $u$ level's only access
to the lexical field is through the $m$ **lexical projections**

$$\kappa_y=B^{\mathsf T}W\varphi_{L,y}\in\mathbb R^m,
\tag{29}$$

so two utterances whose projections coincide are indistinguishable to that level, however different
their exclusion sets. $m$ is therefore a substantive commitment.

**The letter is $\kappa$ and not $c$.** Appendix B writes
$c_y=B^{\mathsf T}W(\ell_0-\varphi_{L,y})$ for the vector that enters the closed forms of
Eqs. (15)–(16) and the stationarity equation Eq. (B2). Both are vectors in $\mathbb R^m$ and both
would naturally be called "the projections", so they are given different letters here. They differ by
an utterance-independent offset,

$$c_y=B^{\mathsf T}W\ell_0-\kappa_y,\qquad B^{\mathsf T}W\ell_0=[\,0,\;-18.5996\,]
\tag{30}$$

at the settings used throughout. The parity argument below turns on $\kappa_y$, because $\ell_0$
contributes the same amount to every entry and therefore cannot separate them; everything in
Appendix B turns on $c_y$.

#### <a id="tc5-m1-degenerate" name="tc5-m1-degenerate"></a>$m=1$ is degenerate

The three exclusion sets of Eq. (5) are built from a single pair of thresholds
$\pm\theta_L$. (The endpoints $s=0,1$ lie outside the open interval and carry no quadrature mass, so they
are represented as the neighbourhoods $|\zeta|\ge\theta_L$; $\theta_L$ is semantic, and
Appendix A gives its denotation.) Their pairwise differences are consequently very symmetric:

$$\chi_{\textit{all}}-\chi_{\textit{some}}=\mathbb 1_{(-\theta_L,\,\theta_L)},
\qquad
\chi_{\textit{no}}-\chi_{\textit{all}}=\mathbb 1_{(\theta_L,\,\infty)}-\mathbb 1_{(-\infty,\,-\theta_L)}.
\tag{31}$$

The first is an interval symmetric about the origin; the second is a mirror-image pair carrying
opposite signs. On a grid symmetric about $\zeta=0$, therefore, **any single basis function of definite
parity annihilates one of them**. An odd $b$ gives $\langle\chi_{\textit{all}}-\chi_{\textit{some}},
b\rangle=0$ and cannot separate *some* from *all*; an even $b$ gives
$\langle\chi_{\textit{no}}-\chi_{\textit{all}},b\rangle=0$ and cannot separate *no* from *all*. The
identities are exact, and the argument is one about parity rather than about any particular profile:
$\zeta$, $\zeta^3$ and $\sin\zeta$ all collapse $\{\textit{some},\textit{all}\}$, while $\zeta^2$,
$|\zeta|$ and $\cos\zeta$ all collapse $\{\textit{no},\textit{all}\}$.

The consequence reaches past the $u$ level. That level is the only site in the chain at which an
implicative reading could be computed, since $\varphi_L$ carries the lexical entry and no other quantity does.
With $m=1$ and a basis of definite parity, two of the three utterances receive *identical* utility
states, at every $\theta_u$, $\Lambda$ and $\mu_u$. Their beliefs still differ, since Eq. (15) carries
$\ell_0-\varphi_L$ into $\varphi_S^\ast$ without passing through the $u$ level; what is lost is the
utility level's purchase on the pair, so the contrast between them no longer depends on $\theta_u$.
Appendix C §6 measures that independence and records it as Eq. (C3).

A profile of no definite parity escapes the argument and does separate all three, but it confounds
into one coordinate the two things a basis can do to a log-density, which Part B §2 separates as
tilt and width. The implementation therefore takes $m=2$, the least the parity argument permits.
Its two columns are a linear and a quadratic profile in $\zeta$: the linear one separates *no* from
$\{\textit{some},\textit{all}\}$ and the quadratic separates *some* from
$\{\textit{no},\textit{all}\}$, so all three utterances receive distinct projections. Dropping the
constant column is deliberate, since a constant direction shifts $\varphi_S$ uniformly and is invisible to
the read-out (Eq. 12), so it would consume a degree of freedom without affecting the belief.

Since $m=2$, the terminating prior is an ordered pair $\mu_u=(\mu_{u,1},\mu_{u,2})\in\mathbb R^2$, one
weight per basis direction. Part B below reads what each component does.

---

### <a id="tc5-partb" name="tc5-partb"></a>Part B. probe report on the terminating prior $\mu_u$

Six settings of $\mu_u$ run against Code Cell 1. The configuration is held fixed
throughout: Gaussian base world prior, $\Lambda=8$, $\theta_L=2.9444$ ($n=10$), $m=2$,
every $\sigma=1$, $K=101$, and $\theta_u$ **held at $1$ as a control**. Holding it is the point of
the report: $\theta_u^\ast$ is itself a function of $\mu_u$ (Appendix B), so a table at each
setting's own $\theta_u^\ast$ mixes $\mu_u$'s direct path into $\varphi_S$ with its path through
the learned parameter, and only a fixed $\theta_u$ isolates the first. The invariance of §1 holds at
*any* fixed value, and $1$ is chosen for legibility, not because the model sits there; it does not,
and it does not start there either, since the flow starts at $\theta_u(0)=0$. Every table below is
therefore reported twice, once at the control and once with $\theta_u^\ast$ re-learned for each
setting, and the two readings are different in kind. Every quantity below is evaluated in closed form (Eqs. 15-16, exact by
§8.6), so no integration step enters. Reported statistics are as defined at Eqs. (25)-(27). The base world prior is grid-renormalized, as Code Cell 1 renormalizes it.

#### <a id="tc5-1" name="tc5-1"></a>1. The structural result

$\mu_u$ is a **pure common-mode shift**. It translates all three beliefs together and can
move no contrast between them.

From Eq. (16), and writing $S=\sigma_L+\sigma_S$, $G=B^{\mathsf T}WB$:

$$\frac{\partial\varphi_u^\ast}{\partial\mu_u}
=\Big(I+\tfrac{\sigma_u\theta_u^2}{S}G\Big)^{-1},
\qquad
\frac{\partial\varphi_S^\ast}{\partial\mu_u}
=\frac{\sigma_L\theta_u}{S}\,B\Big(I+\tfrac{\sigma_u\theta_u^2}{S}G\Big)^{-1}.
\tag{32}$$

Neither expression contains $\varphi_L$, and $\varphi_L$ is the only utterance-dependent
quantity in the model. At the settings above ($\sigma=1$, $\theta_u=1$, $G=I$, $S=2$) this
collapses to

$$\frac{\partial\varphi_S^\ast}{\partial\mu_u}=\tfrac13 B,
\tag{33}$$

so raising $\mu_{u,j}$ by one unit adds exactly $\tfrac13 b_j$ to the settled log-density, *the same field for every utterance*. Eq. (33) is the control's coefficient. At the default's own
$\theta_u^\ast=-28.4375$ the same expression gives $-0.0351$, so a unit of $\mu_{u,j}$ then adds
$-0.0351\,b_j$: about a tenth of the size and of the opposite sign, because the $\theta_u^2$ in the
denominator of Eq. (32) grows faster than the $\theta_u$ in its numerator. Whatever $\mu_u$ does
directly, learning makes it an order of magnitude weaker and turns it around. Verified numerically to $2.6\times10^{-15}$ (Code Cell 3).

The consequence is exact and worth stating as a test rather than a remark: the differences
$\varphi_S^\ast(u_1)-\varphi_S^\ast(u_2)$ are **bit-identical** across all six settings,
agreeing to $3.6\times10^{-15}$, which is machine epsilon and not a tolerance.

**At fixed $\theta_u$, $\mu_u$ therefore moves no contrast between utterances**, and this is a
measurement, not a reading. The qualifier is necessary: $\theta_u^\ast$ depends on $\mu_u$
(Appendix B), so once $\theta_u$ is learned, $\mu_u$ reaches contrasts through it. Sweeping
$\mu_u=v\,[1,1]$ from $v=0.2$ to $v=3.0$ carries $\theta_u^\ast$ from $-143.2577$ to $-8.8983$, and
with it $\max\lvert\varphi_S^\ast(\textit{all})-\varphi_S^\ast(\textit{some})\rvert$ from $5.8608$
to $5.8152$, a movement of $0.8\%$ that is nevertheless fifteen orders of magnitude above the
$3.6\times10^{-15}$ measured at fixed $\theta_u$. Code Cell 3 measures the same thing across the six
settings rather than along a ray: re-learning $\theta_u^\ast$ for each of them moves the $\varphi_S$
contrasts by $3.19\times10^{-2}$, against the $3.6\times10^{-15}$ they move at fixed $\theta_u$. The
tables in this report are the control, and the learned tables beside them are the model.

**What the structural result does not settle.** The mass shift criterion reported in Text cell 4 Part C is not a contrast between
utterances: $\Delta_y$ compares one utterance's belief against its own literal listener
$q_{\mathrm{lit},y}$ (Eq. 37), and it is a $q$-mass (Eq. 27), nonlinear in $\varphi_S$. The
invariance above therefore does not transfer to it, and measurement confirms that it does not:
across the same six settings at $\theta_u=1$, $\Delta_{\textit{some}}$ runs $+0.0080$ to $+0.0109$
and $\Delta_{\textit{all}}$ runs $-0.3598$ to $-0.2914$, each varying by more than a fifth of its
own size, while the $\varphi_S$ contrasts stay bit-identical throughout. With $\theta_u^\ast$
re-learned per setting the same two quantities run $+0.0008$ to $+0.0009$ and $-0.4514$ to
$-0.4641$: the spread in $\Delta_{\textit{some}}$ collapses by an order of magnitude, and what the
learned network shows is not a $\mu_u$ effect but its absence. So §1 does
not settle in closed form whether $\mu_u$ is worth sweeping for an implicature under that criterion.
What it does settle is narrower and still worth having: $\mu_u$ moves nothing the model represents *about the relation between* utterances.
§4 makes that argument in full.

#### <a id="tc5-2" name="tc5-2"></a>2. What each position of the ordered pair does

The two basis columns have **definite and opposite parity**, which is what makes $m=2$ able
to separate all three utterances at all (Part A):

| | parity | $\langle\mathbf 1,b_j\rangle$ | $b_j(-6)$ | $b_j(0)$ | $b_j(+6)$ | shape |
|---|---|---|---|---|---|---|
| $b_1$ | **odd** | $-1.4\times10^{-17}$ | $-0.500$ | $0.000$ | $+0.500$ | monotone ramp |
| $b_2$ | **even** | $-3.6\times10^{-16}$ | $+0.645$ | $-0.323$ | $+0.645$ | symmetric "U" |

**Position 1: $\mu_{u,1}$ sets TILT.** Adding $\tfrac13b_1$ is adding a monotone ramp in
$\zeta$, which slides log-density from one end of the scale to the other. It moves every
utterance in the *same* direction and leaves spread essentially untouched.

**Position 2: $\mu_{u,2}$ sets WIDTH.** Adding $\tfrac13b_2$ raises both tails and lowers
the centre, pushing $q$-mass outward. It moves each utterance *away from the centre*, so its
effect on $\mathbb E[s]$ has opposite sign for beliefs sitting below and above $s=0.5$.

Both readings are the control's. At a learned $\theta_u^\ast$ the coefficient is $-0.0351$ rather
than $+\tfrac13$, so each effect runs backwards and at about a tenth of the size: tilt still tilts
and width still widens, but a positive $\mu_{u,1}$ now lowers the scale rather than raising it. The
sign of $\mu_u$ is invisible altogether under learning, for a reason §6 makes exact: the network
re-learns $\theta_u^\ast(-\mu_u)=-\theta_u^\ast(\mu_u)$, and the product $\theta_uB\mu_u$ that
reaches $\varphi_S$ is unchanged, so $\mathbb E[s]$ at $-v$ equals $\mathbb E[s]$ at $+v$ exactly
(relative gap $0.0$ in every row Code Cell 3 reports).

**The two entries share one scale.** Each is a coefficient on an orthonormal column, so
$\theta_u\mu_{u,j}b_j$ is a log-density field of quadrature norm $\theta_u\lvert\mu_{u,j}\rvert$:
a unit of either adds the same amount of log-density, distributed differently. One unit of
$\mu_{u,1}$ buys $1.00$ nat of end-to-end tilt; one unit of $\mu_{u,2}$ buys $0.97$ nats of
tail-minus-centre. The entries are therefore commensurable, which is what licenses the isotropic
default $[1,1]$ despite the two meaning different things.

Against the fields it competes with, $\mu_u$ is small:

| $\lVert\ell_0\rVert$ | $\lVert\varphi_L\rVert$ (*no*, *all*) | $\lVert\varphi_L\rVert$ (*some*) | $\lVert\mu_u\rVert$ |
|---|---|---|---|
| $30.34$ | $23.92$ | $13.99$ | $1.41$ |

At the control's $\theta_u=1$ the terminating prior sits some $21\times$ below $\ell_0$ in norm,
and $\varphi_u$ is accordingly set by $\varepsilon_S$ from below rather than by $\mu_u$ from above.
That reading belongs to the control and does not survive learning: the field that terminates the
chain is $\theta_uB\mu_u$, not $\mu_u$, and at $\theta_u^\ast=-28.4375$ its norm is $40.2167$,
$1.33\times$ the norm of $\ell_0$ rather than a twenty-first of it. What stays true at either value
is why the degeneracy at $\mu_u=0$ (§5, Appendix B) needed a closed-form analysis to detect: at
$\theta_u=1$ the quantity whose vanishing destroys identifiability is an order of magnitude below
everything around it, and no reading of the tables would have shown it.

**The midpoint of both scales is $0$, which Text cell 3 §3 forbids.** Each effect inverts under
$\mu_{u,j}\to-\mu_{u,j}$, and since $\varphi_S^\ast$ is affine in $\mu_u$ (§1) a signed pair
straddles the zero value exactly:

$$\varphi_S^\ast(+v\,e_j)+\varphi_S^\ast(-v\,e_j)=2\,\varphi_S^\ast(0),
\tag{34}$$

verified to $7.1\times10^{-15}$ for $j=1,2$ and $v\in\{0.25,0.5,1,2\}$ (Code Cell 3), again at the
$\theta_u=1$ control. Under learning the identity cannot even be posed: its right-hand side needs
$\varphi_S^\ast(0)$, and at $\mu_u=0$ there is no $\theta_u^\ast$ to evaluate it at (Appendix B).
What the learned sweep shows instead is the symmetry, $\theta_u^\ast(-v)=-\theta_u^\ast(+v)$ and
identical beliefs at $\pm v$. Reading $\mathbb E[s]$ across the sweep at the control separates the
two shapes cleanly:

| entry | $v$ | *no* $(-v)$ | *no* $(+v)$ | *some* $(-v)$ | *some* $(+v)$ | *all* $(-v)$ | *all* $(+v)$ |
|---|---|---|---|---|---|---|---|
| $\mu_{u,1}$ | $0.5$ | 0.2573 | 0.2713 | 0.5328 | 0.5401 | 0.7287 | 0.7427 |
| $\mu_{u,1}$ | $2.0$ | 0.2371 | 0.2932 | 0.5217 | 0.5511 | 0.7068 | 0.7629 |
| $\mu_{u,2}$ | $0.5$ | 0.2693 | 0.2592 | 0.5360 | 0.5369 | 0.7307 | 0.7408 |
| $\mu_{u,2}$ | $2.0$ | 0.2843 | 0.2438 | 0.5348 | 0.5383 | 0.7157 | 0.7562 |

$\mu_{u,1}$ moves all three utterances the same way, up for $+v$ and down for $-v$. $\mu_{u,2}$
moves *no* and *all* in opposite directions, outward from the centre for $+v$ and inward for $-v$.

Zero being both the natural centre and the forbidden value is a property of the architecture rather
than an inconvenience: a tonic node carrying weight zero terminates nothing, so the chain has to end
in a commitment. Bogacz's $v_p$ carries the same requirement, never tested at zero. A sweep of
$\mu_u$ should therefore be symmetric about zero with zero excluded, and §6 does it on that shape, which turns out to remove rather more than a point.

#### <a id="tc5-3" name="tc5-3"></a>3. The six settings

| $\mu_u$ | $\mathbb E[s]$ *no* | *some* | *all* | $\mathrm{sd}(s)$ \| *some* | *all*−*some* | *some*−*no* |
|---|---|---|---|---|---|---|
| $[0,0]$ *(degenerate; see §5)* | 0.2643 | 0.5364 | 0.7357 | 0.2309 | $+0.1993$ | $+0.2722$ |
| $[0,1]$ | 0.2541 | 0.5373 | 0.7459 | 0.2327 | $+0.2086$ | $+0.2833$ |
| $[1,0]$ | 0.2785 | 0.5438 | 0.7496 | 0.2306 | $+0.2058$ | $+0.2653$ |
| $[1,1]$ **(default)** | 0.2683 | 0.5448 | 0.7597 | 0.2324 | $+0.2149$ | $+0.2765$ |
| $[1,2]$ | 0.2580 | 0.5459 | 0.7699 | 0.2342 | $+0.2239$ | $+0.2879$ |
| $[2,1]$ | 0.2830 | 0.5523 | 0.7729 | 0.2320 | $+0.2206$ | $+0.2693$ |

Computed by Code Cell 3 at $n=10$ ($\theta_L=2.9444$) and $\theta_u=1$, the control; the row
$[0,0]$ is retained only as the degenerate reference $\mu_u$ may not take.

Code Cell 3 prints the same six settings at their own $\theta_u^\ast$ beside this table, and the two
read differently. The learned $\theta_u^\ast$ swings from $-35.993$ to $+127.940$ across the five
admissible settings, and yet $\mathbb E[s]$ under *some* moves only between $0.5671$ and $0.5692$,
against the $0.5373$ to $0.5523$ the control spreads over. The $[0,0]$ row has no learned
counterpart at all: Code Cell 3 reports it as degenerate rather than as a number, since no
$\theta_u^\ast$ exists there. So the axes read below are the control's; under learning $\mu_u$ is
nearly inert, and the tilt and width it buys are spent on $\theta_u^\ast$ instead.

Reading the two axes out of the table, holding the other position fixed:

**Tilt**, read along $[0,1]\to[1,1]\to[2,1]$, gives *no* $0.2541\to0.2683\to0.2830$, *some*
$0.5373\to0.5448\to0.5523$, *all* $0.7459\to0.7597\to0.7729$. Every utterance rises, by
$\approx\!+0.0075$ per unit for *some*. Spread is nearly static ($0.2327\to0.2324\to0.2320$).
This is a location effect.

**Width**, along $[1,0]\to[1,1]\to[1,2]$: *no* $0.2785\to0.2683\to0.2580$ (falls), *all*
$0.7496\to0.7597\to0.7699$ (rises), and spread grows monotonically
($0.2306\to0.2324\to0.2342$). The two extreme utterances move in *opposite* directions,
outward from the centre, which is the signature of a width change rather than a shift and the
behaviour the even parity of $b_2$ predicts.

#### <a id="tc5-4" name="tc5-4"></a>4. Why we do not read the contrast drift as a pragmatic effect

The last two columns drift: the *all*$-$*some* figure ranges $0.1993$–$0.2239$, a spread of
$0.0246$ on a mean of $0.2122$, or about 12% of its own size. That is large enough that it cannot be
waved past, which is why the argument below is made in full. Whether a movement of that kind
counts as a pragmatic effect is not something this project has settled, so what follows is an
argument rather than a classification.

The premises are three, and the first two are measurements rather than judgements.

1. Across the six rows, the only quantity that varies is $\mu_u$, with $\theta_u$ held at $1$ as
   the control of Part B's preamble. The argument below is about that table; at each setting's own
   $\theta_u^\ast$ the drift it explains is an order of magnitude smaller, and premise 2 fails, so
   §6 takes the learned case up separately.

2. The differences $\varphi_S^\ast(u_1)-\varphi_S^\ast(u_2)$ are invariant to $\mu_u$ to
   $3.6\times10^{-15}$ (§1), which is machine epsilon. Whatever the model represents about the relation
   *between* utterances is therefore bit-for-bit unchanged across the table.

3. $\mathbb E[s]$ (Eq. 25) is a nonlinear functional of $\varphi_S$: a softmax, then $\varsigma$.
   Adding one and the same field to three different read-outs consequently moves their means by
   three different amounts.

Premises 1 and 2 leave the nonlinearity of premise 3 as the only available source of the drift, and
since premise 2 is exact to machine epsilon it is not merely the *likeliest* source but the only
one available at any size, this one included. We therefore do not think the drift should be read as
a pragmatic effect: it is a property of the statistic we chose to report, and the model computed no
corresponding change in how it relates the three utterances.

**What the argument does not establish.** This claim presupposes that a pragmatic effect ought to show up as a
contrast in $\varphi_S$, and not in a summary of $\varphi_S$ alone. Someone who located pragmatic content directly in $\mathbb E[s]$, or in
some other functional of the read-out, could read the same six rows differently, and nothing measured
here would contradict them. The argument also does not make a claim about whether the model produces a
pragmatic effect anywhere *else*.

The practical consequence is the same either way, and is the reason the section exists: the table
should not be read without §1 beside it, because the columns move for a reason that has nothing to do
with $\mu_u$ acting on the utterances differently.

#### <a id="tc5-5" name="tc5-5"></a>5. Disposition

$\mu_u$ is fixed at $[1,1]$. It **cannot** be $0$: the terminating contribution $\theta_u B\mu_u$
would vanish for every $\theta_u$, leaving $\theta_u$ unidentifiable (Text cell 3 §3, Appendix B).
At $\mu_u=0$ the terminating level stops regularizing $\theta_u$ altogether.

Nothing is stipulated about $\varepsilon_u$ by this choice: Eq. (11) gives $r_u=\varphi_u-\mu_u$, so
the terminating error unit reports the offset of the utility state from $\mu_u$, which Eq. (19) then
opposes. This is Bogacz's $v_p$ prior, and $\mu_u\neq0$ is his premise as much as ours.

Two things this probe does **not** settle. It says what the two components of $\varphi_u$ do
to the belief, but not whether $m=2$ is *sufficient* as well as necessary, since Part A establishes necessity by a parity argument, within bases of definite parity, and leaves sufficiency open. (The *shape* of the sweep is settled in
§6, which replaces the six lattice points with the plane.) And it generalizes no
further than $\mu_u$. What $\mu_u$ shares with $\ell_0$ is a position in the closed forms rather than a term in $g_S$, which
carries neither of them, since $\ell_0$ sits in $g_L$ (Eq. 9): $\mu_u$ enters Eq. (16) as the terminating prior and $\ell_0$ enters
Eqs. (15)–(16) only inside the combination $\ell_0-\varphi_L$, and $\varphi_S^\ast$ is affine in
each with an utterance-independent coefficient. That is why neither can move a $\varphi_S$ contrast. The
property is theirs alone: every other quantity touching an utterance-dependent one can move a
contrast, and measurably does: $\Lambda$ and $\theta_L$ through $\chi_y$, the variances through the weighting of $\varphi_L$
against $\ell_0$ in Eq. (15), $B$ through the lexical projections $\kappa_y$ of Eq. (29), and the slow parameter
$\theta_u$ through $B\varphi_u$. Note that $\theta_u$ is *not* a fixed quantity: it is
variational on the slow timescale (Eq. 20), and belongs in this list for its effect, not its
status.

#### <a id="tc5-6" name="tc5-6"></a>6. The sweep on the shape the constraints license

The six settings above are lattice points in one quadrant, which is the wrong shape for this
parameter on three counts. $\mu_u$ lives in $\mathbb R^2$, so a sweep of it is a sweep of a *plane*.
The excluded set is not a point: Text cell 3 §3 requires $\mu_u\neq0$, and Appendix B sharpens that to
$\langle\mu_u,\sum_y c_y\rangle\neq0$, which is a requirement on **direction** and removes two whole
rays. And the pair $(\theta_u,\mu_u)$ carries an exact symmetry, so half the plane is redundant.
Code Cell 3 sweeps it in polar coordinates, at five radii and twelve directions.

**The symmetry, first, because it halves the work.** $\varphi_u^\ast$ solves
$(I+\sigma_u\theta_u^2G/S)\,\varphi_u^\ast=\mu_u+\sigma_u\theta_u c/S$. The matrix sees only
$\theta_u^2$, so $(\theta_u,\mu_u)\mapsto(-\theta_u,-\mu_u)$ negates the right-hand side, sends
$\varphi_u^\ast\mapsto-\varphi_u^\ast$, and leaves $g_S=\theta_u B\varphi_u^\ast$, and therefore $\varphi_S^\ast$, untouched:

$$\varphi_S^\ast(\theta_u,\mu_u)=\varphi_S^\ast(-\theta_u,-\mu_u),
\tag{35}$$

verified to $3.6\times10^{-15}$ over the whole sweep. So the plane need only be covered at one sign
of $\theta_u$; the other sign is a reflection, not a second experiment. It shows up directly in the
sweep as $\theta_u^\ast(\text{angle}+180^\circ)=-\theta_u^\ast(\text{angle})$, exact down the table.

**The excluded rays.** At the settings here $\sum_y c_y=[+9.1191,\,-50.0310]$, so
$\langle\mu_u,\sum_y c_y\rangle=0$ on the two rays at $10.33^\circ$ and $190.33^\circ$. Approaching
them, $\theta_u^\ast$ diverges: at $r=1$ the sweep reads $127.94$ at $0^\circ$ against $23.40$ at
$90^\circ$, and at $r=0.25$ it reads $514.17$ at $0^\circ$. That is Appendix B's requirement made visible: what
$\mu_u\neq0$ excludes is a point, what identifiability actually excludes is a direction, and the
default $\mu_u=[1,1]$ sits $34.67^\circ$ from the nearer ray. That margin is a property of this
$\ell_0$ and this $\Lambda$, since $\sum_y c_y$ moves with both.

**What the plane confirms and what it adds.** At the fixed $\theta_u=1$ of the control the $\varphi_S$ contrast is
invariant over all sixty settings to $3.6\times10^{-15}$, so the
six-row result of §1 was a sample of a plane-wide fact. What varies is $\theta_u^\ast$, and it varies by a factor of $90$ across the
plane, from $5.729$ at $r=4,\,90^\circ$ to $514.17$ at $r=0.25,\,0^\circ$; the *all*$-$*some*
contrast it carries ranges $13.6510$ to $13.9242$. Set against $3.6\times10^{-15}$ at fixed
$\theta_u$, that is the whole of $\mu_u$'s reach: it touches the utterances through the slow
parameter and by no other route.

<a id="code3" name="code3"></a>

````python
# === Code Cell 3: the probes reported in Text cell 5. EVALUATION ===
#
# Text cell 5 states two results in prose and tables. This cell computes them, so that
# the numbers there are reproducible rather than quoted.
#
#   Part A, the utility basis dimension: any single basis function of definite
#             parity collapses a pair of utterances, so m = 1 cannot separate all three.
#   Part B, the terminating prior mu_u: at fixed theta_u it moves no contrast, and
#             the drift in the reported E[s] contrasts is read-out curvature. That
#             is a CONTROL, reported at theta_u = 1 as before: the invariance holds at
#             any fixed theta_u, and fixing it isolates mu_u's direct path. Beside
#             it, each setting at its own learned theta_u*, where mu_u reaches the
#             contrasts through theta_u*.
#
# Everything is evaluated in closed form (Eqs. 15-16, exact by Sec. 8.6).

import math
import torch


# ============================================================================
# PART A. THE BASIS DIMENSION
# ============================================================================

def basis_parity_probe(net, verbose=True):
    """Text cell 5 Part A: which utterance pairs a single basis function collapses.

    The u level sees the lexical field only through the lexical projections
    kappa_y = B^T W phi_L (Eq. 29, not Appendix B's c_y = B^T W (ell_0 - phi_L)), so
    two utterances with equal projections are indistinguishable to it.
    The three exclusion sets are built from one pair of thresholds +-theta_L, so their
    pairwise differences are an interval symmetric about the origin and a mirror-image
    pair; an ODD basis function annihilates the first and an EVEN one the second.
    """
    zeta, weights = net.zeta, net.weights
    chi = {y: exclusion_indicator(y, zeta, net.theta_L) for y in UTTERANCES}
    candidates = [
        ("zeta", zeta, "odd"), ("zeta^3", zeta ** 3, "odd"), ("sin(zeta)", torch.sin(zeta), "odd"),
        ("zeta^2", zeta ** 2, "even"), ("|zeta|", zeta.abs(), "even"), ("cos(zeta)", torch.cos(zeta), "even"),
        ("exp(0.3 zeta)", torch.exp(0.3 * zeta), "none"), ("sigmoid(zeta)", torch.sigmoid(zeta), "none"),
    ]
    rows = []
    for name, profile, parity in candidates:
        projections = {y: float((weights * chi[y] * profile).sum()) for y in UTTERANCES}
        collapsed = [
            (a, b) for a in UTTERANCES for b in UTTERANCES
            if a < b and abs(projections[a] - projections[b]) < 1e-9
        ]
        rows.append((name, parity, projections, collapsed))

    if verbose:
        print("PART A. BASIS PARITY AND UTTERANCE SEPARATION")
        print(f"  theta_L = {net.theta_L:.4f} (n = {net.num_atoms})")
        print()
        print(f"  {'basis function':<16}{'parity':<8}"
              + "".join(f"{y:>13}" for y in UTTERANCES) + "   collapsed pair")
        for name, parity, projections, collapsed in rows:
            pair = f"{collapsed[0][0]} = {collapsed[0][1]}" if collapsed else "none"
            print(f"  {name:<16}{parity:<8}"
                  + "".join(f"{projections[y]:>13.5f}" for y in UTTERANCES)
                  + f"   {pair}")
        print()
        print("  Every ODD profile collapses {some, all}; every EVEN profile collapses")
        print("  {no, all}. m = 1 with a parity-definite basis therefore cannot separate")
        print("  all three, whatever the profile. The implementation takes m = 2, one of")
        print("  each parity, verified below.")
        print()
        for j in range(net.num_basis):
            column = net.basis[:, j]
            reversed_column = torch.flip(column, [0])
            parity = ("odd" if float((column + reversed_column).abs().max()) < 1e-9
                      else "even" if float((column - reversed_column).abs().max()) < 1e-9
                      else "none")
            print(f"    B column {j + 1}: parity {parity:<5} "
                  f"<1,b> = {float((net.weights * column).sum()):+.2e}  "
                  f"b(-Z) = {float(column[0]):+.4f}  b(0) = {float(column[len(column) // 2]):+.4f}  "
                  f"b(+Z) = {float(column[-1]):+.4f}")
        separations = {
            y: tuple(round(v, 9) for v in net.project(net.lexical_field(y)).tolist())
            for y in UTTERANCES
        }
        distinct = len(set(separations.values())) == len(UTTERANCES)
        print(f"    projections distinct across all three utterances: {distinct}")
        for y, value in separations.items():
            print(f"      {y:<6} kappa = {value}")
        offset = net.project(net.base_log_prior)
        print("    B^T W ell_0 = [" + ", ".join(f"{round(float(v), 4) + 0.0:+.4f}" for v in offset)
              + "], so c_y = B^T W ell_0 - kappa_y (Eq. 30)")
        print()
    return rows


# ============================================================================
# PART B. THE TERMINATING PRIOR mu_u
# ============================================================================

def mu_u_probe(net, settings=None, theta_u=1.0, verbose=True):
    """Text cell 5 Part B: what each component of mu_u does, and what it cannot do.

    The first table holds theta_u fixed, as a control. That qualifier is essential:
    theta_u* depends on mu_u (Sec. 8.5), so through a learned theta_u the invariance
    below does not hold. The second table re-learns theta_u* for every setting, as the
    model does, and reports how far the contrasts then move.
    """
    settings = settings or [(0., 0.), (0., 1.), (1., 0.), (1., 1.), (1., 2.), (2., 1.)]
    proportion = logistic(net.zeta)
    upper = (net.zeta >= net.theta_L).to(net.dtype)

    def spawn(pair):
        # theta_u held fixed: mu_u = 0 has no theta_u* (Appendix B) but is a valid control
        return net.respawn(mu_u=torch.tensor(pair, dtype=net.dtype), theta_u=theta_u)

    # === sensitivity: d phi_S* / d mu_u, against the analytic prediction ===
    gram = net.basis.T @ (net.weights[:, None] * net.basis)
    total_sigma = net.sigma_lexical + net.sigma_state
    predicted = (net.sigma_lexical * theta_u / total_sigma) * (
        net.basis @ torch.linalg.inv(
            torch.eye(net.num_basis, dtype=net.dtype)
            + net.sigma_utility * theta_u ** 2 * gram / total_sigma
        )
    )
    zero = spawn((0., 0.))
    worst_sensitivity = 0.0
    for j in range(net.num_basis):
        bump = [0.0, 0.0]; bump[j] = 1.0
        moved = spawn(tuple(bump)).closed_form_fixed_point("some", theta_u=theta_u)[0]
        base = zero.closed_form_fixed_point("some", theta_u=theta_u)[0]
        worst_sensitivity = max(
            worst_sensitivity, float((moved - base - predicted[:, j]).abs().max())
        )

    # === the six settings ===
    rows = []
    for pair in settings:
        probe = spawn(pair)
        means, spreads = {}, {}
        for y in UTTERANCES:
            phi_S, _ = probe.closed_form_fixed_point(y, theta_u=theta_u)
            q = probe.read_out(phi_S)
            mean = float((probe.weights * q * proportion).sum())
            means[y] = mean
            spreads[y] = float(((probe.weights * q * (proportion - mean) ** 2).sum()) ** 0.5)
        rows.append((pair, means, spreads))

    # === the invariance: phi_S contrasts must not move with mu_u ===
    def contrast(pair):
        probe = spawn(pair)
        return (probe.closed_form_fixed_point("all", theta_u=theta_u)[0]
                - probe.closed_form_fixed_point("some", theta_u=theta_u)[0])
    reference = contrast(settings[0])
    worst_invariance = max(
        float((contrast(pair) - reference).abs().max()) for pair in settings[1:]
    )

    # === the same settings at each one's own learned theta_u* ===
    learned_rows = []
    for pair in settings:
        try:
            probe = net.respawn(mu_u=torch.tensor(pair, dtype=net.dtype))
        except ValueError:
            learned_rows.append((pair, None, None, None, None))     # no theta_u* exists
            continue
        means, shifts = {}, {}
        fields = {}
        for y in UTTERANCES:
            fields[y] = probe.closed_form_fixed_point(y)[0]
            q = probe.read_out(fields[y])
            means[y] = float((probe.weights * q * proportion).sum())
            literal = probe.read_out(probe.literal_fixed_point(y))
            shifts[y] = float((probe.weights * (q - literal) * upper).sum())
        learned_rows.append((pair, probe.theta_u, means, shifts, fields["all"] - fields["some"]))
    learned_contrasts = [row[4] for row in learned_rows if row[4] is not None]
    learned_deviation = max(
        float((field - learned_contrasts[0]).abs().max()) for field in learned_contrasts[1:]
    )
    star = net.learned_theta_u()
    learned_coefficient = (net.sigma_lexical * star / total_sigma) / (
        1.0 + net.sigma_utility * star ** 2 / total_sigma)

    if verbose:
        print("PART B. THE TERMINATING PRIOR mu_u   (theta_u fixed at %g)" % theta_u)
        print()
        print(f"  sensitivity d(phi_S*)/d(mu_u) matches the analytic "
              f"(sigma_L theta_u / S) B (I + sigma_u theta_u^2 G / S)^-1")
        print(f"    max discrepancy = {worst_sensitivity:.2e}")
        print()
        print(f"  {'mu_u':<12}" + "".join(f"{y + ' E[s]':>13}" for y in UTTERANCES)
              + f"{'sd|some':>10}{'all-some':>11}{'some-no':>10}")
        for pair, means, spreads in rows:
            print(f"  {str(list(pair)):<12}"
                  + "".join(f"{means[y]:>13.4f}" for y in UTTERANCES)
                  + f"{spreads['some']:>10.4f}"
                  + f"{means['all'] - means['some']:>+11.4f}"
                  + f"{means['some'] - means['no']:>+10.4f}")
        print()
        print(f"  phi_S contrasts across all settings: max deviation = {worst_invariance:.2e}")
        print("  (machine epsilon: mu_u translates the common baseline and moves no")
        print("  contrast, AT FIXED theta_u.)")
        print()
        contrasts = [m['all'] - m['some'] for _, m, _ in rows]
        mean_contrast = sum(contrasts) / len(contrasts)
        spread = max(contrasts) - min(contrasts)
        print(f"  reported E[s] contrast all-some ranges {min(contrasts):.4f} to "
              f"{max(contrasts):.4f}, a spread of {spread:.4f} on a mean of "
              f"{mean_contrast:.4f}, {100.0 * spread / mean_contrast:.1f}% of its own size.")
        print("  Since the phi_S contrasts are invariant to machine epsilon, that spread")
        print("  is produced entirely by the nonlinearity of E[s] (Eq. 25: a softmax,")
        print("  then varsigma) acting on a common additive shift, which is the argument")
        print("  of Text cell 5 Part B Sec. 4, not an effect of mu_u on the utterances.")
        print()
        control_shifts = {y: [] for y in UTTERANCES}
        for pair in settings:
            probe = spawn(pair)
            for y in UTTERANCES:
                q = probe.read_out(probe.closed_form_fixed_point(y, theta_u=theta_u)[0])
                literal = probe.read_out(probe.literal_fixed_point(y))
                control_shifts[y].append(float((probe.weights * (q - literal) * upper).sum()))
        print(f"  the shift of Eq. (37) across the same settings at theta_u = {theta_u:g}: "
              + "; ".join(f"{y} {min(values):+.4f} to {max(values):+.4f}"
                          for y, values in control_shifts.items()))
        by_setting = {pair: means for pair, means, _ in rows}
        tilt = [(0., 1.), (1., 1.), (2., 1.)]
        if all(pair in by_setting for pair in tilt):
            moves = [by_setting[tilt[i + 1]]["some"] - by_setting[tilt[i]]["some"] for i in range(2)]
            print("  tilt, read along [0, 1] -> [1, 1] -> [2, 1]: E[s] | some moves by "
                  + " and ".join(f"{move:+.4f}" for move in moves) + " per unit of mu_u_1")
        print()
        print("  THE SAME SETTINGS, EACH AT ITS OWN LEARNED theta_u*")
        print()
        print(f"  {'mu_u':<12}{'theta_u*':>11}" + "".join(f"{y + ' E[s]':>13}" for y in UTTERANCES)
              + f"{'all-some':>11}{'shift some':>12}{'shift all':>11}")
        for pair, theta, means, shifts, _ in learned_rows:
            if theta is None:
                print(f"  {str(list(pair)):<12}   degenerate: no theta_u* (Appendix B)")
                continue
            print(f"  {str(list(pair)):<12}{theta:>11.3f}"
                  + "".join(f"{means[y]:>13.4f}" for y in UTTERANCES)
                  + f"{means['all'] - means['some']:>+11.4f}"
                  + f"{shifts['some']:>+12.4f}{shifts['all']:>+11.4f}")
        print()
        print(f"  phi_S contrasts across the learned settings: max deviation = "
              f"{learned_deviation:.2e}")
        print(f"  (against {worst_invariance:.1e} at fixed theta_u: re-learning theta_u* per")
        print("  setting is the path by which mu_u moves the contrasts.)")
        ray = []
        for scale in (0.2, 3.0):
            probe = net.respawn(mu_u=torch.full((net.num_basis,), scale, dtype=net.dtype))
            gap = float((probe.closed_form_fixed_point("all")[0]
                         - probe.closed_form_fixed_point("some")[0]).abs().max())
            ray.append((scale, probe.theta_u, gap))
        print("  along the ray mu_u = v [1, 1], each at its own theta_u*: "
              + "; ".join(f"v = {scale:g}, theta_u* = {theta:.4f}, "
                          f"max |phi_S*(all) - phi_S*(some)| = {gap:.4f}"
                          for scale, theta, gap in ray))
        print(f"  At the default's theta_u* = {star:.4f}, the coefficient of Eq. (33) is")
        print(f"  (sigma_L theta_u*/S) / (1 + sigma_u theta_u*^2 / S) = {learned_coefficient:+.4f},")
        print(f"  so one unit of mu_u_j adds {learned_coefficient:+.4f} b_j at theta_u*, against")
        print(f"  +{1.0 / 3.0:.4f} b_j at theta_u = 1, and with the opposite sign.")
        print()
    return rows, worst_sensitivity, worst_invariance


# ============================================================================
# RUN
# ============================================================================

basis_rows = basis_parity_probe(evaluation_network)
mu_rows, mu_sensitivity, mu_invariance = mu_u_probe(evaluation_network)

# ============================================================================
# PART B2. THE SCALES OF mu_u, AND THE SIGN OF EACH ENTRY
# ============================================================================

def mu_u_scale_probe(net, theta_u=1.0, verbose=True):
    """What units the two entries of mu_u sit on, and where the middle of each is.

    Both are coefficients on ORTHONORMAL columns of B, so theta_u * mu_u_j * b_j is a
    log-density field of quadrature-norm theta_u * |mu_u_j|: the two entries are
    commensurable and differ in shape, not in size. The shapes are reported as the
    characteristic swing each unit buys.
    """
    columns = [net.basis[:, j] for j in range(net.num_basis)]
    middle = len(net.zeta) // 2
    rows = []
    for j, column in enumerate(columns):
        rows.append({
            "norm": float(net.squared_norm(column)) ** 0.5,
            "at_low": float(column[0]),
            "at_mid": float(column[middle]),
            "at_high": float(column[-1]),
            "end_to_end": float(column[-1] - column[0]),
            "tail_minus_centre": float(column[0] - column[middle]),
        })
    references = {
        "||ell_0||": float(net.squared_norm(net.base_log_prior)) ** 0.5,
        "||mu_u||": float(net.mu_u.norm()),
    }
    for y in UTTERANCES:
        references[f"||phi_L|| ({y})"] = float(net.squared_norm(net.lexical_field(y))) ** 0.5

    if verbose:
        print("PART B2. SCALES OF THE TWO ENTRIES OF mu_u")
        print()
        print("  mu_u_j is a coefficient on an orthonormal column of B, so")
        print("  theta_u * mu_u_j * b_j is a log-density field of norm theta_u * |mu_u_j|.")
        print()
        print(f"  {'column':<10}{'||b_j||':>10}{'b(-Z)':>10}{'b(0)':>10}{'b(+Z)':>10}"
              f"{'end-to-end':>13}{'tail-centre':>14}")
        for j, r in enumerate(rows):
            print(f"  b_{j + 1}{'':<8}{r['norm']:>10.6f}{r['at_low']:>10.4f}{r['at_mid']:>10.4f}"
                  f"{r['at_high']:>10.4f}{r['end_to_end']:>13.4f}{r['tail_minus_centre']:>14.4f}")
        print()
        print("  one unit of mu_u_1 buys %.2f nats of end-to-end tilt;" % rows[0]["end_to_end"])
        print("  one unit of mu_u_2 buys %.2f nats of tail-minus-centre." % rows[1]["tail_minus_centre"])
        print()
        print("  reference magnitudes in the same norm:")
        for name, value in references.items():
            print(f"    {name:<20} {value:9.4f}")
        print(f"    -> at theta_u = {theta_u:g} the terminating prior is "
              f"{references['||ell_0||'] / references['||mu_u||']:.0f}x below ell_0 in norm.")
        star = net.learned_theta_u()
        terminating = abs(star) * references["||mu_u||"]
        print(f"    -> at the learned theta_u* = {star:.4f} its field theta_u* B mu_u has norm "
              f"{terminating:.4f},")
        print(f"       {terminating / references['||ell_0||']:.2f}x the norm of ell_0.")
        print()
    return rows, references


def mu_u_sign_probe(net, magnitudes=(0.25, 0.5, 1.0, 2.0), theta_u=1.0, verbose=True):
    """Signed sweep of each entry about zero, testing that zero is the midpoint.

    phi_S* is affine in mu_u (Sec. 1), so a signed pair must straddle the mu_u_j = 0
    value exactly:  phi_S*(+v e_j) + phi_S*(-v e_j) = 2 phi_S*(0).  Verifying that
    identity is what establishes 0 as the midpoint of each scale, the value Sec. 3
    forbids, since the terminating contribution theta_u B mu_u vanishes there.

    The other entry is held at 0 while one is swept, which is legitimate here because
    only the whole vector is required to be nonzero.

    theta_u is held fixed in the sweep, as a control: the identity needs phi_S* at
    mu_u = 0, where no theta_u* exists. A second table reports each signed setting at
    its own learned theta_u*, where the midpoint identity cannot be posed.
    """
    proportion = logistic(net.zeta)
    zero_field = net.respawn(
        mu_u=torch.zeros(net.num_basis, dtype=net.dtype), theta_u=theta_u
    ).closed_form_fixed_point("some", theta_u=theta_u)[0]

    rows, worst_midpoint = [], 0.0
    for j in range(net.num_basis):
        for magnitude in magnitudes:
            entry = {"component": j + 1, "magnitude": magnitude, "means": {}}
            fields = {}
            entry["learned"] = {}
            for sign in (+1, -1):
                vector = torch.zeros(net.num_basis, dtype=net.dtype)
                vector[j] = sign * magnitude
                probe = net.respawn(mu_u=vector, theta_u=theta_u)
                fields[sign] = probe.closed_form_fixed_point("some", theta_u=theta_u)[0]
                entry["means"][sign] = {
                    y: float((probe.weights
                              * probe.read_out(probe.closed_form_fixed_point(y, theta_u=theta_u)[0])
                              * proportion).sum())
                    for y in UTTERANCES
                }
                learned = net.respawn(mu_u=vector)
                entry["learned"][sign] = {
                    "theta_u": learned.theta_u,
                    **{y: float((learned.weights
                                 * learned.read_out(learned.closed_form_fixed_point(y)[0])
                                 * proportion).sum())
                       for y in UTTERANCES},
                }
            entry["midpoint_error"] = float(
                (fields[+1] + fields[-1] - 2.0 * zero_field).abs().max()
            )
            worst_midpoint = max(worst_midpoint, entry["midpoint_error"])
            rows.append(entry)

    if verbose:
        print("  SIGNED SWEEP ABOUT ZERO")
        print()
        print(f"  {'entry':<7}{'|v|':>6}" + "".join(f"{y + ' -v':>12}{y + ' +v':>12}" for y in UTTERANCES)
              + f"{'midpoint err':>15}")
        for entry in rows:
            cells = "".join(
                f"{entry['means'][-1][y]:>12.4f}{entry['means'][+1][y]:>12.4f}" for y in UTTERANCES
            )
            print(f"  mu_u_{entry['component']:<2}{entry['magnitude']:>6.2f}{cells}"
                  f"{entry['midpoint_error']:>15.1e}")
        print()
        print(f"  phi_S*(+v) + phi_S*(-v) = 2 phi_S*(0) to {worst_midpoint:.1e} in every case.")
        print("  Zero is therefore the exact midpoint of both scales. Sec. 3 forbids it:")
        print("  the terminating contribution theta_u B mu_u vanishes there and theta_u")
        print("  loses identifiability (Sec. 8.5). A terminating prior sits off-centre.")
        print(f"  (theta_u held at {theta_u:g} above, as a control.)")
        print()
        print("  THE SAME SIGNED SETTINGS, EACH AT ITS OWN LEARNED theta_u*")
        print()
        print(f"  {'entry':<7}{'|v|':>6}{'theta* -v':>11}{'theta* +v':>11}"
              + "".join(f"{y + ' -v':>12}{y + ' +v':>12}" for y in UTTERANCES))
        for entry in rows:
            learned = entry["learned"]
            cells = "".join(
                f"{learned[-1][y]:>12.4f}{learned[+1][y]:>12.4f}" for y in UTTERANCES
            )
            print(f"  mu_u_{entry['component']:<2}{entry['magnitude']:>6.2f}"
                  f"{learned[-1]['theta_u']:>11.3f}{learned[+1]['theta_u']:>11.3f}{cells}")
        print()
        flip = max(abs(e["learned"][-1]["theta_u"] + e["learned"][+1]["theta_u"])
                   / abs(e["learned"][+1]["theta_u"]) for e in rows)
        mirror = max(abs(e["learned"][-1][y] - e["learned"][+1][y])
                     for e in rows for y in UTTERANCES)
        print(f"  theta_u*(-v) = -theta_u*(+v) in every row (relative gap {flip:.1e}), and")
        print(f"  E[s] at -v equals E[s] at +v to {mirror:.1e}: under learning the sign of mu_u")
        print("  is invisible, since the network re-learns the opposite theta_u* and the")
        print("  symmetry (theta, mu) -> (-theta, -mu) of Part B3 does the rest. The midpoint")
        print("  identity is not testable here: at mu_u = 0 no theta_u* exists (Appendix B),")
        print("  so there is no phi_S*(0) to straddle.")
        print()
    return rows, worst_midpoint


scale_rows, scale_references = mu_u_scale_probe(evaluation_network)
sign_rows, sign_midpoint_error = mu_u_sign_probe(evaluation_network)


# ============================================================================
# PART B3. THE mu_u PLANE, SWEPT PROPERLY
# ============================================================================

def mu_u_plane_probe(net, *, radii=(0.25, 0.5, 1.0, 2.0, 4.0), sectors=12, verbose=True):
    """A 2-D sweep of mu_u, on the shape Sec. 3 and Appendix B actually license.

    Three things make the six hand-picked settings of Text cell 5 Part B Sec. 3 the
    wrong shape, and all three are structural rather than matters of taste.

    1. mu_u lives in R^2, so a sweep of it is a sweep of a PLANE. The six settings
       are lattice points in one quadrant.
    2. The origin is excluded, not merely awkward: Sec. 3 requires mu_u != 0, and
       Appendix B sharpens that to <mu_u, sum_y c_y> != 0, which is a requirement on
       DIRECTION and removes two whole rays rather than one point. Polar coordinates
       are the ones the constraint is stated in, so the sweep uses them.
    3. The pair (theta_u, mu_u) carries an exact symmetry. phi_u* solves
       (I + sigma_u theta^2 G / S) phi_u* = mu_u + sigma_u theta c / S; the matrix
       sees only theta^2, so (theta, mu) -> (-theta, -mu) sends phi_u* -> -phi_u* and
       leaves g_S = theta B phi_u* and hence phi_S* alone. Half the plane is therefore
       redundant given both signs of theta_u, and the sweep is taken modulo it: the
       circle is swept at one sign of theta_u, and the identity is verified rather
       than assumed.
    """
    proportion = logistic(net.zeta)
    coupling = sum(
        (net.project(net.base_log_prior - net.lexical_field(y)) for y in UTTERANCES),
        torch.zeros(net.num_basis, dtype=net.dtype),
    )
    degenerate = math.degrees(math.atan2(-float(coupling[0]), float(coupling[1]))) % 180.0

    def spawn(radius, degrees):
        # theta_u is passed explicitly everywhere below, so the probes are built as
        # fixed-theta_u controls, which also keeps a setting on a degenerate ray
        # constructible; theta_u* is read from each probe where it is wanted.
        angle = math.radians(degrees)
        return net.respawn(mu_u=radius * torch.tensor(
            [math.cos(angle), math.sin(angle)], dtype=net.dtype), theta_u=1.0)

    angles = [360.0 * k / sectors for k in range(sectors)]

    # === (3) the exact symmetry, verified over the plane ===
    worst_symmetry = 0.0
    for radius in radii:
        for degrees in angles:
            plus, minus = spawn(radius, degrees), spawn(radius, degrees + 180.0)
            for theta in (0.7, 1.0, 4.0):
                for y in UTTERANCES:
                    worst_symmetry = max(worst_symmetry, float((
                        plus.closed_form_fixed_point(y, theta_u=theta)[0]
                        - minus.closed_form_fixed_point(y, theta_u=-theta)[0]
                    ).abs().max()))

    # === (1) at fixed theta_u the whole plane is common-mode ===
    reference = None
    worst_invariance = 0.0
    for radius in radii:
        for degrees in angles:
            probe = spawn(radius, degrees)
            contrast = (probe.closed_form_fixed_point("all", theta_u=1.0)[0]
                        - probe.closed_form_fixed_point("some", theta_u=1.0)[0])
            if reference is None:
                reference = contrast
            worst_invariance = max(
                worst_invariance, float((contrast - reference).abs().max()))

    # === (2) what does vary: theta_u*, and the contrast it carries ===
    rows = []
    for radius in radii:
        for degrees in angles:
            probe = spawn(radius, degrees)
            inner = float(probe.mu_u @ coupling)
            entry = {"radius": radius, "degrees": degrees, "inner": inner}
            if abs(inner) < 1e-9:
                entry["theta_star"] = float("nan")
                entry["contrast"] = float("nan")
            else:
                star = theta_u_stationary_point(probe)
                entry["theta_star"] = star
                fields = {y: probe.closed_form_fixed_point(y, theta_u=star)[0]
                          for y in UTTERANCES}
                entry["contrast"] = float(
                    probe.squared_norm(fields["all"] - fields["some"])) ** 0.5
                entry["expected_s"] = {
                    y: float((probe.weights * probe.read_out(fields[y]) * proportion).sum())
                    for y in UTTERANCES
                }
            rows.append(entry)

    if verbose:
        print("PART B3. THE mu_u PLANE")
        print()
        print(f"  sum_y c_y = [{coupling[0]:+.4f}, {coupling[1]:+.4f}], so Appendix B's")
        print(f"  requirement <mu_u, sum_y c_y> != 0 fails on the two rays at")
        print(f"  {degenerate:.2f} deg and {degenerate + 180:.2f} deg. Those are excluded, and so is")
        print("  the origin. What is removed is a DIRECTION, not a point.")
        print()
        print(f"  exact symmetry  phi_S*(theta, mu) = phi_S*(-theta, -mu), an identity in theta,")
        print(f"  checked at theta_u in (0.7, 1, 4):")
        print(f"    max deviation over {len(radii)}x{sectors} settings, 3 |theta_u|, 3 utterances "
              f"= {worst_symmetry:.1e}")
        print("    (bit-identical, so the sweep below covers the plane at one sign of")
        print("     theta_u and the other sign is obtained by reflection, not re-run.)")
        print()
        print(f"  common-mode at fixed theta_u = 1 (a control; it holds at any fixed theta_u):")
        print(f"    max deviation of the all-some phi_S contrast over the plane "
              f"= {worst_invariance:.1e}")
        print("    (the six settings of Sec. 3 were a sample of this; the plane confirms it.)")
        print()
        print("  theta_u* over the plane (the quantity that DOES move), by direction:")
        print(f"    {'angle':>8}" + "".join(f"{f'r={r:g}':>12}" for r in radii))
        for degrees in angles:
            cells = []
            for radius in radii:
                entry = next(e for e in rows
                             if e["radius"] == radius and e["degrees"] == degrees)
                value = entry["theta_star"]
                cells.append("  degenerate" if value != value else f"{value:>12.3f}")
            near = "  <-- near a degenerate ray" if min(
                abs(degrees - degenerate), abs(degrees - degenerate - 180)) < 20 else ""
            print(f"    {degrees:>8.0f}" + "".join(cells) + near)
        print()
        finite = [e for e in rows if e["theta_star"] == e["theta_star"]]
        widest = max(finite, key=lambda e: abs(e["theta_star"]))
        narrowest = min(finite, key=lambda e: abs(e["theta_star"]))
        print(f"    |theta_u*| ranges {abs(narrowest['theta_star']):.3f} "
              f"(r={narrowest['radius']:g}, {narrowest['degrees']:.0f} deg) to "
              f"{abs(widest['theta_star']):.3f} "
              f"(r={widest['radius']:g}, {widest['degrees']:.0f} deg), a factor of "
              f"{abs(widest['theta_star']) / abs(narrowest['theta_star']):.0f}.")
        contrasts = [e["contrast"] for e in finite]
        print(f"    the all-some contrast it carries ranges {min(contrasts):.4f} to "
              f"{max(contrasts):.4f}, a spread of {max(contrasts) - min(contrasts):.4f}")
        print(f"    against {worst_invariance:.1e} at fixed theta_u. mu_u reaches the")
        print("    utterances only through the slow parameter, and that is the whole of it.")
        print()
        default = math.degrees(math.atan2(1.0, 1.0))
        gap = min(abs(default - degenerate), abs(default - degenerate - 180.0),
                  abs(default - degenerate + 180.0))
        print(f"  Note theta_u*(angle + 180) = -theta_u*(angle) exactly down the table:")
        print("  that is the symmetry above, read off the sweep rather than assumed.")
        print()
        print(f"  The default mu_u = [1, 1] sits at r = {math.sqrt(2):.3f}, {default:.0f} deg,")
        print(f"  {gap:.2f} deg from the nearer degenerate ray at {degenerate:.2f} deg, comfortably")
        print("  clear of it, but the margin is a property of this ell_0 and this Lambda,")
        print("  since sum_y c_y moves with both.")
        print()

    return rows, worst_symmetry, worst_invariance


plane_rows, plane_symmetry, plane_invariance = mu_u_plane_probe(evaluation_network)
````

## <a id="tc6" name="tc6"></a>Text cell 6: the $\Lambda\times\alpha$ sweep

Text cell 5 Part B showed that $\mu_u$ and $\ell_0$ move no $\varphi_S$ contrast **at fixed
$\theta_u$**. Neither does so by sitting in $g_S$, which carries neither of them
since $\ell_0$ is in $g_L$ (Eq. 9), but by entering the closed forms only through
utterance-independent terms: $\mu_u$ as the terminating prior in Eq. (16), $\ell_0$ only inside the
combination $\ell_0-\varphi_L$ of Eqs. (15)–(16), and $\varphi_S^\ast$ is affine in each with a
coefficient that does not depend on the utterance. (Through a learned $\theta_u$ they *do* reach
contrasts, since $\theta_u^\ast$ depends on both, and that is the route every cell below takes:
each of the $121$ cells is its own configuration and carries its own $\theta_u^\ast$, which runs
from $-557.67$ to $+12228.42$ across the plane and is negative at three cells only, $(1,2)$, $(1,4)$
and $(2,2)$.) That property is theirs alone:
$\Lambda$, $\theta_L$, $B$, the variances, and the slow parameter $\theta_u$ all touch
utterance-dependent quantities and can move contrasts. This cell takes $\Lambda$, the one that mediates directly between the lexicon and
the world belief, and maps it against the concentration of $\ell_0$, the quantity it competes with in
Eq. (15).

The motivating case is a base world prior certain of the top state, $P_0(s{=}1)=1$. It is
not representable here: $s$ lies in the **open** interval $(0,1)$, $\zeta=\operatorname{logit}(1)$
is off the grid, and $\ell_0=\log p_0$ would be $-\infty$ almost everywhere, which makes the
Gaussian residual $r_S$ infinite. The limit family $\mathrm{Beta}(\alpha,1)$ concentrates at
$s=1$ as $\alpha$ grows and is swept in its place.

---

### <a id="tc6-quantities-swept" name="tc6-quantities-swept"></a>Quantities swept

**Concentration $\alpha$ of the base world prior.** This is the sweep's row coordinate. For $\alpha>0$ the base world prior is
$p_0=\mathrm{Beta}(\alpha,1)$ over the proportion, pushed forward to $\zeta$ by the change of
variables of Text cell 4 Part D. Writing $\beta$ for the second shape parameter of the Beta and
$B(\cdot,\cdot)$ for its normalizing constant, the push-forward of $\mathrm{Beta}(\alpha,\beta)$ to
$\zeta$ is $\alpha\log s+\beta\log(1-s)-\log B(\alpha,\beta)$, which at $\beta=1$, where
$B(\alpha,1)=1/\alpha$, collapses to

$$\ell_0(\zeta)=\alpha\log\varsigma(\zeta)+\log\!\big(1-\varsigma(\zeta)\big)+\log\alpha .
\tag{36}$$

$\alpha=1$ is $\mathrm{Uniform}(0,1)$, the flat base world prior of Part D. As $\alpha$ grows the family
concentrates at $s=1$: its mean is $\alpha/(\alpha+1)$, and its mode *in $\zeta$* sits at the same
place, since $\partial_\zeta\ell_0=0$ gives $\alpha(1-s)=s$. That mode is not the mode in $s$, which
$\mathrm{Beta}(\alpha,1)$ places at the boundary; the Jacobian moves it. $\alpha\to\infty$ is the
limit standing in for the unrepresentable $P_0(s{=}1)=1$.

*Grid caveat.* The concentration a truncated grid can represent is bounded, and the bound bites
gradually rather than at a threshold. On the default grid ($|\zeta|\le6$) the base world prior mass carried by
the top node runs $0.0095$ at $\alpha=64$, $0.0445$ at $\alpha=300$, $0.1512$ at $\alpha=1024$ and
$0.5477$ at $\alpha=4096$; what fails first is not the representation of the base world prior but the model's
ability to tell the utterances apart, since the spread $D$ of Eq. (38) collapses $0.0057\to0.0012\to
0.0003$ over $\alpha=64,256,1024$ at $\Lambda=8$, each at its own $\theta_u^\ast$. $\mathbb E[s]$ tends to about $0.980$ for every utterance rather than to
$\varsigma(6)=0.9975$, since the belief spreads over the top several nodes rather than concentrating
on the last one. The top rows of each map therefore measure the grid as much as the model; widening
the grid moves the ceiling without changing the picture.

**Lexical strength $\Lambda$.** This is the sweep's column coordinate, entering
$\varphi_L=\Lambda\chi_y$ exactly as in Eq. (6). No other quantity is varied: $\mu_u=[1,1]$,
$n=10$ so $\theta_L=\log19=2.9444$ (Eq. A5), $m=2$, every $\sigma=1$, and $\theta_u$ learned per
cell. $\mu_u\neq0$ is required by Text cell 3 §3, and at fixed $\theta_u$ it is common-mode across
utterances.

**Under learning $\mu_u$ is nearly inert, and at a fixed $\theta_u$ it is not.** The contrast is
worth stating, because the two readings point in opposite directions. Hold $\theta_u$ at $1$ and a
hundredfold range of $\lvert\mu_u\rvert$, both signs included, moves $\mathbb E[s]$ at
$(\alpha,\Lambda)=(1,8)$ from $0.5774$ to $0.7191$ and $\Delta_{\textit{some}}$ from $+0.0422$ to
$+0.2457$, and at $(16,128)$ it carries $\Delta_{\textit{some}}$ across zero, from $-0.1473$ to
$+0.0029$: that is the failure mode of Text cell 5 Part B §4, a common-mode field moving no
$\varphi_S$ contrast but moving a statistic taken after the softmax. Re-learn $\theta_u^\ast$ for
each setting and the same range leaves $\mathbb E[s]$ within $0.0026$ at $(1,8)$ and $0.0002$ at
$(16,128)$, with $\Delta_{\textit{some}}$ running $+0.0278$ to $+0.0305$ and $-0.2033$ to
$-0.2003$, no sign change anywhere. What absorbs $\mu_u$ is $\theta_u^\ast$ itself, which swings
over four orders of magnitude across the same settings. So the sensitivity the fixed-$\theta_u$
control exhibits is the control's, not the model's, and the maps below, drawn at $\mu_u=[1,1]$, do
not depend on that choice.

### <a id="tc6-statistics" name="tc6-statistics"></a>Statistics

Recalled from Text cell 4: the **leak** is the $q$-mass over the region the observed entry excludes;
$\mathbb E[s]$ is Eq. (25); and $P(R)$, the $q$-mass over a region, is Eq. (27), written $P(R;q)$ below where the read-out it is taken under must be named. The *all-region* is $R=\{\zeta>\theta_L\}$.

Four further statistics appear here. The first is Part C's shift, restated for the maps; the other
three are used only in this cell. Write $q_y$ for the settled read-out under utterance $y$ and
$q_{\mathrm{lit},y}$ for the literal listener of Part C, untempered,
$\varphi_S=\ell_0-\varphi_L$.

We call

$$\Delta_y=P(\text{all-region};q_y)-P(\text{all-region};q_{\mathrm{lit},y})
\tag{37}$$

the **shift** for $y$: how much $q$-mass separates the settled belief from the literal listener
over the region where *all* is true. Two things in the hierarchy produce it, and they come apart
at $\theta_u=0$, where the field is $(\ell_0-\varphi_L)/2$ and $\Delta_y$ is already non-zero by
the tempering alone; the remainder is the utility level. The tempering is the larger of the two in
$59$ of the $121$ cells of this plane, and they carry opposite signs often enough that
$\Delta_{\textit{some}}$ and the utility level's own contribution disagree in sign in $53$ of the
$121$ cells. So $\Delta_y$ is the pragmatic-literal gap and not a measure of the utility level;
Eq. (39) below is the statistic that isolates the latter. Part C's mass shift criterion is
$\Delta_{\textit{some}}<0$. The baseline is the one Part D tabulates, so a cell of this plane and
a row of that table are the same measurement; the delta-like row of Text cell 4b, $\alpha=64$ and
$\Lambda=512$, is such a cell and reads $-0.5217$ in both places. Part C's mass position criterion,
$P(\text{all-region};q_y)<\tfrac12$, is not a difference and so is not mapped; it is reported at
the end of this cell.

We call

$$D=\max_y\mathbb E[s\mid y]-\min_y\mathbb E[s\mid y]
\tag{38}$$

the **spread**: how far apart the three utterances' expected proportions lie. $D\to0$ says the model
has stopped distinguishing them, whatever it believes.

We call

$$\Delta_{\textit{some}}-\Delta_{\textit{no}}
\tag{39}$$

the **differential**. The tempering enters $\Delta_y$ at nearly the same size for every utterance,
so subtracting the *no* shift removes most of it and leaves the utterance-specific part: between
*some* and *no* the tempering differs by a median $0.0007$ and at most $0.0988$, against a median
$0.0405$ and a maximum $1.1884$ for the utility level. The differential's sign accordingly agrees
with the utility level's own in $104$ of the $121$ cells, where $\Delta_{\textit{some}}$ agrees in
$84$, and it is the differential that reports what $\theta_u$ does. It remains a **control** on
$\Delta_{\textit{some}}$ and not a second criterion, since the mass shift criterion is stated on the raw
shift. On this plane it discriminates in one region and not elsewhere: over most of the
plane $\Delta_{\textit{no}}$ is small and the differential simply *is* the raw shift, but on the
override diagonal $\Delta_{\textit{no}}$ reaches $1.0000$, the entry for *no* having been beaten outright by $\ell_0$, and there the two statistics say different things. Both are mapped, and both readings are
reported below.

We call

$$\Lambda_{\mathrm{crit}}(\alpha,\theta_L)=\inf\{\Lambda>0:\ \text{leak}(\textit{no};\Lambda,\alpha,\theta_L)<\tfrac12\}
\tag{40}$$

the **override threshold**, the least lexical strength at which the entry for *no* still holds against $\ell_0$. It is located by geometric bisection, and it is the diagonal the maps show. On the
truth region the settled log-density is $\ell_0/2$ and on the excluded region $(\ell_0-\Lambda)/2$
(Eq. 15), so the entry survives only while $\Lambda$ exceeds the gap in $\ell_0$ between the two. For
Eq. (36) that gap is dominated by $-\alpha\log s$ at the truth region's best point
$s=\varsigma(-\theta_L)$, predicting

$$\Lambda_{\mathrm{crit}}\;\approx\;\lvert\log\varsigma(-\theta_L)\rvert\,\alpha ,
\tag{41}$$

This is linear in $\alpha$, with a slope fixed by the endpoint threshold alone. Since $\theta_L=\log(2n-1)$ is
itself derived from the predicate's granularity (Eq. A5), $\lvert\log\varsigma(-\theta_L)\rvert=\log 2n$
and the law reads $\Lambda_{\mathrm{crit}}\approx\alpha\log 2n$: the lexicon must be stronger, by a
logarithm of the set size, to hold against a confident base world prior over a finer predicate.

Eq. (41) is derived with the utility level severed, and there it holds: at $\alpha=1024$ the measured
slope is $1.3675$, $2.1572$, $3.0435$ against predictions of $1.3133$, $2.1269$, $3.0486$ for $\theta_L=1,2,3$, giving ratios $1.041$, $1.014$ and $0.998$. So the diagonal is the gap in $\ell_0$ and nothing else.

**Restoring the utility level raises the threshold.** The measured slopes at the $\theta_u=1$
control are $1.6122$, $2.4839$, $3.5480$, so the lexicon needs $15$–$18\%$ *more* strength to hold.
The model's own column re-learns $\theta_u^\ast$ at every $\Lambda$ the bisection visits, since each
is a configuration of its own, and there the slopes are $1.9890$, $2.9334$, $4.3102$: $36$–$45\%$
more. The $\theta_u^\ast$ found at the threshold itself is large, $13378$ to $14937$ across the
three boundaries, which is why the learned column sits at the saturation value the control only
approaches. The utility level reinforces $\ell_0$ against the entry, and learning it reinforces it
further.

Which mechanism does that matters, because the same one runs through Text cell 4 Part C. It is not
$\varphi_u$ regressing toward $\mu_u$: holding $\theta_u=1$ and moving
$\mu_u$ over a hundredfold range in magnitude and both signs changes the slope by under $0.1\%$,
and with $\theta_u^\ast$ re-learned for each of those $\mu_u$ the slope is $1.9890$ in every case,
identical to four decimals. The effect is identical at $\theta_u$ and $-\theta_u$ ($1.6122$ against
$1.6114$) and saturates at $+45.4\%$ as $\lvert\theta_u\rvert$ grows, which is where the learned
column already sits. Those are the signatures of the $\theta_u^{2}$ term.
What is happening is the amplification Part C measures: raising
$\lvert\theta_u\rvert$ doubles the $\operatorname{span}B$ component of $\ell_0-\varphi_L$, and under
$\mathrm{Beta}(\alpha,1)$ that component is dominated by the tilt of $\ell_0$, so amplifying it pushes $\ell_0$ harder against the entry. Both claims are checked in the cell rather than asserted.

---

**What the sweep is for.** $\Lambda$ and $\ell_0$ are both additive in the same log-density
(Eq. 15), so they compete directly, and a sufficiently confident base world prior overrides the lexical
entry outright, so that the model, told *no*, comes to believe $s\approx1$ with certainty. This is the
cost of the soft mask: a hard mask makes semantics inviolable by construction, and
finite $\Lambda$ buys differentiability and the $\Lambda\to\infty$ limit in exchange for making
the lexicon **defeasible**.

**Method.** With $\varphi_L$ clamped the model is linear-Gaussian and Eqs. (15)–(16) are exact
(Sec. 8.6), so the plane is evaluated in closed form, one small linear solve per cell rather
than thousands of Euler steps. The shortcut is verified rather than assumed, but the verification is
narrow, and the reason is the timescale commitment: at each cell's own $\theta_u^\ast$ only $2$ of
the $121$ cells keep $\lambda_{\max}(H)$ within what the evaluation can integrate, and both of them
are re-run through the full `infer` dynamics, agreeing with the closed form to $1.48\times10^{-9}$, which is the tolerance those runs stop at
rather than a limit of the closed form: a tolerance keyed to $\lambda_{\max}(H)$, so it is a
property of the cells being integrated and not a constant (I3). The cells where both mass criteria are met are not among them at $\theta_u^\ast$;
they are all integrable at the much smaller $\theta_u$ the conjunction itself needs, which the last
section of this cell reports.

**Reading the maps.** Rows are $\alpha$ (upward), columns are $\Lambda$ (rightward), both on
powers of two. Leak and $\mathbb E[s]$ are magnitudes on a fixed $0$–$1$ scale and take single-hue
sequential ramps; the shift is signed with a meaningful zero and takes a diverging pair with
limits symmetric about it, so that the neutral midpoint reads as "no difference from the literal
listener".

The third figure's last panel maps the differential (39). Under Part C's mass shift criterion
an implicature appears as $q$-mass moved *away* from the *all* region **for *some* specifically**, and
the differential is the control on that "specifically". It is a control and not a proof: the common
component need not be identical across utterances, and the criterion it serves is a stipulation of
Part C rather than a result.

**Across the plane the differential is negative almost everywhere.** $94$ of the $121$ cells are
negative; $72$ fall below $-10^{-3}$ and
$62$ below $-10^{-2}$; the most negative is $-1.2470$ (at $\alpha=64$, $\Lambda=256$) against a
maximum of $+0.0656$ (at $\alpha=1$, $\Lambda=64$). The raw *some* shift is negative in $74$ cells
and reaches $-0.9466$. A further $10$ cells, the whole $\alpha=1024$ row, are zero to roundoff and
are counted separately: there $\ell_0$ has overridden every entry, so the shifts are differences of
equal numbers of order $10^{-16}$, and reading their signs would be reading roundoff. The
differential is not flat against zero, and Part C's mass shift criterion is met over most of the plane.

The geography is orderly, and it is what makes the numbers readable.

- **The $17$ positive cells all sit at $\alpha\le4$**, under diffuse base world priors at or near uniform.
  This is the region Part D samples with its *flat* and $\mathrm{Beta}(3,1)$ rows.
  $\Delta_{\textit{some}}$ is positive over a wider region still, $37$ cells reaching $\alpha=64$,
  and there it is the tempering that carries it: little base world prior mass sits on the *all* region for
  the utility level to move either way, so the flattening dominates the difference.

- **The $32$ cells with $\lvert\text{differential}\rvert<10^{-3}$ are mostly override cells**: $30$
  of them have $\Lambda\le\alpha$, where Eq. (41) says $\ell_0$ has already beaten the entry and
  every utterance settles on the same belief. Flatness there is saturation, not the absence of an
  effect, which is the same point Text cell 4b makes for its rows: the headroom has to be read
  before the number.

- **The effect is largest in the band $16\le\alpha\le128$ with $\Lambda$ several times $\alpha$**:
  the base world prior sharp enough to concentrate on the *all* region, the lexicon strong enough to hold
  against it. Text cell 4b's delta-like configuration, $\alpha=64$ and $\Lambda=512$, is a cell of this
  plane and sits in that band, reading $\Delta_{\textit{some}}=-0.5217$ with
  $\Delta_{\textit{no}}=0.0000$, so there the differential and the raw shift coincide to four
  decimals and the two statistics agree. That $-0.5217$ is also the number Text cell 4b's table carries,
  which is the check that the two cells measure one quantity.

So Part C's mass shift criterion is met over most of this plane on either statistic, and the choice
between the two is not load-bearing.

One region does separate them, and it is the one the leak map already marks. On the diagonal where
$\Lambda$ is about four times $\alpha$ $\ell_0$ overrides the entry for *no* outright, and the
shift for *no* runs to $+1.0000$ there: the belief given *no* has been carried into the all-region
entire. Along that diagonal it reads $+0.8348$ at $(32,128)$, $+0.9996$ at $(64,256)$ and exactly
$+1.0000$ at $(128,512)$, $(256,1024)$ and $(512,2048)$, and the third figure shows it as the bright
diagonal in the *no* panel. Subtracting a shift of $+1$ from
the *some* shift makes the differential large and negative for a reason that has nothing to do with
*some*, so the raw shift is the statistic to read there. The leak map gives the same boundary from
the other side: the least $\Lambda$ at which the entry for *no* still holds is $4$, $8$, $16$, $32$,
$64$, $256$, $512$, $1024$ and $2048$ as $\alpha$ runs $1$ to $256$, and past $\alpha=512$ no
$\Lambda$ on this grid holds it.

### <a id="tc6-conjunction" name="tc6-conjunction"></a>Where both of Part C's mass criteria hold

The mass position criterion, $q_H(\text{all}\mid\textit{some})<\tfrac12$, is a statement about the
settled belief rather than about a difference, so it does not appear in the maps. Measured across
the same plane it holds in $59$ of the $121$ cells, and **both mass criteria hold together in $33$** (at $K=101$; Appendix A prints the plane at other
node counts).

The two run against each other in $\alpha$, which is what makes the region a band. Each criterion
holds above a floor in $\Lambda$ and keeps holding above it. The mass shift criterion's floor **falls**
as the base world prior sharpens, from $\Lambda\ge512$ at $\alpha=1$ to $\Lambda\ge64$ at $\alpha=8$,
$\Lambda\ge16$ at $\alpha=64$ and $\Lambda\ge2$ from $\alpha=128$ to $\alpha=512$: the more base world prior
mass sits on the *all* region, the less lexical strength the utility level needs in order to take
some of it away. (At $\alpha=1024$ the floor reads $2048$, which is the saturated row again: only
the last column resolves above roundoff there.) The mass position criterion's floor **rises**, from
$\Lambda\ge2$ up to $\alpha=8$ to $\Lambda\ge64$ at $\alpha=16$, $\Lambda\ge256$ at
$\alpha=32$, $\Lambda\ge512$ at $\alpha=64$ and $\Lambda\ge1024$ at $\alpha=128$, and past
$\alpha=256$ it is not reachable on this grid at all: the sharper the base world prior, the more strength the
entry needs before what is left below the *all* boundary is a majority. Both hold where the higher
floor is still on the plane, a band running from $(\alpha,\Lambda)=(1,512)$ to $(128,2048)$.

**Part D's four rows sit outside that band, and at $\Lambda=512$ Text cell 4b's flat and
delta-like rows sit inside it.** Part D holds $\Lambda=8$, which clears the mass position criterion's
floor of $2$ at $\alpha\le8$ but the mass shift criterion's floor only at $\alpha\ge128$; the flat base world prior,
$\mathrm{Beta}(1,1)$, is the cell $(1,8)$ and meets the mass position criterion alone. At $\Lambda=512$
the flat base world prior meets the mass shift criterion's floor of $512$ and clears the mass position criterion's, which puts it at
the band's corner, and the delta-like base world prior clears the mass shift criterion's floor of $16$ and meets
the mass position criterion's of $512$. $\mathrm{Beta}(3,1)$ belongs to the family but falls between the plane's
$\alpha=2$ and $\alpha=4$, and the Gaussian and $\mathrm{Beta}(1,3)$ base world priors are not members of it.
Holding the delta-like base world prior and raising $\Lambda$ further deepens both: $q_H$ falls from $0.4351$ at $\Lambda=512$ to $0.1352$
at $1024$ and $0.0102$ at $2048$, with the shift deepening from $-0.5217$ to $-0.8217$ and
$-0.9466$, and $\theta_u^\ast$ rising from $1407.77$ to $5531.15$ along the way. The delta-like
row is at the band's lower edge in $\Lambda$, not outside it.

**What the conjunction costs, where it holds.** The band is reported in closed form, since at their
own $\theta_u^\ast$ these cells carry $\lambda_{\max}(H)$ between $3.5\times10^{4}$ and
$3.6\times10^{7}$. That is the cost of $\theta_u^\ast$ and not of the conjunction, and the two are
measured apart here: the least $\lvert\theta_u\rvert$ at which both mass criteria hold runs from
$0.100$ to $4.250$ over the $33$ cells, with $\lambda_{\max}(H)$ between $2.0$ and $20.1$ there, so
**every cell of the band is integrable at the $\theta_u$ its conjunction needs**, and the conjunction,
once met, holds from that $\theta_u$ all the way to $\theta_u^\ast$ in all $33$. Eq. (20) started
from $\theta_u(0)=0$ passes the threshold within one or two updates. Code Cell 2b runs three rows of
this kind end to end, two of them cells of the band, without the closed form at any step (Text cell
4, *Integration cost and conditioning*).

**The same plane under the delta read-out.** Code Cell 4 also reads the mode shift and mode position
criteria of Text cell 4 Part D off every cell. The mode position criterion holds in the same $59$
cells as the mass position criterion. The mode shift criterion holds in $67$ cells against the mass shift
criterion's $74$, and the two disagree in $35$: the mass shift criterion alone is met in $21$, all at
$\alpha\le16$, and the mode shift criterion alone in $14$, which are ten cells of the saturated
$\alpha=1024$ row and $(32,16)$, $(64,2)$, $(64,4)$ and $(64,8)$. Both mode criteria hold together in
$13$ cells at $K=101$, every one of them inside the band of $33$, so the two conjunctions part in $20$ cells,
all of them met under q alone.

Under the delta read-out the band keeps its right arm and loses its left. The mode position
criterion's floor is the mass position criterion's at every $\alpha$. The mode shift criterion is never
met at $\alpha\le8$; from $\alpha=16$ its floor is $256$, then $16$ at $\alpha=32$, then $2$ from
$\alpha=64$ on. The floor of both mode criteria is therefore absent at $\alpha\le8$ and reads $256$,
$256$, $512$ and $1024$ at $\alpha=16$, $32$, $64$ and $128$, against the mass conjunction's $64$,
$256$, $512$ and $1024$. The right arm, set by the position criteria, is shared from $\alpha=32$; the
left arm, where the mass shift criterion's floor falls as the base world prior sharpens, exists only under q.
Appendix C §8 measures what the mode does there. Every cell where the mode moves up has the tempered
control's mode at $s\le0.9405$, and every cell where it moves down has it at $s\ge0.9405$; at
$\alpha=1$, $2$, $4$ and $8$ the mode of $\ell_0$, which is also the tempered control's, sits at
$s=0.5000$, $0.6726$, $0.8085$ and $0.8849$.

<a id="code4" name="code4"></a>

````python
# === Code Cell 4: the Lambda x alpha sweep. EVALUATION ===
#
# Two fixed quantities compete additively inside the same log-density (Eq. 15): the
# lexical field enters as -Lambda*chi_y, and the base world prior as ell_0. A base world prior
# concentrated near s = 1 therefore does not merely bias the belief; past some
# concentration it OVERRIDES the lexical entry outright. This cell maps that
# competition over the (alpha, Lambda) plane.
#
# The base world prior family is Beta(alpha, 1), which concentrates at s = 1 as alpha grows. A
# literal delta at s = 1 is not representable: s lies in the OPEN interval (0, 1),
# zeta = logit(1) is off the grid, and ell_0 = log p_0 would be -infinity almost
# everywhere, which makes the Gaussian residual r_S infinite. Beta(alpha, 1) is the
# limit family that approaches it.
#
# METHOD. With phi_L clamped the model is linear-Gaussian and Eqs. (15)-(16) are
# exact (Sec. 8.6), so the sweep is evaluated in closed form, a small linear solve
# per cell instead of some thousands of Euler steps. Every cell is its own
# configuration and carries its own learned theta_u* (Eq. B2), which runs into the
# thousands here, so the closed form is also the only affordable route under the
# timescale commitment. The shortcut is verified, not assumed: `verify_samples` cells
# among those where integration is feasible (lambda_max(H) <= FEASIBLE_STIFFNESS,
# Code Cell 2) are re-run through the full `infer` dynamics and the worst
# disagreement is reported.

import math
import numpy as np
import torch


def lambda_alpha_sweep(
    net,
    *,
    alphas=None,
    lambdas=None,
    verify_samples=6,
    verbose=True,
):
    """Sweep the concentration of the base world prior against lexical strength.

    Returns a dict of (len(alphas), len(lambdas)) arrays, rows indexed by alpha and
    columns by Lambda, matching the heat-map layout.

    The shift is Eq. (37): P(all; full) - P(all; q_lit), the same baseline Code Cell 2
    uses, so the values here and the ones Text cell 4 Part D tabulates are the same
    quantity. `literal_fixed_point` returns ell_0 - phi_L, the base world prior restricted by the
    entry and nothing else; the theta_u = 0 control is a DIFFERENT field, tempered by
    one half (Text cell 4 Part C), and is not what the shift is taken against. It is
    computed too, as the tempering part of the shift, so that the shift can be split
    into the tempering and the utility level's own contribution.
    """
    # Equal counts make each map square, and the offset ranges put Lambda = 2*alpha on
    # the main diagonal. That diagonal is NOT the override threshold: Eq. (41) puts
    # Lambda_crit at |log sigma(-theta_L)| * alpha = log(2n) * alpha = 2.996 * alpha at
    # n = 10, and the utility level at its learned theta_u* raises it further
    # (`override_threshold`). So the main diagonal runs BELOW the threshold and every
    # cell on it has the entry overridden; the threshold is a parallel diagonal higher
    # up, which `override_threshold` locates.
    alphas = [float(a) for a in (alphas if alphas is not None else [2.0 ** k for k in range(11)])]
    lambdas = [float(l) for l in (lambdas if lambdas is not None else [2.0 ** k for k in range(1, 12)])]

    shape = (len(alphas), len(lambdas))
    fields = ["leak", "expected_s", "shift", "tempering", "full_upper"]
    grids = {f: {y: np.full(shape, np.nan) for y in UTTERANCES} for f in fields}
    grids["discriminability"] = np.full(shape, np.nan)
    grids["theta_u"] = np.full(shape, np.nan)
    grids["stiffness"] = np.full(shape, np.nan)
    # the delta read-out under "some" (B8): the mode nodes of phi_S*, ell_0 and ell_0 - phi_L, the
    # modes of phi_S* and ell_0 in s, whether phi_S*'s lies outside the cell of "all", and the
    # smallest gap between the two largest nodes of the three fields (argmax takes the first node)
    for field in ("mode_model", "mode_prior", "mode_literal", "mode_s_model", "mode_s_prior",
                  "mode_outside", "prior_mode_outside", "mode_gap"):
        grids[field] = np.full(shape, np.nan)
    # what Part C's conjunction needs, cell by cell (Code Cell 2's realizability block):
    # the threshold theta_u, the stiffness there, whether it then holds to theta_u*, and
    # the Eq. (20) updates from theta_u(0) = 0 that pass it. NaN where the conjunction
    # fails at theta_u* itself, so there is no threshold to report.
    for field in ("theta_crit", "crit_rate", "crit_holds", "crit_updates"):
        grids[field] = np.full(shape, np.nan)

    proportion = logistic(net.zeta)

    for i, alpha in enumerate(alphas):
        prior = beta_world_prior(alpha, 1.0)
        for j, strength in enumerate(lambdas):
            probe = net.respawn(base_prior=prior, lexical_strength=strength)  # own theta_u*
            grids["theta_u"][i, j] = probe.theta_u
            grids["stiffness"][i, j] = probe.stiffest_state_rate()
            upper = (probe.zeta >= probe.theta_L).to(probe.dtype)
            means = []
            for utterance in UTTERANCES:
                excluded = exclusion_indicator(
                    utterance, probe.zeta, probe.theta_L,
                    sharpness=probe.exclusion_sharpness,
                )
                phi_S = probe.closed_form_fixed_point(utterance)[0]
                literal = probe.literal_fixed_point(utterance)
                full = probe.read_out(phi_S)
                base = probe.read_out(literal)
                if utterance == "some":
                    settled_fields = (phi_S, probe.base_log_prior, literal)
                    nodes = [int(torch.argmax(field)) for field in settled_fields]
                    grids["mode_model"][i, j], grids["mode_prior"][i, j], grids["mode_literal"][i, j] = nodes
                    grids["mode_s_model"][i, j] = float(logistic(probe.zeta[nodes[0]]))
                    grids["mode_s_prior"][i, j] = float(logistic(probe.zeta[nodes[1]]))
                    grids["mode_outside"][i, j] = float(float(probe.zeta[nodes[0]]) < probe.theta_L)
                    # the same test applied to ell_0's own mode, so the two can be compared:
                    # a criterion its own baseline meets is reported separately below
                    grids["prior_mode_outside"][i, j] = float(
                        float(probe.zeta[nodes[1]]) < probe.theta_L)
                    grids["mode_gap"][i, j] = min(float(torch.topk(field, 2).values.diff().abs())
                                                  for field in settled_fields)
                tempered = probe.read_out(
                    probe.closed_form_fixed_point(utterance, theta_u=0.0)[0])
                mean = float((probe.weights * full * proportion).sum())
                means.append(mean)
                grids["expected_s"][utterance][i, j] = mean
                grids["leak"][utterance][i, j] = float(
                    (probe.weights * full * excluded).sum()
                )
                grids["shift"][utterance][i, j] = float(
                    (probe.weights * full * upper).sum()
                    - (probe.weights * base * upper).sum()
                )
                grids["tempering"][utterance][i, j] = float(
                    (probe.weights * tempered * upper).sum()
                    - (probe.weights * base * upper).sum()
                )
                grids["full_upper"][utterance][i, j] = float(
                    (probe.weights * full * upper).sum())
            grids["discriminability"][i, j] = max(means) - min(means)
            threshold = criterion_threshold(probe)
            if threshold is not None:
                arrival = updates_to_criterion(probe)
                grids["theta_crit"][i, j] = threshold["theta_u"]
                grids["crit_rate"][i, j] = threshold["rate"]
                grids["crit_holds"][i, j] = float(threshold["holds_to_star"])
                if arrival is not None:
                    grids["crit_updates"][i, j] = arrival["updates"]

    grids["alphas"] = alphas
    grids["lambdas"] = lambdas

    # === verify the closed-form shortcut against the full dynamics ===
    # Only cells whose own theta_u* keeps lambda_max(H) within FEASIBLE_STIFFNESS can be
    # integrated under the timescale commitment; the sample is drawn from those.
    feasible = [(i, j) for i in range(len(alphas)) for j in range(len(lambdas))
                if grids["stiffness"][i, j] <= FEASIBLE_STIFFNESS]
    rng = np.random.default_rng(0)
    picks = (rng.choice(len(feasible), size=min(verify_samples, len(feasible)),
                        replace=False) if feasible else [])
    worst = 0.0
    checked = []
    for pick in picks:
        i, j = feasible[int(pick)]
        probe = net.respawn(
            base_prior=beta_world_prior(alphas[i], 1.0), lexical_strength=lambdas[j]
        )
        utterance = UTTERANCES[int(rng.integers(len(UTTERANCES)))]
        settled = probe.infer(utterance, record_history=False)
        target = probe.closed_form_fixed_point(utterance)[0]
        worst = max(worst, float((settled["phi_S"] - target).abs().max()))
        checked.append((alphas[i], lambdas[j], utterance, settled["converged"]))
    grids["verification"] = {"worst_error": worst, "checked": checked,
                             "feasible": len(feasible)}

    if verbose:
        print("LAMBDA x ALPHA SWEEP")
        print(f"  alpha  in [{alphas[0]:.0f}, {alphas[-1]:.0f}]  ({len(alphas)} values)")
        print(f"  Lambda in [{lambdas[0]:.0f}, {lambdas[-1]:.0f}]  ({len(lambdas)} values)")
        print(f"  evaluated in closed form (Eqs. 15-16, exact by Sec. 8.6), each cell at")
        print(f"  its own learned theta_u*: from {np.nanmin(grids['theta_u']):.2f} to "
              f"{np.nanmax(grids['theta_u']):.2f}, negative at "
              + ", ".join(f"({alphas[i]:.0f}, {lambdas[j]:.0f})"
                          for i, j in zip(*np.where(grids["theta_u"] < 0))))
        print(f"  verification: integration is feasible where lambda_max(H) <= "
              f"{FEASIBLE_STIFFNESS:.0e} under the timescale commitment,")
        print(f"  which holds in {len(feasible)} of {len(alphas) * len(lambdas)} cells; "
              f"{len(checked)} of them re-run through the full dynamics:")
        print("    " + ", ".join(f"({a:.0f}, {l:.0f}, {y})" for a, l, y, _ in checked))
        print(f"    worst |phi_S - phi_S*| = {worst:.2e}, "
              f"all converged = {all(c[3] for c in checked)}")
        print()

        # where does the lexicon lose? first Lambda at which "no" still leaks > 1/2
        print("  smallest Lambda holding the entry for \"no\" (leak < 0.5):")
        for i, alpha in enumerate(alphas):
            row = grids["leak"]["no"][i]
            ok = [lambdas[j] for j in range(len(lambdas)) if row[j] < 0.5]
            print(f"    alpha = {alpha:>6.0f}   " +
                  (f"Lambda >= {min(ok):.0f}" if ok else f"none up to {lambdas[-1]:.0f}"))
        print()

    return grids


# === the plane in numbers ==================================================

def plane_summary(grids, verbose=True):
    """The counts and floors Text cell 6 reads off the plane, computed here.

    The shift for each utterance splits into the TEMPERING, the theta_u = 0 control
    against q_lit, and the UTILITY LEVEL's own contribution, the rest. The
    differential (39) is shift(some) - shift(no). Part C's mass shift criterion is
    shift(some) < 0 and its mass position criterion P(all-region | "some") < 1/2. A criterion's floor
    at a given alpha is the least Lambda from which it holds at every larger Lambda
    on this grid.

    Signs are read with a zero band of ROUNDOFF. Where ell_0 overrides the entry
    every utterance settles on the same belief, and the shifts there are differences of
    equal numbers, of order 1e-16 with either sign; counting their signs would count
    roundoff. Those cells are reported separately as saturated.
    """
    ROUNDOFF = 1e-12

    def sign(x):
        return np.where(np.abs(x) < ROUNDOFF, 0.0, np.sign(x))

    alphas, lambdas = grids["alphas"], grids["lambdas"]
    cells = [(i, j) for i in range(len(alphas)) for j in range(len(lambdas))]
    at = lambda i, j: f"({alphas[i]:.0f}, {lambdas[j]:.0f})"
    shift = {y: grids["shift"][y] for y in UTTERANCES}
    tempering = {y: grids["tempering"][y] for y in UTTERANCES}
    utility = {y: shift[y] - tempering[y] for y in UTTERANCES}
    differential = shift["some"] - shift["no"]
    first = sign(shift["some"]) < 0
    second = grids["full_upper"]["some"] < 0.5
    both = first & second

    def floor(holds, i):
        value = None
        for j in reversed(range(len(lambdas))):
            if not holds[i, j]:
                break
            value = lambdas[j]
        return value

    low = min(cells, key=lambda c: differential[c])
    high = max(cells, key=lambda c: differential[c])
    positive = [c for c in cells if sign(differential[c]) > 0]
    positive_some = [c for c in cells if sign(shift["some"][c]) > 0]
    saturated = [c for c in cells if sign(differential[c]) == 0]
    flat = [c for c in cells if abs(differential[c]) < 1e-3]
    tempering_gap = np.abs(tempering["some"] - tempering["no"]).ravel()
    utility_gap = np.abs(utility["some"] - utility["no"]).ravel()
    utility_sign = sign(utility["some"] - utility["no"])
    band = [c for c in cells if both[c]]
    summary = {
        "differential_negative": int((sign(differential) < 0).sum()),
        "both": int(both.sum()),
        "second": int(second.sum()),
        "first_floor": {alphas[i]: floor(first, i) for i in range(len(alphas))},
        "second_floor": {alphas[i]: floor(second, i) for i in range(len(alphas))},
    }

    # the delta read-out's two criteria under "some" (B8), beside the mass criteria. The mode is a
    # grid node, so the mode shift criterion is a node comparison; a mode that does not move has
    # a shift of 0 steps and does not meet it.
    mode_shift = grids["mode_model"] < grids["mode_prior"]
    mode_position = grids["mode_outside"] > 0.5
    mode_both = mode_shift & mode_position
    criteria = {"mass shift": first, "mass position": second, "mass both": both,
                "mode shift": mode_shift, "mode position": mode_position, "mode both": mode_both}
    summary.update({
        "mode_shift": int(mode_shift.sum()),
        "mode_position": int(mode_position.sum()),
        "mode_both": int(mode_both.sum()),
        "floors": {name: {alphas[i]: floor(holds, i) for i in range(len(alphas))}
                   for name, holds in criteria.items()},
    })

    if verbose:
        print("THE PLANE IN NUMBERS (each cell at its own learned theta_u*)")
        print()
        print("  differential (39), shift(some) - shift(no):")
        print(f"    negative in {int((sign(differential) < 0).sum())} of {len(cells)} cells; "
              f"below -1e-3 in {int((differential < -1e-3).sum())}; "
              f"below -1e-2 in {int((differential < -1e-2).sum())}")
        print(f"    zero to roundoff (|differential| < {ROUNDOFF:.0e}, saturated) in "
              f"{len(saturated)} cells: "
              + ", ".join(at(*c) for c in saturated))
        print(f"    most negative {differential[low]:+.4f} at {at(*low)}; "
              f"maximum {differential[high]:+.4f} at {at(*high)}")
        print(f"    positive in {len(positive)} cells, at alpha in "
              f"{sorted({int(alphas[i]) for i, _ in positive})}")
        print(f"    |differential| < 1e-3 in {len(flat)} cells, "
              f"{sum(lambdas[j] <= alphas[i] for i, j in flat)} of them with Lambda <= alpha")
        print("  raw shift for \"some\" (Part C's mass shift criterion):")
        print(f"    negative in {int(first.sum())} cells, minimum "
              f"{float(shift['some'].min()):+.4f}; positive in {len(positive_some)} cells, "
              f"up to alpha = {max(int(alphas[i]) for i, _ in positive_some) if positive_some else '-'}")
        print(f"    max |shift(no)| = {float(np.abs(shift['no']).max()):.4f}")
        print("  the \"some\" shift split into tempering (theta_u = 0 against q_lit) and utility:")
        resolved = np.maximum(np.abs(tempering["some"]), np.abs(utility["some"])) >= ROUNDOFF
        print(f"    tempering larger in magnitude in "
              f"{int(((np.abs(tempering['some']) > np.abs(utility['some'])) & resolved).sum())}"
              f" of the {int(resolved.sum())} cells where either is above roundoff; "
              f"shift and utility part of opposite sign in "
              f"{int((sign(shift['some']) * sign(utility['some']) < 0).sum())}")
        print(f"    some - no difference: tempering median {np.median(tempering_gap):.4f}, "
              f"max {tempering_gap.max():.4f}; utility median {np.median(utility_gap):.4f}, "
              f"max {utility_gap.max():.4f}")
        print(f"    sign agrees with the utility level's some - no difference: the "
              f"differential in {int((sign(differential) == utility_sign).sum())} cells, "
              f"the raw shift in {int((sign(shift['some']) == utility_sign).sum())}")
        print("  Part C's mass conjunction:")
        print(f"    mass position criterion, P(all-region | \"some\") < 1/2, in {int(second.sum())} "
              f"cells; both mass criteria in {int(both.sum())}"
              + (f", from {at(*min(band))} to {at(*max(band))}" if band else ""))
        # Both position criteria are met in many cells by ell_0 alone, before the network runs.
        # The two counts below leave those cells out: they ask, of the cells where the baseline
        # does NOT meet the criterion, in how many the settled state meets it anyway. Reported
        # as measurements, with no position taken on how to read them (B10).
        baseline_upper = grids["full_upper"]["some"] - shift["some"]
        baseline_majority = baseline_upper >= 0.5
        carried_q = baseline_majority & second
        prior_inside = grids["prior_mode_outside"] < 0.5
        carried_mode = prior_inside & (grids["mode_outside"] > 0.5)
        mode_shift_here = grids["mode_model"] < grids["mode_prior"]
        print("    where the baseline does not already meet a position criterion:")
        print(f"      q_lit holds at least half the all-region in {int(baseline_majority.sum())} "
              f"cells; the settled belief holds less than half in {int(carried_q.sum())} of them, "
              f"and the mass shift criterion is met in {int((carried_q & first).sum())} of those")
        print(f"      the mode of ell_0 lies inside the cell of \"all\" in {int(prior_inside.sum())} cells; "
              f"the mode of phi_S* lies outside it in {int(carried_mode.sum())} of them, and the mode shift "
              f"criterion is met in {int((carried_mode & mode_shift_here).sum())} of those")
        print(f"      the mode count is a subset of the q count: "
              f"{bool(int((carried_mode & ~carried_q).sum()) == 0)}; alpha over the q count runs "
              f"{sorted({int(alphas[i]) for i, _ in zip(*np.where(carried_q))})}, "
              f"over the mode count {sorted({int(alphas[i]) for i, _ in zip(*np.where(carried_mode))})}")
        met = ~np.isnan(grids["theta_crit"])
        if met.any():
            crit = grids["theta_crit"][met]
            rate = grids["crit_rate"][met]
            updates = grids["crit_updates"][met]
            print(f"    what the mass conjunction NEEDS, over those {int(met.sum())} cells:")
            print(f"      |theta_crit| from {np.abs(crit).min():.3f} to "
                  f"{np.abs(crit).max():.3f}, with lambda_max(H) there from "
                  f"{rate.min():.1f} to {rate.max():.1f},")
            print(f"      {int((rate <= FEASIBLE_STIFFNESS).sum())} of them within "
                  f"{FEASIBLE_STIFFNESS:.0e} and so integrable, against "
                  f"{np.nanmin(grids['stiffness'][met]):.1e} to "
                  f"{np.nanmax(grids['stiffness'][met]):.1e} at theta_u*;")
            print(f"      Eq. (20) from theta_u(0) = 0 passes the threshold in at most "
                  f"{np.nanmax(updates):.0f} update(s), and the mass conjunction then holds")
            print(f"      to theta_u* in {int(grids['crit_holds'][met].sum())} of them. "
                  f"What costs hours is theta_u*, not the mass conjunction.")
        print("    floor in Lambda by alpha (mass shift / mass position criterion):")
        print("      " + "  ".join(
            f"{a:.0f}: {summary['first_floor'][a] or '-':>}/{summary['second_floor'][a] or '-'}"
            for a in alphas))
        # the override diagonal (Lambda = 4 alpha): where ell_0 has beaten the
        # entry for "no" outright, the shift for "no" runs to +1 and the differential
        # stops reporting anything about "some" (Text cell 6 reads this).
        diagonal = [(i, j) for i, j in cells if shift["no"][i, j] > 0.5]
        if diagonal:
            print("  the override diagonal, where shift(no) > 1/2:")
            print("    " + ", ".join(f"{at(i, j)} {shift['no'][i, j]:+.4f}"
                                     for i, j in diagonal))
            print("    (there the differential subtracts a saturated shift, so the raw")
            print("     shift is the statistic to read; the leak map gives the same line.)")

        # the grid ceiling Text cell 6 warns about, at the sweep's own Lambda = 8 column
        if 8.0 in lambdas:
            column = lambdas.index(8.0)
            print(f"  the grid ceiling at Lambda = 8, each cell at its own theta_u*:")
            print(f"    {'alpha':>8}{'spread D':>12}{'max E[s]':>12}")
            for i, value in enumerate(alphas):
                if value in (64.0, 256.0, 1024.0):
                    means = [grids["expected_s"][y][i, column] for y in UTTERANCES]
                    print(f"    {value:>8.0f}{max(means) - min(means):>12.4f}"
                          f"{max(means):>12.4f}")
            print("    (D collapses as the base world prior sharpens: the top rows measure the grid")
            print("     as much as the model, and E[s] tends to about 0.98 rather than to")
            print("     sigma(6) = 0.9975.)")

        delta = alphas.index(64.0) if 64.0 in alphas else None
        if delta is not None:
            print("    the delta-like base world prior, alpha = 64, as Lambda rises:")
            for strength in (512.0, 1024.0, 2048.0):
                if strength in lambdas:
                    j = lambdas.index(strength)
                    print(f"      Lambda = {strength:>6.0f}   P(all-region | \"some\") = "
                          f"{grids['full_upper']['some'][delta, j]:.4f}   shift "
                          f"{shift['some'][delta, j]:+.4f}   theta_u* = "
                          f"{grids['theta_u'][delta, j]:.2f}")
        print()

        print("  THE MODE SHIFT AND MODE POSITION CRITERIA ON THE PLANE, under \"some\" (B8), beside the mass criteria")
        print("    mode shift: mode(phi_S*) - mode(ell_0) < 0;  mode position: mode(phi_S*) outside the cell of")
        print("    \"all\", zeta < theta_L. The mode is the grid node where the field is largest.")
        print(f"    the mode shift criterion is met in {int(mode_shift.sum())} of {len(cells)} cells, the mode "
              f"position criterion in {int(mode_position.sum())}, both in {int(mode_both.sum())}; the mass shift "
              f"criterion in {int(first.sum())}, the mass position criterion in {int(second.sum())}, both in "
              f"{int(both.sum())}")
        unmoved = int((grids["mode_model"] == grids["mode_prior"]).sum())
        print(f"    the mode of phi_S* is the node of mode(ell_0) in {unmoved} cells, a shift of 0 steps; "
              f"mode(ell_0 - phi_L) differs from mode(ell_0) in "
              f"{int((grids['mode_literal'] != grids['mode_prior']).sum())} cells")
        print(f"    smallest gap between the two largest nodes of phi_S*, ell_0 and ell_0 - phi_L: "
              f"{np.nanmin(grids['mode_gap']):.2e} (no ties)")
        print("    agreement, cell by cell:")
        for label, q, mode in (("the q and mode shift criteria", first, mode_shift),
                               ("the q and mode position criteria", second, mode_position),
                               ("the two conjunctions", both, mode_both)):
            print(f"      {label}: both met {int((q & mode).sum())}, only under q {int((q & ~mode).sum())}, "
                  f"only under the mode {int((~q & mode).sum())}, neither {int((~q & ~mode).sum())}")
        disagree = [c for c in cells if first[c] != mode_shift[c] or second[c] != mode_position[c]]
        yes = lambda flag: "yes" if flag else "no"
        print(f"    the {len(disagree)} cells where a mass criterion and its mode counterpart disagree:")
        print(f"      {'alpha':>6}{'Lambda':>7}{'theta_u*':>12}{'P(all|some)':>12}{'shift':>9}{'tempering':>10}"
              f"{'utility':>9}{'mode ell_0':>11}{'mode phi_S*':>12}{'mass shift':>11}{'mode shift':>11}"
              f"{'mass position':>14}{'mode position':>14}")
        for i, j in disagree:
            print(f"      {alphas[i]:>6.0f}{lambdas[j]:>7.0f}{grids['theta_u'][i, j]:>12.2f}"
                  f"{grids['full_upper']['some'][i, j]:>12.4f}{shift['some'][i, j]:>+9.4f}"
                  f"{tempering['some'][i, j]:>+10.4f}{utility['some'][i, j]:>+9.4f}"
                  f"{grids['mode_s_prior'][i, j]:>11.4f}{grids['mode_s_model'][i, j]:>12.4f}"
                  f"{yes(first[i, j]):>11}{yes(mode_shift[i, j]):>11}{yes(second[i, j]):>14}"
                  f"{yes(mode_position[i, j]):>14}")
        print("    floors by alpha under both read-outs: the least Lambda from which a criterion holds at every")
        print("    larger Lambda on this grid; - where it does not hold at the largest, * where it also holds")
        print("    at some Lambda below its floor")
        names = list(criteria)
        print(f"      {'alpha':>6} |" + "".join(f"{name:>14}" for name in names[:3]) + " |"
              + "".join(f"{name:>14}" for name in names[3:]))
        for i, alpha in enumerate(alphas):
            marks = []
            for name in names:
                value = summary["floors"][name][alpha]
                cut = len(lambdas) if value is None else lambdas.index(value)
                marks.append(("-" if value is None else f"{value:.0f}")
                             + ("*" if criteria[name][i, :cut].any() else ""))
            print(f"      {alpha:>6.0f} |" + "".join(f"{m:>14}" for m in marks[:3]) + " |"
                  + "".join(f"{m:>14}" for m in marks[3:]))
        print()
    return summary


# === heat maps =============================================================

def plot_lambda_alpha_sweep(grids):
    """Three figures. Rows are alpha (upward), columns are Lambda (rightward).

    Colour follows the job each statistic does. Leak and E[s] are magnitudes on a
    fixed 0-1 scale, so each takes a single-hue sequential ramp, light to dark.
    The shift is signed and its zero is meaningful, so it takes a diverging pair
    with a neutral midpoint and limits placed symmetrically about zero, since otherwise
    the midpoint colour would land somewhere other than "no effect".

    The shift is Eq. (37), measured against q_lit exactly as in Code Cell 2.
    """
    import matplotlib.pyplot as plt
    from matplotlib.colors import TwoSlopeNorm

    alphas, lambdas = grids["alphas"], grids["lambdas"]

    def draw(axis, values, *, cmap, norm=None, vmin=None, vmax=None, title=""):
        image = axis.imshow(
            values, origin="lower", aspect="equal", cmap=cmap,
            norm=norm, vmin=vmin, vmax=vmax, interpolation="nearest",
        )
        axis.set_xticks(range(len(lambdas)))
        axis.set_xticklabels([f"{l:.0f}" for l in lambdas], fontsize=6.5, rotation=90)
        axis.set_yticks(range(len(alphas)))
        axis.set_yticklabels([f"{a:.0f}" for a in alphas], fontsize=6.5)
        axis.set_title(title, fontsize=9)
        axis.set_xlabel("lexical strength $\\Lambda$", fontsize=8)
        for spine in axis.spines.values():          # recessive frame
            spine.set_color("0.8")
        axis.tick_params(length=2, color="0.8")
        return image

    # 1. leak: how much belief survives where the entry excludes
    figure, axes = plt.subplots(1, 3, figsize=(11.5, 4.9), sharey=True, layout="constrained")
    for axis, utterance in zip(axes, UTTERANCES):
        image = draw(axis, grids["leak"][utterance], cmap="Reds", vmin=0.0, vmax=1.0,
                     title=f'leak  |  "{utterance}"')
    axes[0].set_ylabel("concentration $\\alpha$ of the base world prior", fontsize=8)
    figure.colorbar(image, ax=axes, location="bottom", fraction=0.055, pad=0.02,
                    label="q-mass over the excluded region")
    figure.suptitle(
        "Where the lexical entry holds. Dark = the entry is overridden by ell_0.",
        fontsize=10)
    plt.show()

    # 2. E[s] and discriminability
    figure, axes = plt.subplots(1, 4, figsize=(14.5, 4.9), sharey=True, layout="constrained")
    for axis, utterance in zip(axes[:3], UTTERANCES):
        image = draw(axis, grids["expected_s"][utterance], cmap="Blues", vmin=0.0, vmax=1.0,
                     title=f'$\\mathbb{{E}}[s]$  |  "{utterance}"')
    figure.colorbar(image, ax=axes[:3], location="bottom", fraction=0.055, pad=0.02,
                    label="$\\mathbb{E}[s]$")
    separation = draw(axes[3], grids["discriminability"], cmap="Greens", vmin=0.0, vmax=1.0,
                      title="$\\max_y\\mathbb{E}[s]-\\min_y\\mathbb{E}[s]$")
    figure.colorbar(separation, ax=axes[3], location="bottom", fraction=0.055, pad=0.02,
                    label="spread")
    axes[0].set_ylabel("concentration $\\alpha$ of the base world prior", fontsize=8)
    figure.suptitle(
        "What the model believes. Rightmost: how far apart the three utterances are: "
        "pale means the model has stopped distinguishing them.", fontsize=10)
    plt.show()

    # 3. implicature shift: signed, so diverging about zero
    figure, axes = plt.subplots(1, 4, figsize=(14.5, 4.9), sharey=True, layout="constrained")
    stack = np.concatenate([grids["shift"][y].ravel() for y in UTTERANCES])
    limit = float(np.nanmax(np.abs(stack))) or 1.0
    norm = TwoSlopeNorm(vmin=-limit, vcenter=0.0, vmax=limit)
    for axis, utterance in zip(axes[:3], UTTERANCES):
        image = draw(axis, grids["shift"][utterance], cmap="RdBu_r", norm=norm,
                     title=f'shift  |  "{utterance}"')
    figure.colorbar(image, ax=axes[:3], location="bottom", fraction=0.055, pad=0.02,
                    label="shift in $P(\\mathrm{all})$")

    differential = grids["shift"]["some"] - grids["shift"]["no"]
    d_limit = float(np.nanmax(np.abs(differential))) or 1.0
    diff_image = draw(axes[3], differential, cmap="PuOr_r",
                      norm=TwoSlopeNorm(vmin=-d_limit, vcenter=0.0, vmax=d_limit),
                      title="shift(some) $-$ shift(no)")
    figure.colorbar(diff_image, ax=axes[3], location="bottom", fraction=0.055, pad=0.02,
                    label="difference")
    axes[0].set_ylabel("concentration $\\alpha$ of the base world prior", fontsize=8)
    figure.suptitle(
        "Implicature probe across the plane, each cell at its own learned $\\theta_u^\\ast$. "
        "Blue = q-mass moved AWAY from the \"all\" region (the implicature direction).\n"
        "Rightmost is the differential (39), a control on the \"some\" panel; Text cell 6 "
        "reports where the two coincide.",
        fontsize=10)
    plt.show()


# === how much mu_u moves a cell ============================================

MU_SENSITIVITY = ([1.0, 1.0], [0.05, 0.05], [5.0, 5.0], [-1.0, -1.0], [1.0, 2.0])


def mu_u_sensitivity(net, cells=((1.0, 8.0), (16.0, 128.0)), *, verbose=True):
    """What a hundredfold range of mu_u, both signs, does to two cells of the plane.

    At a FIXED theta_u this is the failure mode of Text cell 5 Part B Sec. 4: mu_u is
    common-mode across utterances and moves no phi_S contrast, yet it moves E[s] and
    Delta_some, which are taken after the softmax. Under learning the same range moves
    theta_u* instead, and the statistics barely move: the reading inverts, so both are
    reported. The theta_u = 1 column is the control.
    """
    rows = []
    for alpha, strength in cells:
        entry = {"cell": (alpha, strength), "fixed": [], "learned": []}
        for mu in MU_SENSITIVITY:
            overrides = dict(base_prior=beta_world_prior(alpha, 1.0),
                             lexical_strength=strength,
                             mu_u=torch.tensor(mu, dtype=net.dtype))
            for key, theta in (("fixed", 1.0), ("learned", None)):
                probe = net.respawn(**overrides, **({"theta_u": theta} if theta else {}))
                belief = probe.read_out(probe.closed_form_fixed_point("some")[0])
                literal = probe.read_out(probe.literal_fixed_point("some"))
                upper = (probe.zeta >= probe.theta_L).to(probe.dtype)
                proportion = logistic(probe.zeta)
                entry[key].append({
                    "mu_u": mu,
                    "theta_u": probe.theta_u,
                    "expected_s": float((probe.weights * belief * proportion).sum()),
                    "shift": float((probe.weights * belief * upper).sum()
                                   - (probe.weights * literal * upper).sum()),
                })
        rows.append(entry)

    if verbose:
        print("MU_U SENSITIVITY AT TWO CELLS (Text cell 6 reads this)")
        print(f"  mu_u over {MU_SENSITIVITY[1]} to {MU_SENSITIVITY[2]}, both signs included")
        print()
        print(f"  {'cell':<14}{'theta_u':<10}{'theta_u* range':>22}"
              f"{'E[s] | some':>22}{'E[s] width':>12}{'shift, some':>26}{'sign change':>13}")
        for entry in rows:
            label = f"({entry['cell'][0]:.0f}, {entry['cell'][1]:.0f})"
            for key, name in (("fixed", "1 (control)"), ("learned", "learned")):
                values = entry[key]
                thetas = [v["theta_u"] for v in values]
                means = [v["expected_s"] for v in values]
                shifts = [v["shift"] for v in values]
                span = (f"{min(thetas):.2f} to {max(thetas):.2f}" if key == "learned"
                        else "held at 1")
                changed = min(shifts) < 0.0 < max(shifts)
                print(f"  {label:<14}{name:<10}{span:>22}"
                      f"{f'{min(means):.4f} to {max(means):.4f}':>22}"
                      f"{max(means) - min(means):>12.4f}"
                      f"{f'{min(shifts):+.4f} to {max(shifts):+.4f}':>26}"
                      f"{str(changed):>13}")
        print("  At a fixed theta_u a common-mode mu_u still moves statistics taken after")
        print("  the softmax, and at (16, 128) it moves the shift across zero. With theta_u*")
        print("  re-learned per setting, mu_u spends itself on theta_u* and the statistics")
        print("  barely move: the same range leaves the sign of the shift alone.")
        print()
    return rows


# === the override threshold =================================================

def override_threshold(net, *, alpha=1024.0, boundaries=(1.0, 2.0, 3.0), verbose=True):
    """Locate the diagonal the heat maps show, and check what sets it.

    On the truth region the settled log-density is ell_0/2; on the excluded region it
    is (ell_0 - Lambda)/2 (Eq. 15). The entry therefore survives only while Lambda
    exceeds the gap in ell_0 between the two regions. For Beta(alpha, 1) that gap is
    dominated by -alpha*log s at the truth region's best point s = sigma(-theta_L),
    predicting

        Lambda_crit  ~  |log sigma(-theta_L)| * alpha,                       (41)

    linear in alpha with a slope fixed by the endpoint threshold alone.

    Eq. (41) is derived with the utility level severed, and holds to a few percent
    there. Restoring it RAISES the required Lambda: the lexicon needs MORE strength to
    hold against ell_0, not less. The model's own value is the LEARNED column, which
    re-learns theta_u* at every Lambda the bisection visits, since each is its own
    configuration; theta_u = 0, +1, -1 and the magnitude sweep are controls that hold
    theta_u fixed in order to separate amplification from regression toward mu_u. The mechanism is the one Text cell 4 Part C
    measures: raising |theta_u| amplifies the span(B) component of (ell_0 - phi_L),
    and under Beta(alpha, 1) that component is dominated by the tilt of ell_0, so
    amplifying it reinforces ell_0 against the entry. Two things follow, and both
    are checked below rather than asserted: the effect is quadratic in theta_u, so it
    is the SAME for either sign and saturates as |theta_u| grows; and it is
    independent of mu_u, which rules out reading it as regression toward mu_u.
    """
    def critical(boundary, theta, mu=None):
        """Lambda_crit at a fixed theta_u, or at each probe's own theta_u* if None."""
        def leak(strength):
            overrides = dict(
                base_prior=beta_world_prior(alpha, 1.0),
                lexical_strength=strength,
                theta_L=boundary,
            )
            if mu is not None:
                overrides["mu_u"] = torch.tensor(mu, dtype=net.dtype)
            if theta is not None:
                overrides["theta_u"] = theta        # a fixed-theta_u control
            probe = net.respawn(**overrides)
            belief = probe.read_out(probe.closed_form_fixed_point("no")[0])
            excluded = exclusion_indicator("no", probe.zeta, boundary)
            return float((probe.weights * belief * excluded).sum())

        low, high = 1e-3, 1e7
        for _ in range(90):                       # geometric bisection
            middle = math.sqrt(low * high)
            if leak(middle) < 0.5:
                high = middle
            else:
                low = middle
        return math.sqrt(low * high)

    rows = []
    for boundary in boundaries:
        predicted = -math.log(float(torch.sigmoid(torch.tensor(-boundary))))
        slope_learned = critical(boundary, None) / alpha
        at_threshold = net.respawn(base_prior=beta_world_prior(alpha, 1.0),
                                   lexical_strength=slope_learned * alpha,
                                   theta_L=boundary)
        rows.append({
            "boundary": boundary,
            "predicted_slope": predicted,
            "slope_with_u": critical(boundary, 1.0) / alpha,
            "slope_without_u": critical(boundary, 0.0) / alpha,
            "slope_negated": critical(boundary, -1.0) / alpha,
            "slope_learned": slope_learned,
            "theta_at_threshold": at_threshold.theta_u,
        })

    if verbose:
        print("OVERRIDE THRESHOLD (the diagonal in the leak map)")
        print(f"  Lambda at which leak(\"no\") = 1/2, divided by alpha, at alpha = {alpha:.0f}")
        print()
        print(f"  {'theta_L':>8}{'Eq. (41)':>11}{'theta_u=0':>11}{'(41) ratio':>10}"
              f"{'theta_u=1':>11}{'theta_u=-1':>12}{'learned':>10}{'change at 1':>13}"
              f"{'change, learned':>17}{'theta_u* there':>16}")
        for row in rows:
            print(f"  {row['boundary']:>8.0f}{row['predicted_slope']:>11.4f}"
                  f"{row['slope_without_u']:>11.4f}"
                  f"{row['slope_without_u'] / row['predicted_slope']:>10.3f}"
                  f"{row['slope_with_u']:>11.4f}{row['slope_negated']:>12.4f}"
                  f"{row['slope_learned']:>10.4f}"
                  f"{row['slope_with_u'] / row['slope_without_u'] - 1.0:>+12.1%}"
                  f"{row['slope_learned'] / row['slope_without_u'] - 1.0:>+17.1%}"
                  f"{row['theta_at_threshold']:>16.2f}")
        print("  (learned: theta_u* re-learned at every Lambda the bisection visits; the last")
        print("   column is theta_u* at the threshold itself. theta_u = 0, +1, -1: controls.)")
        print()
        at_one = [row["slope_with_u"] / row["slope_without_u"] - 1.0 for row in rows]
        learned = [row["slope_learned"] / row["slope_without_u"] - 1.0 for row in rows]
        print("  With the utility level severed, Eq. (41) is accurate to a few percent,")
        print("  so the diagonal is the gap in ell_0 and nothing else. Restoring the level")
        print(f"  RAISES the required Lambda, by {min(at_one):.0%}-{max(at_one):.0%} at the "
              f"theta_u = 1 control and by {min(learned):.0%}-{max(learned):.0%} at the")
        print("  learned theta_u*. The utility level reinforces ell_0 against the")
        print("  lexical entry; it does not oppose it.")
        print()
        print("  Two checks that this is amplification (Text cell 4 Part C) and not")
        print("  regression toward mu_u:")
        print()
        baseline = critical(boundaries[0], 0.0) / alpha
        print(f"    (a) sign and magnitude, at theta_L = {boundaries[0]:.0f}:")
        print(f"        {'theta_u':>10}{'slope':>10}{'vs theta_u=0':>15}")
        for theta in (1.0, -1.0, 3.0, -3.0, 10.0, 100.0, 10000.0):
            value = critical(boundaries[0], theta) / alpha
            print(f"        {theta:>10.0f}{value:>10.4f}{value / baseline - 1.0:>+15.1%}")
        print("        +theta and -theta agree; the effect saturates. Both are")
        print("        signatures of the theta_u^2 term, not of a mu_u-carried one.")
        print()
        print(f"    (b) mu_u-independence, at theta_L = {boundaries[0]:.0f}, at the theta_u = 1")
        print("        control and with theta_u* re-learned for each mu_u (and each Lambda):")
        print(f"        {'mu_u':>16}{'slope, 1':>10}{'vs theta_u=0':>15}"
              f"{'slope, learned':>16}{'vs theta_u=0':>15}")
        for mu in ([1.0, 1.0], [0.05, 0.05], [5.0, 5.0], [-1.0, -1.0], [1.0, -1.0]):
            value = critical(boundaries[0], 1.0, mu) / alpha
            try:
                learned_value = critical(boundaries[0], None, mu) / alpha
                learned_cells = (f"{learned_value:>16.4f}"
                                 f"{learned_value / baseline - 1.0:>+15.1%}")
            except ValueError:
                learned_cells = f"{'degenerate on the way (Appendix B)':>31}"
            print(f"        {str(mu):>16}{value:>10.4f}{value / baseline - 1.0:>+15.1%}"
                  + learned_cells)
        print("        At theta_u = 1, a hundredfold change in |mu_u|, and both signs, move")
        print("        the slope by less than a tenth of a percent. mu_u is not what does this.")
        print()

    return rows


def grid_caveat_report(net, alphas=(64.0, 300.0, 1024.0, 4096.0)):
    """Text cell 6's grid caveat: how much base world prior mass a truncated grid puts on its top node.

    The plane's alpha sharpens ell_0 as Beta(alpha, 1) in s, whose mass piles toward s = 1.
    A grid truncated at |zeta| <= Z cannot represent an arbitrarily concentrated base world prior: past
    some alpha the top node carries a mass that a finer or wider grid would have spread over
    the states beyond it. The bound bites GRADUALLY rather than at a threshold, which is why
    it is reported as a ladder and not as a cutoff. The alphas here run past the plane's own
    ladder (which stops at 1024) so the trend is visible; the plane is not evaluated here.

    What fails first is not the base world prior but the model's ability to tell the utterances apart,
    which the spread D of Eq. (38) reports on the plane itself. Reported, not interpreted.
    """
    print("THE GRID CAVEAT: PRIOR MASS ON THE TOP NODE AS alpha SHARPENS (Text cell 6)")
    print(f"  grid half-width Z = {float(net.zeta[-1]):g}, {net.num_nodes} nodes, "
          f"ell_0 = Beta(alpha, 1) in s, grid-renormalized as code cell 1 renormalizes it")
    print(f"    {'alpha':>8}{'top-node mass':>16}{'top-two mass':>15}{'E[s] under ell_0':>19}")
    for alpha in alphas:
        probe = net.respawn(base_prior=beta_world_prior(alpha, 1.0))
        density = torch.exp(probe.base_log_prior)
        mass = probe.weights * density
        total = float(mass.sum())
        top = float(mass[-1]) / total
        top_two = float(mass[-2:].sum()) / total
        mean_s = float((mass * torch.sigmoid(probe.zeta)).sum()) / total
        print(f"    {alpha:>8.0f}{top:>16.4f}{top_two:>15.4f}{mean_s:>19.4f}")
    print("  The mass is a fraction of the base world prior's own total on this grid, so it is what the")
    print("  implementation actually holds, not what Beta(alpha, 1) holds on [0, 1].")
    print()


sweep_grids = lambda_alpha_sweep(evaluation_network)
plane_numbers = plane_summary(sweep_grids)
mu_sensitivity = mu_u_sensitivity(evaluation_network)
override_threshold(evaluation_network)
grid_caveat_report(evaluation_network)
plot_lambda_alpha_sweep(sweep_grids)
````

## <a id="appa" name="appa"></a>Appendix A: the utterance level and the threshold $\theta_L$

**Realizing $E_y$ on the grid: the threshold $\theta_L$.** The scale on which Eq. (5) defines the
entries is the closed interval $[0,1]$. The network carries the state in logit coordinates (Eq. 1),
and $\operatorname{logit}$ is asymptotic at $s=0$ and $s=1$, so it is a bijection only on the **open**
interval $(0,1)$: the two endpoints at which Eq. (5) anchors *no* and *all* have no $\zeta$ coordinate.
Realizing $E_y$ on the grid consequently requires a threshold $\theta_L>0$, under which

$$E_{\textit{no}}\mapsto\{\zeta>-\theta_L\},\qquad
E_{\textit{some}}\mapsto\{\zeta\le-\theta_L\},\qquad
E_{\textit{all}}\mapsto\{\zeta<\theta_L\}.
\tag{A1}$$

**$g_y$, and $\theta_L$ as its generative weight.** $g_y$ maps lexical content to word form. Write
$A\in\mathbb R^{|\mathcal Y|\times K}$ for a fixed profile matrix, one projection per word-form unit,
as $b$ is one projection at the $S$ level (§3), and let $W$ be the quadrature measure of Eq. (3). Then

$$\hat\varphi_y=g_y(\varphi_L)=\theta_L\,A W\varphi_L.
\tag{A2}$$

This is $\Theta h(\varphi)$ with $\Theta=\theta_L A$ and $h=\mathrm{id}$: the form of $g_S$ (Eq. 10)
with the direction reversed, $g_S$ expanding $\mathbb R^m$ into a field where $g_y$ contracts a field
onto the $|\mathcal Y|$ word-form units. $A$ is fixed and $\theta_L$ is the whole of the parameter,
exactly as $B$ is fixed and $\theta_u$ is the whole of the parameter in $g_S$.

The three fields $\Lambda\chi_y(\theta_L)$ of Eq. (A1) are linearly independent under $W$ (their Gram
matrix has rank $3$), so an $A$ exists for which $g_y$ returns the observed word form exactly at each
clamp. Under it Eq. (8) is a *consequence*, as §4 states, rather than a stipulation.

**The free energy this appendix differentiates.** Eq. (13) drops the $y$ level on the grounds that
$r_y\equiv0$ contributes nothing. That is true of the term's *value* and not of its *gradient*: a term
pinned at zero by a clamp still has a derivative in the parameter that would move it off the clamp.
Read literally, Eq. (13) gives $\partial\mathcal F/\partial\theta_L=0$ identically, which is the right number at this phase, for the wrong reason, and with no direction attached to it. What is differentiated
below is Eq. (13) with the $y$ level restored:

$$\mathcal F=-\frac{\lVert r_y\rVert^{2}}{2\sigma_y}
-\frac{\lVert r_L\rVert^{2}}{2\sigma_L}
-\frac{\lVert r_S\rVert^{2}}{2\sigma_S}
-\frac{\lVert r_u\rVert^{2}}{2\sigma_u}
-\tfrac12\big(\lvert\mathcal Y\rvert\log\sigma_y+K\log\sigma_L+K\log\sigma_S+m\log\sigma_u\big)+C,
\qquad r_y=\varphi_y-\theta_L\,AW\varphi_L .
\tag{A3}$$

Here $\lVert r_y\rVert^{2}$ is the ordinary squared norm on $\mathbb R^{\lvert\mathcal Y\rvert}$
(Eq. 4), as $\lVert r_u\rVert^{2}$ is on $\mathbb R^m$; the two field norms remain those of Eq. (2).
The $y$ level contributes one $\log\sigma$ per component on the same rule as the other three, hence
$\lvert\mathcal Y\rvert$ and not $K$.

**Eq. (A3) is not a different model, which is why Text cell 3 may state Eq. (13) as it does.** $r_y$
depends on $\varphi_y$, $\theta_L$ and $\varphi_L$, and on nothing below $\varphi_L$. So the restored
term is constant in $x=(\varphi_S,\varphi_u)$, leaving §8's concavity,
uniqueness and convergence results untouched together with every closed form of §6, and at the
clamp its value is $0$, so
$\mathcal F$ takes the same numbers everywhere the implementation evaluates it. The truncation loses
exactly one thing, and it is the thing this appendix needs: the direction $\theta_L$ would move in.

**Its gradient, and its flow.** Differentiating Eq. (A3) in the gain of Eq. (A2), with
$\partial r_y/\partial\theta_L=-AW\varphi_L$ and $\varepsilon_y=r_y/\sigma_y$ on the rule of
Eq. (14),

$$\tau_{\theta_L}\dot\theta_L=\frac{\partial\mathcal F}{\partial\theta_L}
=\frac{1}{\sigma_y}\,r_y\cdot AW\varphi_L
=\varepsilon_y\cdot AW\varphi_L\,,
\tag{A4}$$

The right-hand side is the error below times the activity above, and each of its terms is a product of pre- and
post-synaptic activity of the form of Bogacz's Eq. (25). $\theta_L$ takes a time
constant of its own at the slow end of the model's one ordering,
$\tau_\varepsilon\le\tau_\varphi/(4\lambda_{\max}(H))\ll\tau_\varphi\ll\tau_\theta$ (Text cell 3
commitment 7), beside $\theta_u$. Eq. (A4) involves no normalization across the word-form units, which would be a further
non-locality, but it does sum over them, which the paragraph closing *$\theta_L$ enters twice* below takes up. Note that Eq. (A4) is the partial through the gain alone; $\theta_L$'s second appearance, in
the threshold that builds $\chi_y$, is taken up under *$\theta_L$ enters twice* below.

Eq. (A4) vanishes identically while $\varepsilon_y\equiv0$ (Eq. 8). **$\theta_L$ is therefore fixed because $\varphi_L$ is clamped and $A$ is calibrated to the clamps, not by stipulation**, and unclamping $\varphi_L$ sets it running (§4). Until then it is carried as a fixed
quantity with a variational quantity's status.

**$\theta_L$ enters twice.** It is the gain of Eq. (A2) and, through Eq. (A1), the threshold building
$\chi_y$ and so the clamp itself, which makes it the only parameter of the model appearing in both the
generative map and the observation model. That is what identifies it: $\theta_L\,AWX(\theta_L)=I$, with $X$ the matrix of clamped fields $\Lambda\chi_y$ of Eq. (4), is a
fixed-point equation in $\theta_L$, and any departure from it leaves an $\varepsilon_y$ for Eq. (A4)
to act on.

Each term of Eq. (A4) is local, as shown under Eq. (A4). **Eq. (A4) as a whole nevertheless does not satisfy the
locality constraint for the three-member inventory stated here**: $\theta_L$ gains all $\lvert\mathcal Y\rvert$ rows
of $A$, so the sum runs over products formed at $\lvert\mathcal Y\rvert$ different word-form units; this is the same
violation Eq. (B4) states for $\theta_u$ at $m>1$, and a relay would not resolve it, for the reason Appendix B gives for $\theta_u$. On the other hand, under the alternatives level which the accompanying paper proposes
(§5.2.2), each level would carry two lexical entries, and if each pair is carried by a single
log-odds unit, as that proposal has it, Eq. (A4) has a single term and the question does not arise.

**$n$: the denotation of $\theta_L$.** A threshold at $\varsigma(-\theta_L)$ commits the predicate to
a finest distinguishable quantity. Let the predicate resolve $n$ atoms, so that the scale carries
$n+1$ quantities, the counts $k=0,\dots,n$ at proportions $s_k=k/n$, with resolution $1/n$. The
**Voronoi cell** of a count is the band of width $1/n$ about it, the proportions closer to that count
than to any other. **This phase stipulates** that the region each entry leaves open is the Voronoi cell
of the count it anchors, $s_0$ for *no* and $s_n$ for *all*, symmetrically at both ends. Since
$\varsigma$ is strictly increasing, that fixes the threshold uniquely: the cell's boundary lies half a
cell from the endpoint, so

$$\varsigma(-\theta_L)=\frac{1}{2n}
\qquad\Longleftrightarrow\qquad
\theta_L=\log(2n-1),
\qquad n=\frac{e^{\theta_L}+1}{2},
\tag{A5}$$

and Eq. (A1) then excludes precisely the states lying outside the cell of the count each entry
anchors. Eq. (A5) gives $\theta_L$ its denotation: **$n$ is the number of atoms the predicate
resolves**. It also settles how a threshold can serve as the gain of Eq. (A2): $e^{\theta_L}=2n-1$, so
$\theta_L$ is a *log*-gain, and since $\zeta=\operatorname{logit}s$ is a log-odds coordinate,
translating by $\theta_L$ in $\zeta$ is multiplying by $e^{\theta_L}$ in odds. $\theta_L$ is
multiplicative on the odds scale and appears as a location in $\zeta$ only because $\zeta$ is
logarithmic. At the canonical $n=10$, $\theta_L=\log19=2.9444$ and $\varsigma(-\theta_L)=0.05$.

*$\theta_L$ is semantic.* Three consequences follow.

First, this design predicts that **exact zero is not distinguished** from proportions below half an atom; if the
language user's resolution is the architecture's, natural language shares that limit. There is no resolution limit
without a finite set to resolve, and a predicate over $n$ atoms cannot distinguish "none" from "fewer
than half an atom". *No* means *below the resolution limit*, which is what a finite perceptual system
can mean by it. Eq. (5) states the semantics; Eq. (A1) states what a system of finite granularity does
with it. I invite the reader to have their own judgment on whether this prediction is a desideratum, as one might
intuitively think zero is representable. However, it is useful to distinguish representation from imagination per induction.

Second, $\theta_L$ grows like $\log 2n$, so the grid must leave $E_{\textit{no}}$ some unexcluded
node: the implementation requires $\theta_L<Z$ for grid half-width $Z$ (§1, where $Z$ is
distinguished from $K$ as the grid parameter that is not an accuracy parameter), i.e.
$n<(e^{Z}+1)/2=202.21$, capping $n$ at $202$ when $Z=6$ ($\theta_L=5.99894$; $n=203$ gives
$6.00389$ and is rejected). Sharpness is bounded by the semantics rather than free. **The same inequality has a second
consequence, and it is not a cap.** $\theta_L<Z$ leaves the cell of *all* an extent of
$Z-\theta_L$, so $n$ and $Z$ set its two ends between them: raising $n$ lifts the near end,
widening the grid pushes the far one out. §1 states what follows for a statistic read over it.

*The gradable case: the same formula, a different source for $n$.* A gradable predicate supplies
no count, so $\theta_L$ needs a different source. It does not need a different *formula*, and it
does not need a different name. Read Eq. (A5) back to what it asserts: the threshold sits at half
the scale's resolution, and for a counted predicate the resolution happens to be $1/n$ because $n$
equally spaced atoms is what generates it. Where nothing counts, $n$ is **the number of
distinguishable steps the scale affords** (the reciprocal of a just-noticeable difference, expressed
as a fraction of the scale), and Eq. (A5) is read with a real $n$ rather than an integer
one. The two cases differ in where $n$ comes from, not in the formula: a count where one exists, a
resolution reciprocal where it does not. Nothing in §§4–8 changes, since $\theta_L$ enters only
through $\chi_y$ and Eq. (A2), and neither asks where it came from. The grid cap above covers both
readings unchanged: $n<202.21$ bounds distinguishable steps whether or not they are countable
things, so a resolution finer than about half a percent of the scale is unrepresentable on the
default grid.

**What fixes $n$ is not settled here, and does not need to be.** This phase *stipulates* $n$, at the
canonical $10$ above, and models no predicate whose atoms are unstable, so nothing it reports turns
on an answer. Two readings of what a resolution could mean are available and they are not
equivalent. A resolution constant in $s$, a fixed absolute discrimination width, is what the
counting case supplies, since atoms are equally spaced in the proportion. A resolution constant in
$\zeta$ is the Weber-type alternative, a fixed ratio in odds, and it is the one the coordinate of
Eq. (1) is natural for: $\zeta$ is a log-odds scale and $\theta_L$ is already a log-gain, so a
constant-in-$\zeta$ resolution would make $\theta_L$ a step count rather than a function of one.
The two disagree most near the endpoints, which is exactly where Eq. (A1) places the boundaries.
This appendix states both and reserves its position between them.

**A later phase does not fix $n$ either; it reads it off.** Where $\varphi_L$ is unclamped, $n$ is
not a parameter to be stipulated but a **read-out of a learned $\theta_L$**, through Eq. (A5) run
backwards. Eq. (A4) says why this phase cannot do that: $\theta_L$'s gradient vanishes identically
while $\varepsilon_y\equiv0$, so nothing here learns it. That reframes the question rather than
answering it: not *which* resolution rule is right, but what exposure would have to teach
$\theta_L$, which is where §9.2's dense-scale extension reconnects.

**What $n$ does to what the evaluation reports.** Code Cell A runs all four criteria of Text cell 4
at every combination of $n$ on the ladder, both lexical strengths and all four base world priors, and
prints the status of each beside the quantity it tests. Three things do not move. The three
$\chi_y$ keep rank $3$ at every $n$, so Appendix C's count of thresholds, and with it $m=2$, is
unaffected. The mode shift criterion is met in a single row of the whole table, under *skewed low*
at $\Lambda=8$ and $n=201$: on this sweep the mode of $\varphi_S^\ast$ almost always sits *above* the mode of $\ell_0$, whatever the mass criteria report. And $n=1$ remains the only degenerate case (there
$\theta_L=0$ and the entries for *some* and *all* coincide, which the constructor refuses), while
$n=2$ is admitted with two distinct thresholds. The separation between the $\kappa_y$ is best
near $n=15$, and the implementation's $n=10$ sits near that maximum.

**What does move, and how far.** The mass shift criterion changes status somewhere on the ladder in
every one of the eight $(\Lambda,\ell_0)$ blocks. The mass position criterion does not: under $N(0,1)$
and under *skewed low* at $\Lambda=8$ it is met at every rung. Where it does change, the mode
position criterion changes at the same $n$, and the two return the same status in every row of the
table but one: *flat* at $\Lambda=8$ and $n=2$, where the mode of $\varphi_S^\ast$ lies outside the cell of *all* while more than half the mass lies inside it.

Two qualifications belong with any of this, and neither is visible in a status column. The first is
magnitude. Under the default $N(0,1)$ base world prior at $\Lambda=8$ the mass shift reaches its threshold only
after falling from $+1.017\times10^{-1}$ at $n=2$ to $-7.593\times10^{-7}$, so what changes status
there is the sign of a quantity that has decayed to nothing; under *skewed high* at $\Lambda=512$
the same criterion changes status while the quantity is $-9.546\times10^{-2}$. Those are not findings
of the same weight, and the table is meant to be read with its printed quantities and not by its
statuses alone. The second is that a ladder **brackets** rather than locates: two rungs with
different statuses place the change somewhere between them, and no rung is a measurement of where it
happens.

**Granularity and lexical strength are not separable.** Where a criterion changes status depends on
$\Lambda$. The mass shift criterion changes at $n=50$ under three of the base world priors at $\Lambda=8$ and by
$n=15$ or $n=10$ under all four at $\Lambda=512$; the mass position criterion, met throughout the
$\Lambda=8$ ladder under $N(0,1)$, instead fails at $n\le5$ when $\Lambda=512$. $\theta_u^\ast$ is
not monotone in $n$ either, and under the flat base world prior at $\Lambda=8$ it is not even bounded: the
coupling $\langle\mu_u,\sum_y c_y\rangle$ the cell prints beside it changes sign twice along the
ladder, and since the two roots of Eq. (B2) have the constant product $-S/\sigma_u$,
$\theta_u^\ast$ diverges exactly where that coupling vanishes. That is Appendix B's degenerate ray,
and it is reached here by varying $n$ alone. So the granularity a predicate resolves is not a unit
convention that the rest of the evaluation is invariant to. The cell prints it; this appendix states
it and draws nothing from it.

**The cell of *all* has two ends, and $n$ fixes only one of them.** The cell runs from $\theta_L$ to
the grid's cut at $+Z$, so its width in log-odds is $Z-\theta_L$. Everything above moves the lower
end by varying $n$. The half-width moves the upper end, and it is stipulated in the same way: Text
cell 3 Sec. 1 separates it from the node count $K$, which is an accuracy parameter that the
specification check reports converging, whereas Eq. (1) carries the state onto the whole of
$\mathbb{R}$ and cutting it at $\pm Z$ bounds the state space instead. The cell runs a second
ladder over $Z$ at fixed node spacing, so that truncation is varied and resolution is not, and
reports the same four criteria and the same plane over it.

**What the second ladder shows, stated and not interpreted.** The four criteria change status along
it as they do along the ladder in $n$, and the mechanism is visible in the column beside them: the mode of $\varphi_S^\ast$ moves up the scale as $Z$ grows, because the basis $B$ is orthonormalized
under the grid's own measure and so is rebuilt whenever the grid is. That mode is the mode position criterion's own left-hand side. Under the flat base world prior at $\Lambda=512$ it crosses the start of the
cell of *all* between $Z=6$ and $Z=7$, which is where the two mass criteria stop being met. The plane
block counts the same thing cell by cell, including the two counts that condition on $\ell_0$ not
already meeting a position criterion. Because the dependence enters through $B$, it sits below the
criteria rather than in them. Eq. (24)'s limit doubles the field's component in the span of $B$, a
projection taken under the grid's own measure on $[-Z,Z]$, so no restatement of a criterion over
$\varphi_S^\ast$ removes it. A projection under a measure fixed independently of $Z$ would remove
it, and would give up the orthonormality $B^{\mathsf T}WB=I$ on which the scalar form of
Eqs. (B1)–(B2) rests.

**The two cautions of the $n$ ladder carry over unchanged**, and one is added. A ladder brackets a
change rather than locating it, and a change of status may be the sign flip of a quantity that has
already decayed to nothing. The addition is that a count conditioned on the baseline failing is
bounded above by the shift count, since a belief that falls below half of what the baseline held
above half has fallen, so the conditioned counts cannot outlast the shift criterion along this
ladder. As with $n$, the cell prints all of it and this appendix draws nothing from it.

**The least $n$ at which the mode position criterion is met.** A ladder brackets a change of status, and the cell's last block locates one: the least $n$ at which, under *some*, the mode of $\varphi_S^\ast$ lies outside the cell of *all*. The block runs every $n$ from $2$ to $40$, at four lexical strengths with $Z=6$ and at three further half-widths with $\Lambda=512$, under Part D's four base world priors and Text cell 4b's delta-like base world prior, and it marks a least $n$ at which an entry of the lexicon does not hold. At $\Lambda=512$ and $Z=6$ the least $n$ reads $7$, $8$, $7$, $8$ and $10$, and every entry holds at each of them. Under Part D's four base world priors the least $n$ is no larger at a smaller $\Lambda$ ($5$, $7$, $5$, $7$ at $\Lambda=128$) and is larger at a larger $Z$ ($10$, $12$, $10$, $12$ at $Z=7$); under the delta-like base world prior the least $n$ at a smaller $\Lambda$ is larger, and an entry does not hold there. In no setting is the criterion, once met, unmet at a larger $n$ that was run. Below the least $n$ at $\Lambda=512$ and $Z=6$, from $n=3$ to $n=6$, the literal listener's mode meets the criterion under each of Part D's four base world priors and the mode of $\varphi_S^\ast$ does not; under the default base world prior $P(\text{all}\mid\textit{some})$ reads $0.9999$ at $n=4$, $0.4259$ at $n=7$ and $0.0088$ at $n=10$. The cell prints it, and this appendix draws nothing from it.

**Finding the mode by comparing neighbours.** The mode criteria of Text cell 4 read the node where
$\varphi_S^\ast$ is largest, and the block after the half-width ladder reports whether, under *some*, that node can be
found without a comparison across the grid. A node larger than both its neighbours is a local
maximum; a run of equal values counts once, at its first node, as the argmax breaks ties. The block
runs Part D's nine rows (the four base world priors at $\Lambda=8$ and at $\Lambda=512$, and Text cell 4b's delta-like base world prior) along both ladders of this appendix and at three node counts, and Code
Cell 4's plane at every half-width. In every one of those configurations $\varphi_S^\ast$ has a
single local maximum, so the mode is the one node that exceeds its neighbours; $\ell_0$ has a single
local maximum too, so the mode shift criterion, $k^\ast<k_0$, is the sign of $\ell_0$'s slope at
that node; and both neighbour rules return the argmax's answer throughout. The single local maximum is measured and not shown. A control makes the limit concrete: $\mathrm{Beta}(1,\beta)$ base world priors, which the evaluation does not use, put their mode inside the region *some* excludes, $s<1/2n$, once
$\beta$ is large, and at $\Lambda=8$ the $\mathrm{Beta}(1,64)$ and $\mathrm{Beta}(1,256)$ base world priors give two local maxima, while at $\Lambda=512$ each gives one. Whether reading the mode is an operation
the network itself must perform is not taken up here.

**A third axis: the node count.** Text cell 3 §1 calls $K$ an accuracy parameter, and the
specification check reports $\mathbb E[s]$ converging as $K$ is refined. The cell's last block asks
the same of the four criteria, with $Z$ held at $6$ so that only the spacing moves. On the grid the
all-region is the nodes at $\zeta\ge\theta_L$; under the grid's trapezoidal weights (Code Cell 1's `build_grid`) the first
of them carries the interval down to the midpoint with the node below it, and the block prints that
midpoint, the region's lower edge on the grid, against $\theta_L$ at every $K$. Toward finer grids,
Part D's nine rows keep all four statuses at every $K$ from $101$ to $1601$, while
$P(\text{all}\mid\textit{some})$ moves (the delta-like row reads $0.4351$ at $K=101$ and between
$0.3625$ and $0.4248$ on the finer grids). Code Cell 4's plane moves by a few cells: the mass
conjunction holds in $33$, $36$, $36$ and $34$ and the mode conjunction in $13$, $15$, $15$ and $15$
at $K=101$, $201$, $401$ and $801$, seven cells change status in some column, and the mode
conjunction stays inside the mass conjunction at every $K$. The fine grids' own agreement, $K=401$
against $801$, is the floor the coarser grids are read against. Toward coarser grids, agreement with
$K=801$ stays near that floor ($119$ to $121$ of $121$ per column) down to $K=81$ ($117$ to $120$), and at $K=61$ only $33$ of Part D's $36$
statuses agree. Below that the counts do not fall steadily with $K$; they sort by the sign of the
lower-edge column. Where the edge lies below $\theta_L$ the mass conjunction shrinks as the gap grows,
and holds nowhere from $K=21$; where it lies $0.056$ above, the mass conjunction holds in $32$ to $39$
cells at every such rung, and agreement there still falls as the spacing grows. The smooth mask, run
as a control, softens the pattern without removing it. This suggests a lower bound on $K$ without
establishing one, since the ladder does not separate where the nodes fall against $\theta_L$ from
how finely they resolve the field. The cell prints it, and this appendix draws nothing further from
it.

````python
# === Code Cell A: the numbers Appendix A quotes, and those of Text cell 3 Sec. 2. EVALUATION ===
#
# g_y and its profile matrix A are defined in Appendix A and exist nowhere in the
# architecture cell, so the numbers Text cell 3 Sec. 2 quotes about them are printed here
# too.

import math
import torch


def utterance_level_report(net):
    """Text cell 3 Sec. 2 and Appendix A: the clamped fields, the A of Eq. (A2), and theta_L."""
    fields = torch.stack([net.lexical_field(y) for y in UTTERANCES], dim=1)     # X, K x |Y|
    gram = fields.T @ (net.weights[:, None] * fields)                            # X^T W X
    eigenvalues = torch.linalg.eigvalsh(gram)
    # the A for which g_y returns each observed word form exactly: theta_L A W X = I
    profile = torch.linalg.solve(gram, fields.T) / net.theta_L
    identity_gap = float((net.theta_L * profile @ (net.weights[:, None] * fields)
                          - torch.eye(len(UTTERANCES), dtype=net.dtype)).abs().max())
    print("TEXT CELL 3 SEC. 2: THE CLAMPED FIELDS, AND THE A OF Eq. (A2)")
    print(f"  Gram matrix X^T W X of the fields Lambda chi_y: rank "
          f"{int(torch.linalg.matrix_rank(gram))}, condition number "
          f"{float(eigenvalues.max() / eigenvalues.min()):.1f}")
    print(f"  A = (X^T W X)^-1 X^T / theta_L gives theta_L A W X = I to {identity_gap:.1e}")
    print()

    half_width = float(net.zeta[-1])
    print("APPENDIX A: theta_L AND ITS DENOTATION n (Eq. A5)")
    print(f"  n = {net.num_atoms:g}: theta_L = log(2n - 1) = {net.theta_L:.4f}, "
          f"sigma(-theta_L) = {float(torch.sigmoid(torch.tensor(-net.theta_L))):.4f}")
    print(f"  grid half-width Z = {half_width:g}: theta_L < Z requires n < (e^Z + 1)/2 = "
          f"{(math.exp(half_width) + 1.0) / 2.0:.2f}")
    for atoms in (202, 203):
        try:
            net.respawn(num_atoms=atoms, theta_u=1.0)
            outcome = "accepted"
        except ValueError:
            outcome = "rejected"
        print(f"    n = {atoms}: theta_L = {math.log(2 * atoms - 1):.5f}, {outcome} by the constructor")
    print()


def granularity_report(net, atom_ladder=(2, 3, 4, 5, 10, 15, 20, 50, 100, 201),
                       strengths=(8.0, 512.0), priors=None):
    """O10: where the four criteria of Text cell 4 stand at every (n, Lambda, ell_0).

    n = 10 is a STIPULATION (O1), not a neutral choice of units, and this block is what
    makes that consequential rather than cosmetic. It reports the status of ALL FOUR
    criteria at EVERY combination of n, Lambda and ell_0 on the ladder, so that what the
    stipulation settles can be read off the rows themselves. Both lexical strengths the
    evaluation uses are run, and all four base world priors of BASE_WORLD_PRIORS, because the
    probe that raised this measured Lambda = 8 under one base world prior only.

    The four criteria are the evaluation's own (Text cell 4 Parts C and D):

        mass shift     P(all-region | "some") - P(all-region | q_lit) < 0
        mass position  P(all-region | "some") < 1/2
        mode shift     mode(phi_S*) - mode(ell_0) < 0, counted in grid steps
        mode position  mode(phi_S*) outside the cell of "all", i.e. zeta < theta_L

    The mode shift is measured against the mode of ell_0, the control Code Cell 2 and
    Code Cell 4 use, so that the name carries one quantity across the notebook.

    The mass shift is printed in scientific notation: over this ladder it spans ten orders
    of magnitude, and a fixed-decimal column would print values of both signs as 0.0000.
    The magnitude is part of what a row reports, since a status is a threshold test and a
    test on a quantity that has decayed to 1e-06 is not the finding that the same test is
    on a quantity of order 1.

    <mu_u, sum_y c_y> is printed beside theta_u* because it is what that column does.
    Eq. (B2)'s two roots have product -S / sigma_u, a constant, so the maximizer theta_u*
    diverges exactly as this coupling passes through zero: Appendix B's degenerate ray,
    which `theta_u_stationary_points` refuses only at exact zero. Where the coupling is
    small the theta_u* beside it is large for that reason and not for a numerical one.

    Two rows whose theta_L fall between the same two grid nodes are the SAME network but
    for theta_L itself, since chi_y is a threshold on the grid: n is resolved here only as
    finely as the grid resolves theta_L. No two rungs of this ladder collide.

    This cell reports statuses and takes no position on what any of them means
    (Text cell 4, B10).
    """
    if priors is None:
        priors = BASE_WORLD_PRIORS
    upper = lambda p: (p.zeta > p.theta_L).to(p.dtype)
    met = lambda flag: "met" if flag else "no"

    def readings(p):
        w, up = p.weights, upper(p)
        literal = p.read_out(p.literal_fixed_point("some"))
        settled = p.closed_form_fixed_point("some")[0]
        mass = lambda q: float((w * q * up).sum())
        position = mass(p.read_out(settled))
        peak = int(torch.argmax(settled))
        coupling = float(p.dot(p.mu_u, torch.stack(
            [p.project(p.base_log_prior - p.lexical_field(y)) for y in UTTERANCES]).sum(dim=0)))
        return (coupling, float(p.theta_u), position - mass(literal), position,
                peak - int(torch.argmax(p.base_log_prior)), float(p.zeta[peak]))

    print("APPENDIX A: THE FOUR CRITERIA AT EVERY (n, Lambda, ell_0) (Eq. A5; decision O10)")
    print("  n is STIPULATED at 10 in every other cell (O1). theta_L = log(2n - 1) throughout,")
    print("  and every row is the closed-form settled state under \"some\" (Eqs. 15-16).")
    print("    mass shift     P(all-region | \"some\") - P(all-region | q_lit) < 0")
    print("    mass position  P(all-region | \"some\") < 1/2")
    print("    mode shift     mode(phi_S*) - mode(ell_0) < 0, in grid steps")
    print("    mode position  mode(phi_S*) outside the cell of \"all\", zeta < theta_L")
    default = net.respawn(num_atoms=10, lexical_strength=8.0)
    named = net.respawn(num_atoms=10, lexical_strength=8.0, base_prior=priors["gaussian"])
    print(f"  \"gaussian\" below is the constructor's default ell_0, agreeing to "
          f"{float((default.base_log_prior - named.base_log_prior).abs().max()):.1e} in log density.")
    for strength in strengths:
        for name, prior in priors.items():
            print()
            print(f'  "some", Lambda = {strength:g}, ell_0 = {name}'
                  f'{" (the default)" if name == "gaussian" else ""}')
            print(f"    {'n':>4}{'theta_L':>9}{'<mu_u,sum c>':>14}{'theta_u*':>12}"
                  f"{'shift':>12}{'P(all)':>10}{'steps':>7}   {'mass shift':>11}"
                  f"{'mass position':>14}{'mode shift':>12}{'mode position':>15}")
            for atoms in atom_ladder:
                probe = net.respawn(num_atoms=atoms, lexical_strength=strength,
                                    base_prior=prior)
                coupling, theta, shift, position, step, peak = readings(probe)
                print(f"    {atoms:>4}{probe.theta_L:>9.4f}{coupling:>+14.3e}{theta:>12.3f}"
                      f"{shift:>+12.3e}{position:>10.4f}{step:>+7d}   {met(shift < 0.0):>11}"
                      f"{met(position < 0.5):>14}{met(step < 0):>12}"
                      f"{met(peak < probe.theta_L):>15}")
    print()
    print("  SEPARATION AND RANK AGAINST n, which do NOT depend on ell_0 or on Lambda:")
    print(f"    {'n':>5}{'theta_L':>9}{'min kappa separation':>22}{'rank of the three chi_y':>25}")
    best = (0.0, None)
    for atoms in atom_ladder:
        probe = net.respawn(num_atoms=atoms)
        fields = [probe.lexical_field(y) / probe.lexical_strength for y in UTTERANCES]
        kappa = [probe.project(f) for f in fields]
        separation = min(float((kappa[i] - kappa[j]).norm())
                         for i in range(len(kappa)) for j in range(i + 1, len(kappa)))
        stacked = torch.stack(fields, dim=1)
        rank = int(torch.linalg.matrix_rank(stacked))
        if separation > best[0]:
            best = (separation, atoms)
        print(f"    {atoms:>5}{probe.theta_L:>9.4f}{separation:>22.5f}{rank:>25}")
    print(f"    best separated over the ladder: n = {best[1]} at {best[0]:.5f}; the "
          f"implementation's n = 10 is near that maximum.")
    print("    The rank stays 3, so Appendix C's count of thresholds, and with it m = 2,")
    print("    does not move with n.")
    print()
    print("  n = 1 is refused by the constructor, and is the only degenerate case: there")
    print("  theta_L = 0 and the entries for \"some\" and \"all\" coincide.")
    for atoms in (1, 2):
        try:
            net.respawn(num_atoms=atoms, theta_u=1.0)
            outcome = "accepted"
        except ValueError:
            outcome = "rejected"
        print(f"    n = {atoms}: theta_L = {math.log(max(2 * atoms - 1, 1e-300)):.4f}, "
              f"{outcome} by the constructor")
    print()



# Appendix A's cell of "all" is {zeta >= theta_L} cut at +Z, so theta_L fixes its lower
# end and the grid's half-width fixes its upper one. The ladder above varies n and so
# theta_L; this one varies Z. Both are stipulated (O1 for n, I6 for Z), and neither is
# measured, so the appendix reports what each end does to the four criteria and takes no
# position on them (B10).
HALF_WIDTHS = (5.0, 6.0, 7.0, 8.0, 10.0)
NODE_SPACING = 0.12          # the default grid's spacing, 2 * 6.0 / 100


def half_width_report(net, half_widths=HALF_WIDTHS, spacing=NODE_SPACING):
    """Appendix A: the four criteria at every half-width Z, and the plane at each.

    The node count is scaled with Z so that the spacing stays at the default grid's,
    which separates TRUNCATION from RESOLUTION. Code Cell 2's specification check varies
    nodes at fixed half-width and so cannot see this axis; the two ladders together are
    what Text cell 3 Sec. 1 means by K and Z not being parameters of the same kind.
    Every row is the closed-form settled state under "some" (Eqs. 15-16), each
    configuration at its own theta_u* (A9).
    """
    nodes_for = lambda Z: int(round(2 * Z / spacing)) + 1
    roundoff = 1e-12

    def verdicts(probe):
        upper = (probe.zeta >= probe.theta_L).to(probe.dtype)
        mass = lambda field: float((probe.weights * probe.read_out(field) * upper).sum())
        literal = mass(probe.literal_fixed_point("some"))
        phi_S = probe.closed_form_fixed_point("some")[0]
        settled = mass(phi_S)
        node = int(torch.argmax(phi_S))
        prior_node = int(torch.argmax(probe.base_log_prior))
        shift = settled - literal
        return dict(
            literal=literal, settled=settled, shift=shift,
            peak=float(logistic(probe.zeta[node])),
            q_shift=(abs(shift) >= roundoff and shift < 0.0), q_position=settled < 0.5,
            mode_shift=node < prior_node,
            mode_position=float(probe.zeta[node]) < probe.theta_L,
        )

    print("APPENDIX A: THE FOUR CRITERIA AT EVERY HALF-WIDTH Z (Eq. A5's cell, from the other end)")
    print("  Z is STIPULATED at 6.0 in every other cell (I6), as n is at 10 (O1). The cell of")
    print("  \"all\" runs from theta_L to the cut at +Z, so its width in log-odds is Z - theta_L:")
    print("  the ladder above moves its lower end and this one moves its upper end.")
    print(f"  Nodes are scaled with Z to hold the spacing at {spacing}, which separates truncation")
    print("  from resolution. 'mode s' is the mode of phi_S* in s, which is the mode position")
    print("  criterion's own left-hand side; the cell of \"all\" starts at "
          f"s = {float(logistic(torch.tensor(net.theta_L))):.4f}.")
    say = lambda flag: "met" if flag else "no"
    for strength in (8.0, 512.0):
        for prior_name in ("gaussian", "flat"):
            print()
            print(f'  "some", Lambda = {strength:g}, ell_0 = {prior_name}')
            print(f"    {'Z':>6}{'K':>5}{'Z-theta_L':>11}{'theta_u*':>12}{'shift':>12}"
                  f"{'P(all)':>10}{'mode s':>9}{'mass shift':>11}{'mass position':>14}"
                  f"{'mode shift':>12}{'mode position':>15}")
            for half_width in half_widths:
                probe = net.respawn(
                    grid_half_width=half_width, num_nodes=nodes_for(half_width),
                    lexical_strength=strength,
                    base_prior=None if prior_name == "gaussian"
                    else BASE_WORLD_PRIORS[prior_name])
                row = verdicts(probe)
                print(f"    {half_width:>6.1f}{nodes_for(half_width):>5}"
                      f"{half_width - probe.theta_L:>11.4f}{probe.theta_u:>12.3f}"
                      f"{row['shift']:>+12.3e}{row['settled']:>10.4f}{row['peak']:>9.4f}"
                      f"{say(row['q_shift']):>11}{say(row['q_position']):>14}"
                      f"{say(row['mode_shift']):>12}{say(row['mode_position']):>15}")
    print()

    # Code Cell 4's plane at each half-width. The two "carried" columns are its
    # baseline-conditioned counts: the cells where ell_0 does NOT already meet a position
    # criterion and the settled state meets it anyway.
    alphas = [2.0 ** k for k in range(11)]
    lambdas = [2.0 ** k for k in range(1, 12)]
    print("  CODE CELL 4's PLANE AT EACH HALF-WIDTH. Beta(alpha,1), alpha = 2^0..2^10,")
    print(f"  Lambda = 2^1..2^11, {len(alphas) * len(lambdas)} cells, each at its own theta_u*.")
    print(f"    {'Z':>6}{'Z-theta_L':>11}{'mass shift':>12}{'mass position':>15}{'mass both':>11}"
          f"{'mode shift':>12}{'mode position':>15}{'mode both':>11}{'mode not mass':>15}"
          f"{'carried mass':>14}{'carried mode':>14}")
    for half_width in half_widths:
        counts = dict(qs=0, qp=0, qb=0, ms=0, mp=0, mb=0, rev=0, cq=0, cm=0)
        for alpha in alphas:
            prior = beta_world_prior(alpha, 1.0)
            for strength in lambdas:
                probe = net.respawn(grid_half_width=half_width,
                                    num_nodes=nodes_for(half_width),
                                    lexical_strength=strength, base_prior=prior)
                row = verdicts(probe)
                prior_node = int(torch.argmax(probe.base_log_prior))
                counts["qs"] += row["q_shift"]; counts["qp"] += row["q_position"]
                counts["qb"] += row["q_shift"] and row["q_position"]
                counts["ms"] += row["mode_shift"]; counts["mp"] += row["mode_position"]
                counts["mb"] += row["mode_shift"] and row["mode_position"]
                # Sec. 4.1.5 argues from the delta conjunction being a SUBSET of the q
                # conjunction, so the reversals are counted rather than the gap alone
                counts["rev"] += ((row["mode_shift"] and row["mode_position"])
                                  and not (row["q_shift"] and row["q_position"]))
                counts["cq"] += row["literal"] >= 0.5 and row["q_position"]
                counts["cm"] += (float(probe.zeta[prior_node]) >= probe.theta_L
                                 and row["mode_position"])
        print(f"    {half_width:>6.1f}{half_width - net.theta_L:>11.4f}{counts['qs']:>12}"
              f"{counts['qp']:>15}{counts['qb']:>11}{counts['ms']:>12}{counts['mp']:>15}"
              f"{counts['mb']:>11}{counts['rev']:>15}{counts['cq']:>14}{counts['cm']:>14}")
    print("    ('mode not mass' counts the cells where the mode conjunction holds and the mass")
    print("     conjunction does not, so a zero column is the subset relation, cell by cell.")
    print("    (the Z = 6.0 row is Code Cell 4's own, and every column there is the count it")
    print("     prints. 'carried' counts the cells where ell_0 does not already meet the")
    print("     position criterion and the settled state meets it: it is bounded above by the")
    print("     shift count, since a belief that falls below half of what the baseline held")
    print("     above half has fallen.)")
    print()



# Text cell 4's mode criteria read the argmax of phi_S*; this block reports whether, under
# "some", the argmax can be found by comparing neighbours alone. It reads STRONG_LAMBDA and
# delta_like_row from Code Cell 2b, so that Part D's rows are the ones Text cell 4b reports.
def peak_locality_report(net, half_widths=HALF_WIDTHS, spacing=NODE_SPACING,
                         atom_ladder=(2, 3, 4, 5, 10, 15, 20, 50, 100, 201),
                         node_counts=(101, 201, 401)):
    """Where the mode of phi_S* under "some" can be found by comparing neighbours alone.

    The mode criteria read the grid node where phi_S* is largest (Text cell 4). A unit
    that exceeds both its neighbours is a local maximum; where phi_S* has exactly one,
    that unit is the mode and no comparison across the grid is needed. And where ell_0
    has exactly one, the mode shift criterion, k* < k0, is the sign of ell_0's slope at
    the mode: k* < k0 exactly when ell_0 still rises from k* to k* + 1. This block counts
    both, and checks the two neighbour rules against the argmax, over every configuration
    the evaluation reports and over the ladders of this cell. It then runs base world priors the
    evaluation does not use, with their mode inside the region "some" excludes, s < 1/2n.
    """
    nodes_for = lambda Z: int(round(2 * Z / spacing)) + 1

    def local_maxima(field):
        # a run of equal values counts once, at its first node (the argmax's tie-break,
        # I12): on a grid with an even node count a base world prior symmetric about zeta = 0 has
        # two equal central nodes, and a strict test would find no maximum at all
        v, peaks, start = field.tolist(), [], 0
        while start < len(v):
            end = start
            while end + 1 < len(v) and v[end + 1] == v[start]:
                end += 1
            if (start == 0 or v[start - 1] < v[start]) and (end == len(v) - 1 or v[end + 1] < v[end]):
                peaks.append(start)
            start = end + 1
        return peaks

    def inspect(probe):
        phi_S = probe.closed_form_fixed_point("some")[0]
        ell_0 = probe.base_log_prior
        peaks, prior_peaks = local_maxima(phi_S), local_maxima(ell_0)
        node, prior_node = int(torch.argmax(phi_S)), int(torch.argmax(ell_0))
        row = dict(peaks=peaks, one=len(peaks) == 1, prior_one=len(prior_peaks) == 1,
                   node=node, position_rule=None, shift_rule=None)
        if row["one"]:
            k = peaks[0]
            row["position_rule"] = ((float(probe.zeta[k]) < probe.theta_L)
                                    == (float(probe.zeta[node]) < probe.theta_L))
            rises = k + 1 < len(ell_0) and float(ell_0[k + 1]) > float(ell_0[k])
            row["shift_rule"] = rises == (node < prior_node)
        return row

    def tally(label, probes):
        rows = [inspect(p) for p in probes]
        print(f"    {label:<46}{len(rows):>6}{sum(r['one'] for r in rows):>10}"
              f"{sum(r['prior_one'] for r in rows):>10}"
              f"{sum(bool(r['position_rule']) for r in rows):>11}"
              f"{sum(bool(r['shift_rule']) for r in rows):>9}")
        return rows

    rows = {**{f"{name}, Lambda = 8": (prior, {"lexical_strength": 8.0})
               for name, prior in BASE_WORLD_PRIORS.items()},
            **{f"{name}, Lambda = {STRONG_LAMBDA:g}": (prior, {"lexical_strength": STRONG_LAMBDA})
               for name, prior in BASE_WORLD_PRIORS.items()},
            **delta_like_row()}
    configurations = list(rows.values())

    print('APPENDIX A: FINDING THE MODE OF phi_S* UNDER "some" BY NEIGHBOUR COMPARISON')
    print("  A local maximum is a node larger than both neighbours (an end node, than its one;")
    print("  a run of equal values counts once, at its first node, as the argmax breaks ties).")
    print("  'one max' counts configurations where phi_S* has exactly one, 'ell_0 one' the same")
    print("  for the base world prior; the last two columns count where the neighbour rule for each")
    print("  mode criterion returns the argmax's answer (a configuration with more than one")
    print("  local maximum counts as a disagreement).")
    print(f"    {'configurations':<46}{'count':>6}{'one max':>10}{'ell_0 one':>10}"
          f"{'position':>11}{'shift':>9}")
    total = []
    total += tally(f"Part D's {len(rows)} rows, Z ladder ({len(half_widths)} rungs)",
                   [net.respawn(base_prior=p, grid_half_width=Z, num_nodes=nodes_for(Z), **o)
                    for p, o in configurations for Z in half_widths])
    total += tally(f"Part D's {len(rows)} rows, n ladder ({len(atom_ladder)} rungs)",
                   [net.respawn(base_prior=p, num_atoms=n, **o)
                    for p, o in configurations for n in atom_ladder])
    total += tally(f"Part D's {len(rows)} rows, K = {node_counts}",
                   [net.respawn(base_prior=p, num_nodes=K, **o)
                    for p, o in configurations for K in node_counts])
    alphas = [2.0 ** k for k in range(11)]
    lambdas = [2.0 ** k for k in range(1, 12)]
    total += tally(f"Code Cell 4's plane ({len(alphas) * len(lambdas)} cells) at each Z",
                   [net.respawn(base_prior=beta_world_prior(a, 1.0), lexical_strength=l,
                                grid_half_width=Z, num_nodes=nodes_for(Z))
                    for Z in half_widths for a in alphas for l in lambdas])
    print(f"    {'all of the above':<46}{len(total):>6}{sum(r['one'] for r in total):>10}"
          f"{sum(r['prior_one'] for r in total):>10}"
          f"{sum(bool(r['position_rule']) for r in total):>11}"
          f"{sum(bool(r['shift_rule']) for r in total):>9}")
    print()

    cut = float(logistic(torch.tensor(-net.theta_L)))
    print(f"  A CONTROL: PRIORS THE EVALUATION DOES NOT USE, with their mode inside the region \"some\"")
    print(f"  excludes, s < 1/2n = {cut:.4f}: Beta(1, beta), whose mode in zeta is at s = 1/(1 + beta).")
    print(f"    {'p_0':<14}{'Lambda':>8}{'ell_0 mode s':>14}{'local maxima of phi_S* in s':>40}{'argmax s':>10}")
    for beta in (3.0, 16.0, 64.0, 256.0):
        for strength in (8.0, STRONG_LAMBDA):
            probe = net.respawn(base_prior=beta_world_prior(1.0, beta), lexical_strength=strength)
            row = inspect(probe)
            where = ", ".join(f"{float(logistic(probe.zeta[k])):.4f}" for k in row["peaks"])
            print(f"    {f'Beta(1, {beta:g})':<14}{strength:>8g}{1.0 / (1.0 + beta):>14.4f}"
                  f"{where:>40}{float(logistic(probe.zeta[row['node']])):>10.4f}")
    print()




# Text cell 3 Sec. 1 holds Z and K apart: Z truncates the scale and K resolves it. The two
# ladders above vary n and Z; this one varies K at Z = 6.0, so that the spacing is the only
# thing that moves. K is STIPULATED at 101 in every other cell (I6), and the block reports
# what the four criteria do along the ladder and takes no position on it (B10).
FINE_NODE_COUNTS = (101, 201, 401, 801, 1601)
COARSE_NODE_COUNTS = (101, 81, 61, 51, 41, 31, 25, 21, 17, 15, 13, 11, 9, 7, 5)


def node_count_report(net, fine=FINE_NODE_COUNTS, coarse=COARSE_NODE_COUNTS,
                      reference=801, control_sharpness=0.1):
    """Appendix A: the four criteria at every node count K, at the default half-width.

    Every configuration is the closed-form settled state under "some" (Eqs. 15-16), each
    at its own theta_u* (A9). The statistics are the ones Text cell 4 defines, and the
    all-region they integrate over is {zeta >= theta_L} on the grid, so the region's
    lower edge moves with the nodes. The coarse ladder takes odd K only: an even K ties
    the two central nodes of a base world prior symmetric about zeta = 0.
    """
    roundoff = 1e-12
    keys = ("mas", "map", "mab", "mos", "mop", "mob")
    alphas = [2.0 ** k for k in range(11)]
    lambdas = [2.0 ** k for k in range(1, 12)]
    part_d = {**{f"{name}, Lambda = 8": (prior, {"lexical_strength": 8.0})
                 for name, prior in BASE_WORLD_PRIORS.items()},
              **{f"{name}, Lambda = {STRONG_LAMBDA:g}": (prior, {"lexical_strength": STRONG_LAMBDA})
                 for name, prior in BASE_WORLD_PRIORS.items()},
              **delta_like_row()}

    def verdicts(probe):
        upper = (probe.zeta >= probe.theta_L).to(probe.dtype)
        mass = lambda field: float((probe.weights * probe.read_out(field) * upper).sum())
        literal = mass(probe.literal_fixed_point("some"))
        phi_S = probe.closed_form_fixed_point("some")[0]
        settled = mass(phi_S)
        node = int(torch.argmax(phi_S))
        shift = settled - literal
        row = dict(settled=settled,
                   mas=abs(shift) >= roundoff and shift < 0.0, map=settled < 0.5,
                   mos=node < int(torch.argmax(probe.base_log_prior)),
                   mop=float(probe.zeta[node]) < probe.theta_L)
        row["mab"], row["mob"] = row["mas"] and row["map"], row["mos"] and row["mop"]
        return row

    def plane(base, K):
        return {(a, l): verdicts(base.respawn(num_nodes=K, base_prior=beta_world_prior(a, 1.0),
                                              lexical_strength=l))
                for a in alphas for l in lambdas}

    def rows(base, K):
        return {label: verdicts(base.respawn(num_nodes=K, base_prior=p, **o))
                for label, (p, o) in part_d.items()}

    agree = lambda A, B, k: sum(A[c][k] == B[c][k] for c in A)
    agree_d = lambda A, B: sum(A[r][k] == B[r][k] for r in A for k in ("mas", "map", "mos", "mop"))
    say = lambda flag: "met" if flag else "no"
    half_width = float(net.zeta[-1])

    print("APPENDIX A: THE FOUR CRITERIA AT EVERY NODE COUNT K (the third grid axis)")
    print(f"  K is STIPULATED at {len(net.zeta)} in every other cell (I6). Z is held at {half_width:g}, so the")
    print(f"  spacing is 2Z/(K - 1). The all-region is the nodes at zeta >= theta_L = {net.theta_L:.4f};")
    print("  under the grid's trapezoidal weights (build_grid, Code Cell 1) the first of them carries")
    print("  the interval down to the midpoint with the node below it, and that midpoint is the")
    print("  region's lower edge on the grid.")
    print("  Columns mas map mab | mos mop mob are the mass shift, mass position and mass conjunction |")
    print("  the mode shift, mode position and mode conjunction.")
    print()

    references = {K: plane(net, K) for K in (401, reference)}
    floor = {k: agree(references[401], references[reference], k) for k in keys}
    print(f"  THE FLOOR: Code Cell 4's plane at K = 401 against K = {reference}, cells agreeing of "
          f"{len(alphas) * len(lambdas)}")
    print("    " + "  ".join(f"{k} {floor[k]}" for k in keys))
    print()

    # the finer end: Part D's rows, and the plane
    fine_rows = {K: rows(net, K) for K in fine}
    first = fine[0]
    print(f"  THE FINER END, PART D'S {len(part_d)} ROWS: the four statuses at K = {first}, the number of")
    print("  the four that agree with them at each finer K, and P(all | \"some\") at every K.")
    print(f"    {'row':<24}{'mass shift':>11}{'mass position':>14}{'mode shift':>12}{'mode position':>15}"
          + "".join(f"{f'K={K}':>8}" for K in fine[1:]))
    for label in part_d:
        r = fine_rows[first][label]
        same = [sum(fine_rows[K][label][k] == r[k] for k in ("mas", "map", "mos", "mop")) for K in fine[1:]]
        print(f"    {label:<24}{say(r['mas']):>11}{say(r['map']):>14}{say(r['mos']):>12}{say(r['mop']):>15}"
              + "".join(f"{v:>8}" for v in same))
    heading = 'P(all | "some")'
    print(f"    {heading:<24}" + "".join(f"{f'K={K}':>9}" for K in fine))
    for label in part_d:
        print(f"    {label:<24}" + "".join(f"{fine_rows[K][label]['settled']:>9.4f}" for K in fine))
    print()

    planes = {K: (references[K] if K in references else plane(net, K))
              for K in fine if K <= reference}
    print(f"  THE FINER END, CODE CELL 4's PLANE ({len(alphas) * len(lambdas)} cells) at each K.")
    print(f"    {'K':>6}{'mass shift':>12}{'mass position':>15}{'mass both':>11}{'mode shift':>12}"
          f"{'mode position':>15}{'mode both':>11}{'mode not mass':>15}")
    for K, g in planes.items():
        c = {k: sum(v[k] for v in g.values()) for k in keys}
        reversals = sum(v["mob"] and not v["mab"] for v in g.values())
        print(f"    {K:>6}{c['mas']:>12}{c['map']:>15}{c['mab']:>11}{c['mos']:>12}{c['mop']:>15}"
              f"{c['mob']:>11}{reversals:>15}")
    print("    ('mode not mass' as in the half-width plane above: a zero is the subset relation.)")
    moved = {}
    for K, g in planes.items():
        for cell in g:
            for k in keys:
                if g[cell][k] != planes[first][cell][k]:
                    moved.setdefault(cell, set()).add(k)
    print(f"    Cells whose status differs from K = {first} at any finer K, and in which columns:")
    for (a, l), which in sorted(moved.items()):
        print(f"      alpha = {a:g}, Lambda = {l:g}: {' '.join(k for k in keys if k in which)}")
    print()

    # the coarse end, and the same ladder with a smooth exclusion
    def ladder(base, refs, ref_rows, title):
        if title:
            print(title)
        print(f"    {'K':>4}{'spacing':>9}{'nodes in':>10}{'first node':>12}{'lower edge':>12}"
              f"{'mass both':>10}{'mode both':>11}   agree with K = {reference}"
              f"{'':>6}{'Part D':>8}")
        print(f"    {'':>4}{'':>9}{'region':>10}{'- theta_L':>12}{'- theta_L':>12}"
              f"{'':>10}{'':>11}   {'mas map mab | mos mop mob':>26}{'of 36':>9}")
        for K in coarse:
            g, d = plane(base, K), rows(base, K)
            probe = base.respawn(num_nodes=K)
            spacing = float(probe.zeta[1] - probe.zeta[0])
            inside = probe.zeta >= probe.theta_L
            first_node = float(probe.zeta[inside][0])
            a = [agree(g, refs, k) for k in keys]
            print(f"    {K:>4}{spacing:>9.3f}{int(inside.sum()):>10}{first_node - probe.theta_L:>+12.3f}"
                  f"{first_node - spacing / 2 - probe.theta_L:>+12.3f}"
                  f"{sum(v['mab'] for v in g.values()):>10}{sum(v['mob'] for v in g.values()):>11}   "
                  f"{a[0]:>3} {a[1]:>3} {a[2]:>3} | {a[3]:>3} {a[4]:>3} {a[5]:>3}"
                  f"{agree_d(d, ref_rows):>9}")

    ladder(net, references[reference], rows(net, reference),
           f"  THE COARSE END, against K = {reference} (its floor is the first line of this block).")
    print()
    smooth = net.respawn(exclusion_sharpness=control_sharpness)
    smooth_refs = {K: plane(smooth, K) for K in (401, reference)}
    smooth_floor = {k: agree(smooth_refs[401], smooth_refs[reference], k) for k in keys}
    print(f"  A CONTROL: THE SAME LADDER WITH A SMOOTH EXCLUSION, sharpness {control_sharpness:g}, the")
    print("  control of Code Cell 2's Sec. 1 check, against the smooth network at "
          f"K = {reference}.")
    print(f"  Its floor, K = 401 against K = {reference}: "
          + "  ".join(f"{k} {smooth_floor[k]}" for k in keys))
    ladder(smooth, smooth_refs[reference], rows(smooth, reference), "")
    print()


def h1_domain_report(net, alphas=tuple(2.0 ** k for k in range(11)),
                     lambdas=tuple(2.0 ** k for k in range(1, 12))):
    """Sec. 4.1's hypothesis over its domain: on Code Cell 4's Beta(alpha, 1) plane, the cells
    where the entry for "no" holds (leak < 1/2, Eq. 40), and in how many of them the mode of
    phi_S* under "some" lies outside the cell of "all". Then what moves that mode as Z widens.

    The five readings of the second table, at each Z with the spacing held:
      model      the network at its own theta_u*(Z)                      (Eqs. 15-16, B2)
      theta6     theta_u held at its Z = 6 value
      limit      the large-gain limit of Eq. (24), which has no theta_u
      Z6 field   the Z = 6 utility field, same polynomials and coefficients, extended onto
                 the wider grid and added to the tempered control
      Z6 basis   the Z = 6 basis polynomials, not re-orthonormalized, with the limit's
                 projection recomputed over the wider grid (general Gram matrix)."""
    def poly_basis(zeta, weights):
        # Gram-Schmidt of 1, zeta, zeta^2 under W, as code cell 1 builds B, kept as
        # coefficients on (1, zeta, zeta^2) so that it can be evaluated on another grid
        raw = [torch.ones_like(zeta), zeta, zeta ** 2]
        unit = torch.eye(3, dtype=zeta.dtype)
        vectors, coefs = [], []
        for v, c in zip(raw, unit):
            for p, pc in zip(vectors, coefs):
                a = torch.sum(weights * v * p)
                v, c = v - a * p, c - a * pc
            norm = torch.sqrt(torch.sum(weights * v * v))
            vectors.append(v / norm)
            coefs.append(c / norm)
        return coefs[1:]

    def evaluate(coefs, zeta):
        return torch.stack([c[0] + c[1] * zeta + c[2] * zeta ** 2 for c in coefs], dim=1)

    def outside(probe, field):
        return float(probe.zeta[int(torch.argmax(field))]) < probe.theta_L

    def holds(probe):
        belief = probe.read_out(probe.closed_form_fixed_point("no")[0])
        excluded = exclusion_indicator("no", probe.zeta, probe.theta_L)
        return float((probe.weights * belief * excluded).sum()) < 0.5

    grids = [(Z, int(round(2 * Z / NODE_SPACING)) + 1) for Z in HALF_WIDTHS]
    grids += [(6.0, K) for K in (201, 401)]
    print("APPENDIX A: SEC. 4.1's HYPOTHESIS OVER ITS DOMAIN. Code Cell 4's Beta(alpha, 1) plane; the")
    print("  domain is the cells where the entry for \"no\" holds, leak < 1/2 (Eq. 40); 'outside' counts")
    print("  the domain cells whose mode of phi_S* under \"some\" lies outside the cell of \"all\";")
    print("  'literal inside' those where the literal listener's mode lies inside it, and 'moved out'")
    print("  those of these whose settled mode lies outside.")
    print(f"    {'Z':>6}{'K':>5}{'domain':>8}{'outside':>9}{'literal inside':>16}{'moved out':>11}   domain cells not outside")
    for Z, K in grids:
        base = net.respawn(grid_half_width=Z, num_nodes=K)
        dom = out = lit_in = moved = 0
        misses = []
        for a in alphas:
            for lam in lambdas:
                probe = base.respawn(base_prior=beta_world_prior(a, 1.0), lexical_strength=lam)
                if not holds(probe):
                    continue
                dom += 1
                o = outside(probe, probe.closed_form_fixed_point("some")[0])
                li = not outside(probe, probe.literal_fixed_point("some"))
                out += o
                lit_in += li
                moved += (li and o)
                if not o:
                    misses.append((a, lam))
        listed = (", ".join(f"({a:g}, {l:g})" for a, l in misses) if len(misses) <= 3
                  else f"{len(misses)} cells")
        print(f"    {Z:>6.1f}{K:>5}{dom:>8}{out:>9}{lit_in:>16}{moved:>11}   {listed or 'none'}")
    print()

    print("  WHAT MOVES THE MODE AS Z WIDENS, over the same domain at each Z (spacing held): the")
    print("  domain cells whose mode lies outside the cell of \"all\", by reading (see the docstring).")
    print(f"    {'Z':>6}{'domain':>8}{'model':>8}{'theta6':>8}{'limit':>8}{'Z6 field':>10}{'Z6 basis':>10}")
    ref = net.respawn(grid_half_width=6.0, num_nodes=101)
    coefs6 = poly_basis(ref.zeta, ref.weights)
    counts = {Z: [0] * 6 for Z, _ in grids[:len(HALF_WIDTHS)]}
    for a in alphas:
        for lam in lambdas:
            six = ref.respawn(base_prior=beta_world_prior(a, 1.0), lexical_strength=lam)
            theta6 = six.theta_u
            literal6 = six.base_log_prior - six.lexical_field("some")
            k6 = evaluate(coefs6, six.zeta).T @ (six.weights * literal6)
            for Z, K in grids[:len(HALF_WIDTHS)]:
                probe = net.respawn(grid_half_width=Z, num_nodes=K,
                                    base_prior=beta_world_prior(a, 1.0), lexical_strength=lam)
                if not holds(probe):
                    continue
                literal = probe.base_log_prior - probe.lexical_field("some")
                tempered = literal / 2
                held = probe.respawn(theta_u=theta6)
                B = probe.basis
                frozen = evaluate(coefs6, probe.zeta)
                gram = frozen.T @ (probe.weights[:, None] * frozen)
                readings = (
                    probe.closed_form_fixed_point("some")[0],
                    held.closed_form_fixed_point("some")[0],
                    tempered + 0.5 * B @ (B.T @ (probe.weights * literal)),
                    tempered + 0.5 * frozen @ k6,
                    tempered + 0.5 * frozen @ torch.linalg.solve(
                        gram, frozen.T @ (probe.weights * literal)),
                )
                c = counts[Z]
                c[0] += 1
                for i, field in enumerate(readings):
                    c[i + 1] += outside(probe, field)
    for Z, _ in grids[:len(HALF_WIDTHS)]:
        d, *r = counts[Z]
        print(f"    {Z:>6.1f}{d:>8}" + "".join(f"{x:>8}" for x in r[:3])
              + "".join(f"{x:>10}" for x in r[3:]))
    print()


# `granularity_report` runs a ladder in n, and a ladder brackets a change of status. The block
# below runs every n on a range and reports where the mode position criterion is first met
# under "some". It reads delta_like_row from Code Cell 2b, as `peak_locality_report` does.
ATOM_SETTINGS = ((512.0, 6.0), (128.0, 6.0), (32.0, 6.0), (8.0, 6.0),
                 (512.0, 5.0), (512.0, 7.0), (512.0, 8.0))      # (Lambda, Z)


def atoms_threshold_report(net, settings=ATOM_SETTINGS, top=40, detail_top=12,
                           spacing=NODE_SPACING):
    """The least n at which the mode position criterion of Text cell 4 is met under "some".

    Every n from 2 to `top` that the grid admits (theta_L < Z) is run, at each setting
    (Lambda, Z) and under Part D's four base world priors and Text cell 4b's delta-like one. A
    setting's entry is the least n at which the mode of phi_S* lies outside the cell of "all",
    zeta < theta_L. The half-width is varied at the default grid's node spacing, as in
    `half_width_report`, so that truncation moves and resolution does not.

    An entry of the lexicon holds where its leak is below 1/2 (Eq. 40). '*' marks a least n at
    which the largest leak over the three utterances is not below 1/2.

    n is STIPULATED at 10 in every other cell (O1), and n is resolved here only as finely as
    the grid resolves theta_L (see `granularity_report`).

    This cell reports statuses and takes no position on what any of them means
    (Text cell 4, B10).
    """
    priors = {**BASE_WORLD_PRIORS,
              **{name: prior for name, (prior, _) in delta_like_row().items()}}
    met = lambda flag: "met" if flag else "no"
    to_s = lambda value: 1.0 / (1.0 + math.exp(-value))

    def probe_at(atoms, strength, Z, prior):
        return net.respawn(num_atoms=atoms, lexical_strength=strength, base_prior=prior,
                           grid_half_width=Z, num_nodes=int(round(2 * Z / spacing)) + 1)

    def outside(p, field):
        return float(p.zeta[int(torch.argmax(field))]) < p.theta_L

    def holds(p):
        leaks = []
        for y in UTTERANCES:
            belief = p.read_out(p.closed_form_fixed_point(y)[0])
            excluded = exclusion_indicator(y, p.zeta, p.theta_L)
            leaks.append(float((p.weights * belief * excluded).sum()))
        return max(leaks) < 0.5

    def least(strength, Z, prior):
        admitted = [a for a in range(2, top + 1) if math.log(2 * a - 1) < Z]
        status = []
        for atoms in admitted:
            p = probe_at(atoms, strength, Z, prior)
            status.append(outside(p, p.closed_form_fixed_point("some")[0]))
        if not any(status):
            return None, True, False
        first = status.index(True)
        return (admitted[first], holds(probe_at(admitted[first], strength, Z, prior)),
                not all(status[first:]))

    print("APPENDIX A: THE LEAST n AT WHICH THE MODE POSITION CRITERION IS MET UNDER \"some\" (Eq. A5)")
    print(f"  n is STIPULATED at 10 in every other cell (O1). Every n from 2 to {top} that the grid admits")
    print("  (theta_L < Z) is run. An entry is the least n at which the mode of phi_S* lies outside the")
    print("  cell of \"all\", zeta < theta_L. '*' marks a least n at which an entry of the lexicon does")
    print("  not hold (largest leak over the three utterances not below 1/2, Eq. 40). K keeps the")
    print("  default grid's node spacing.")
    print(f"    {'Lambda':>6}{'Z':>6}{'K':>5}" + "".join(f"{name:>13}" for name in priors))
    unmet_again = 0
    for strength, Z in settings:
        cells = []
        for prior in priors.values():
            atoms, held, again = least(strength, Z, prior)
            unmet_again += again
            cells.append(f"{'none' if atoms is None else atoms}{'' if held else '*'}")
        print(f"    {strength:>6g}{Z:>6.1f}{int(round(2 * Z / spacing)) + 1:>5}"
              + "".join(f"{cell:>13}" for cell in cells))
    print(f"  entries at which the criterion, once met, is unmet at a larger n up to {top}: "
          f"{unmet_again} of {len(settings) * len(priors)}")
    print()

    strength, Z = settings[0]
    print(f"  AROUND THE LEAST n, at Lambda = {strength:g}, Z = {Z:g}: the mode position criterion for the")
    print("  mode of phi_S* under \"some\", and in brackets the same for the literal listener's mode.")
    print(f"    {'n':>4}{'theta_L':>9}" + "".join(f"{name:>13}" for name in priors))
    for atoms in range(2, detail_top + 1):
        cells = []
        for prior in priors.values():
            p = probe_at(atoms, strength, Z, prior)
            settled = outside(p, p.closed_form_fixed_point("some")[0])
            literal = outside(p, p.literal_fixed_point("some"))
            cells.append(f"{met(settled)} [{met(literal)}]")
        print(f"    {atoms:>4}{p.theta_L:>9.4f}" + "".join(f"{cell:>13}" for cell in cells))
    print()

    print(f"  THE DEFAULT BASE WORLD PRIOR at Lambda = {strength:g}, Z = {Z:g}: the mode of phi_S* and the")
    print("  lower end of the cell of \"all\", both in s, and the q-mass on the cell.")
    print(f"    {'n':>4}{'theta_L':>9}{'mode (s)':>10}{'cell from (s)':>15}{'P(all | some)':>15}{'theta_u*':>11}")
    for atoms in range(2, detail_top + 1):
        p = probe_at(atoms, strength, Z, priors["gaussian"])
        settled = p.closed_form_fixed_point("some")[0]
        upper = (p.zeta > p.theta_L).to(p.dtype)
        mass = float((p.weights * p.read_out(settled) * upper).sum())
        print(f"    {atoms:>4}{p.theta_L:>9.4f}{to_s(float(p.zeta[int(torch.argmax(settled))])):>10.4f}"
              f"{to_s(p.theta_L):>15.4f}{mass:>15.4f}{float(p.theta_u):>11.1f}")
    print()


utterance_level_report(evaluation_network)
granularity_report(evaluation_network)
half_width_report(evaluation_network)
peak_locality_report(evaluation_network)
node_count_report(evaluation_network)
h1_domain_report(evaluation_network)
atoms_threshold_report(evaluation_network)
````

## <a id="appb" name="appb"></a>Appendix B: the stationary point of $\theta_u$, its locality, and alternative spaces

**The premise §8.5 rests on is $\mu_u\neq0$.** The terminating
contribution to $g_S$ is $\theta_u B\mu_u$. At $\mu_u=0$, the contribution vanishes for every $\theta_u$. Thus, varying
$\theta_u$ cannot move the terminating prediction at all, the chain stops rather than terminates, whereby
the only remaining use for $\theta_u$ is to shift magnitude out of the priced factor $\varphi_u$ into
the unpriced $\theta_u$ along the exact degeneracy
$(\theta_u,\varphi_u)\mapsto(c\,\theta_u,\varphi_u/c)$, which leaves $g_S$ invariant.
$\tilde{\mathcal F}$ then has **no maximizer**. Its one stationary point is $\theta_u=0$, where it is
smallest, and from there it climbs monotonically in $\lvert\theta_u\rvert$ to a finite asymptote. The
flow of Eq. (20), started at $\theta_u(0)=0$, meets a zero gradient and does not move; started
anywhere else, it sends $\lvert\theta_u\rvert\to\infty$.

**The reduced objective.** Eliminating $x$ analytically (legitimate by §8.6) collapses
$\tilde{\mathcal F}$ to a ratio of quadratics in the single scalar $\theta_u$:

$$\tilde{\mathcal F}(\theta_u)=\mathrm{const}-\frac{\lVert\theta_u\mu_u-c\rVert^{2}}
{2\big(\sigma_u\theta_u^{2}+S\big)},\qquad S=\sigma_L+\sigma_S,\qquad
c=B^{\mathsf T}W\big(\ell_0-\varphi_L\big),
\tag{B1}$$

$\mathrm{const}$ collecting what $\theta_u$ cannot reach. Stationarity is therefore the quadratic

$$\sigma_u\langle\mu_u,c\rangle\,\theta_u^{2}
+\big(S\lVert\mu_u\rVert^{2}-\sigma_u\lVert c\rVert^{2}\big)\,\theta_u
-S\langle\mu_u,c\rangle=0,
\tag{B2}$$

That is the stationarity equation for a single utterance. Here $c_y=B^{\mathsf T}W(\ell_0-\varphi_{L,y})$ is that utterance's target
projected onto the utility basis, one vector in $\mathbb R^m$ per entry, and
$\sum_y$ denotes a
**sum over the three utterances**, not the variance $\sigma_y$ of §3:

| | $c_y$ | $\langle\mu_u,c_y\rangle$ |
|---|---|---|
| *no* | $[-9.11909,\;-12.83178]$ | $-21.9509$ |
| *some* | $[+9.11909,\;-24.36744]$ | $-15.2484$ |
| *all* | $[+9.11909,\;-12.83178]$ | $-3.7127$ |
| **sum** | $[+9.11909,\;-50.03100]$ | $-40.9119$ |

For the flow of Eq. (20), which visits all three, the same quadratic governs with
$\lVert\mu_u\rVert^{2}\mapsto3\lVert\mu_u\rVert^{2}$, $c\mapsto\sum_y c_y$ and
$\lVert c\rVert^{2}\mapsto\sum_y\lVert c_y\rVert^{2}$. The three utterances share the denominator
$\sigma_u\theta_u^{2}+S$, so only their numerators combine:

$$\sum_y\lVert\theta_u\mu_u-c_y\rVert^{2}
=3\theta_u^{2}\lVert\mu_u\rVert^{2}
-2\theta_u\Big\langle\mu_u,\sum_y c_y\Big\rangle
+\sum_y\lVert c_y\rVert^{2}.
\tag{B3}$$

The individual $c_y$ reach $\theta_u$ only through the term linear in it, that is, only through their sum.

**What $\theta_u^\ast$ is a function of.** Everything in Eq. (B2) is fixed by the configuration, so
$\theta_u^\ast$ is too: $\ell_0$ and $\Lambda$ through each $c_y$, $\mu_u$ through both the linear
term and $\lVert\mu_u\rVert^2$, $\theta_L$ (hence $n$) through $\chi_y$, the grid $K$ and its
half-width through the quadrature, the mask sharpness through $\chi_y$ again, $B$ and $m$ through
the projection, every $\sigma$ through $S$ and $\sigma_u$, and the exposure ensemble through which
$c_y$ enter the sum and with what weights. A configuration is therefore not a network with a
parameter set for it but a network with a parameter *implied* by it, and two configurations that
differ in any of these carry two different $\theta_u^\ast$. This is why every evaluation re-learns
it wherever it respawns, and why the value ranges from $-557.67$ to $+12228.42$ across the plane of
Text cell 6 and reaches $+5950.63$ under a flat base world prior at $\Lambda=8$. Those extremes are not
samples from a continuum: they are neighbourhoods of the ray named below, where $\theta_u^\ast$
diverges and changes sign.

**The standing requirement, stated exactly.** §3 requires $\mu_u\neq0$. What Eq. (B2) actually needs is
$\langle\mu_u,\sum_y c_y\rangle\neq0$, which is strictly stronger: $\mu_u=0$ violates it
automatically, but so would any nonzero $\mu_u$ orthogonal to $\sum_y c_y$. There the leading and
constant coefficients of Eq. (B2) vanish together, so one root reaches $\theta_u=0$ and the other
diverges, which is the ray taken up below. What $\theta_u=0$ then is depends on the linear
coefficient: a maximizer of $\tilde{\mathcal F}$ when $3S\lVert\mu_u\rVert^2>\sigma_u\sum_y\lVert c_y\rVert^2$,
a minimizer when the inequality is reversed, and one point of a constant $\tilde{\mathcal F}$ at
equality. In each case the gradient vanishes at the shared start, so the flow does not leave it. At $\mu_u=\mathbf 1$ the quantity is $-40.9119$ under the default base world prior, and how near the ray the other configurations come is taken up below. The weights in that sum are moreover a **prior over utterances**: the flow averages
the three gradients equally, so the operative quantity is $3\,\mathbb E_{p(y)}[c_y]$ at uniform
$p(y)$, and a non-uniform exposure moves $\theta_u^\ast$ accordingly.

**That uniformity is a stipulation of this phase, and is labelled as one.** It is the same move as
fixing every $\sigma$ at $1$ (Text cell 3, preamble item 3), and for the same reason: a quantity
this phase does not model is set to the value that treats every channel alike (there every connection,
here every utterance), and that value is the one which makes the quantity vanish from
the formula, leaving the bare $\sum_y c_y$ that Eq. (B3) already writes. Nothing here measures how
far a non-uniform $p(y)$ would move $\theta_u^\ast$; the claim is only that it would.

**What the ensemble contains is stipulated too**, and the two stipulations belong together. This
phase takes the inventory to hold at least one entry $\chi$ together with $\ker\chi$ (Appendix D §1).
The evaluation's ensemble meets it with *some* and *no*, each the other's kernel; *all* is a further
member, whose kernel, the O corner of Appendix C §4, is not in the ensemble (Appendix D §1). Membership does not follow from what a participant was shown: presence in the inventory
follows from exposure somewhere, not from exposure in a particular experiment, and the converse
does not hold. Under the alternatives level the accompanying paper proposes (§5.2.2), each level holds a pair and only a pair of
entries, each the kernel of the other, so the pair would be a **consequence** rather than a stipulation; at this phase it is a stipulation and is written as
one.

The roots of Eq. (B2) have product $-S/\sigma_u<0$, so whenever $\langle\mu_u,c\rangle\neq0$
there is **exactly one root of each sign**: the maximizer is the one with
$\operatorname{sign}\theta_u^\ast=\operatorname{sign}\langle\mu_u,c\rangle$ and the other is a
minimum, which ascent cannot cross. At $\mu_u=[1,1]$ this is a genuine interior maximum at
$\theta_u^\ast=-28.4375$, with $\partial\tilde{\mathcal F}/\partial\theta_u=3.3\times10^{-16}$ there
and a margin of $+0.2398$ over the asymptote.

**How near that ray this notebook's own rows come.** The $-40.9119$ above is the *default* base world prior's margin, not a bound on the others. $c_y=B^{\mathsf T}W(\ell_0-\varphi_{L,y})$ depends on $\ell_0$
and, through $\varphi_{L,y}$, on $\theta_L$, so the coupling moves with the base world prior and with $n$
while $\mu_u$ stays at its stipulated value. Under the **flat** base world prior at $\Lambda=8$ and $n=10$ (a row
Code Cell 2 prints) it is $+0.0594$, three orders of magnitude smaller and of the opposite
sign, and varying $n$ alone carries it through zero between $n=10$ and $n=11$, taking
$\theta_u^\ast$ from $+5950.63$ to $-3322.90$. **The ray is reachable without touching $\mu_u$**,
which the requirement as stated above, being about $\mu_u$'s orientation, does not suggest. The
$-557.67$ to $+12228.42$ quoted earlier are neighbourhoods of it rather than points of a continuum.

**And it is harmless, for a reason worth stating precisely.** *Exactly* on the ray there is no
maximizer away from $\theta_u=0$ and `theta_u_stationary_points` refuses it. *Near* it $\theta_u^\ast$ is merely
large: by the constant product just given, a maximizer in the thousands means the other root lies
within $10^{-3}$ of zero and the maximizer is about to change sign through infinity. But
$\varphi_S^\ast$ **converges to a single limit as $\lvert\theta_u\rvert\to\infty$, approached from
both signs** (the branches agree to $2.290\times10^{-6}$ at $\lvert\theta_u\rvert=10^{6}$, a gap
falling like $1/\lvert\theta_u\rvert$), so the readings cross the ray continuously. Across the sign
change both q readings for *some* move monotonically and neither criterion changes status.
**The ray is a pole of $\theta_u^\ast$ and not of anything the evaluation reads off**, and that,
rather than a margin measured at one configuration, is what makes it harmless here. Code Cell B
prints all three parts.

**Where the flow is started, and why at zero.** Eq. (20) is integrated from $\theta_u(0)=0$ in every
configuration. Zero is not a neutral default but the one point the root structure makes safe: since
the two roots have opposite signs, zero always lies *between* the minimum and the maximizer, on the
maximizer's rising side, so ascent from it moves toward $\theta_u^\ast$ whatever that sign turns
out to be, and no configuration needs a start chosen for it. The gradient there is exactly
$\langle\mu_u,\sum_y c_y\rangle\big/\big(2\lvert\mathcal Y\rvert\big)$ at
$\sigma_L=\sigma_S$, since $\varphi_u^\ast(0)=\mu_u$ and
$\varepsilon_S(0)=(\ell_0-\varphi_L)/2$ contracts against $b$ to $c_y/2$; at $\mu_u=[1,1]$ that is
$-40.9119/6=-6.8187$, which is the value the learning probe prints at its first update. It is
nonzero by exactly the standing requirement above, so the flow leaves zero at once. A fixed start of $+1$, the value a
constructor default would supply, has no such guarantee: it sits on the far side of the minimum
wherever $\theta_u^\ast<0$, which is the default configuration among many others, and ascent from
it then runs away to $-\infty$ rather than to the maximizer.

At $\theta_u=0$ the network is the **tempered control** of Text cell 4 Part C,
$\varphi_S^\ast=(\ell_0-\varphi_L)/2$, and not the literal listener $q_{\mathrm{lit}}$, which
requires $\sigma_S\to\infty$ as well. The flow therefore begins at a tempered version of the
control and moves away from it; it does not begin at the baseline the mass shift criterion is measured against.

**What the flow shows, and what the closed form supplies.** The flow does not run for a fixed
number of updates and stop; it **halts when its own update falls below a tolerance** (§8.5), and
$\tilde{\mathcal F}$ is shallow near its maximum, so it halts far short of $\theta_u^\ast$: from
zero at $\tau_\theta=20$, a tolerance of $0.1$ halts after $15$ updates at $\theta_u=-5.285227$,
with the gradient still at $-1.997$ and $\tilde{\mathcal F}$ rising monotonically throughout. What
the integration establishes is therefore the direction of the flow, the monotone rise, and a
shrinking gradient.

The value of $\theta_u^\ast$ is supplied by Eq. (B2), which is exact, and cross-checked against an
independent bisection on the reduced gradient (relative gap $4.6\times10^{-14}$). The division of
labour is not one of convenience: $\theta_u^\ast$ is the **asymptote** of this flow and carries no
tolerance, which is why it is what the notebook reports, whereas a $\theta_u$ taken *from* the flow
is **realizable** and carries on its face the tolerance it halted at, a quantity this study does
not fix. No result here depends on the flow arriving: Text cell 4 shows that the mass criteria it
reports are met at a $\lvert\theta_u\rvert$ of a few units, which the flow passes within its first
updates and well below where it halts.

**The sign is not free to assume.** $\theta_u$ is an unconstrained real parameter: Bogacz's Eqs. (25)
and (56) supply a gradient and no domain, and his own $\theta^\ast=u/v_p$ takes whichever sign the
data give. Because $\ell_0$ sits in $g_L$ (Eq. 9), $c_y=B^{\mathsf T}W(\ell_0-\varphi_{L,y})$ carries $\ell_0$
with a positive sign against $\varphi_L$ with a negative one, which puts
$\sum_y\langle\mu_u,c_y\rangle$ at $-40.912$ and the maximizer at $-28.437$. The sign is a
consequence of that placement, not a choice: were both signs reversed the maximizer would be
$+28.437$. A search bracketed on $\theta_u>0$ reports, in consequence, a
divergence that is an artefact of the bracket and not a property of the model.

This is **not** a departure from Bogacz. His Eq. (25) model diverges identically at $v_p=0$: with his
own Exercise 1 numbers ($u=2$, $\Sigma=1$) the optimum is $\theta^\ast=u/v_p$, so $v_p=3$ gives
$0.667$, $v_p=1$ gives $2.0$, and $v_p=0$ gives none. **$v_p\neq0$ is an unstated premise of his worked example's finite
optimum**, which takes $v_p=3$; his rule, Eq. (25), remains defined at $v_p=0$. §3's $\mu_u\neq0$ is
the same premise stated, and it is a stipulation of this implementation, not a requirement of the
construction.

**What pins $\theta_u$ here is not what pins $\theta$ there, and the difference runs in our favour.**
In Bogacz's scalar model $\theta v_p=u$ is solvable for any $u$ once $v_p\neq0$, so at $\theta^\ast$
*both residuals vanish* and $F=0$: it is **exact solvability** that identifies $\theta$. Our $\theta_u$ is
one scalar facing $c\in\mathbb R^m$, so $\theta_u\mu_u=c$ would need $c\parallel\mu_u$, a coincidence
of measure zero, and here $\cos(\mu_u,c)=-0.986,\,-0.414,\,-0.167$ for the three utterances. The
residuals stay far from zero at the optimum ($\lVert r_S\rVert=22.5,\,15.8,\,22.5$, and
$\tilde{\mathcal F}=-420.209$ rather than $0$). Promoting $\theta_u$ to the matrix of §8.4 would not
repair this: one $\theta_u$ is shared across utterances, so zeroing all three at once would require
$c_{\textit{no}}=c_{\textit{some}}=c_{\textit{all}}$, and those differ by up to $21.58$.

$\theta_u$ is identified regardless, because Eq. (B1) is a ratio of quadratics and not a residual to
be zeroed: it trades fit against the cost of the terminating prior, $\lVert\varphi_u-\mu_u\rVert^{2}$, and that trade has
an interior optimum whenever $\langle\mu_u,c\rangle\neq0$. The argument therefore has a **weaker premise and a wider reach** than Bogacz's, since it never asks the generative model to be able to fit the
situation level at all. What $\mu_u\neq0$ buys is not exact solvability but a nonzero
$\langle\mu_u,c\rangle$: a cost of the terminating prior that $\theta_u$ has something to trade against.

We therefore claim **global** convergence for $(\varphi_S,\varphi_u)$ and, for $\theta_u$, convergence
of $\tilde{\mathcal F}$ together with convergence to a stationary point that is only guaranteed
**local**, since $\tilde{\mathcal F}$ is not concave. The division matches Bogacz's in form only. He
argues that $\varphi$ converges to values maximizing $F$ once the error nodes are faster (his §6),
but makes no concavity claim, and his running example, $g(v)=v^2$, would not support one; for
parameters he gives the values at which the expected change vanishes, since his are updated after
every observation and never converge (his §2.4).

### <a id="appb-locality" name="appb-locality"></a>Locality

Bogacz states two constraints in his Introduction:

> **Local computation.** A neuron performs computations only on the basis of the activity of its input
> neurons and synaptic weights associated with these inputs (rather than information encoded in other
> parts of the circuit).
>
> **Local plasticity.** Synaptic plasticity is only based on the activity of pre-synaptic and
> post-synaptic neurons.

**The sum over $y$ is not a locality matter.** No equation the network integrates contains
$\sum_y c_y$. Eq. (20) is evaluated at one presented utterance, after the fast subsystem has
equilibrated, from the two local factors $\varphi_u$ and $\langle\varepsilon_S,b\rangle$; the sum is
an accumulation across successive presentations, and appears only in the closed-form analysis of
where the accumulated flow settles. The implementation batches the three for numerical convenience,
but the batching is not load-bearing: updating once per presentation instead, one local update of a third the size
three times per cycle, reaches the same fixed point to within $3\times10^{-4}$: the two trajectories
differ by $2.7\times10^{-3}$ after $1{,}000$ cycles ($\theta_u=-14.48$) and by $1.8\times10^{-4}$ after
$10{,}000$ ($\theta_u=-22.46$), and after $400{,}000$ the per-presentation flow sits $2.8\times10^{-4}$ from
$\theta_u^\ast$. What the sum carries is an assumption about **exposure**, not about circuitry: the
ensemble must be visited within one $\tau_\theta$, at the frequencies $p(y)$ that weight it.

This is where the parallel with $\sigma$ stops being exact, and the difference is worth stating
because it marks where a later phase would be felt. $\sigma$ is a variable of the implementation
with a default of $1$, so a precision-bearing phase changes defaults. $p(y)$ is **not a variable at
all** (it is implicit in the batched sum), so a frequency-bearing phase would have to add a weight
vector, and this per-presentation check, the one place the ensemble is visited an utterance at a
time, is where that addition would show.

**Eq. (20) itself is the real question, and it has nothing to do with alternatives.** The forward
weights from the $u$ level to the $S$ level are $M=\theta_u B\in\mathbb R^{K\times m}$. Were $M$ free, as
Bogacz's $\Theta$ of Eq. (56) is, each entry would update as
$\dot M_{kj}\propto\varepsilon_{S,k}\varphi_{u,j}$: one error neuron times one state neuron, Hebbian,
and local without further comment. We have instead tied all $K\times m$ entries to a single scalar
times a fixed $B$, so moving $\theta_u$ requires pooling the local products across every one of them.
That is **weight sharing**: the update of a single parameter reads products formed at synapses on
several different neurons, which is what the constraint of local plasticity quoted above forbids.
The ranking therefore inverts against expectation: **Bogacz's full matrix is local, and it
is our scalar restriction that needs defending.**

**The commitment: $b$ is a single projection.** We take $b$ to be the fixed spatial profile of one connection, one projection whose terminal field on the $S$ population
has shape $b$ and which pools error back through that same fixed profile, and *not* $K$ independently modifiable synapses sharing a gain. $\theta_u$ is the gain of that
projection as a whole, applied at the $\varphi_u$ unit: the unit's output $\theta_u\varphi_u$ is
distributed through $b$, and its ascending input is pooled through the same $b$ into
$\langle\varepsilon_S,b\rangle$ before the gain is applied, which is the order Eq. (19) writes. The
plastic quantity is therefore a gain of the projection rather than a synaptic weight. That extends
Bogacz's statement of local plasticity, made for synapses, and is a commitment of this phase.
Unit-wide multiplicative factors are established, as in synaptic scaling (Turrigiano et al., 1998) and
gain modulation (Salinas & Thier, 2000), but those are driven by activity level or modulatory input,
not by the product Eq. (20) requires. Under that reading $\langle\varepsilon_S,b\rangle$ is a single signal physically present at a
single synapse, $\varphi_u$ is that synapse's presynaptic activity, and Eq. (20),
$\tau_\theta\dot\theta_u=\varphi_u\langle\varepsilon_S,b\rangle$, is a product of pre- and
post-synaptic activity: Hebbian, and local in Bogacz's sense.

Nothing extra is assumed for plasticity that inference did not already require. Eq. (19) reads
$\tau_\varphi\dot\varphi_u=-\varepsilon_u+\theta_u\langle\varepsilon_S,b\rangle$, so the $\varphi_u$
neuron already receives the pooled $\langle\varepsilon_S,b\rangle$ as dendritic input: **local
computation** already obliges that signal to exist as a physical quantity, and local plasticity then
gets it for free. Bogacz reasons the same way about his own $\theta$, which likewise sits on two
connections at once:

> "the rule also satisfies the constraint of local plasticity … as $\phi$ fully determines $h(\phi)$,
> so the change in weight is fully determined by the activity of pre-synaptic and post-synaptic
> neurons."

Our $b$ extends that argument from two connections to $K$, on the same grounds.

**The price.** This is what forecloses the matrix case of §8.4 as a *physical* generalization: the
scalar is what makes the projection single, and the single projection is what makes Eq. (20) local.

**A free matrix is local, but has no maximizer.** Were the $u\to S$ weights a free
$M\in\mathbb R^{K\times m}$, $\varphi_u$ would enter $\mathcal F$ only through the prediction
$M\varphi_u$ and the cost of the terminating prior, $\lVert\varphi_u-\mu_u\rVert^{2}$. For any invertible
$C\in\mathbb R^{m\times m}$ with $C\mu_u=\mu_u$, the substitution
$(M,\varphi_u)\mapsto(MC^{-1},C\varphi_u)$, applied for every utterance at once, leaves every
prediction unchanged and turns the cost of the terminating prior into $\lVert C(\varphi_u-\mu_u)\rVert^{2}$. At $m>1$,
$\mu_u$ fixes one direction of $\mathbb R^m$ and no more. Take $C$ to be the identity on $\mu_u$ and
$c\,I$ on its orthogonal complement: as $c\to0$ the cost of the terminating prior falls wherever a settled $\varphi_u$
departs from $\mu_u$ off its ray, so $\tilde{\mathcal F}$ rises strictly, while the columns of
$MC^{-1}$ off that ray grow as $1/c$. $\tilde{\mathcal F}$ therefore has no maximizer in $M$ unless
every settled $\varphi_u$ lies on $\mu_u$'s ray, and learning drives $M$ off to infinity. The scalar
$\theta_u$ admits only $C=cI$, which fixes $\mu_u$ only at $c=1$, so the degeneracy does not arise:
$\mu_u$ anchors the one direction the scalar has. This is the second reason for the scalar, beside the
single projection: it gives $\theta_u^\ast$ a value the configuration fixes (Eq. B2).

**What fails at $m>1$.** At $m=1$, Eq. (20) reads $\varphi_u\langle\varepsilon_S,b\rangle$: one
presynaptic activity, one pooled error signal, one synapse. At $m>1$ the same gradient is

$$\frac{\partial\mathcal F}{\partial\theta_u}
=\sum_{j=1}^{m}\varphi_{u,j}\big\langle\varepsilon_S,b_j\big\rangle.
\tag{B4}$$

Every summand is itself a legitimate Hebbian product, formed where column $j$ meets the error field.
**The violation is the sum.** $\theta_u$ is one scalar gaining all $m$ columns, so moving it requires
the whole of Eq. (B4), and the $j$-th synapse holds $\varphi_{u,j}$ and
$\langle\varepsilon_S,b_j\rangle$ and nothing else: the remaining $m-1$ terms are formed at synapses
on other neurons. A synapse cannot read them, which is what the constraint of local plasticity
forbids.

The single-projection commitment does not reach this case. It licenses pooling across the $K$
terminals of *one* projection, and it does so because Eq. (19) already obliges
$\langle\varepsilon_S,b\rangle$ to exist as one signal at one site. Nothing obliges the sum over $j$
to exist anywhere, so at $m>1$ the pooling is across $m$ sites and the argument gives out. At $m=1$
the sum has one term and the question does not arise, which is why Eq. (20) is stated there.

**A relay does not repair it.** The natural repair is Bogacz's move in his §5: put the quantity a
rule needs into one neuron's activity. Here that means a descending relay of $K$ units,
$r=B\varphi_u$, each forming $\sum_jB_{kj}\varphi_{u,j}$ in its own dendrite, with $\theta_u$ moved
onto the relay's projection to $\varepsilon_S$. The update then reads $\langle\varepsilon_S,r\rangle$.
But $\langle\varepsilon_S,r\rangle=\sum_j\varphi_{u,j}\langle\varepsilon_S,b_j\rangle$, which is
Eq. (B4) again, and no unit receives it: the ascending path delivers $\langle\varepsilon_S,b_j\rangle$
to each $\varphi_{u,j}$ separately. The relay's projection runs from $K$ units to $K$ units, so a gain
shared across it is shared across synapses on different neurons, which is the weight sharing defined
above. And the sum the relay relocates, the forward prediction, was already local computation, since
each error unit sums its own afferents. A relay therefore makes nothing local that was not local
already. The relay was implemented and is retired; it moves no number.

**The departure, stated.** At the $m=2$ this implementation runs, the update of $\theta_u$ is not
local, and this phase does not repair it. What would remove it is one opposition per level, with one
utility dimension, $m=1$, which the accompanying paper proposes as its alternatives level (§5.2.2);
Appendix C shows why $m=1$
cannot serve the present three-member inventory at one level.

### <a id="appb-alternative-spaces" name="appb-alternative-spaces"></a>Alternative spaces

§3 fixes $E_y$ by the utterance entry alone, with no reference to an alternatives space. Since
$\theta_u^\ast$ depends on $\sum_y c_y$, the two appear to be in conflict. They are not, but the reason commits the model to a position, so it is stated rather than assumed.

**At inference there is no conflict whatever.** With $\theta_u$ held fixed, Eqs. (15)–(16) at
utterance $y$ involve $c_y$ and nothing else: reconstructing $\varphi_u^\ast$ from $c_y$ alone
reproduces the full fixed point exactly, to $0$ at double precision. One utterance in, one belief out; no
other entry is consulted, and §3's claim holds exactly.

**At learning, $\theta_u$ does see the utterance set**, through exposure and substantially:

| exposure | $\theta_u^\ast$ |
|---|---|
| *no* | $-11.2844$ |
| *some* | $-44.1766$ |
| *all* | $-65.7004$ |
| *no*, *some* | $-24.7250$ |
| *no*, *all* | $-19.1053$ |
| *some*, *all* | $-48.3901$ |
| all three | $-28.4375$ |

A factor of nearly six separates the single-utterance optima.

**But this is not an alternatives set in the sense the neo-Gricean literature uses.** Grice (1975, pp. 45–46)
states maxims and requires of an implicature only that it be capable of being worked out, though it
may in fact be intuitively grasped (p. 50); he does not say by what process. It is the neo-Gricean
formalizations that make alternatives a *within-trial* comparison: a lexical scale supplies the competitors
(Horn, 1972), and what the speaker declined to say is used, on that occasion, to strengthen what was
said (Gazdar, 1979; Levinson, 2000; Sauerland, 2004). The strengthening step is not uniform across
those accounts, and nothing here needs it to be: Gazdar (1979) derives directly the
strong inference that the speaker is certain the stronger alternative is false (as reported by
Sauerland, 2004, p. 369). Sauerland (2004, pp. 383, 388) derives in the first instance the weak
inference that the speaker is not certain of the stronger alternative, and reaches its negation only
by a further step. Levinson (2000, p. 36) takes the negation itself as the default. What they share is the
*occasion*: on all of them a listener hearing *some* consults what *all* would have conveyed
**on this trial**. What
$\theta_u$ holds is a *cross-trial average*, one shared weight, shaped by the frequencies at which
the entries have been met and consulted identically whatever is heard. No two entries are compared at any moment: $\theta_u$ is one scalar, updated from signals averaged across
presentations. At $m=1$ there is moreover no anatomy in which such a comparison could take place: one
projection, one scalar, one pooled error signal. At $m=2$ the update is not local (*Locality*), but it
is still one scalar averaged across trials.

The position this commits us to is that any implicature reaching the belief through $\theta_u^\ast$
is **conventionalized rather than computed**, lexicalized into a weight by exposure and not derived
online from a reasoning step. That is a substantive claim. What the neo-Gricean (Horn, 1972;
Sauerland, 2004; Geurts, 2010), grammatical (Chierchia, 2004; Chierchia et al., 2012)
and probabilistic (Frank & Goodman, 2012; Goodman & Frank, 2016) accounts share, whatever else
divides them, is that the strengthening is *computed on the occasion of use* from a set of
alternatives. What does the computing differs sharply, being pragmatic reasoning about the speaker for the
first, compositional rules or a covert exhaustification operator applying in the parse of that very
utterance for the second, and
recursive Bayesian inference over a literal listener for the third, but all three put the
alternatives inside the trial. Ours is not computed at the occasion at all, but read off a weight.
Levinson's (2000) default generalized implicatures are the nearest neighbour, a default doing less online work than a full derivation, but a default is still keyed to an alternatives set at the
utterance, and Levinson holds a GCI to be a defeasible inference and not a part of conventional
meaning, where $\theta_u^\ast$ is a single scalar keyed to nothing but exposure frequency and, once
learned, not defeasible within the trial at all. The position should be argued rather than
inherited. It also bounds what the present
architecture can deliver: within-trial alternatives would require a mechanism that compares entries
inside a single inference pass, and Eq. (20) is not one.

<a id="codeb" name="codeb"></a>

````python
# === Code Cell B: the numbers Appendix B quotes. EVALUATION ===
#
# Everything is in closed form (Eqs. 15-16, exact by Sec. 8.6). Numbers Appendix B
# attributes to another cell are printed there and not repeated: the learning probe's
# first gradient and endpoint, Part A's bisection gap and margin, and Code Cell 4's range
# of theta_u* over the plane.

import math
import torch


def equilibrium_gradient(net, theta_u, utterances=UTTERANCES):
    """dF~/dtheta_u of Eq. (20) at the equilibrium of Eqs. (15)-(16), averaged over `utterances`."""
    total = 0.0
    for y in utterances:
        phi_S, phi_u = net.closed_form_fixed_point(y, theta_u=theta_u)
        _, r_state, _ = net.residuals(phi_S, phi_u, net.lexical_field(y), theta_u=theta_u)
        total += float(net.dot(phi_u, net.project(r_state / net.sigma_state)))
    return total / len(utterances)


def stationary_maximizer(net, couplings):
    """The maximizing root of Eq. (B2), summed over `couplings` as in Eq. (B3), with G = I."""
    total_sigma = net.sigma_lexical + net.sigma_state
    total = sum(couplings)
    a = net.sigma_utility * float(net.dot(net.mu_u, total))
    b = (len(couplings) * total_sigma * float(net.dot(net.mu_u, net.mu_u))
         - net.sigma_utility * sum(float(net.dot(c, c)) for c in couplings))
    c0 = -total_sigma * a / net.sigma_utility
    q = -0.5 * (b + math.copysign(math.sqrt(b * b - 4.0 * a * c0), b))
    roots = (q / a, c0 / q)
    return roots[0] if (roots[0] > 0) == (a > 0) else roots[1]


def appendix_b_report(net):
    couplings = {y: net.project(net.base_log_prior - net.lexical_field(y)) for y in UTTERANCES}
    total = sum(couplings.values())
    print("APPENDIX B: THE COUPLINGS c_y = B^T W (ell_0 - phi_L)")
    for y, c in couplings.items():
        print(f"  {y:<5} c = [{float(c[0]):+.5f}, {float(c[1]):+.5f}]   "
              f"<mu_u, c> = {float(net.dot(net.mu_u, c)):+.4f}")
    print(f"  sum   c = [{float(total[0]):+.5f}, {float(total[1]):+.5f}]   "
          f"<mu_u, c> = {float(net.dot(net.mu_u, total)):+.4f}")
    print(f"  theta_u* = {net.theta_u:.4f}; dF~/dtheta_u there = "
          f"{equilibrium_gradient(net, net.theta_u):.1e}")
    print()

    print("  THE SIGN OF theta_u*, against the signs with which ell_0 and phi_L enter c_y")
    model = stationary_maximizer(net, list(couplings.values()))
    for label, prior_sign, lexical_sign in (
            ("ell_0 - phi_L  (Eq. 9, the model)", 1.0, -1.0),
            ("-ell_0 + phi_L (both signs reversed)", -1.0, 1.0),
            ("-ell_0 - phi_L (ell_0's sign reversed)", -1.0, -1.0),
            ("ell_0 + phi_L  (phi_L's sign reversed)", 1.0, 1.0)):
        variant = [net.project(prior_sign * net.base_log_prior + lexical_sign * net.lexical_field(y))
                   for y in UTTERANCES]
        print(f"    {label:<40} sum <mu_u, c> = {float(net.dot(net.mu_u, sum(variant))):+8.3f}   "
              f"maximizer {stationary_maximizer(net, variant):+.4f}")
    print(f"    (the model's row matches the network's theta_u* to {abs(model - net.theta_u):.1e})")
    print("  Bogacz's Eq. (25) model at u = 2: theta* = u / v_p = "
          + ", ".join(f"{2.0 / v_p:.3f} at v_p = {v_p:g}" for v_p in (3.0, 1.0))
          + "; no maximizer at v_p = 0")
    print()

    print("  EXACT SOLVABILITY AT theta_u*")
    cosines, residual_norms, energies = [], [], []
    for y in UTTERANCES:
        c = couplings[y]
        cosines.append(float(net.dot(net.mu_u, c) / (net.mu_u.norm() * c.norm())))
        phi_S, phi_u = net.closed_form_fixed_point(y)
        residual_norms.append(float(net.squared_norm(phi_S - net.predict_state(phi_u))) ** 0.5)
        energies.append(float(net.free_energy(phi_S, phi_u, net.lexical_field(y))))
    print("    cos(mu_u, c_y) = " + ", ".join(f"{v:+.3f}" for v in cosines)
          + "   (no, some, all)")
    print("    ||r_S|| = " + ", ".join(f"{v:.2f}" for v in residual_norms)
          + f";  F~(theta_u*) = {sum(energies) / len(energies):.3f} (mean over the three)")
    print("    largest ||c_a - c_b|| = "
          f"{max(float((couplings[a] - couplings[b]).norm()) for a in UTTERANCES for b in UTTERANCES):.2f}")
    print()

    print("  ALTERNATIVE SPACES")
    gram = net.basis.T @ (net.weights[:, None] * net.basis)
    total_sigma = net.sigma_lexical + net.sigma_state
    rebuilt = 0.0
    for y in UTTERANCES:
        _, phi_u = net.closed_form_fixed_point(y)
        alone = torch.linalg.solve(
            torch.eye(net.num_basis, dtype=net.dtype)
            + net.sigma_utility * net.theta_u ** 2 * gram / total_sigma,
            net.mu_u + net.sigma_utility * net.theta_u * couplings[y] / total_sigma)
        rebuilt = max(rebuilt, float((phi_u - alone).abs().max()))
    print(f"    phi_u* rebuilt from c_y alone agrees with the fixed point to {rebuilt:.1e}")
    print("    theta_u* by exposure:")
    for exposure in (["no"], ["some"], ["all"], ["no", "some"], ["no", "all"],
                     ["some", "all"], list(UTTERANCES)):
        print(f"      {', '.join(exposure):<16}{net.learned_theta_u(exposure):>10.4f}")
    print()


def batching_report(net, *, tau_theta=20.0, tau_state=1.00, cycles=400000):
    """Appendix B, Locality: the batched update against one local update per presentation.

    The batched flow is `learn_theta_u`'s, theta_u += (tau_state / tau_theta) times the
    gradient averaged over the three utterances. The per-presentation flow takes one
    update of a third of that size after each utterance in turn, so a cycle of three
    carries the same total step. Both read the gradient at the closed-form equilibrium,
    in the reduced form Eq. (20) takes there, phi_u* . (c_y - theta_u G phi_u*) / S."""
    gram = net.basis.T @ (net.weights[:, None] * net.basis)
    if float((gram - torch.eye(net.num_basis, dtype=net.dtype)).abs().max()) > 1e-10:
        raise NotImplementedError("the reduced gradient below assumes B^T W B = I")
    total_sigma = net.sigma_lexical + net.sigma_state
    mu = [float(v) for v in net.mu_u]
    coupling = {y: [float(v) for v in net.project(net.base_log_prior - net.lexical_field(y))]
                for y in UTTERANCES}

    def gradient(y, theta):
        c = coupling[y]
        denominator = 1.0 + net.sigma_utility * theta * theta / total_sigma
        phi_u = [(mu[j] + net.sigma_utility * theta * c[j] / total_sigma) / denominator
                 for j in range(len(mu))]
        return sum(phi_u[j] * (c[j] - theta * phi_u[j]) for j in range(len(mu))) / total_sigma

    check = max(abs(sum(gradient(y, theta) for y in UTTERANCES) / len(UTTERANCES)
                    - equilibrium_gradient(net, theta)) for theta in (-1.0, -10.0, net.theta_u))
    rate = tau_state / tau_theta
    batched = local = 0.0
    marks = (10, 100, 1000, 10000, 100000, cycles)
    print("  BATCHED AGAINST PER-PRESENTATION UPDATES OF Eq. (20), from theta_u(0) = 0")
    print(f"    (reduced gradient checked against the fixed point to {check:.1e}; "
          f"tau_theta = {tau_theta:g})")
    print(f"    {'cycles':>8}{'batched':>12}{'per-presentation':>18}{'gap':>10}")
    for cycle in range(1, cycles + 1):
        batched += rate * sum(gradient(y, batched) for y in UTTERANCES) / len(UTTERANCES)
        for y in UTTERANCES:
            local += (rate / len(UTTERANCES)) * gradient(y, local)
        if cycle in marks:
            print(f"    {cycle:>8}{batched:>12.4f}{local:>18.4f}{abs(batched - local):>10.1e}")
    print(f"    theta_u* = {net.theta_u:.4f}: the batched flow ends {abs(batched - net.theta_u):.1e} "
          f"from it and the per-presentation flow {abs(local - net.theta_u):.1e}")
    print()


def degenerate_ray_report(net, priors=None, atom_ladder=(8, 9, 10, 11, 12, 15),
                          magnitudes=(1.0e2, 1.0e3, 1.0e4, 1.0e5, 1.0e6)):
    """Appendix B: how near the ray <mu_u, sum_y c_y> = 0 this notebook's own rows come,
    and what happens to the readings when theta_u* passes through it.

    The standing requirement above is stated at the default configuration, where the
    coupling is -40.9119. That is the best case, not a general margin: the coupling is a
    function of ell_0 and of theta_L, so it moves with the base world prior and with n while
    mu_u stays at its stipulated value, and this block prints how close the rows the
    evaluation actually runs come to zero.

    Nothing here is a hazard, and the last two parts are why. The two roots of Eq. (B2)
    have the constant product -S / sigma_u, so a large theta_u* means the OTHER root is
    near zero and the maximizer is about to change sign through infinity; but phi_S*
    converges to a single limit as |theta_u| -> inf, approached from both signs, so the
    readings pass through the ray continuously. Reported, not interpreted (B10).
    """
    if priors is None:
        priors = BASE_WORLD_PRIORS
    upper = lambda p: (p.zeta > p.theta_L).to(p.dtype)

    def coupling(p):
        return float(p.dot(p.mu_u, sum(p.project(p.base_log_prior - p.lexical_field(y))
                                       for y in UTTERANCES)))

    def q_readings(p):
        w, up = p.weights, upper(p)
        mass = lambda f: float((w * p.read_out(f) * up).sum())
        position = mass(p.closed_form_fixed_point("some")[0])
        return position - mass(p.literal_fixed_point("some")), position

    print("APPENDIX B: THE RAY <mu_u, sum_y c_y> = 0, AND theta_u* PASSING THROUGH IT")
    print("  Eq. (B2)'s two roots have the product -S / sigma_u, a constant, so theta_u* and the")
    print("  other stationary point are reciprocal up to that factor: theta_u* is large exactly")
    print("  where the coupling is small, and changes sign with it.")
    print(f"    {'ell_0':<14}{'<mu_u,sum c>':>14}{'theta_u*':>13}{'theta_min':>13}"
          f"{'product':>10}{'-S/sigma_u':>12}")
    for name, prior in priors.items():
        probe = net.respawn(num_atoms=10, lexical_strength=8.0, base_prior=prior)
        top, bottom = probe.theta_u_stationary_points()
        print(f"    {name:<14}{coupling(probe):>+14.4f}{top:>13.4f}{bottom:>13.6f}"
              f"{top * bottom:>10.5f}"
              f"{-(probe.sigma_lexical + probe.sigma_state) / probe.sigma_utility:>12.5f}")
    print("  So the -40.9119 the appendix quotes is the DEFAULT base world prior's margin, not a bound:")
    print("  at n = 10 and Lambda = 8 the flat base world prior's coupling is three orders of magnitude")
    print("  smaller and of the opposite sign, and that row is one Code Cell 2 prints.")
    print()

    print("  THE RAY IS CROSSED BY VARYING n ALONE, with mu_u at its stipulated value.")
    print("  Under the flat base world prior at Lambda = 8 the coupling changes sign between two rungs, so")
    print("  theta_u* diverges between them. The mass criteria are printed beside it:")
    print(f"    {'n':>4}{'theta_L':>9}{'<mu_u,sum c>':>14}{'theta_u*':>12}{'shift':>12}"
          f"{'P(all)':>10}   {'mass shift':>11}{'mass position':>14}")
    for atoms in atom_ladder:
        probe = net.respawn(num_atoms=atoms, lexical_strength=8.0, base_prior=priors["flat"])
        shift, position = q_readings(probe)
        print(f"    {atoms:>4}{probe.theta_L:>9.4f}{coupling(probe):>+14.4e}"
              f"{float(probe.theta_u):>12.2f}{shift:>+12.4e}{position:>10.4f}   "
              f"{('met' if shift < 0.0 else 'no'):>9}"
              f"{('met' if position < 0.5 else 'no'):>12}")
    print("  theta_u* changes sign across the ray while both q readings vary monotonically and")
    print("  neither criterion changes status. The next block is why.")
    print()

    print("  WHY THE READINGS DO NOT FEEL IT: phi_S* HAS ONE LIMIT AS |theta_u| -> inf,")
    print("  REACHED FROM BOTH SIGNS. Under the flat base world prior at Lambda = 8, for \"some\":")
    probe = net.respawn(num_atoms=10, lexical_strength=8.0,
                        base_prior=priors["flat"], theta_u=1.0)
    print(f"    {'|theta_u|':>11}{'max|phi_S*(+) - phi_S*(-)|':>28}{'mode zeta (+)':>15}"
          f"{'mode zeta (-)':>15}")
    for size in magnitudes:
        plus = probe.closed_form_fixed_point("some", theta_u=+size)[0]
        minus = probe.closed_form_fixed_point("some", theta_u=-size)[0]
        print(f"    {size:>11.0e}{float((plus - minus).abs().max()):>28.3e}"
              f"{float(probe.zeta[int(torch.argmax(plus))]):>15.3f}"
              f"{float(probe.zeta[int(torch.argmax(minus))]):>15.3f}")
    print("    The gap falls like 1/|theta_u|, so the two branches meet in one limit field and")
    print("    the ray is a pole of theta_u* and not of anything the evaluation reads off.")
    print("    Eq. (B2) still REQUIRES the coupling to be nonzero: exactly on the ray there is")
    print("    no interior maximizer to report, which is what the constructor refuses.")
    print()


appendix_b_report(evaluation_network)
degenerate_ray_report(evaluation_network)
batching_report(evaluation_network)
````

## <a id="appc" name="appc"></a>Appendix C: how many utility directions an inventory needs

Text cell 5 Part A shows that $m=1$ is
degenerate and takes $m=2$ as the bare minimum. That leaves open what fixes the number in general:
whether $2$ is a fact about this inventory of three, or about the architecture. This appendix
answers that, and the answer is neither: it is a fact about the **thresholds**, and $m=2$ is the
least that retains every direction of the entries' span, because the scale has two ends. That is a
rationale for $m=2$ rather than a proof that no other architecture could serve: separation alone
can be had with one column of no definite parity (§9).

Throughout, $\Lambda$ is factored out of the projections, since $\varphi_L=\Lambda\chi_y$ makes it a
common scalar. The two indicators $\chi_{\textit{no}}=\mathbb 1[\zeta>-\theta_L]$ and
$\chi_{\textit{all}}=\mathbb 1[\zeta<\theta_L]$ are reflections of each other on a grid symmetric
about $\zeta=0$.

---

### <a id="appc-1" name="appc-1"></a>1. Two readings of the criterion, which come apart

Eq. (29) says the $u$ level reaches the lexicon only through
$\kappa_y=B^{\mathsf T}W\varphi_{L,y}=\Lambda\,B^{\mathsf T}W\chi_y$, since
$\varphi_{L,y}=\Lambda\chi_y$ by Eq. (6). Every $\kappa$ tabulated in this appendix is written
**per unit $\Lambda$**, that is as $B^{\mathsf T}W\chi_y$, so that the entries carry the geometry
of the exclusion sets and not the strength of the lexicon; Code Cell 3 prints the same quantities at
the default $\Lambda=8$ and its numbers are eight times these ($\pm9.11909$ against the
$\pm1.13989$ below). Nothing in the argument depends on the scale, since every requirement below is a
statement about distinctness or collinearity, both invariant under a positive factor.
Two requirements can be asked of $B$, and they are not the same requirement:

- **Separation.** The $\kappa_y$ are pairwise distinct, so no two utterances are confused at the $u$
  level. This is Part A's criterion.

- **Spanning.** The projection $B^{\mathsf T}W$ is injective on the space the $\chi_y$ span modulo
  the constant, so nothing the lexicon can express is invisible to the $u$ level. The word is used
  in this sense throughout the appendix. The columns of $B$ are smooth profiles and do not
  themselves span the step-shaped $\chi_y$: projecting a $\chi_y$ onto them leaves a residual, and
  what spanning requires is only that no direction of $\operatorname{span}\{\chi_y\}/\langle\mathbf 1\rangle$
  be lost in the projection.

Spanning implies separation and not conversely. Section 7 below exhibits the gap; until then the
two agree.

### <a id="appc-2" name="appc-2"></a>2. Complementation costs nothing

$E_{\textit{some}}$ is the complement of $E_{\textit{no}}$, so $\chi_{\textit{some}}=\mathbf 1-\chi_{\textit{no}}$.
Part A drops the constant from $B$ deliberately, since a constant direction shifts $\varphi_S$
uniformly and is invisible to the read-out (Eq. 12), so $\langle\mathbf 1,b_j\rangle=0$ for every
column, and

$$E_{y'}=[0,1]\setminus E_y\quad\Longrightarrow\quad
\kappa_{y'}=B^{\mathsf T}W(\mathbf 1-\chi_y)=-\kappa_y .
\tag{C1}$$

Complementation is **negation** in projection space, not a new direction. Two consequences run
through everything below: a complementary pair costs no dimension at all, and it is the *easiest*
kind of pair to separate, since $\kappa$ and $-\kappa$ are antipodal and every functional that sees
one sees the difference doubled.

**The same holds on any domain of the scale.** Eq. (C1) uses one fact about the basis,
$\langle\mathbf 1,b_j\rangle=0$. Suppose a pair of entries partitions a domain $D$ of the scale rather
than the whole of it, so that $\chi_{y'}=\chi_D-\chi_y$, and the columns of a basis are taken
orthogonal to $\chi_D$ in place of $\mathbf 1$, $\langle\chi_D,b_j\rangle=0$. Substituting $\chi_D$
for $\mathbf 1$ in Eq. (C1) then gives $\kappa_{y'}=B^{\mathsf T}W(\chi_D-\chi_y)=-\kappa_y$, and
nothing else in the derivation changes. Such a pair also costs no dimension, and a single column
with $\langle\chi_y,b\rangle\neq0$ separates it. The result therefore holds at every level of a
cascade whose levels are complementary pairs, each on the domain the level above it leaves. No such
cascade is built in this notebook.

### <a id="appc-3" name="appc-3"></a>3. The count is thresholds, not utterances

With $t$ distinct cut points the scale is divided into $t+1$ intervals; every $\chi_y$ is a union of
them, so the $\chi_y$ span at most $t+1$ dimensions including the constant, hence

$$\dim\operatorname{span}\{\chi_y\}\big/\langle\mathbf 1\rangle\;\le\;t .
\tag{C2}$$

Here $t=2$, at $-\theta_L$ and $+\theta_L$, and the reason is Eq. (A1): the scale's two
endpoints have no $\zeta$ coordinate, and *no* is anchored at $s=0$ while *all* is anchored at $s=1$.
There is one threshold per end, and **two ends give two thresholds and $m=2$**, which is the whole
of the coincidence.

$m=2$ is moreover exactly right rather than merely sufficient. Solving for what $B$ cannot see, in
coordinates on the intervals:

| cuts | intervals | dim of the unseen subspace | an unseen direction |
|---|---|---|---|
| $\{-\theta_L,+\theta_L\}$ | 3 | 1 | $[1,\,1,\,1]$ |
| $\{-\theta_L,0,+\theta_L\}$ | 4 | 2 | $[-0.325,\,1.000,\,-0.974,\,0.298]$, orthogonal to the constant |

At two thresholds the only thing $m=2$ misses is the constant, which the read-out misses
too, so nothing is lost. Part A's parity argument is the symmetric instance of this count: with the cuts at
$\pm\theta_L$ the two-dimensional space splits into one even and one odd direction, so a single $b$ of
definite parity spans exactly half of it.

### <a id="appc-4" name="appc-4"></a>4. A fourth utterance: the complement of *all*

Consider adding a word with $E=\{1\}$, true everywhere except $s=1$, that is *not all*. By the
pattern of Eq. (A1) it maps to $\chi=\mathbb 1[\zeta\ge\theta_L]=\mathbf 1-\chi_{\textit{all}}$, verified exactly. By
Eq. (C1) its projection is $-\kappa_{\textit{all}}$, and the four utterances land on the corners of a
rectangle about the origin:

| | $\kappa_{y,1}$ (tilt) | $\kappa_{y,2}$ (width) |
|---|---|---|
| *no* | $+1.13989$ | $-0.72098$ |
| *some* | $-1.13989$ | $+0.72098$ |
| *all* | $-1.13989$ | $-0.72098$ |
| *not all* | $+1.13989$ | $+0.72098$ |

All four are distinct, so **$m=2$ still suffices at $\lvert\mathcal Y\rvert=4$**, and the two pairs of contradictories, *no* against *some* and *all* against *not all*, sit at opposite
corners. The rank confirms that nothing was added:
$\operatorname{span}\{\chi_{\textit{no}},\chi_{\textit{some}},\chi_{\textit{all}}\}$ has rank $3$,
adjoining $\chi_{\textit{not all}}$ leaves it at $3$, and adjoining $\mathbf 1$ leaves it at $3$ again, since $\mathbf 1=\chi_{\textit{no}}+\chi_{\textit{some}}$ was already inside.
Modulo the constant the rank is $2$. The fourth word introduces no threshold, so by Eq. (C2) it cannot introduce a
dimension.

*An aside worth recording.* $E=\{1\}$ is the **O corner of the Aristotelian square**, the vertex
that is not lexicalized as one word (Horn, 1972, as reported by Katzir & Singh, 2013). Under this model, the gap
is not representational: the O corner is exactly as expressible as the other three and costs the
architecture nothing. Whatever explains the gap, nothing here supplies a reason for it.

### <a id="appc-5" name="appc-5"></a>5. Inventories of two, and the parity each demands

Dropping one utterance leaves three inventories. Each needs only **one** dimension to separate, since two points always admit a
separating functional, but *which* direction is forced, and the answer differs:

| inventory | thresholds | span mod $\mathbf 1$ | odd $b_1$ | even $b_2$ |
|---|---|---|---|---|
| *no*, *some* | 1 | 1 | $2.27977$ | $1.44196$ |
| *no*, *all* | 2 | 2 | $2.27977$ | **collapses** ($6.9\times10^{-18}$) |
| *some*, *all* | 2 | 2 | **collapses** ($6.9\times10^{-18}$) | $1.44196$ |

Read in Part B §2's vocabulary, where $b_1$ is **tilt** and $b_2$ is **width**:

- *no* and *all* differ by *which end of the scale*, an odd difference, and want tilt.

- *some* and *all* differ by *middle versus both ends*, an even difference, and want width.

- *no* and *some* are complementary, so by Eq. (C1) either direction serves; this is the only
  inventory with no parity constraint, and also the only one whose span drops to $1$, since dropping
  *all* removes the $+\theta_L$ threshold outright.

Hence the general statement, of which Part A is the instance:

> Within bases whose columns have definite parity, $m\ge2$ is forced exactly when the inventory
> contains two pairs whose differences have **opposite parity**, and with symmetric thresholds that
> requires three utterances.

$\chi_{\textit{no}}-\chi_{\textit{all}}$ is odd and $\chi_{\textit{all}}-\chi_{\textit{some}}$ is
even, so no single $b$ of definite parity serves both. Section 4 now follows without recomputation:
*not all* adds neither a threshold nor a third parity class.

### <a id="appc-6" name="appc-6"></a>6. What an odd $m=1$ basis costs

$b=\zeta$, a monotone ramp meaning "more versus less", is the natural first choice for a single direction,
and it is odd. Under it $\kappa_{\textit{some}}=\kappa_{\textit{all}}=-1.13989$ and the two utterances
are one point at the $u$ level. Running the network at $m=1$:

| | $\lvert\Delta\varphi_u^\ast\rvert$ | $\lvert\Delta\varphi_S^\ast\rvert$ | $\lvert\Delta\mathbb E[s]\rvert$ |
|---|---|---|---|
| *some* vs *all* | $0.0000$ | $4.0000$ | $0.2534$ |
| *no* vs *all* | $0.6793$ | $8.5464$ | $0.7818$ |

at the $\theta_u^\ast=+26.774$ this configuration learns, which is positive where the $m=2$ default's
is negative: $m$ is one of the inputs $\theta_u^\ast$ is a function of (Appendix B), so changing it
changes the parameter as well as the basis.

$\varphi_S^\ast$ still separates *some* from *all* by $4.0000$, because Eq. (15) carries
$\ell_0-\varphi_L$ directly and not only through the projection. The model is not blind; the utility
level is.

That distinction is exactly what makes the failure serious rather than cosmetic. With
$\varphi_u^\ast(\textit{some})=\varphi_u^\ast(\textit{all})$, the term $\theta_u B\varphi_u^\ast$ is
the *same field* for both, so the utility level acts on the pair purely common-mode and

$$\varphi_S^\ast(\textit{all})-\varphi_S^\ast(\textit{some})\ \text{ is independent of }\theta_u ,
\tag{C3}$$

measured to $8.9\times10^{-16}$ over $\theta_u\in\{0,1,-1,5,\theta_u^\ast\}$, the learned value
included, against a spread of $3.87$ in the same contrast at $m=2$ over $\{0,1,-1,5\}$ and $4.21$
with its $\theta_u^\ast$ included (quadrature norm, the $9.6995\to13.9124$ Code Cell 2 prints). Under an odd one-dimensional basis $\theta_u$ becomes to $\{\textit{some},\textit{all}\}$
precisely what $\mu_u$ is to all three utterances at $m=2$ (Part B §1): a pure common-mode shift that
can move no contrast between them. The mechanism Text cell 4 Part C measures, amplification of
the $\operatorname{span}B$ component carried by $\theta_u$, is therefore unavailable for the one pair a
scalar implicature is about, at every $\theta_u$, $\Lambda$ and $\mu_u$.

$\Delta_y$ nevertheless differs between them, $+0.0604$ against $-0.0473$, and differs against the
$\theta_u=0$ control as well, $+0.0429$ against $+0.2715$. That is not a residual
effect surviving Eq. (C3) but the failure mode of Part B §4 again: $\Delta_y$ is a $q$-mass (Eq. 27)
taken after the softmax, so a field added identically to two differently-placed beliefs moves their
masses by different amounts. Nothing the model represents *about the relation between* the two
utterances has moved.

### <a id="appc-7" name="appc-7"></a>7. Separation is not spanning

Section 3 gives $m=t$ as the design rule, but that rule is about spanning, and separation of a
particular inventory can survive $m<t$. Adding *most*, true where $s>\tfrac12$ and so a third cut at
$\zeta=0$, raises the span from $3$ to $4$, so $m=2$ no longer spans and there is a genuinely unseen
direction beyond the constant. Yet **$m=2$ collapses no contrast between exclusion sets**: no non-constant $\{-1,0,1\}$ combination of the four intervals
lies in the unseen subspace, and all $14$ nonempty proper unions
of the four intervals still separate under $m=2$, with a minimum $\kappa$ distance of $0.72097$ and
no collisions.

The contrast with Part A's degeneracy is the point. At two thresholds the direction $m=1$ misses is the middle interval
$\chi_{\textit{all}}-\chi_{\textit{some}}=\mathbb 1_{(-\theta_L,\theta_L)}$, a difference of two exclusion sets, so
$m=1$ *does* collapse a contrast, the one between *some* and *all*, and that is why $m=1$ actually fails. Whether $m<t$ bites is a question about the integer
lattice of exclusion sets, not about dimension alone. The margin does erode: at asymmetric cuts
$\{-4,-1,2.5\}$ the minimum separation falls to $0.13100$.

### <a id="appc-8" name="appc-8"></a>8. Which part of the utility field moves the mode

By Eq. (23) the field the utility level adds under *some*, $\varphi_S^\ast-\varphi_S^\ast(\theta_u{=}0)$,
lies in $\operatorname{span}B$, so at $m=2$ it is $k_1b_1+k_2b_2$: a tilt part on the odd column and
a width part on the even column, in §5's vocabulary. Unlike the projections of §§1–7, $k_1$ and
$k_2$ are not per unit $\Lambda$: each is taken at its configuration's own $\Lambda$ and
$\theta_u^\ast$, and $\theta_u^\ast$ moves with $\Lambda$. Code Cell C adds each part alone to the
tempered control ($\theta_u=0$, field $\tfrac12(\ell_0-\varphi_L)$, whose mode node is that of
$\ell_0$ in every configuration below) and reads off the mode and the all-region $q$-mass of
Eq. (27). Adding one part alone is a **counterfactual manipulation** of the settled field and not a
manipulation of the model: no setting of $\theta_u$, $\Lambda$ or $\mu_u$ makes the network settle
on either field, and *control* is kept for a configuration the model is run in, such as the
$\theta_u$ control itself.

**Part D's base world priors at $\Lambda=8$.** The width coefficient is negative under all four base world priors, and
the tilt coefficient is positive under three and $-1.50$ under $\mathrm{Beta}(1,3)$. Width alone
lowers the all-region mass under all four, from $0.2356$ to $0.0230$ under $\mathrm{Beta}(3,1)$.
Under the Gaussian and flat base world priors the tempered control's mode sits at $s=0.5000$, where the width
part's slope is $0$: width alone leaves the mode there, tilt alone moves it up $6$ and $17$ grid
steps, and the model moves it up $3$ and $6$. Under $\mathrm{Beta}(3,1)$ tilt alone moves the mode up
$41$ steps and width alone down $4$, and the model moves it up $5$. Under $\mathrm{Beta}(1,3)$ the
directions are exchanged, tilt alone down $3$ and width alone up $4$, and the model moves it up $3$.

**The plane.** Over the $121$ cells of Code Cell 4, the tilt coefficient is positive and the width
coefficient negative in every cell. The model's mode moves up in $50$ cells, down in $67$, and not
at all in $4$. In every up cell tilt alone moves it up and width alone does not; in every down cell
width alone moves it down and tilt alone does not. In all $117$ cells where the mode moves, the
direction is the sign of the two parts' summed slope at the tempered control's mode, which is the
first-order displacement of a smooth maximum: adding $g$ to a field with a maximum of curvature $f''<0$ moves the maximum by about $g'/(-f'')$. The up cells are those whose tempered mode sits at
$s\le0.9405$, the down cells those at $s\ge0.9405$. Width alone lowers the all-region mass in all
$121$ cells and tilt alone in none. Code Cell C tabulates the $21$ cells where the $q$ shift
criterion is met and the mode shift criterion is not.

**Where the signs come from.** At $\sigma=1$ and $B^{\mathsf T}WB=I$, Eqs. (15)–(16) give
$k=\tfrac{\theta_u}{2}\big(\mu_u+\tfrac{\theta_u}{2}c\big)\big/\big(1+\tfrac{\theta_u^2}{2}\big)$
with $c=B^{\mathsf T}W(\ell_0-\varphi_L)$, so $k\to c/2$ as $\lvert\theta_u\rvert$ grows, which is
Eq. (24). The fitted $k$ matches that form to $1.2\times10^{-15}$. The relative distance from $c/2$
has median $8.6\times10^{-7}$ in the tilt column and $4.8\times10^{-6}$ in the width column, both
columns are within $10^{-3}$ of it in $111$ cells, and the largest distance is $17\%$, in the tilt
column at $(\alpha,\Lambda)=(1,2)$, where $\theta_u^\ast=-6.51$. $k$ and $c$ agree in sign in both
columns in all $121$ cells. The limit field of Eq. (24) has the model's mode node, and gives the
same reading on each of the four criteria, in all $121$ cells and under Part D's four base world priors. The
signs of the tilt and width parts are therefore the signs of the literal field's own projection.

Modes here are grid nodes, and the direction argument is first order, checked only by its sign.

### <a id="appc-9" name="appc-9"></a>9. What this appendix does not settle

Sections 1–7 concern separation and spanning. They do not address whether $m=2$ is *sufficient*
in the wider sense Part A leaves open. Part A's own preference for $m=2$ over a single $b$ of no
definite parity is a different argument, that one coordinate confounds tilt and width, and that
argument still favours $m=2$ in the two-utterance cases of §5 where one dimension separates. Nor does
Eq. (C2) say anything about a scale with more than one dimension, which §9.3 takes up separately.

````python
# === Code Cell C: the numbers Appendix C quotes. EVALUATION ===
#
# Projections are per unit Lambda, B^T W chi_y, wherever Appendix C writes them so. The
# m = 1 networks of Sec. 6 are respawned with basis_degree = 1 and carry their own
# theta_u*.

import itertools
import math
import torch


def interval_indicators(net, cuts):
    """Indicators of the intervals the cut points divide the grid into, each open below."""
    edges = [-math.inf] + list(cuts) + [math.inf]
    return [((net.zeta > low) & (net.zeta <= high)).to(net.dtype)
            for low, high in zip(edges[:-1], edges[1:])]


def appendix_c_report(net):
    zeta, theta_L = net.zeta, net.theta_L
    chi = {y: exclusion_indicator(y, zeta, theta_L) for y in UTTERANCES}
    chi["not all"] = (zeta >= theta_L).to(net.dtype)
    ones = torch.ones_like(zeta)

    def rank(columns):
        return int(torch.linalg.matrix_rank(torch.stack(columns, dim=1)))

    def shown(value):
        return f"{value:+.5f}" if abs(value) > 1e-9 else f"{value:+.1e}"

    print("APPENDIX C, Secs. 1 and 4: LEXICAL PROJECTIONS PER UNIT LAMBDA, B^T W chi_y")
    for y, field in chi.items():
        value = net.project(field)
        print(f"  {y:<8}[{float(value[0]):+.5f}, {float(value[1]):+.5f}]")
    print(f"  (Code Cell 3 prints them at Lambda = {net.lexical_strength:g}.)")
    print(f"  max |chi_(not all) - (1 - chi_all)| = "
          f"{float((chi['not all'] - (1.0 - chi['all'])).abs().max()):.1e}")
    three = [chi[y] for y in UTTERANCES]
    with_one = rank(three + [chi["not all"], ones])
    print(f"  rank of the three chi_y {rank(three)}; with chi_(not all) "
          f"{rank(three + [chi['not all']])}; with the constant as well {with_one}, "
          f"so {with_one - 1} modulo the constant")
    print()

    print("  Sec. 3: WHAT B CANNOT SEE, in coordinates on the intervals")
    for cuts, label in (([-theta_L, theta_L], "-theta_L, +theta_L"),
                        ([-theta_L, 0.0, theta_L], "-theta_L, 0, +theta_L")):
        pieces = interval_indicators(net, cuts)
        seen = torch.stack([net.project(piece) for piece in pieces], dim=1)      # m x intervals
        _, singular, right = torch.linalg.svd(seen)
        unseen = right[int((singular > 1e-10).sum()):]
        constant = torch.ones(len(pieces), dtype=net.dtype) / math.sqrt(len(pieces))
        if unseen.shape[0] == 1:
            direction = unseen[0] / float(unseen[0][unseen[0].abs().argmax()])
            note = "(the constant)"
        else:
            beyond = [row - float(row @ constant) * constant for row in unseen]
            direction = max(beyond, key=lambda row: float(row.norm()))
            direction = direction / float(direction[direction.abs().argmax()])
            note = "(the unseen direction orthogonal to the constant)"
        hidden = [pattern for pattern in itertools.product((-1, 0, 1), repeat=len(pieces))
                  if len(set(pattern)) > 1
                  and float((seen @ torch.tensor(pattern, dtype=net.dtype)).abs().max()) < 1e-10]
        print(f"    cuts {label}: {len(pieces)} intervals, unseen subspace of dimension "
              f"{unseen.shape[0]}: [" + ", ".join(f"{float(v):+.3f}" for v in direction)
              + f"] {note}")
        print(f"      non-constant {{-1, 0, 1}} combinations of the intervals that B cannot see: "
              f"{len(hidden)}")
    print()

    print("  Sec. 5: INVENTORIES OF TWO, <chi_a - chi_b, b_j> per unit Lambda")
    print(f"    {'pair':<12}{'thresholds':>11}{'span mod 1':>12}{'odd b_1':>14}{'even b_2':>14}")
    for a, b in (("no", "some"), ("no", "all"), ("some", "all")):
        difference = chi[a] - chi[b]
        changes = (torch.stack([chi[a], chi[b]]).diff(dim=1) != 0).any(dim=0)
        print(f"    {a + ' - ' + b:<12}{int(changes.sum()):>11}{rank([chi[a], chi[b], ones]) - 1:>12}"
              f"{shown(float(net.inner(difference, net.basis[:, 0]))):>14}"
              f"{shown(float(net.inner(difference, net.basis[:, 1]))):>14}")
    print()

    def expected(model, field):
        return float((model.weights * model.read_out(field) * logistic(model.zeta)).sum())

    def all_mass(model, field):
        return float((model.weights * model.read_out(field)
                      * (model.zeta >= model.theta_L).to(model.dtype)).sum())

    print("  Sec. 6: AN ODD m = 1 BASIS, b = zeta")
    for label, overrides in (("theta_u = 1, a control", {"theta_u": 1.0}), ("learned", {})):
        model = net.respawn(basis_degree=1, mu_u=1.0, **overrides)
        fixed = {y: model.closed_form_fixed_point(y) for y in UTTERANCES}
        kappa = [float(model.project(exclusion_indicator(y, model.zeta, model.theta_L))[0])
                 for y in UTTERANCES]
        print(f"    {label}: theta_u = {model.theta_u:+.4f}; kappa per unit Lambda "
              + ", ".join(f"{y} {k:+.5f}" for y, k in zip(UTTERANCES, kappa)))
        for a, b in (("some", "all"), ("no", "all")):
            print(f"      {a} vs {b}: |d phi_u*| {float((fixed[a][1] - fixed[b][1]).abs().max()):.4f}"
                  f"  |d phi_S*| {float((fixed[a][0] - fixed[b][0]).abs().max()):.4f}"
                  f"  |d E[s]| {abs(expected(model, fixed[a][0]) - expected(model, fixed[b][0])):.4f}")
        literal = {y: all_mass(model, model.literal_fixed_point(y)) for y in ("some", "all")}
        tempered = {y: all_mass(model, model.closed_form_fixed_point(y, theta_u=0.0)[0])
                    for y in ("some", "all")}
        full = {y: all_mass(model, fixed[y][0]) for y in ("some", "all")}
        print(f"      shift against q_lit: some {full['some'] - literal['some']:+.4f}, "
              f"all {full['all'] - literal['all']:+.4f}; against the theta_u = 0 control: "
              f"some {full['some'] - tempered['some']:+.4f}, all {full['all'] - tempered['all']:+.4f}")
    for degree in (1, 2):
        model = net.respawn(basis_degree=degree, mu_u=1.0)
        thetas = (0.0, 1.0, -1.0, 5.0, model.theta_u)
        contrasts = [model.closed_form_fixed_point("all", theta_u=t)[0]
                     - model.closed_form_fixed_point("some", theta_u=t)[0] for t in thetas]
        change = max(float((c - contrasts[0]).abs().max()) for c in contrasts[1:])
        norms = [float(model.squared_norm(c)) ** 0.5 for c in contrasts]
        print(f"    (C3) m = {degree}, theta_u in {{0, 1, -1, 5}} and theta_u* = {model.theta_u:+.3f}: "
              f"largest change of phi_S*(all) - phi_S*(some) {change:.1e}; its quadrature norm "
              f"ranges {max(norms[:4]) - min(norms[:4]):.2f} without theta_u* and "
              f"{max(norms) - min(norms):.2f} with it")
    print()

    print("  Sec. 7: SEPARATION WITHOUT SPANNING, every nonempty proper union of intervals")
    for cuts, label in (([-theta_L, 0.0, theta_L], "-theta_L, 0, +theta_L"),
                        ([-4.0, -1.0, 2.5], "-4, -1, 2.5")):
        pieces = interval_indicators(net, cuts)
        unions = [sum(pieces[i] for i in chosen) for size in range(1, len(pieces))
                  for chosen in itertools.combinations(range(len(pieces)), size)]
        points = [net.project(union) for union in unions]
        distances = [float((points[i] - points[j]).norm())
                     for i in range(len(points)) for j in range(i + 1, len(points))]
        print(f"    cuts {label}: {len(unions)} unions, least distance between projections "
              f"{min(distances):.5f}, collisions {sum(d < 1e-9 for d in distances)}")
    print()


# ---- Sec. 8: the utility field under "some", split into tilt and width ----
#
# By Eq. (23) the utility field phi_S* - phi_S*(theta_u = 0) lies in span B: one coefficient on B's
# odd column (tilt) and one on its even column (width). Each part is added alone to the tempered
# control: a counterfactual manipulation of the settled field (Sec. 8), not a manipulation of the
# model, since no setting of theta_u, Lambda or mu_u produces either field. The limit field is the
# right-hand side of Eq. (24). Coefficients are at each configuration's own Lambda and theta_u*, not
# per unit Lambda. Counts over the plane read signs with the 1e-12 zero band.


def utility_split_report(net, priors, alphas, lambdas):
    ROUNDOFF = 1e-12
    parity = []
    for j in range(net.num_basis):
        column = net.basis[:, j]
        parity.append("odd" if float((column + column.flip(0)).abs().max()) < 1e-9
                      else "even" if float((column - column.flip(0)).abs().max()) < 1e-9 else "mixed")
    if sorted(parity) != ["even", "odd"]:
        raise RuntimeError(f"Sec. 8 needs one odd and one even column of B, found {parity}")
    TILT, WIDTH = parity.index("odd"), parity.index("even")
    PARTS = ("tempered", "tilt", "width", "model")
    CRITERIA = ("mode shift", "mode position", "mass shift", "mass position")

    def split(probe):
        if not probe.sigma_lexical == probe.sigma_state == probe.sigma_utility == 1.0:
            raise RuntimeError("Sec. 8's exact form is written at sigma = 1")
        B, w, theta = probe.basis, probe.weights, float(probe.theta_u)
        gram = B.T @ (w[:, None] * B)
        literal = probe.literal_fixed_point("some")
        phi_S = probe.closed_form_fixed_point("some")[0]
        tempered = probe.closed_form_fixed_point("some", theta_u=0.0)[0]
        utility = phi_S - tempered
        k = torch.linalg.solve(gram, probe.project(utility))
        c = probe.project(literal)
        exact = (theta / 2) * (probe.mu_u + (theta / 2) * c) / (1 + theta ** 2 / 2)
        fields = {"tempered": tempered, "tilt": tempered + B[:, TILT] * k[TILT],
                  "width": tempered + B[:, WIDTH] * k[WIDTH], "model": phi_S,
                  "limit": tempered + B @ (c / 2)}
        upper = (probe.zeta >= probe.theta_L).to(probe.dtype)

        def mass(field):
            return float((w * probe.read_out(field) * upper).sum())

        node = {key: int(torch.argmax(field)) for key, field in fields.items()}
        masses = {key: mass(field) for key, field in fields.items()}
        literal_mass = mass(literal)
        k0 = int(torch.argmax(probe.base_log_prior))
        at = min(max(node["tempered"], 1), len(probe.zeta) - 2)

        def slope(field):
            return float(field[at + 1] - field[at - 1]) / float(probe.zeta[at + 1] - probe.zeta[at - 1])

        def criteria(key):
            return (node[key] < k0, float(probe.zeta[node[key]]) < probe.theta_L,
                    masses[key] - literal_mass < -ROUNDOFF, masses[key] < 0.5)

        return {
            "theta": theta, "k": (float(k[TILT]), float(k[WIDTH])), "c": (float(c[TILT]), float(c[WIDTH])),
            "relative": tuple(abs(float(k[j] - c[j] / 2)) / abs(float(c[j] / 2)) for j in (TILT, WIDTH)),
            "residual": float((utility - B @ k).abs().max()),
            "gram_error": float((gram - torch.eye(gram.shape[0], dtype=gram.dtype)).abs().max()),
            "exact_error": float((k - exact).abs().max()) / max(1.0, float(k.abs().max())),
            "node": node, "k0": k0, "s": {key: float(logistic(probe.zeta[n])) for key, n in node.items()},
            "mass": masses, "literal_mass": literal_mass,
            "slope": (slope(B[:, TILT] * k[TILT]), slope(B[:, WIDTH] * k[WIDTH])),
            "criteria": {key: criteria(key) for key in ("model", "limit")},
        }

    def direction(a, b):
        return (a > b) - (a < b)

    step = float(net.zeta[1] - net.zeta[0])
    part_d = {name: split(net.respawn(base_prior=prior)) for name, prior in priors.items()}
    plane = []
    for alpha in alphas:
        for strength in lambdas:
            row = split(net.respawn(base_prior=beta_world_prior(alpha, 1.0), lexical_strength=strength))
            row["alpha"], row["Lambda"] = alpha, strength
            plane.append(row)

    print("APPENDIX C, Sec. 8: THE UTILITY FIELD UNDER \"some\", SPLIT INTO TILT (odd b_1) AND WIDTH (even b_2)")
    print("  Each configuration at its own Lambda and theta_u*. \"+ tilt\" and \"+ width\" add that part alone to")
    print("  the tempered control (theta_u = 0): counterfactual manipulations of the settled field, which no")
    print("  setting of the model produces.")
    print()

    print(f"  Block 1. PART D'S PRIORS AT Lambda = {net.lexical_strength:g}")
    print(f"    {'p_0':<13}{'theta_u*':>10}{'tilt':>9}{'width':>9}{'span-B residual':>17}"
          f"{'slope of tilt':>15}{'slope of width':>16}   (slopes at the tempered control's mode)")
    for name, row in part_d.items():
        print(f"    {name:<13}{row['theta']:>10.2f}{row['k'][0]:>+9.2f}{row['k'][1]:>+9.2f}{row['residual']:>17.1e}"
              f"{row['slope'][0]:>+15.3f}{row['slope'][1]:>+16.3f}")
    print("    Modes in s, and in grid steps from the tempered control's mode; multiplying the steps by the")
    print(f"    grid spacing, {step:.2f}, recovers the distance in zeta.")
    print(f"    {'p_0':<13}{'tempered':>9}{'+ tilt':>9}{'steps':>7}{'+ width':>9}{'steps':>7}{'model':>9}{'steps':>7}")
    for name, row in part_d.items():
        s, n = row["s"], row["node"]
        print(f"    {name:<13}{s['tempered']:>9.4f}"
              + "".join(f"{s[key]:>9.4f}{n[key] - n['tempered']:>+7d}" for key in ("tilt", "width", "model")))
    print("    All-region q-mass, zeta >= theta_L (Eq. 27)")
    print(f"    {'p_0':<13}{'q_lit':>9}{'tempered':>10}{'+ tilt':>9}{'+ width':>9}{'model':>9}")
    for name, row in part_d.items():
        print(f"    {name:<13}{row['literal_mass']:>9.4f}"
              + "".join(f"{row['mass'][key]:>{10 if key == 'tempered' else 9}.4f}" for key in PARTS))
    print(f"    the tempered control's mode node is ell_0's under every base world prior: "
          f"{all(row['node']['tempered'] == row['k0'] for row in part_d.values())}")
    print()

    n = len(plane)
    print(f"  Block 2. THE PLANE, {n} cells (alpha, Lambda) of Code Cell 4")
    print(f"    largest span-B residual of the utility field: {max(row['residual'] for row in plane):.1e}")
    sign = lambda x: "+" if x > ROUNDOFF else "-" if x < -ROUNDOFF else "0"
    print("    signs of the coefficients (tilt, width): "
          + ", ".join(f"({a},{b}) {sum(1 for row in plane if (sign(row['k'][0]), sign(row['k'][1])) == (a, b))}"
                      for a in "+-" for b in "+-"))
    print(f"    the tempered control's mode node is ell_0's in {sum(1 for row in plane if row['node']['tempered'] == row['k0'])}"
          f" of {n} cells")
    groups = {label: [row for row in plane if direction(row["node"]["model"], row["node"]["tempered"]) == d]
              for label, d in (("up", 1), ("down", -1), ("unmoved", 0))}
    print("    the model's mode against the tempered control's: "
          + ", ".join(f"{label} {len(group)}" for label, group in groups.items()))
    for label, d in (("up", 1), ("down", -1)):
        group = groups[label]
        tilt = sum(1 for row in group if direction(row["node"]["tilt"], row["node"]["tempered"]) == d)
        width = sum(1 for row in group if direction(row["node"]["width"], row["node"]["tempered"]) == d)
        first = sum(1 for row in group if direction(row["slope"][0] + row["slope"][1], 0.0) == d)
        print(f"      {label}: the tilt part alone moves it the same way in {tilt}, the width part alone in {width};"
              f" the sign of the two parts' summed slope agrees in {first} of {len(group)}")
    print("    all-region q-mass below the tempered control's: "
          + ", ".join(f"{label} {sum(1 for row in plane if row['mass'][key] < row['mass']['tempered'] - ROUNDOFF)}"
                      for label, key in (("+ tilt", "tilt"), ("+ width", "width"), ("model", "model"))))
    for label in ("up", "down"):
        print(f"    where the tempered control's mode sits in the {label} cells (s): "
              + ", ".join(f"{s:.4f}" for s in sorted({round(row['s']['tempered'], 4) for row in groups[label]})))
    split_cells = [row for row in plane if row["criteria"]["model"][2] and not row["criteria"]["model"][0]]
    print(f"    the {len(split_cells)} cells where the mass shift criterion is met and the mode shift criterion is not:")
    print(f"      {'alpha':>6}{'Lambda':>7}{'tilt':>9}{'width':>9} | {'mode, s:':<9}{'tempered':>9}{'+ tilt':>8}"
          f"{'+ width':>8}{'model':>8} | {'q-mass:':<8}{'tempered':>9}{'+ tilt':>8}{'+ width':>8}{'model':>8}")
    for row in split_cells:
        print(f"      {row['alpha']:>6.0f}{row['Lambda']:>7.0f}{row['k'][0]:>+9.2f}{row['k'][1]:>+9.2f} | {'':<9}"
              + "".join(f"{row['s'][key]:>{9 if key == 'tempered' else 8}.4f}" for key in PARTS) + f" | {'':<8}"
              + "".join(f"{row['mass'][key]:>{9 if key == 'tempered' else 8}.4f}" for key in PARTS))
    print()

    print("  Block 3. THE HALVING OF Eq. (24): the utility coefficients k against half the literal field's, c/2,")
    print("  with c = B^T W (ell_0 - phi_L). At sigma = 1 and B^T W B = I, Eqs. (15)-(16) give exactly")
    print("  k = (theta_u/2)(mu_u + (theta_u/2) c) / (1 + theta_u^2/2). The limit field is ell_0 - phi_L, halved,")
    print("  plus B c/2.")
    print(f"    B^T W B = I to {max(row['gram_error'] for row in plane):.1e}; k against the exact form to "
          f"{max(row['exact_error'] for row in plane):.1e} (relative)")
    tilt_rel = sorted(row["relative"][0] for row in plane)
    width_rel = sorted(row["relative"][1] for row in plane)
    print(f"    |k - c/2| / |c/2|: tilt largest {tilt_rel[-1]:.2e}, median {tilt_rel[n // 2]:.2e}; "
          f"width largest {width_rel[-1]:.2e}, median {width_rel[n // 2]:.2e}")
    for bound in (1e-1, 1e-2, 1e-3):
        print(f"      cells with both columns within {bound:g} of c/2: "
              f"{sum(1 for row in plane if max(row['relative']) < bound)}")
    print(f"    sign of k = sign of c in both columns: "
          f"{sum(1 for row in plane if all(sign(a) == sign(b) for a, b in zip(row['k'], row['c'])))} of {n}")
    print("    signs of c (tilt, width): "
          + ", ".join(f"({a},{b}) {sum(1 for row in plane if (sign(row['c'][0]), sign(row['c'][1])) == (a, b))}"
                      for a in "+-" for b in "+-"))
    print(f"    the limit field's mode node is the model's in "
          f"{sum(1 for row in plane if row['node']['limit'] == row['node']['model'])} of {n} cells; largest gap "
          f"{max(abs(row['node']['limit'] - row['node']['model']) for row in plane)} nodes")
    for index, label in enumerate(CRITERIA):
        print(f"    the {label} criterion reads the same off the limit field as off the model in "
              f"{sum(1 for row in plane if row['criteria']['limit'][index] == row['criteria']['model'][index])} of {n}")
    print(f"    all four the same: {sum(1 for row in plane if row['criteria']['limit'] == row['criteria']['model'])} of {n}")

    def halving_row(label, row):
        print(f"      {label}{row['theta']:>11.2f}{row['c'][0]:>+10.2f}{row['c'][1]:>+10.2f}{row['k'][0]:>+10.2f}"
              f"{row['k'][1]:>+10.2f}{row['relative'][0]:>9.1e}{row['relative'][1]:>9.1e}{row['s']['model']:>8.4f}"
              f"{row['s']['limit']:>9.4f}{row['mass']['model']:>8.4f}{row['mass']['limit']:>9.4f}"
              f"{'yes' if row['criteria']['limit'] == row['criteria']['model'] else 'no':>12}")

    columns = (f"{'theta_u*':>11}{'c tilt':>10}{'c width':>10}{'k tilt':>10}{'k width':>10}{'rel tilt':>9}"
               f"{'rel wid':>9}{'mode':>8}{'mode lim':>9}{'mass':>8}{'mass lim':>9}{'same four':>12}")
    print("    the ten cells furthest from the halving (the larger of the two relative distances):")
    print(f"      {'alpha':>6}{'Lambda':>7}" + columns)
    for row in sorted(plane, key=lambda row: -max(row["relative"]))[:10]:
        halving_row(f"{row['alpha']:>6.0f}{row['Lambda']:>7.0f}", row)
    print(f"    Part D's base world priors at Lambda = {net.lexical_strength:g} (modes in s; \"lim\" is the limit field):")
    print(f"      {'p_0':<13}" + columns)
    for name, row in part_d.items():
        halving_row(f"{name:<13}", row)
    print()


appendix_c_report(evaluation_network)
utility_split_report(evaluation_network, part_d_priors(), sweep_grids["alphas"], sweep_grids["lambdas"])
````

## <a id="appd" name="appd"></a>Appendix D: why emission is exclusion, and where $\ell_0$ enters

Eq. (5) fixes each entry by the states it **rules out**. Fixing it by the states it admits is the
more immediate choice, and this appendix says why the model does not take it. The reason cannot be
that the two representations give different beliefs, since Section 1 shows they do not; nor is it a
property of the two families of sets, which Section 2 shows behave symmetrically. It is a property of the
map that carries an entry into a field, and Section 3 states it.

---

### <a id="appd-1" name="appd-1"></a>1. The two representations, and what the network sees of the difference

Write $T_y\subseteq[0,1]$ for the **truth set** of an utterance, the states at which it is true, keeping $E_y$ for the exclusion set of Eq. (5). The two are complements, $E_y=[0,1]\setminus T_y$, so
on the scale of Eq. (5)

$$T_{\textit{no}}=\{0\},\qquad T_{\textit{some}}=(0,1],\qquad T_{\textit{all}}=\{1\}.
\tag{D1}$$

Read a set as the mask that keeps it, acting on propositions, the subsets $A\subseteq[0,1]$: for
$X\subseteq[0,1]$, the map $h_X(A)=A\cap X$ restricts each proposition to $X$, a homomorphism of
Boolean algebras from the subsets of $[0,1]$ onto those of $X$. (Acting on states instead, as the
identity on $X$ and $0$ elsewhere, the mask would send the state $0$ to $0$ wherever $0\in X$, and so
into its own kernel.) Its **kernel** is what the mask destroys, the propositions it sends to
$\varnothing$: the subsets of $[0,1]\setminus X$, an ideal generated by the set's complement. Write
$\ker X=[0,1]\setminus X$ for that generator, so each family is the other's kernel,

$$E_y=\ker T_y,\qquad T_y=\ker E_y .
\tag{D2}$$

On the grid, where an entry is its indicator $\chi_y$ (Eq. 6), the same kernel is linear.
Multiplication by $\chi_y$, $\operatorname{diag}(\chi_y)$ on $\mathbb R^K$, sends to zero exactly the
fields that vanish where $\chi_y=1$, and $\operatorname{diag}(\mathbf 1-\chi_y)$ is the orthogonal
projector onto them, under $W$'s inner product as well, since $W$ is diagonal. Write
$\ker\chi_y=\mathbf 1-\chi_y$ for the indicator of the nodes those fields occupy, the grid form of
$\ker E_y$, so that $\chi+\ker\chi=\mathbf 1$.

**What the inventory contains, stipulated.** This phase takes the inventory to hold at least one
entry $\chi$ together with $\ker\chi$, the pair $\{\chi,\ker\chi\}$. That is a stipulation of this phase and is labelled
as one: nothing here derives why a lexicon contains the entries it does. Two things follow that the
prose should not blur. An **antonym is never needed to define an entry**: $\ker$ is, and $\ker$ is
an involution where antonymy is not, since $\mathrm{ant}(\mathrm{ant}(x))$ need not be $x$. And the
pair belongs to the stipulated inventory whether or not a word lexicalizes its second member.

**The evaluation's ensemble meets the stipulation with *some* and *no*.** `UTTERANCES`
is $\{\textit{no},\textit{some},\textit{all}\}$, and $\ker E_{\textit{some}}=E_{\textit{no}}$, so
$\{\textit{some},\textit{no}\}$ is a pair; *all* is a further member, its kernel
$\ker E_{\textit{all}}=\{1\}$ absent. That kernel is the O corner, which no natural language
lexicalizes and which costs the architecture no dimension (Appendix C §4); added to the ensemble,
it would still change $\sum_y c_y$, and so $\theta_u^\ast$ (Eq. B2). Appendix F's ensembles hold one pair per class. Where this phase's single $\theta_u$ makes $\mathrm{ant}(x)$ coincide with
$\ker(x)$ for the absolute classes, that is a **prediction of this implementation**, not a thesis
about antonymy.

**No quantity the model reports separates them.** Writing the
entry as $\varphi_L=\Lambda\chi_{T_y}$ and making $g_L$ order-preserving, $g_L(\varphi_S)=\varphi_S-\ell_0$,
reproduces the model: $\varphi_u^\ast$ agrees to $5.6\times10^{-16}$, $c_y$ to $1.4\times10^{-14}$, and
$\theta_u^\ast$ is $-28.43749$ under both. Eq. (21) is untouched, since $g_L'=+I$ contributes
$\lVert v\rVert^{2}/\sigma_L$ to the Hessian exactly as $-I$ does, so every proof of §8 stands. The
settled field moves by a constant, writing $\vert_E$ and $\vert_T$ for a quantity under the two
conventions,

$$\varphi_S^\ast\big\vert_{T}=\varphi_S^\ast\big\vert_{E}
+\frac{\sigma_S\Lambda}{\sigma_L+\sigma_S}\,\mathbf 1 ,
\tag{D3}$$

$+4.000000$ at the settings of Text cell 3 §3, which Eq. (12) normalizes away: the read-out agrees to
$8.9\times10^{-16}$, and with it every statistic of Eqs. (25)–(27). The settled *states* do move:
at the learned $\theta_u^\ast=-28.43749$, $\lVert r_S\rVert$ runs $22.47,\,15.76,\,22.47$ under
$E$ against $8.88,\,3.18,\,8.88$ under $T$ (at a fixed $\theta_u=1$ they are
$23.19,\,18.08,\,23.10$ and $10.56,\,9.43,\,10.36$, so the gap the convention opens is wider at the
learned value, not narrower). Eq. (D3) confines that movement to the constant direction, and
Section 3 turns on exactly this.

The other visible difference is the sign of $g_L'$: the lexical error reaches $\varphi_S$
excitatorily where Eq. (9) has it reach inhibitorily. Eq. (19) reads
$\tau_\varphi\dot\varphi_S=-\varepsilon_S\mp\varepsilon_L$ under the two, so the sign sits in the
return weight rather than in what the unit reads, and both are local on the terms §7 already
establishes. The choice is not settled there either.

### <a id="appd-2" name="appd-2"></a>2. As sets, $E$ and $T$ are symmetric

Complementation is an order-anti-isomorphism of the Boolean algebra of subsets of $[0,1]$
(Davey & Priestley, 2002, pp. 14, 22), so every structural property of one family transfers to the other with
$\cup$ and $\cap$ exchanged and $\subseteq$ reversed. Three instances:

- **Entailment.** *all* entails *some*, which is $T_{\textit{all}}\subseteq T_{\textit{some}}$ and
  equally $E_{\textit{some}}\subseteq E_{\textit{all}}$. The remaining relations of the Aristotelian square read
  the same either way: *no* and *some* are contradictories, being disjoint and exhaustive, and *no*
  and *all* are contraries, being disjoint and not exhaustive.

- **Complementation.** Each family holds exactly one complementary pair, *no* with *some*, and one
  member whose complement falls outside it: $E_{\textit{all}}$, whose complement $\{1\}$ is the
  *not all* of Appendix C §4, and $T_{\textit{all}}$, whose complement is $[0,1)$.

- **The kernel.** By Eq. (D2) the two families close under $\ker$ to precisely the same extent:

| | closes inside the family | escapes it |
|---|---|---|
| $E$ | $\ker E_{\textit{no}}=E_{\textit{some}}$, $\ker E_{\textit{some}}=E_{\textit{no}}$ | $\ker E_{\textit{all}}=\{1\}$ |
| $T$ | $\ker T_{\textit{no}}=T_{\textit{some}}$, $\ker T_{\textit{some}}=T_{\textit{no}}$ | $\ker T_{\textit{all}}=[0,1)$ |

Two of three close in each family, and in each it is the *all* entry that escapes.

**No property of the two families as sets can prefer one of them.** That is a result rather than a
gap: it rules out every argument from the set algebra, and says where the justification has to be
found instead.

### <a id="appd-3" name="appd-3"></a>3. The asymmetry lies in the representation map

By Eq. (D3) the two conventions differ by $\Lambda\mathbf 1=\Lambda\chi_{[0,1]}$, which is the field of a **tautology**, the entry that rules out nothing. Three readings follows from this fact.

**The empty entry is the zero vector.**

| entry | | $\lVert\varphi_L\rVert$ under $E$ | under $T$ |
|---|---|---|---|
| tautology | $T=[0,1]$, $E=\varnothing$ | $0$ | $27.7128$ |
| contradiction | $T=\varnothing$, $E=[0,1]$ | $27.7128$ | $0$ |

Under $E$, "this entry carries no lexical information" and "this entry contributes no lexical field"
are one statement. Under $T$ the semantically empty entry carries the largest field the lexicon can
supply, and the entry true of nothing carries none.

**Field magnitude orders the entailment.**

| | $\lVert\varphi_L\rVert$ under $E$ | under $T$ |
|---|---|---|
| *no* | $23.9199$ | $13.9943$ |
| *some* | $13.9943$ | $23.9199$ |
| *all* | $23.9199$ | $13.9943$ |

*all* and *some* are the one entailment-comparable pair, and under $E$ the entailing member carries
the larger field where under $T$ it carries the smaller. $\Lambda$ times the root quadrature measure
of $E_y$ is not a canonical measure of informativity, so the claim is the limited one: it orders that
pair, and it orders it in opposite directions under the two conventions.

**A tautologous entry leaves the belief where it found it, as a state and not only as a belief.**
Against what the network settles on with no lexical field at all,

$$\max_k\big\lvert\varphi^\ast_{S,k}(\text{tautology})-\varphi^\ast_{S,k}(\text{no lexical field})\big\rvert
=\begin{cases}0 & \text{under }E,\\[2pt] 4.0000 & \text{under }T.\end{cases}
\tag{D4}$$

Under $E$ the two coincide by construction, a tautology having $\varphi_L=0$. Under $T$ they agree in
$q$ and differ in $\varphi_S$, and the difference is invisible only because $B\perp\mathbf 1$, Part A having dropped the constant
column, and because Eq. (12) normalizes.

**The third reading is the decisive one.** §9.1 establishes that $\varphi_S$ is not a log-density defined
up to an additive constant: $\nabla^2\mathcal F$ has no flat direction, the absolute scale is pinned,
and this scale functions as the temperature under Eq. (15). A convention whose adequacy rests on an
additive constant being invisible relies on the property §9.1 declines to grant. Under $E$ the
constant is not invisible; it is absent.

Text cell 3 §4's account of $g_L$, item 2, turns on the same absent constant. The complement is $x\mapsto\mathbf 1-x$ on
indicators and $x\mapsto\ell_0-x$ on log-weights, both order-reversing involutions. An involution has
a fixed point, and $g_L$'s is $\varphi_S=\ell_0/2$, which is what Eq. (15) returns at
$\varphi_L=0,\ \theta_u=0$, so the centre of the reflection is a field the model settles on. Under $T$
the corresponding map is order-preserving, hence a translation, which has no fixed point at all, and
$\ell_0$ enters as an offset that no structure of the map locates. Eq. (9) is derived under $E$ from
where its own centre sits; under $T$ its $-\ell_0$ would have to be argued on separate grounds.

### <a id="appd-4" name="appd-4"></a>4. What the exclusion convention settles, and what it leaves

Secs. 1–3 settle the E–T asymmetry over the representation map. Exclusion is the convention under
which $y\mapsto\varphi_L$ sends the entry that says nothing to the field that does nothing, and the
inhibitory $g_L'=-I$ is understood as a consequence of that choice. A second question about the same
map is left open, and Sec. 5 takes it up: $g_L$ carries $\ell_0$, and nothing above says it had to.

### <a id="appd-5" name="appd-5"></a>5. Where $\ell_0$ enters

Text cell 3 §4, item 2, derives Eq. (9) from the reflection $x\mapsto\ell_0-x$, and Sec. 3 above turns on
where that reflection's centre sits. Neither fixes the *level* at which $\ell_0$ enters. The chain
has a second generative map below the situation level, and $\ell_0$ could be carried there:

$$g_L(\varphi_S)=-\varphi_S,\qquad g_S(\varphi_u)=\ell_0+\theta_u B\varphi_u .
\tag{D5}$$

Eq. (9) and Eq. (D5) are the whole of the choice. A bias on $\varphi_S$ belonging to no $g$ is not a
prediction, so no error unit forms its residual; $\mu_u$ lives in $\mathbb R^m$ and cannot carry a
field; and $g_y$ takes no part in the dynamics while $\varphi_L$ is clamped (Eq. 8). Eq. (D5) has the
Bogacz status Eq. (9) has, his $\Theta h(\varphi)$ acting on $(\varphi_u,1)$ with the tonic unit
supplying $\ell_0$ to the other error node, so one register entry covers both.

Under Eq. (D5) the stationary point of Eq. (13) is

$$\varphi_S^\ast=\frac{\sigma_L\big(\ell_0+\theta_u B\varphi_u\big)-\sigma_S\varphi_L}{S},
\qquad
\Big(I+\frac{\sigma_u\theta_u^{2}}{S}G\Big)\varphi_u^\ast=\mu_u+\frac{\sigma_u\theta_u}{S}c_y,
\qquad c_y=-B^{\mathsf T}W\big(\ell_0+\varphi_L\big),
\tag{D6}$$

with $S=\sigma_L+\sigma_S$ and $G=B^{\mathsf T}WB$ as in Eqs. (15)–(16). Code Cell D builds Eq. (D5)
as a network of its own and solves it from Eq. (13) by the same Newton step Sec. 1 uses, so nothing
is carried over: Eq. (D6) agrees with that solution to $3.6\times10^{-15}$, and its $\theta_u^\ast$
agrees with a bisection on Eq. (20)'s gradient to five decimals.

**One quantity differs.** Set against Eqs. (15)–(16), Eq. (D6) changes the coupling and nothing else:

$$c_y\big\vert_{(9)}-c_y\big\vert_{(D5)}=2B^{\mathsf T}W\ell_0,\qquad
c_y\big\vert_{(9)}+c_y\big\vert_{(D5)}=-2B^{\mathsf T}W\varphi_L .
\tag{D7}$$

Both hold over the three entries, to $2.8\times10^{-14}$ and $1.1\times10^{-14}$. The second says the
entry reaches the utility level alike under either placement; the first says the residue is $\ell_0$ alone, and is one vector whatever was uttered. At $\sigma_L=\sigma_S$ the two placements return the
same $\varphi_S$ from the same $\varphi_u$, to $1.8\times10^{-15}$, and part company only through
$\varphi_u^\ast$; away from $\sigma_L=\sigma_S$ they differ in the field directly, by
$(\sigma_L-\sigma_S)\ell_0/S$.

**What Eq. (9) makes available.** Under Eq. (9), $\sigma_S\to\infty$ sends $\varphi_S^\ast$ to
$\ell_0-\varphi_L$, and Text cell 4 Part C reads its baseline off that limit. Under Eq. (D5),
$\sigma_S\to\infty$ leaves $-\varphi_L$ and $\sigma_L\to\infty$ leaves
$\ell_0+\theta_u B\varphi_u$; $\ell_0-\varphi_L$ is the value of neither limit, nor of any
$\sigma$ between them.

**What the coupling does not carry.** Under either placement $c_y$ is a projection onto $B$, and
$B^{\mathsf T}W\mathbf 1=0$, Part A having dropped the constant column, so $c_y$ is blind to the
constant component of the field it reads. That component is not one number: $\langle\mathbf
1,\ell_0-\varphi_L\rangle_W$ is $-154.5617$ under *no* and *all* and $-107.5217$ under *some*,
which is Sec. 3's ordering of the field norms seen from the constant direction. Eq. (15) carries the
component into $\varphi_S^\ast$ with weight $\sigma_S/S$, and §9.1 leaves no flat direction for it
to vanish along. The model is not blind to it; Eq. (16) is.

**Part D under both placements.** Code Cell D repeats Part D's five base world priors under Eq. (D5), at
Part D's $\Lambda$ and at Text cell 4b's, scoring both against the same $\ell_0-\varphi_L$, which
Eq. (D5) does not itself produce. We note the rows and reserve our position on them.

### <a id="appd-6" name="appd-6"></a>6. A context set and a Bayesian prior as kernels

The accompanying paper (§1.4.3) compares the context set of the grammatical accounts with the Bayesian prior of the probabilistic accounts. This section gives the construction that the comparison rests on. It concerns the two objects of the literature and no quantity of the model.

**The two objects.** A context set is a set of worlds, $C\subseteq W$. A Bayesian prior is a probability measure $P$ on (a $\sigma$-algebra over) $W$. They are different kinds of object, so by definition neither is an instance of the other.

**A many-to-one map.** Every Bayesian prior determines a set, its support (in a finite $W$, the worlds with $P(w)>0$). A set determines only the family of Bayesian priors supported on it, not one of them. The context set is thus the support-level coarsening of a Bayesian prior, and a Bayesian prior is a refinement of a context set by a measure. A uniform Bayesian prior over $C$ is one refinement among many, which the context set does not choose.

**Update commutes with the map.** On a finite $W$, update commutes with the map. Stalnakerian update intersects: $C\mapsto C\cap[\![p]\!]$. Bayesian conditioning gives $P(\cdot\mid[\![p]\!])$, whose support is $\operatorname{supp}P\cap[\![p]\!]$ whenever $P([\![p]\!])>0$. Context-set update is the support of Bayesian conditioning. The two part where $P([\![p]\!])=0$, that is, where $[\![p]\!]$ lies in the kernel the two share when $C=\operatorname{supp}P$ (defined below): conditioning is undefined, while updating the context set with $p$ empties it, which is where contradiction lives. Presupposition failure is different: $p$ is presupposed when $C\subseteq[\![p]\!]$, and the presupposition fails when that inclusion fails while worlds remain in $C$.

**One kernel.** The context set and the Bayesian prior share the same kernel. The kernel of a homomorphism is the part of its domain that it sends to zero, and both objects act as homomorphisms on propositions, the subsets of $W$. The context set acts by restriction, $h_C(A)=A\cap C$, a homomorphism of Boolean algebras from the propositions of $W$ onto those of $C$, and $h_C$ is Stalnakerian update itself. Its kernel is the propositions it sends to $\varnothing$, those incompatible with $C$: the subsets of $W\setminus C$, an ideal generated by $W\setminus C$. A probability measure is not itself a Boolean homomorphism, but the propositions it sends to zero, those with $P(A)=0$, form an ideal, which is the kernel of the quotient homomorphism onto the measure algebra of $P$ (Givant & Halmos, 2009, pp. 153, 178, 291–292). On a finite $W$ that homomorphism is restriction to the support, $h_{\operatorname{supp}P}$, so its kernel is generated by the worlds $P$ assigns probability zero. When the context set is the support of the Bayesian prior, $C=\operatorname{supp}P$, the two kernels are one ideal. The context set and the Bayesian prior then exclude the same propositions and differ only on the worlds they admit, which the Bayesian prior grades and the context set does not. On a continuous space the null ideal need not be generated by a single set of worlds: under a uniform density on $[0,1]$ every world is null but $[0,1]$ is not. The comparison is therefore stated for a finite $W$.

**The same in linear algebra.** On the real-valued functions on a finite $W$, read the context set as multiplication by its indicator, $f\mapsto\mathbf 1_C f$, and the Bayesian prior as multiplication by its mass function, $f\mapsto Pf$. Both maps are linear. The kernel of the first is the functions that vanish on $C$; the kernel of the second is the functions that vanish wherever $P>0$. When $C=\operatorname{supp}P$, both kernels are the subspace of functions supported on $W\setminus C$.

<a id="coded" name="coded"></a>

````python
# === Code Cell D: the numbers Appendix D quotes. EVALUATION ===
#
# Two alternatives to the model are built here, each as a network of its own and each
# solved from its own free energy, Eq. (13): the truth-set convention of Secs. 1-3, and
# the g_S placement of ell_0 of Sec. 5. Both are counterfactual manipulations, not
# controls: no setting of this model produces either.
#
# The truth-set convention is built with phi_L = Lambda chi_T and
# the order-preserving g_L(phi_S) = phi_S - ell_0, and solved from its own free energy,
# Eq. (13), by Newton's method, which is exact for a quadratic. Nothing is carried over
# from the exclusion convention's closed forms, so the agreements below are measured, not
# built in. Its theta_u* is found by bisection on the gradient of Eq. (20) at that
# equilibrium.

import math
import torch


class TruthSetNetwork(LexicalPredictiveCodingNetwork):
    """Appendix D's alternative: each entry fixed by its truth set T_y (Eq. D1).

    Construct it with a numeric theta_u: the parent's closed forms and its learned
    theta_u* are the exclusion convention's, and are not used here."""

    def lexical_field(self, utterance):
        return self.lexical_strength * (1.0 - exclusion_indicator(
            utterance, self.zeta, self.theta_L, sharpness=self.exclusion_sharpness))

    def predict_lexical(self, phi_S):
        return phi_S - self.base_log_prior


def settle_by_newton(net, phi_L, theta_u):
    """The maximizer of Eq. (13) over (phi_S, phi_u): a Newton step and one refinement."""
    size = net.num_nodes

    def energy(x):
        return net.free_energy(x[:size], x[size:], phi_L, theta_u=theta_u)

    x = torch.zeros(size + net.num_basis, dtype=net.dtype)
    hessian = torch.autograd.functional.hessian(energy, x, vectorize=True)
    for _ in range(2):
        x = x - torch.linalg.solve(hessian, torch.autograd.functional.jacobian(energy, x))
    return x[:size], x[size:]


def learned_by_bisection(net, bisections=40):
    """theta_u* as the zero of Eq. (20)'s gradient at the Newton equilibrium."""
    def gradient(theta):
        total = 0.0
        for y in UTTERANCES:
            phi_L = net.lexical_field(y)
            phi_S, phi_u = settle_by_newton(net, phi_L, theta)
            _, r_state, _ = net.residuals(phi_S, phi_u, phi_L, theta_u=theta)
            total += float(net.dot(phi_u, net.project(r_state / net.sigma_state)))
        return total / len(UTTERANCES)

    sign = 1.0 if gradient(0.0) > 0 else -1.0
    low, high = 1e-3, 1e6
    for _ in range(bisections):
        middle = math.sqrt(low * high)
        if sign * gradient(sign * middle) > 0:
            low = middle
        else:
            high = middle
    return sign * math.sqrt(low * high)


def appendix_d_report(net):
    star = net.theta_u
    truth = TruthSetNetwork(theta_u=star)
    solver_gap = max(float((settle_by_newton(net, net.lexical_field(y), star)[0]
                            - net.closed_form_fixed_point(y)[0]).abs().max()) for y in UTTERANCES)
    print("APPENDIX D, Sec. 1: THE TRUTH-SET CONVENTION AGAINST THE EXCLUSION CONVENTION")
    print(f"  (Newton's method reproduces the exclusion convention's closed form to {solver_gap:.1e})")
    print(f"  theta_u*: {learned_by_bisection(truth):.5f} under T, {star:.5f} under E")

    utility_gap = coupling_gap = readout_gap = 0.0
    shifts = []
    for y in UTTERANCES:
        phi_S_E, phi_u_E = net.closed_form_fixed_point(y)
        phi_S_T, phi_u_T = settle_by_newton(truth, truth.lexical_field(y), star)
        utility_gap = max(utility_gap, float((phi_u_T - phi_u_E).abs().max()))
        coupling_gap = max(coupling_gap, float((
            truth.project(truth.base_log_prior + truth.lexical_field(y))
            - net.project(net.base_log_prior - net.lexical_field(y))).abs().max()))
        readout_gap = max(readout_gap, float((truth.read_out(phi_S_T)
                                              - net.read_out(phi_S_E)).abs().max()))
        shifts.append(phi_S_T - phi_S_E)
    shifts = torch.cat(shifts)
    predicted = net.sigma_state * net.lexical_strength / (net.sigma_lexical + net.sigma_state)
    print(f"  at theta_u*: phi_u* agrees to {utility_gap:.1e}; c_y, as B^T W (ell_0 + phi_L) under T,"
          f" to {coupling_gap:.1e}")
    print(f"  phi_S*(T) - phi_S*(E) = {float(shifts.mean()):+.6f} everywhere to "
          f"{float(shifts.max() - shifts.min()):.1e}, against Eq. (D3)'s "
          f"sigma_S Lambda / (sigma_L + sigma_S) = {predicted:+.6f}")
    print(f"  the read-out agrees to {readout_gap:.1e}")
    for label, theta in (("learned theta_u*", star), ("theta_u = 1, a control", 1.0)):
        exclusion = net.respawn(theta_u=theta)
        truth_theta = TruthSetNetwork(theta_u=theta)
        under_E, under_T = [], []
        for y in UTTERANCES:
            phi_S, phi_u = exclusion.closed_form_fixed_point(y)
            under_E.append(float(exclusion.squared_norm(phi_S - exclusion.predict_state(phi_u))) ** 0.5)
            phi_S, phi_u = settle_by_newton(truth_theta, truth_theta.lexical_field(y), theta)
            under_T.append(float(truth_theta.squared_norm(phi_S - truth_theta.predict_state(phi_u))) ** 0.5)
        print(f"  ||r_S|| at the {label}: E " + ", ".join(f"{v:.2f}" for v in under_E)
              + "   T " + ", ".join(f"{v:.2f}" for v in under_T) + "   (no, some, all)")
    print()

    print("  Sec. 3: FIELD NORMS UNDER THE TWO CONVENTIONS")
    everything = net.lexical_strength * torch.ones_like(net.zeta)
    print(f"    tautology      E {0.0:.4f}   T {float(net.squared_norm(everything)) ** 0.5:.4f}")
    print(f"    contradiction  E {float(net.squared_norm(everything)) ** 0.5:.4f}   T {0.0:.4f}")
    for y in UTTERANCES:
        print(f"    {y:<14} E {float(net.squared_norm(net.lexical_field(y))) ** 0.5:.4f}   "
              f"T {float(truth.squared_norm(truth.lexical_field(y))) ** 0.5:.4f}")
    nothing = torch.zeros_like(net.zeta)
    gap_E = float((settle_by_newton(net, nothing, star)[0]
                   - settle_by_newton(net, nothing, star)[0]).abs().max())
    gap_T = float((settle_by_newton(truth, everything, star)[0]
                   - settle_by_newton(truth, nothing, star)[0]).abs().max())
    print(f"    Eq. (D4), max |phi_S*(tautology) - phi_S*(no lexical field)|: "
          f"E {gap_E:.4f} (the tautology's field is zero), T {gap_T:.4f}")
    print()



# --- Sec. 5: where ell_0 enters --------------------------------------------------------
#
# Eq. (9) puts ell_0 in g_L. It could have sat in g_S instead (decision A3), and that
# alternative is built below as a network of its own and solved from its own free energy,
# Eq. (13), so the comparisons are measured rather than asserted. It is a COUNTERFACTUAL
# MANIPULATION and not a control: no setting of this model produces it.


class UtilityPlacementNetwork(LexicalPredictiveCodingNetwork):
    """The alternative of Sec. 5: ell_0 carried by g_S rather than by g_L.

        g_L(phi_S) = -phi_S,        g_S(phi_u) = ell_0 + theta_u B phi_u

    Everything else is the model's: the same F, chain, B, mu_u, sigma, read-out and
    exclusion convention. Construct it with a numeric theta_u; the parent's closed forms
    and its learned theta_u* belong to Eq. (9)'s placement, and are not used here."""

    def predict_lexical(self, phi_S):
        return -phi_S

    def predict_state(self, phi_u, theta_u=None):
        theta = self.theta_u if theta_u is None else float(theta_u)
        return self.base_log_prior + theta * torch.mv(self.basis, phi_u)


def utility_placement_coupling(net, utterance):
    """c_y under the alternative: -B^T W (ell_0 + phi_L), against B^T W (ell_0 - phi_L)."""
    return -net.project(net.base_log_prior + net.lexical_field(utterance))


def utility_placement_fixed_point(net, utterance, theta_u):
    """The stationary point of Eq. (13) under the alternative placement.

    dF/dphi_S = 0 gives phi_S* = [sigma_L (ell_0 + theta B phi_u) - sigma_S phi_L] / S and
    dF/dphi_u = 0 gives (I + sigma_u theta^2 G / S) phi_u* = mu_u + sigma_u theta c / S:
    the forms of Eqs. (15)-(16) with c as above, and with ell_0 weighted by sigma_L where
    Eq. (15) weights it by sigma_S. Checked against Newton on Eq. (13) below."""
    phi_L = net.lexical_field(utterance)
    total_sigma = net.sigma_lexical + net.sigma_state
    gram = net.basis.T @ (net.weights[:, None] * net.basis)
    matrix = (torch.eye(net.num_basis, dtype=net.dtype, device=net.device)
              + net.sigma_utility * theta_u ** 2 * gram / total_sigma)
    coupling = utility_placement_coupling(net, utterance)
    rhs = net.mu_u + net.sigma_utility * theta_u * coupling / total_sigma
    phi_u = torch.linalg.solve(matrix, rhs)
    phi_S = (net.sigma_lexical * (net.base_log_prior + theta_u * torch.mv(net.basis, phi_u))
             - net.sigma_state * phi_L) / total_sigma
    return phi_S, phi_u


def theta_star_over_couplings(net, couplings):
    """Eq. (B2)'s maximizer with c_y supplied rather than computed.

    `theta_u_stationary_points` builds c_y = B^T W (ell_0 - phi_L) inline, so the
    alternative placement's c_y cannot be put through it. This form takes the couplings
    as an argument; fed the model's own it must return learned_theta_u() exactly, and the
    report below prints that check before using it on the alternative."""
    total_sigma = net.sigma_lexical + net.sigma_state
    total = torch.stack(couplings).sum(dim=0)
    a = net.sigma_utility * float(net.dot(net.mu_u, total))
    b = (len(couplings) * total_sigma * float(net.dot(net.mu_u, net.mu_u))
         - net.sigma_utility * sum(float(net.dot(c, c)) for c in couplings))
    c0 = -total_sigma * a / net.sigma_utility
    quotient = -0.5 * (b + math.copysign(math.sqrt(b * b - 4.0 * a * c0), b))
    roots = (quotient / a, c0 / quotient)
    return roots[0] if (roots[0] > 0) == (a > 0) else roots[1]


def ell0_placement_report(net):
    couplings = {y: net.project(net.base_log_prior - net.lexical_field(y)) for y in UTTERANCES}
    alternative = {y: utility_placement_coupling(net, y) for y in UTTERANCES}
    constant = torch.ones_like(net.zeta)

    print("  Sec. 5: WHERE ell_0 ENTERS, Eq. (9)'s g_L AGAINST g_S")
    print("    A counterfactual manipulation, not a control: no setting of this model")
    print("    produces g_L(phi_S) = -phi_S with g_S(phi_u) = ell_0 + theta_u B phi_u.")

    # --- self-checks, before anything is read off the alternative ---
    placement = UtilityPlacementNetwork(theta_u=1.0)
    star_here = theta_star_over_couplings(net, [couplings[y] for y in UTTERANCES])
    star_there = theta_star_over_couplings(placement, [alternative[y] for y in UTTERANCES])
    newton_gap = 0.0
    for y in UTTERANCES:
        phi_S, phi_u = utility_placement_fixed_point(placement, y, 1.0)
        by_newton = settle_by_newton(placement, placement.lexical_field(y), 1.0)
        newton_gap = max(newton_gap, float((phi_S - by_newton[0]).abs().max()),
                         float((phi_u - by_newton[1]).abs().max()))
    by_bisection = learned_by_bisection(UtilityPlacementNetwork(theta_u=star_there))
    print(f"    checks: Eq. (B2) over given couplings returns {star_here:.5f} for the model "
          f"({net.learned_theta_u():.5f} from the parent);")
    print(f"            its closed form matches Newton on Eq. (13) to {newton_gap:.1e};")
    print(f"            its theta_u* is {star_there:.5f} in closed form and {by_bisection:.5f} "
          f"by bisection on Eq. (20)")
    print()

    # --- 1. the coupling ---
    print("    (1) THE COUPLING, THE ONE QUANTITY THE PLACEMENT CHANGES (Eq. 16)")
    print(f"        {'utterance':<12}{'c_y under g_L':>26}{'c_y under g_S':>26}")
    for y in UTTERANCES:
        show = lambda c: "[" + ", ".join(f"{float(v):+9.4f}" for v in c) + "]"
        print(f"        {y:<12}{show(couplings[y]):>26}{show(alternative[y]):>26}")
    prior_part = 2.0 * net.project(net.base_log_prior)
    difference = max(float((couplings[y] - alternative[y] - prior_part).abs().max())
                     for y in UTTERANCES)
    summed = max(float((couplings[y] + alternative[y]
                        + 2.0 * net.project(net.lexical_field(y))).abs().max())
                 for y in UTTERANCES)
    print(f"        c(g_L) - c(g_S) = 2 B^T W ell_0 to {difference:.1e}, the same for every entry")
    print(f"        c(g_L) + c(g_S) = -2 B^T W phi_L to {summed:.1e}, so the entry enters both alike")
    tidy = [0.0 if abs(float(v)) < 5e-5 else float(v) for v in prior_part]
    print("        2 B^T W ell_0 = [" + ", ".join(f"{v:+.4f}" for v in tidy)
          + "]: the whole of the difference is the sign ell_0 carries")
    print()

    # --- 2. the coupling is a rank-2 projection ---
    offset = 3.7
    print("    (2) THE COUPLING IS A RANK-2 PROJECTION, AND IT DISCARDS THE CONSTANT")
    print(f"        B^T W 1 = [" + ", ".join(f"{float(v):+.1e}" for v in net.project(constant))
          + "], so c_y cannot see the constant component of the")
    print("        field it reads, and that component is not the same for every entry:")
    heading = f"c_y under ell_0 - phi_L + {offset} * 1"
    print(f"        {'utterance':<12}{'<1, ell_0 - phi_L>_W':>22}{heading:>36}")
    for y in UTTERANCES:
        field = net.base_log_prior - net.lexical_field(y)
        blind = float((net.project(field + offset * constant) - couplings[y]).abs().max())
        print(f"        {y:<12}{float(net.inner(constant, field)):>22.4f}"
              f"{f'unchanged to {blind:.1e}':>36}")
    print(f"        Eq. (15) carries that component into phi_S* with weight sigma_S / S = "
          f"{net.sigma_state / (net.sigma_lexical + net.sigma_state):.4f}; Eq. (16) drops it.")
    print("        The model is not blind to the constant (Sec. 9.1: no flat direction); the")
    print("        coupling is.")
    print()

    # --- 3. is the literal listener a fixed point ---
    print("    (3) IS THE LITERAL LISTENER ell_0 - phi_L A FIXED POINT OF THE NETWORK?")
    print("        max |phi_S* - (ell_0 - phi_L)|, at theta_u = 1, a control:")
    print(f"        {'sigma':<22}{'under g_L':>14}{'under g_S':>14}")
    for label, lexical, state in (("sigma_S = 1e6", 1.0, 1e6), ("sigma_L = 1e6", 1e6, 1.0)):
        limit = net.respawn(sigma_lexical=lexical, sigma_state=state)
        literal = limit.literal_fixed_point("some")
        here = float((limit.closed_form_fixed_point("some", theta_u=1.0)[0] - literal).abs().max())
        there = float((utility_placement_fixed_point(limit, "some", 1.0)[0] - literal).abs().max())
        print(f"        {label:<22}{here:>14.1e}{there:>14.1e}")
    print("        (Under g_S, sigma_S -> infinity leaves -phi_L and sigma_L -> infinity leaves")
    print("         ell_0 + theta_u B phi_u: the baseline of Text cell 4 Part C is reachable at")
    print("         no sigma at all.)")
    print()

    # --- 4. same phi_u, and sigma_L != sigma_S ---
    print("    (4) THE TWO AGREE GIVEN phi_u, AND PART COMPANY IN phi_u* ITSELF")
    for label, lexical in (("sigma_L = sigma_S = 1", 1.0), ("sigma_L = 2, sigma_S = 1", 2.0)):
        pair = net.respawn(sigma_lexical=lexical)
        total_sigma = pair.sigma_lexical + pair.sigma_state
        held = utility_gap = field_gap = 0.0
        for y in UTTERANCES:
            phi_S, phi_u = pair.closed_form_fixed_point(y, theta_u=1.0)
            same_phi_u = (pair.sigma_lexical * (pair.base_log_prior + torch.mv(pair.basis, phi_u))
                          - pair.sigma_state * pair.lexical_field(y)) / total_sigma
            other_S, other_u = utility_placement_fixed_point(pair, y, 1.0)
            held = max(held, float((phi_S - same_phi_u).abs().max()))
            utility_gap = max(utility_gap, float((phi_u - other_u).abs().max()))
            field_gap = max(field_gap, float((phi_S - other_S).abs().max()))
        predicted = float(((pair.sigma_lexical - pair.sigma_state)
                           * pair.base_log_prior / total_sigma).abs().max())
        print(f"        {label:<24} at the same phi_u {held:.4g} (predicted "
              f"(sigma_L - sigma_S)|ell_0|/S {predicted:.4g});")
        print(f"        {'':<24} phi_u* differs by {utility_gap:.4f} and phi_S* by {field_gap:.4f}")
    print("        (at theta_u = 1, a control)")
    print()

    # --- 5. Part D's base world priors under both placements ---
    print(f"    (5) PART D'S PRIORS UNDER BOTH PLACEMENTS, AT LAMBDA = 8 AND {STRONG_LAMBDA:.0f}")
    print("        closed form at each configuration's own theta_u*; shift = q_H - q_lit for")
    print("        \"some\", both placements read against ell_0 - phi_L, which the g_S placement")
    print("        cannot itself produce (3)")
    rows = {**part_d_priors(), "delta-like": beta_world_prior(DELTA_ALL_ALPHA, 1.0)}
    for strength in (net.lexical_strength, STRONG_LAMBDA):
        print(f"        Lambda = {strength:.0f}")
        print(f"        {'p_0':<13}{'theta* g_L':>12}{'theta* g_S':>12}{'q_lit':>9}"
              f"{'q_H g_L':>9}{'q_H g_S':>9}{'shift g_L':>11}{'shift g_S':>11}"
              f"{'both g_L':>10}{'both g_S':>10}")
        for name, prior in rows.items():
            row = net.respawn(base_prior=prior, lexical_strength=strength)
            here = theta_star_over_couplings(
                row, [row.project(row.base_log_prior - row.lexical_field(y)) for y in UTTERANCES])
            there = theta_star_over_couplings(
                row, [utility_placement_coupling(row, y) for y in UTTERANCES])
            under_g_L = criterion_for_some(row, here)
            under_g_S = criterion_for_some(
                row, belief=row.read_out(utility_placement_fixed_point(row, "some", there)[0]))
            print(f"        {name:<13}{here:>12.2f}{there:>12.2f}"
                  f"{under_g_L['upper'] - under_g_L['shift']:>9.4f}"
                  f"{under_g_L['upper']:>9.4f}{under_g_S['upper']:>9.4f}"
                  f"{under_g_L['shift']:>+11.4f}{under_g_S['shift']:>+11.4f}"
                  f"{'met' if under_g_L['both'] else 'not met':>10}"
                  f"{'met' if under_g_S['both'] else 'not met':>10}")
    print()


appendix_d_report(evaluation_network)
ell0_placement_report(evaluation_network)
````

## <a id="appf" name="appf"></a>Appendix F: the scale-class comparison with Xiang et al. (2022)

Appendix A fixes $\theta_L$ by the predicate's own resolution, $\theta_L=\log(2n-1)$ (Eq. A5), and
leaves open what fixes $n$ for a predicate with no atoms to count. This appendix runs the model over
a ladder of such resolutions on the 96 items of Xiang, Kennedy, Xu & Leffel (2022), with those
authors' own elicited priors, and reports where its predictions match their measurements and where
they do not. It states the class-to-entry map it runs, reports the comparison, and stops there.
**Nothing below says what any mismatch is due to.** The two conjectures behind the set-up are stated
in the accompanying paper (§§5.3.1–5.3.2) and not here, and this appendix neither adopts nor rejects
them.

**Why, and not merely that.** $n$ is swept below and not set, because the architecture clamps
$\varphi_L$ and cannot yet vary $n$ within the model: each run fixes one $n$, and the sweep varies it
only across runs. Each class carries one entry of Eq. (A1), and one $\theta_L$ fixes both of that
entry's endpoints. How stably a predicate fixes its resolution, which is what Conjecture I concerns
(the accompanying paper, §5.3.1), is therefore not a quantity of any single run. A result obtained under commitments this project expects
to revisit is evidence about **this** phase of the model, and the tables below are reported in that
spirit.

---

### <a id="appf-1" name="appf-1"></a>1. The class-to-entry map

**The map is stipulated.** Each scale class is given an entry of Eq. (A1), read in that adjective's
own orientation:

$$\textit{maximum-standard}\mapsto\chi_{\textit{all}},\qquad
\textit{minimum-standard}\mapsto\chi_{\textit{some}},\qquad
\textit{relative}\mapsto\mathbb 1[\zeta<t] .
\tag{F1}$$

The ground of the stipulation is given in the accompanying paper (§4.2.1). The relative class's
entry is written for completeness and is not modelled (§2).

**The endpoint orientation is the architecture's scale structure.** Xiang et al. relate the
endpoint-oriented thresholds of the absolute classes to a semantic convention for closed-scale
adjectives (pp. 42, 47).
In this architecture that orientation is not added to the model: it is its scale structure, whose
entries are anchored at the ends of the closed proportion scale as the Voronoi cells of the counts $0$
and $n$. $\theta_L$ itself is not a convention; it is a function of the atomicity $n$ (Eq. A5).

**$n$ and $\Lambda$ are fitted here, and here only.** Nothing else in this notebook is fitted.
Conjecture I (the accompanying paper, §5.3.1) is a claim about atomicity and lexical strength
together. In the model's quantities atomicity is $n$, the count Eq. (A5) turns into a threshold, and
lexical strength is $\Lambda$, the scalar of $g_y(\varphi_L)$ that Text cell 3 §2 carries under that
name, so a comparison that did not vary both could not bear on it. Code Cell F therefore sweeps both, locates the fit per class, and labels the result as
fitted wherever it prints it. The fit is one $(n,\Lambda)$ per class and not one per (class, image
type): a constrained model should carry as few fitted quantities as it can, and the second fit is
declined rather than argued against.

---

### <a id="appf-2" name="appf-2"></a>2. What is instantiated, and what is not

**The relative class is not modelled.** Its entry in Eq. (F1) needs a cut $t$ that is neither
endpoint, and this phase fixes both endpoints with the one $\theta_L$; a context threshold is a
quantity of the entry that Text cell 3 §2's inventory does not have. Its 40 items stay in the data
file, carry no prediction, and appear in no table below. What follows covers the two absolute
classes, 56 items.

**$n$ is swept, not set.** $n$ is the predicate's atomicity. The five-point measurement does not fix
it, and for gradable adjectives it is unstable, in the sense of Chierchia (2010) that the accompanying
paper follows (§5.3.1). Whether the options a task offers can affect an
unstable atomicity is a fair question, but the translation does not assume that they fix it. $n$ and
$\Lambda$ are swept together on powers of two: $n\in\{2^2,\dots,2^7\}$ and $\Lambda\in\{2^2,\dots,2^{10}\}$,
at the grid half-width $Z=6$ of every other cell. $n=2^8$, $2^9$ and $2^{10}$ are outside the ladder,
because Eq. (A5) then gives $\theta_L=6.2364$, $6.9305$ and $7.6241$, none below $Z$ (block F.1). The
default $n=10$ of every other cell in this notebook is untouched.

**The five positions: partition R1.** The authors elicit priors and posterior degrees over five
unlabelled images per scale. On the absolute scales one extreme image is exactly degree zero (a
straight nail, a plain circle): position 5 in the maximum adjective's orientation, position 1 in the
minimum adjective's. That image takes the Voronoi cell of its endpoint count at the given $n$,
$[1-\tfrac1{2n},1]$ in the maximum class's orientation and $[0,\tfrac1{2n}]$ in the minimum class's,
which is exactly the region $\chi_{\textit{all}}$ keeps and $\chi_{\textit{some}}$ excludes. The other
four positions take four cells of equal width over the rest of the scale. At $n=4$ the maximum class's
edges are $0.2188$, $0.4375$, $0.6562$ and $0.8750$. The other extreme image carries no endpoint
status: nothing in the stimuli shows it to be the end of a closed scale. Each minimum item's image set
is its maximum antonym's, with the positions in reverse order, so the two classes' partitions are one
physical partition, mirrored. The partition does two jobs: it is the support of the elicited prior,
and it gives the read-out's five bins. $n$ moves the cells with $\theta_L$, and block F.1 checks at every
$n$ that the five cells partition the grid and that the endpoint cell is the entry's own region, node
for node.

**The ensemble is $\{\chi,\ker\chi\}$.** Eq. (B3) averages over an exposure ensemble, and for each
entry this appendix takes it to hold the entry and its kernel, which is what Appendix D stipulates
for any entry. For $\chi_{\textit{some}}$ the kernel *is* $\chi_{\textit{no}}$, exactly
($\chi_{\textit{no}}+\chi_{\textit{some}}=1$ at every node); for $\chi_{\textit{all}}$ it is a field
no utterance of `UTTERANCES` names. The ensemble is therefore architectural and not lexical: it is
the entry's kernel whether or not English supplies a word for it. For these two classes, in this
phase, the antonym the authors paired each item with **coincides** with that kernel, a consequence
of one $\theta_L$ fixing both endpoints, not a claim about antonymy, which is a lexical accident of
two words sharing a scale.

**The ensemble is the inventory's, not the experiment's.** Membership follows from having met an
entry somewhere, not from having met it in this experiment; the authors' latin square in fact shows
no participant both adjectives of an image set. What a participant was shown constrains nothing
about what their inventory holds.

**The elicited prior becomes $\ell_0$.** A five-cell probability vector is pushed onto the grid as a
density in $s$ that is constant on each cell of R1, times the Jacobian $s(1-s)$ of Eq. (1).

**No published statistic is printed by any cell, and none is compared.** Code Cell F prints this
model's predictions and our own statistics of them against the authors' measurements. The models of
that paper set their thresholds by other means than this architecture, whose $\theta_L$ is fixed by an
atomicity that is swept here, so their statistics and these are not statistics of the same model
quantity and are not set side by side.

### <a id="appf-3" name="appf-3"></a>3. What is measured

For each item, a read-out's profile is its mass in the five cells of R1. Code Cell F reports two
read-outs side by side.

**The literal listener beside the full network.** Experiment 3 asks for a posterior degree ("how ADJ
was the X that the speaker saw?"; Xiang et al. §2.3). Whether that task elicits a literal judgement or
a pragmatic one is not settled here: a question about degree does not by itself call for a utility
level. Every block therefore reports the literal listener $q_{\mathrm{lit}}$, whose field is
$\ell_0-\varphi_L$, beside the full network at $\varphi_S^\ast(\theta_u^\ast)$. A third reference, the
tempered control, is printed where the decomposition needs it (blocks F.4, F.5 and F.10): $\theta_u=0$
at the model's $\sigma$, whose field is $\tfrac12(\ell_0-\varphi_L)$ (Text cell 3 §7). It differs from
$q_{\mathrm{lit}}$ by the finite $\sigma_S$ alone, so the step from $q_{\mathrm{lit}}$ to the tempered
control is the tempering's, and the step from the tempered control to the full network is the utility
level's.

$R^2$ is the squared Pearson correlation over item $\times$ position rows,

$$R^2=\frac{\big(\sum_i (\hat p_i-\bar{\hat p})(p_i-\bar p)\big)^2}
{\sum_i (\hat p_i-\bar{\hat p})^2\ \sum_i (p_i-\bar p)^2},
\tag{F2}$$

computed as the authors compute theirs.

**The plane, and the fit located apart from it.** Block F.2 prints $R^2$ at every $(n,\Lambda)$ of the
two ladders, per class and read-out. Block F.3 then locates the fit between the plane's points: from
the plane's best cell, every integer $n$ from half to twice its $n$, and for each a golden-section
search in $\log\Lambda$ from half to twice its $\Lambda$, both kept within the ladders. The located value
is stated apart from the plane, with a band: the points evaluated within $0.005$ of the located
maximum, the width the search resolves. Block F.3 reads each located coordinate as *interior*, *at
the ladder's end*, *flat over the searched range*, or with its *band reaching the searched range's
edge*. Only an interior coordinate is bounded on both sides by the data. Blocks F.4 to F.8 and F.10
are read at each class's located fit for the full network, the literal listener included.

**One image set drew no Experiment 3 response at all**, for both of its adjectives: `curved_greenline`,
2 of the 56 modelled items. Those rows are scored nowhere: a prediction compared against an absent
measurement read as zero is compared against nothing. This leaves 155 rows in the maximum class and
115 in the minimum class, and Code Cell F prints both counts.

**The half-width.** Block F.11 recomputes the full network's $R^2$ at the located fit for $Z=5$, $6$,
$7$, $8$ and $10$, holding the node spacing as Appendix A does. The maximum class reads $0.992$ to
$0.993$ throughout; the minimum class reads $0.418$ at $Z=5$, $0.456$ at $Z=6$ and $7$, and $0.459$ and
$0.460$ at $Z=8$ and $10$.

### <a id="appf-4" name="appf-4"></a>4. What matches

**The class-to-entry map of Eq. (F1) is what the data also suggest.** Each absolute class's measured
profile is the profile of the entry the map assigns it: the maximum class's mass sits in the
top cell ($0.960$ under shapes, $0.929$ under artifacts), and the minimum class's is spread over every
cell but the bottom one ($0.019$ and $0.015$ there), which is what $\{\zeta\le-\theta_L\}$ excludes.

**The maximum class is fitted almost exactly by either read-out, and fixes neither $n$ nor $\Lambda$.**
The full network's located fit is $R^2=0.993$ at $n=16$, $\Lambda=7.22$, and the literal listener's
is $0.993$ at $n=128$, $\Lambda=7.75$. Neither is located in the sense above: for both, $n$ is flat over
the searched range and the $\Lambda$ band reaches the searched range's upper edge. In the plane, every
cell with $\Lambda\ge16$ reads $0.992$ for both read-outs at every $n$. Below that the full network falls
away at the largest $n$: at $n=128$ it reads $0.829$ at $\Lambda=8$ and $0.276$ at $\Lambda=4$. These data are consistent with the hard entry of the
$\Lambda\to\infty$ limit.

**The minimum class's full network requires a finite $\Lambda$, and its fit sits at the smallest
$n$.** Its located fit is $R^2=0.456$ at $n=4$, the ladder's end, and $\Lambda=23.11$, an interior value
with the band $21.72$ to $26.91$. In the plane the row $n=4$ reaches $0.446$ at $\Lambda=32$, while the
row $n=128$ is still rising at the top of the ladder, $0.342$ at $\Lambda=1024$.

**On the minimum class the full network's fit is earned in two steps, and the utility level's is the
larger.** At that located fit the literal listener reaches $R^2=0.108$, the tempered control $0.253$
and the full network $0.456$: the tempering earns $0.146$ and the utility level $0.203$. On the maximum
class all three reach $0.990$ to $0.993$, which is to say that class is fitted by the entry alone.
Block F.5 also prints the two classes pooled: $0.859$, $0.831$ and $0.769$.

**Every read-out moves up the scale from the elicited prior, as the data do.** In mean scale position
(block F.8), the literal listener moves $+0.67$ and $+1.45$ in the maximum class (shapes, artifacts)
and $+2.24$ and $+1.05$ in the minimum class; the full network $+0.67$, $+1.43$, $+3.05$ and $+2.14$;
the data $+0.60$, $+1.32$, $+2.40$ and $+2.11$. The literal listener's movement barely depends on $n$:
at every $n$ of the ladder, each class at its located $\Lambda$, it stays at $+0.67$ in the maximum class
under shapes, and within $+1.41$ to $+1.45$, $+2.24$ to $+2.26$ and $+1.04$ to $+1.07$ in the other three
conditions.

**The direction of the image-type difference, for the full network.** In the minimum class the
measured mean scale position is lower under shapes than under artifacts, $-0.83$, and the full
network's difference has the same sign, $-0.20$. In the maximum class the data, the literal listener
and the full network are all near zero, $+0.09$, $+0.03$ and $+0.05$.

### <a id="appf-5" name="appf-5"></a>5. What does not match

**The minimum class.** The full network reaches $R^2=0.456$ at its located fit and the literal
listener $0.119$ at its own, against $0.993$ for the maximum class. The literal listener's own fit is
located at $n=9$, with $\Lambda$ flat over the searched range, $32$ to $128$; in the plane it reads
$0.104$ to $0.112$ at every $n$ for every $\Lambda\ge8$. Split by image type, the full network's $R^2$ is
$0.094$ under shapes and $0.879$ under artifacts (block F.10 (c)). The measured profile has its global
maximum at position 3 under shapes and at position 5 under artifacts; the literal listener's is at
position 3 in both conditions, and the full network's at position 5 in both.

**The image-type difference in the minimum class.** Measured, $-0.83$; the full network gives $-0.20$,
the right sign and about a quarter of the magnitude; the literal listener gives $+0.08$, the opposite
sign. The elicited priors differ by $-1.12$.

**The gap between the classes.** Measured, the maximum class sits $+1.33$ above the minimum under
shapes and $+0.41$ under artifacts. The literal listener gives $+1.56$ and $+1.61$, nearly the same in
both conditions, and the full network $+0.75$ and $+0.50$. The elicited priors are $+3.13$ and $+1.21$
apart. All four orderings agree, and the spans do not.

**Displacement from the elicited prior, in the minimum class.** Under shapes the data move $+2.40$ from the
elicited prior, the literal listener $+2.24$ and the full network $+3.05$; under artifacts the data move
$+2.11$, the literal listener $+1.05$ and the full network $+2.14$. The literal listener stays short of
the data in both conditions; the full network overshoots them under shapes and meets them under
artifacts.

These are reported. This appendix offers no account of them.

### <a id="appf-6" name="appf-6"></a>6. How far the utility level reaches

Four facts about $\theta_u^\ast$ in this configuration, all printed in block F.10.

**Its sign is a property of $\ell_0$ alone.** The ensemble is $\{\chi,\ker\chi\}$ and
$\chi+\ker\chi=\mathbf 1$, so $\sum_y\varphi_{L,y}=\Lambda\mathbf 1$ and the $\Lambda$ term
cancels against $B^{\mathsf T}W\mathbf 1=0$ (App. C §2):

$$\sum_y c_y=B^{\mathsf T}W(2\ell_0-\Lambda\mathbf 1)=2B^{\mathsf T}W\ell_0 .
\tag{F3}$$

Eq. (B2) gives $\theta_u^\ast$ the sign of $\langle\mu_u,\sum_y c_y\rangle$, so **that sign is
independent of $n$, of $\Lambda$ and of which entry was uttered**: it is fixed by $\ell_0$. Verified to
$1.9\times10^{-12}$ over six $n$, four $\Lambda$ and all 54 scored items.

**The sign follows $\ell_0$, and a mean over items does not report it.** At the located fit, every
minimum item's $\theta_u^\ast$ is negative (medians $-144.8$ under shapes and $-168.7$ under artifacts),
every maximum item's under shapes is positive (median $+20.9$), and the maximum items under artifacts
split, 7 negative and 9 positive (median $+23.8$). $|\theta_u^\ast|$ diverges near Appendix B's
degenerate ray and reaches $967.7$ here, so one near-ray item can dominate an average; block F.10
reports medians and sign counts instead.

**Where the field has reached Eq. (24)'s limit, the sign cannot reach the read-out**, since that
limit is the same from either sign and the gap to it falls as $1/|\theta_u|$. In the minimum class the
settled field is within $1.7\times10^{-2}$ (shapes) and $1.0\times10^{-2}$ (artifacts) of the limit, in
the maximum class within $3.3\times10^{-2}$ and $9.3\times10^{-2}$, and scoring the limit instead of the
settled field changes no condition's $R^2$ in the third decimal.

**What the amplification does.** Eq. (24)'s limit, $\tfrac12(I+BB^{\mathsf T}W)(\ell_0-\varphi_L)$,
doubles the $\operatorname{span}B$ component of the tempered control $\tfrac12(\ell_0-\varphi_L)$,
which displaces the read-out up the scale. The utility level's displacement is therefore measured
from the tempered control, in mean scale position. In the minimum class it is $+0.58$ under shapes and
$+0.83$ under artifacts, while the data sit $-0.08$ and $+0.80$ from the tempered control; the
tempering's own step, from the literal listener to the tempered control, is $+0.23$ and $+0.25$. In the
maximum class the two steps nearly cancel: the tempering's is $-0.06$ and $-0.22$, the utility level's
$+0.06$ and $+0.20$.
These are reported here; what the accompanying paper makes of them is said there
(§6.1), and not in this appendix.

### <a id="appf-7" name="appf-7"></a>7. The parity of the two entries, at every resolution

Appendix C §5 argues from parity that the extreme-favouring axis is the even coordinate. At every $n$
of the ladder the two absolute entries' loadings $\kappa_y=B^{\mathsf T}W\chi_y$ make that concrete
(block F.9): their tilt (odd) coordinates are identical, differing by at most $1.1\times10^{-16}$, and
their width (even) coordinates sum to zero within $2.8\times10^{-17}$. **The maximum and minimum
entries differ in the even coordinate alone.**

A cut at the midpoint of the log-odds scale carries no even component at all: $\chi=\mathbb
1[\zeta<0]$ is a constant plus a purely odd function, and the basis is orthogonal to the constant by
construction, so its width loading is $1.2\times10^{-17}$ with the node on the cut half-weighted.
Left unweighted the grid reads $0.0194$, which is the node's own mass and not a property of the cut.
Moving the cut off-centre is what gives it a width loading, and both absolute entries have one.

### <a id="appf-8" name="appf-8"></a>8. The antonym relation

**What the entries predict.** At every $n$ of the ladder, $\chi_{\textit{some}}$ is the reflection
$s\mapsto1-s$ of $\ker\chi_{\textit{all}}$, node for node (block F.12 (a): largest difference
$0.0\times10^{0}$). Each minimum item's image set is its maximum antonym's with the positions reversed
(§2), so on one physical scale the minimum adjective's truth region is exactly the maximum adjective's
excluded region: the two regions are complementary, and every degree falls in exactly one. Read at the
five positions of R1, as the share of each cell the entry keeps, the truth values of a maximum
adjective and its minimum antonym sum to $1.000$ at every position and every $n$. Full and empty are
both maximum-standard, so for that pair the entries are contraries: the sum is $1.000$ at the two ends
and $0.000$ between (block F.12 (b)).

**Experiment 2.** Blocks F.12 and F.13 are the only places this notebook reads the truth-value judgements. Reading a
mean judgement as the truth value of the image's cell is this block's linking stipulation. Over the
image sets with both adjectives judged, 11 under shapes and 12 under artifacts (`curved_greenline` has
none), and in the maximum adjective's orientation, the sums at positions 1 to 5 are $0.93$, $0.87$,
$0.84$, $0.80$ and $1.01$ under shapes, and $0.96$, $0.76$, $0.63$, $0.56$ and $1.01$ under artifacts.
The sum is 1 at the shared endpoint and below 1 beneath it, where images are judged to be neither. The
shortfall is the minimum adjective's: at position 4 under artifacts the maximum adjective is judged true
at $0.10$ and the minimum at $0.46$. In full's orientation, the full–empty sums are $1.00$, $0.00$,
$0.00$, $0.18$ and $1.00$ under shapes, and $1.01$, $0.00$, $0.05$, $0.34$ and $0.99$ under artifacts.

**The maximum–minimum pairs do not match the prediction; the full–empty pairs are within $0.05$ of it
except at position 4.** This is reported. This appendix offers no account of it.

### <a id="appf-9" name="appf-9"></a>9. The antonym flip

**The flip.** Each minimum item's image set is its maximum antonym's with the positions reversed
(§2), so the minimum adjective's degree zero is the image at which its antonym is maximal. The
elicited priors are mirrored with it: over the 24 image sets that carry both adjectives, the minimum item's elicited prior is the maximum item's reversed (block F.13 (a): largest difference $0.000$). Full and
empty, both maximum-standard, reverse orientation without a change of class.

**Each class against its own judgements.** Block F.13 builds a literal listener for each item from
the data alone, $L_0^{\mathrm{emp}}$: the elicited prior times the item's mean Experiment 2
judgements, normalized over the five positions. Reading a mean judgement as a truth value is §8's
linking stipulation. No model enters the block. Against Experiment 3, $R^2$ is $0.995$ (shapes) and
$0.960$ (artifacts) in the maximum class, and $0.196$ and $0.261$ in the minimum class (block F.13
(b)). By adjective it is $0.923$ to $0.997$ for every maximum adjective, full ($0.987$) and empty
($0.997$) included, and $0.002$ to $0.571$ for every minimum adjective. In mean scale position, the
measured posterior lies above $L_0^{\mathrm{emp}}$ for every minimum adjective, by $+0.30$ (spotted)
to $+0.70$ (bent), and below it for every maximum adjective, by $0.03$ to $0.13$. By class, the
minimum class's posterior lies $+0.30$ above it under shapes and $+0.78$ under artifacts.

**Through the flip.** Built instead from the antonym's judgements, $1-\mathrm{TVJ}(6-k)$ at position
$k$, the minimum class's $L_0^{\mathrm{emp}}$ reaches $R^2=0.244$ under shapes and $0.062$ under
artifacts, against $0.196$ and $0.261$ from its own judgements. The maximum class's, built from its
minimum antonym's, reaches $0.996$ and $0.655$, against $0.996$ and $0.949$ (block F.13 (c), over
the 11 and 12 antonym pairs, which do not include full and empty).

**The departure goes with the minimum standard, not with the reversal: full and empty reverse
orientation and lie at $L_0^{\mathrm{emp}}$.** Every minimum adjective here has a maximum antonym on
the same images, so these data hold no minimum item without a flip. Each adjective has three to eight
items, Experiments 2 and 3 were run on separate samples of participants, and a mean judgement is itself pooled over
the participants' thresholds. This is reported. This appendix offers no account of it.

### <a id="appf-10" name="appf-10"></a>10. The base world prior varied

class has one predicted profile, the same for both image types. The ground for this control is given
in the accompanying paper (§4.2.1).

**At the located fit.** With $n$ and $\Lambda$ as located on the elicited prior (block F.3), block
F.14 (a) gives $R^2$ under the uniform base world prior beside the elicited one. In the maximum class the full
network reaches $0.992$ against $0.993$, and the literal listener the same. In the minimum class the
full network reaches $0.429$ against $0.456$, the tempered control $0.362$ against $0.253$, and the
literal listener $0.173$ against $0.108$. By image type, the minimum class's full network reaches
$0.069$ under shapes and $0.824$ under artifacts, against $0.094$ and $0.879$.

**Located again.** Block F.14 (b) repeats the plane and the search of §3 under the uniform base world prior. The
maximum class reaches $0.992$ for both read-outs, with $n$ and $\Lambda$ flat over the searched range.
The minimum class's full network reaches $0.478$ at $n=8$ and $\Lambda=11.80$, with $\Lambda$ interior
and $n$ flat over the searched range, $4$ to $8$; its literal listener reaches $0.192$.

**Movement from the base world prior.** The uniform base world prior's mean scale position is $3.00$. At the located fit of
block F.3 the literal listener lies $+1.99$ above it in the maximum class and $+0.48$ in the minimum

### <a id="appf-11" name="appf-11"></a>11. Individual responses

**What is read.** Every block above reads the item aggregate. Block F.15 reads Experiment 2 from the
authors' own file, one row per participant and image (data README). Each participant judged all five
images of an item, so each participant gives one response per item. In the adjective's own orientation
a response is an *interval* if it is true from some position $k$ up to position 5; $k$ is then its cut,
and $k=6$ if it is true of no image. The 48 participant–item cells whose participant id recurs are set
aside wherever a participant's five responses are read together, and kept in the mean judgements, which
reproduce the aggregate's to $4.5\times10^{-7}$. No model enters the block. The ground for reading
individual responses is given in the accompanying paper (§5.3.2).

**The form.** By block F.15 (a), $99.8\%$ of responses are intervals in the maximum class under both
image types, $91.2\%$ (shapes) and $95.9\%$ (artifacts) in the minimum class, and $98.1\%$ and $95.1\%$
in the relative class.

**One place or several.** For each item, block F.15 (b) takes the distribution of its cuts over
$k=1,\dots,6$ and its total-variation distance to the nearest unimodal distribution, one that rises to a
single mode and falls after it. The distance is calibrated by 2,000 draws of the item's size from that
unimodal distribution. At $p<0.05$ unimodality is rejected for 4 of the 12 minimum items under shapes
and 6 of the 12 under artifacts, and for none of the 32 maximum and 40 relative items. Pooled, the
minimum class's cuts lie at $k=2$ with share $0.85$ and at $k=5$ with share $0.12$ under shapes; under
artifacts $0.51$ lie at $k=2$, $0.14$ and $0.12$ at $k=3$ and $4$, and $0.22$ at $k=5$.

**The participants who mark only the top image.** Block F.15 (c) takes the participants who cut at
$k=5$ on at least half the minimum items they answered with an interval: 9 of 56 under shapes, 7 of
them on every such item, and 14 of 57 under artifacts, 8 of them on every such item. They give 30 of
the minimum class's 36 cuts at $k=5$ under shapes and 65 of 71 under artifacts. On relative items they
cut at $k=5$ in 82 of 82 responses (shapes) and 125 of 131 (artifacts); the other participants do so in
227 of 477 and 155 of 411.

**Without them.** By block F.15 (d), unimodality is rejected for no item of any class; one relative
item under artifacts keeps no response. The minimum class's cuts lie at $k=2$ with share $0.94$ under
shapes; under artifacts $0.65$ lie at $k=2$, $0.17$ at $k=3$, $0.13$ at $k=4$ and $0.02$ at $k=5$. The
block lists each minimum item's cuts beside its elicited prior's mean position.

**The antonym sums.** Block F.15 (e) recomputes §8's sums over the same image sets (block F.12 (c)).
Without the participants of (c), positions 1 to 4 sum to $0.98$, $0.97$, $0.95$ and $0.93$ under
shapes, against $0.93$, $0.87$, $0.84$ and $0.80$ with them, and to $0.99$, $0.97$, $0.87$ and $0.77$
under artifacts, against $0.96$, $0.76$, $0.63$ and $0.56$.

**The empirical literal listener.** Block F.15 (f) recomputes §9's $L_0^{\mathrm{emp}}$ (block F.13
(b)). Without the participants of (c), the minimum class's posterior lies $+0.28$ (shapes) and $+0.88$
(artifacts) above it, against $+0.30$ and $+0.78$ with them, and $R^2$ is $0.204$ and $0.145$, against
$0.196$ and $0.261$. In the maximum class the posterior lies $-0.08$ from it under both image types,
against $-0.08$ and $-0.10$.

**Cut by cut.** Block F.15 (g) takes the pairs of (e) cut by cut. If a minimum adjective's truth region
were the complement of its maximum antonym's on the same images, as §8's relation has it, a maximum cut
at $k$ would be a minimum cut at $7-k$. Per image set the block takes the total-variation distance
between the minimum adjective's cuts and the maximum adjective's cuts so mapped, the share of each
adjective's cuts at its modal cut, and whether the two modal cuts differ. Without the participants of
(c) the mean distance is $0.06$ under shapes and $0.35$ under artifacts, against $0.15$ and $0.48$ with
them. The mean share at the modal cut is $0.94$ for the minimum adjectives against $1.00$ for the
maximum adjectives under shapes, and $0.79$ against $0.86$ under artifacts. The two modal cuts differ in
none of the 11 sets under shapes and in 3 of the 12 under artifacts. The block lists the five sets at a
distance of at least $0.40$, all under artifacts. What the comparison bears on is stated in the
accompanying paper (§5.1.3).

This is reported. This appendix offers no account of it.

### <a id="appf-12" name="appf-12"></a>12. Two elicitations of one prior: Degen et al. (2015)

**What is read.** Block F.16 reads the two smoothed elicited priors of Degen, Tessler and Goodman (2015) from the
authors' repository (Degen, 2019), at commit `827ba46`, when the cell runs: the elicited prior of the give-a-number task, which their paper reports and fits, and the elicited prior of the four-step task, which only
the repository holds. The repository carries no licence, so the two files are not kept beside this
notebook, and without a connection the block prints a notice and nothing else. Each elicited prior is a distribution over the sixteen counts, 0 to 15, for each of 90 items. No model enters the block. The
ground for the comparison is given in the accompanying paper (§4.1.1).

**The two elicited priors.** The expected count is $10.21$ on average under the give-a-number task and $10.07$ under the
four-step task; item by item the two expected counts differ by a median of $0.91$ and by at most
$3.94$, of 15. The probability of the all-state, 15 of 15, has a median of $0.352$ under the
give-a-number task and $0.189$ under the four-step task, with largest values $0.828$ and $0.943$, and
exceeds one half in 35 items under the first task and in 16 under the second.

This is reported. This appendix offers no account of it.

````python
# === Code Cell F: the scale classes against Xiang, Kennedy, Xu & Leffel (2022) ===
#
# Prints every number Appendix F and Sec. 4.2 quote about these data, and nothing else. The
# published statistics of that paper are printed by no cell (decision S-5): what is below is
# this model's own predictions, and our own statistics of them against the authors'
# measurements.
#
# SCOPE. The relative class is NOT modelled (decision O14). A relative adjective's cut is
# a context threshold that neither endpoint supplies, and this phase fixes both endpoints
# with the one theta_L; its items stay in the data file but take no model prediction here.
#
# n IS SWEPT, NOT SET (Appendix F Sec. 2). n is the predicate's atomicity, which the
# five-point measurement does not fix; the architecture clamps phi_L and cannot yet vary n
# within the model, so n and Lambda are swept together on powers of two. n stops at 2^7:
# theta_L = log(2n - 1) must stay below the grid half-width Z = 6 (Eq. A5).
#
# THE FIVE POSITIONS (Appendix F Sec. 2, partition R1). The degree-zero image takes the
# Voronoi cell of its endpoint count at the given n; the other four cells are of equal
# width over the rest of the scale. In the maximum class's orientation the endpoint is
# position 5, in the minimum class's position 1. The partition carries the elicited prior
# (constant in s within each cell) and the read-out's bins.
#
# THE PRIOR IS VARIED ONCE (Appendix F Sec. 10, block F.14). Every block but F.14 runs on
# Experiment 1's elicited prior. F.14 is a control: it gives every item the same base world prior,
# uniform over the five positions, and repeats the fits and the movement from the base world prior.
#
# INDIVIDUAL RESPONSES (Appendix F Sec. 11, block F.15). Every block but F.15 reads the item
# aggregate. F.15 reads Experiment 2 participant by participant, from the authors' own file
# (decision B17), and runs no model.
#
# Every helper is local to this cell, so `code cell 1` is untouched.

import csv
import math
import os

XIANG_CSV = os.path.join("data", "xiang_2022", "xiang_items.csv")
EXPT2_CSV = os.path.join("data", "xiang_2022", "Expt2.adjTVJ.csv")   # block F.15 only
# Block F.16 only: Degen, Tessler & Goodman (2015), the authors' repository at a pinned commit.
# The repository carries no licence, so the files are read when the cell runs and are not kept here.
DEGEN_PRIORS_URL = ("https://raw.githubusercontent.com/thegricean/wonky_worlds/"
                    "827ba46f58f18a62d67bbf8c05835bfe590a7b8c/data/")
DEGEN_PRIOR_FILES = (("give-a-number", "priors_gan_smoothed.csv"), ("four-step", "priors_fourstep_smoothed.csv"))

N_LADDER_F = tuple(2 ** k for k in range(2, 8))          # 4 ... 128
LAMBDA_LADDER_F = tuple(2.0 ** k for k in range(2, 11))  # 4 ... 1024
CLASSES_F = ("absolute_max", "absolute_min")
ENTRY_OF_CLASS = {"absolute_max": "all", "absolute_min": "some"}
LEVELS_F = ("q_lit", "model")          # reported side by side (Appendix F Sec. 3)
RESOLUTION_F = 0.005                   # R^2 differences below this are not read
PRIORS_F = ("elicited", "uniform")    # "uniform" is block F.14's control
UNIFORM_PRIOR_F = [0.2] * 5            # the same mass at each of the five positions

_networks_F = {}


def appendix_f_network(n):
    """The evaluation network at n atoms, so theta_L = log(2n - 1) (Eq. A5)."""
    if n not in _networks_F:
        _networks_F[n] = evaluation_network.respawn(num_atoms=n)
    return _networks_F[n]


def partition_edges(cls, n):
    """Partition R1 in s, in the class's own orientation: the endpoint image's cell is
    the Voronoi cell of its endpoint count at n, the other four are of equal width."""
    end = 1.0 / (2 * n)
    width = (1.0 - end) / 4.0
    if cls == "absolute_max":
        return [0.0, width, 2 * width, 3 * width, 1.0 - end, 1.0]
    return [0.0, end, end + width, end + 2 * width, end + 3 * width, 1.0]


def partition_masks(cls, n, zeta):
    """The five cells as node masks on a grid over zeta, with their widths in s."""
    s_edges = partition_edges(cls, n)
    z_edges = ([-math.inf] + [math.log(s / (1.0 - s)) for s in s_edges[1:-1]]
               + [math.inf])
    masks = [(zeta > z_edges[i]) & (zeta <= z_edges[i + 1]) for i in range(5)]
    widths = [s_edges[i + 1] - s_edges[i] for i in range(5)]
    return masks, widths


def elicited_prior(cells, masks, widths, description="elicited (Xiang et al. 2022, Experiment 1)"):
    """A five-cell probability vector as a log-density in zeta: constant in s on each
    cell, times the Jacobian ds/dzeta = s(1 - s) of Eq. (1)."""
    def log_density(zeta):
        s = torch.sigmoid(zeta)
        density = torch.zeros_like(zeta)
        for i, mask in enumerate(masks):
            density[mask] = cells[i] / widths[i]
        return torch.log(density) + torch.log(s) + torch.log1p(-s)
    log_density.description = description
    return log_density


def field_fixed_point(net, phi_L, theta_u):
    """Eqs. (15)-(16) for an arbitrary clamped field, rather than for a named utterance.
    Checked against `closed_form_fixed_point` in block 1."""
    total_sigma = net.sigma_lexical + net.sigma_state
    gram = net.basis.T @ (net.weights[:, None] * net.basis)
    coupling = net.project(net.base_log_prior - phi_L)
    matrix = (torch.eye(net.num_basis, dtype=net.dtype)
              + net.sigma_utility * theta_u ** 2 * gram / total_sigma)
    phi_u = torch.linalg.solve(matrix, net.mu_u + net.sigma_utility * theta_u * coupling
                               / total_sigma)
    phi_S = (net.sigma_lexical * theta_u * torch.mv(net.basis, phi_u)
             + net.sigma_state * (net.base_log_prior - phi_L)) / total_sigma
    return phi_S


def field_stationary_maximizer(net, fields):
    """Eq. (B2)'s maximizing root for an arbitrary ensemble of clamped fields.
    Checked against `theta_u_stationary_points` in block 1."""
    couplings = [net.project(net.base_log_prior - f) for f in fields]
    total = torch.stack(couplings).sum(dim=0)
    total_sigma = net.sigma_lexical + net.sigma_state
    a = net.sigma_utility * float(net.dot(net.mu_u, total))
    b = (len(fields) * total_sigma * float(net.dot(net.mu_u, net.mu_u))
         - net.sigma_utility * sum(float(net.dot(c, c)) for c in couplings))
    c0 = -total_sigma * float(net.dot(net.mu_u, total))
    if a == 0.0:
        raise ZeroDivisionError("<mu_u, sum c> = 0: Appendix B's degenerate ray")
    q = -0.5 * (b + math.copysign(math.sqrt(b * b - 4.0 * a * c0), b))
    roots = (q / a, c0 / q)
    return roots[0] if (roots[0] > 0) == (a > 0) else roots[1]


def r_squared(predicted, observed):
    """Squared Pearson correlation over item x position rows, which is how the authors
    compute theirs (`LG&QF.model.predictions.R`)."""
    n = len(predicted)
    mx = sum(predicted) / n
    my = sum(observed) / n
    sxy = sum((a - mx) * (b - my) for a, b in zip(predicted, observed))
    sxx = sum((a - mx) ** 2 for a in predicted)
    syy = sum((b - my) ** 2 for b in observed)
    return (sxy * sxy) / (sxx * syy) if sxx > 0.0 and syy > 0.0 else float("nan")


def mean_position(profile):
    return sum((i + 1) * x for i, x in enumerate(profile))


# ---------------------------------------------------------------- the items
xiang_items = {}
with open(XIANG_CSV, newline="") as handle:
    for row in csv.DictReader(handle):
        key = (row["adj"], row["img_set"], row["img_type"], row["cls"])
        item = xiang_items.setdefault(key, {"prior": [0.0] * 5, "data": [0.0] * 5,
                                            "observed": [True] * 5, "tvj": [None] * 5})
        i = int(row["pos"]) - 1
        item["prior"][i] = float(row["prior"])
        if row["tvj"]:
            item["tvj"][i] = float(row["tvj"])     # Experiment 2, read by blocks F.12, F.13 and F.15
        if row["posterior"]:
            item["data"][i] = float(row["posterior"])
        else:
            item["observed"][i] = False

MODELLED = {k: v for k, v in xiang_items.items() if k[3] in CLASSES_F}
SCORED = {k: v for k, v in MODELLED.items() if any(v["observed"])}


def by_class(cls, img_type=None):
    return [k for k in SCORED if k[3] == cls and (img_type is None or k[2] == img_type)]


# ---------------------------------------------------------------- one item at (n, Lambda)
_item_nets_F = {}


def item_prior(key, prior="elicited"):
    """The five-position base world prior an item is run on: Experiment 1's, or block F.14's control."""
    return xiang_items[key]["prior"] if prior == "elicited" else UNIFORM_PRIOR_F


def item_network(key, n, prior="elicited"):
    """The network for one item at n: its base world prior on partition R1, its entry's indicator,
    and the partition's node masks. Lambda enters only as a scale on chi."""
    if (key, n, prior) not in _item_nets_F:
        cls = key[3]
        base = appendix_f_network(n)
        masks, widths = partition_masks(cls, n, base.zeta)
        if prior == "elicited":
            log_prior = elicited_prior(item_prior(key), masks, widths)
        else:
            log_prior = elicited_prior(item_prior(key, prior), masks, widths,
                                       description="uniform over the five positions (control)")
        net = base.respawn(base_prior=log_prior)
        chi = exclusion_indicator(ENTRY_OF_CLASS[cls], net.zeta, net.theta_L)
        _item_nets_F[(key, n, prior)] = (net, chi, masks)
    return _item_nets_F[(key, n, prior)]


def settled(key, n, lam, level="model", prior="elicited"):
    """The settled phi_S of one item at (n, Lambda): "q_lit", the literal listener
    ell_0 - phi_L; "tempered", theta_u = 0 at the model's sigma, (1/2)(ell_0 - phi_L); or
    "model", this network at its own theta_u*, learned over the ensemble {chi, ker chi}."""
    net, chi, masks = item_network(key, n, prior)
    phi_L = lam * chi
    if level == "q_lit":
        return net, net.base_log_prior - phi_L, None, phi_L
    if level == "tempered":
        return net, field_fixed_point(net, phi_L, 0.0), 0.0, phi_L
    theta_u = field_stationary_maximizer(net, [phi_L, lam * (1.0 - chi)])
    return net, field_fixed_point(net, phi_L, theta_u), theta_u, phi_L


def profile_of(key, n, lam, level="model", prior="elicited"):
    net, phi_S, _, _ = settled(key, n, lam, level, prior)
    _, _, masks = item_network(key, n, prior)
    q = net.read_out(phi_S) * net.weights
    return [float(q[m].sum()) for m in masks]


def class_r_squared(cls, n, lam, level="model", img_type=None, prior="elicited"):
    predicted, observed = [], []
    for k in by_class(cls, img_type):
        item = xiang_items[k]
        p = profile_of(k, n, lam, level, prior)
        for i in range(5):
            if item["observed"][i]:
                predicted.append(p[i])
                observed.append(item["data"][i])
    return r_squared(predicted, observed), (predicted, observed)


def locate_fit(cls, level, n_best, lam_best, prior="elicited"):
    """Search between the plane's neighbouring points of its best cell (Appendix F Sec. 3):
    every integer n from half to twice n_best, within the ladder, and for each a golden-
    section search in log Lambda over half to twice lam_best, within the ladder."""
    n_lo, n_hi = max(N_LADDER_F[0], n_best // 2), min(N_LADDER_F[-1], 2 * n_best)
    l_lo = math.log(max(LAMBDA_LADDER_F[0], lam_best / 2.0))
    l_hi = math.log(min(LAMBDA_LADDER_F[-1], lam_best * 2.0))
    seen = {}

    def value(n, loglam):
        key = (n, round(loglam, 12))
        if key not in seen:
            seen[key] = class_r_squared(cls, n, math.exp(loglam), level, prior=prior)[0]
        return seen[key]

    best = (-1.0, None, None)
    for n in range(n_lo, n_hi + 1):
        grid = [l_lo + (l_hi - l_lo) * j / 8 for j in range(9)]
        vals = [value(n, x) for x in grid]
        j = max(range(9), key=lambda i: vals[i])
        a, b = grid[max(0, j - 1)], grid[min(8, j + 1)]
        g = (math.sqrt(5.0) - 1.0) / 2.0
        c, d = b - g * (b - a), a + g * (b - a)
        for _ in range(24):
            if value(n, c) >= value(n, d):
                b, d = d, c
                c = b - g * (b - a)
            else:
                a, c = c, d
                d = a + g * (b - a)
        for x in (a, b, c, d, grid[j]):
            if value(n, x) > best[0]:
                best = (value(n, x), n, math.exp(x))
    near = [(n, math.exp(x)) for (n, x), v in seen.items() if best[0] - v <= RESOLUTION_F]
    n_band = (min(p[0] for p in near), max(p[0] for p in near))
    lam_band = (min(p[1] for p in near), max(p[1] for p in near))

    def reading(band, bracket, ladder_ends, located):
        if band[0] <= bracket[0] * (1 + 1e-9) and band[1] >= bracket[1] * (1 - 1e-9):
            return "flat over the searched range"
        if located in ladder_ends:
            return "at the ladder's end"
        if band[0] <= bracket[0] * (1 + 1e-9) or band[1] >= bracket[1] * (1 - 1e-9):
            return "band reaches the searched range's edge"
        return "interior"

    n_reading = reading(n_band, (n_lo, n_hi), (N_LADDER_F[0], N_LADDER_F[-1]), best[1])
    lam_ends = (LAMBDA_LADDER_F[0], LAMBDA_LADDER_F[-1])
    lam_reading = reading(lam_band, (math.exp(l_lo), math.exp(l_hi)), lam_ends,
                          min(lam_ends, key=lambda e: abs(e - best[2]))
                          if min(abs(e - best[2]) for e in lam_ends) < 1e-6 else None)
    return {"r2": best[0], "n": best[1], "lam": best[2],
            "n_reading": n_reading, "lam_reading": lam_reading,
            "n_band": n_band, "lam_band": lam_band,
            "n_bracket": (n_lo, n_hi), "lam_bracket": (math.exp(l_lo), math.exp(l_hi)),
            "evaluations": len(seen)}


def amplification_limit(net, phi_L):
    """Eq. (24): phi_S* -> (1/2)(I + B B^T W)(ell_0 - phi_L) as |theta_u| -> infinity.

    The limit is the SAME from either sign of theta_u (Appendix B), so where the settled field
    has reached it, theta_u*'s sign cannot reach the read-out."""
    f = net.base_log_prior - phi_L
    return 0.5 * (f + torch.mv(net.basis, net.project(f)))


def kept_share(cls, n, masks):
    """The share of each of the five cells that the class's entry keeps (1 - chi), by
    quadrature: the entry's truth value at each position, read off the architecture."""
    net = appendix_f_network(n)
    chi = exclusion_indicator(ENTRY_OF_CLASS[cls], net.zeta, net.theta_L)
    return [float(((1.0 - chi) * net.weights)[m].sum() / net.weights[m].sum()) for m in masks]


def antonym_report():
    """F.12: the relation the architecture's entries predict between the two adjectives of
    one image set, and Experiment 2's truth-value judgements of it."""
    print()
    print("=" * 100)
    print("F.12 THE ANTONYM RELATION: what the entries predict for the two adjectives of one image")
    print("     set, and Experiment 2's truth-value judgements. Positions are put in one physical")
    print("     orientation, the first adjective's: position k of the first is 6 - k of the second.")
    print("=" * 100)
    worst = 0.0
    for n in N_LADDER_F:
        net = appendix_f_network(n)
        chi_all = exclusion_indicator("all", net.zeta, net.theta_L)
        chi_some = exclusion_indicator("some", net.zeta, net.theta_L)
        worst = max(worst, float((chi_some - (1.0 - chi_all.flip(0))).abs().max()))
    print("  (a) The architecture. chi_some is the reflection s -> 1 - s of ker chi_all, node for")
    print(f"      node, at every n of the ladder: largest difference {worst:.1e}. In one physical")
    print("      orientation the minimum entry's truth region is the maximum entry's excluded region:")
    print("      the two regions are complementary, and every degree falls in exactly one.")
    print()
    print("  (b) What that predicts at the five positions: each entry's truth value per cell of R1")
    print("      (the share of the cell the entry keeps), summed across the pair.")
    print(f"      {'n':>5}  {'maximum + minimum':<32}{'full + empty (both maximum)'}")
    for n in N_LADDER_F:
        zeta = appendix_f_network(n).zeta
        k_max = kept_share("absolute_max", n, partition_masks("absolute_max", n, zeta)[0])
        k_min = kept_share("absolute_min", n, partition_masks("absolute_min", n, zeta)[0])
        mm = [k_max[k] + k_min[4 - k] for k in range(5)]
        xx = [k_max[k] + k_max[4 - k] for k in range(5)]
        print(f"      {n:>5}  " + " ".join(f"{v:.3f}" for v in mm) + "   "
              + " ".join(f"{v:.3f}" for v in xx))
    print()
    print("  (c) Experiment 2: the mean over image sets with both adjectives' judgements.")
    sets = {}
    for key in MODELLED:
        sets.setdefault((key[1], key[2]), []).append(key)
    groups = {}
    for (img_set, img), keys in sets.items():
        if any(None in xiang_items[k]["tvj"] for k in keys) or len(keys) != 2:
            continue
        classes = sorted(k[3] for k in keys)
        if classes == ["absolute_max", "absolute_min"]:
            a = next(k for k in keys if k[3] == "absolute_max")
            b = next(k for k in keys if k[3] == "absolute_min")
            groups.setdefault(("maximum + minimum", img), []).append((a, b))
        elif classes == ["absolute_max", "absolute_max"]:
            a, b = sorted(keys, key=lambda k: k[0], reverse=True)
            groups.setdefault(("full + empty", img), []).append((a, b))
    print(f"      {'pair':<19}{'image':<10}{'sets':>5}  {'row':<18}"
          + "".join(f"{k:>7}" for k in range(1, 6)))
    for (label, img), pairs in sorted(groups.items(), key=lambda g: (g[0][0] != "maximum + minimum",
                                                                    g[0][1] != "shape")):
        first = [sum(xiang_items[a]["tvj"][k] for a, _ in pairs) / len(pairs) for k in range(5)]
        second = [sum(xiang_items[b]["tvj"][4 - k] for _, b in pairs) / len(pairs)
                  for k in range(5)]
        rows = (("TVJ first (k)", first), ("TVJ second (6-k)", second),
                ("sum", [x + y for x, y in zip(first, second)]))
        for j, (name, vals) in enumerate(rows):
            head = f"{label:<19}{img:<10}{len(pairs):>5}" if j == 0 else " " * 34
            print(f"      {head}  {name:<18}" + "".join(f"{v:>7.2f}" for v in vals))
    print("      first adjective: the maximum one of each pair; full, of full + empty.")


def empirical_literal_listener(prior, truth):
    """Experiment 1's elicited prior times Experiment 2's mean judgements, normalized over the five
    positions: a literal listener built from the data alone, no model."""
    w = [p * t for p, t in zip(prior, truth)]
    total = sum(w)
    return [x / total for x in w]


def flip_report():
    """F.13: each absolute class's Experiment 3 posterior against the literal listener built
    from its own Experiment 2 judgements, and against the one built from its antonym's
    judgements through the flip. Data against data; no model enters this block."""
    print()
    print("=" * 100)
    print("F.13 THE ANTONYM FLIP: each class's posterior against a literal listener built from the")
    print("     data alone, the elicited prior times the mean truth judgements, normalized (L0_emp).")
    print("     Positions are in each adjective's own orientation.")
    print("=" * 100)
    sets = {}
    for key in MODELLED:
        sets.setdefault((key[1], key[2]), []).append(key)
    pairs = []
    for keys in sets.values():
        maxi = [k for k in keys if k[3] == "absolute_max"]
        mini = [k for k in keys if k[3] == "absolute_min"]
        if len(maxi) == 1 and len(mini) == 1:
            pairs.append((maxi[0], mini[0]))
    mirror = max(abs(xiang_items[a]["prior"][k] - xiang_items[b]["prior"][4 - k])
                 for a, b in pairs for k in range(5))
    print(f"  (a) The flip. {len(pairs)} image sets carry a maximum adjective and its minimum antonym;")
    print(f"      the minimum item's elicited prior is the maximum item's reversed: largest difference"
          f" {mirror:.3f}.")
    print()
    print("  (b) L0_emp against Experiment 3, over item x position rows; mean scale positions.")
    print(f"      {'':<13}{'image':<10}{'items':>6}{'R2':>8}{'L0_emp':>9}{'data':>8}"
          f"{'data - L0_emp':>16}")

    def row(label, img, keys):
        own = [empirical_literal_listener(xiang_items[k]["prior"], xiang_items[k]["tvj"])
               for k in keys]
        r2 = r_squared([x for q in own for x in q],
                       [x for k in keys for x in xiang_items[k]["data"]])
        l0 = sum(mean_position(q) for q in own) / len(keys)
        data = sum(mean_position(xiang_items[k]["data"]) for k in keys) / len(keys)
        print(f"      {label:<13}{img:<10}{len(keys):>6}{r2:>8.3f}{l0:>9.2f}{data:>8.2f}"
              f"{data - l0:>+16.2f}")

    for cls in CLASSES_F:
        for img in ("shape", "artifact"):
            row(cls, img, by_class(cls, img))
    print("      by adjective, both image types:")
    for cls in CLASSES_F:
        for adj in sorted({k[0] for k in by_class(cls)}):
            row(adj, "", [k for k in by_class(cls) if k[0] == adj])
    print()
    print("  (c) Through the flip: L0_emp built from the antonym's judgements, 1 - TVJ(6 - k), beside")
    print("      the one built from the item's own; R2 against the item's Experiment 3 posterior,")
    print("      over the antonym pairs with both items scored (full and empty are not among them).")
    print(f"      {'image':<10}{'pairs':>6}{'minimum: own':>15}{'flipped':>9}"
          f"{'maximum: own':>15}{'flipped':>9}")
    for img in ("shape", "artifact"):
        sel = [(a, b) for a, b in pairs if a[2] == img and a in SCORED and b in SCORED]
        cells = []
        for item, antonym in ((1, 0), (0, 1)):
            own, flipped, observed = [], [], []
            for pair in sel:
                k, j = pair[item], pair[antonym]
                prior = xiang_items[k]["prior"]
                through = [1.0 - xiang_items[j]["tvj"][4 - i] for i in range(5)]
                own += empirical_literal_listener(prior, xiang_items[k]["tvj"])
                flipped += empirical_literal_listener(prior, through)
                observed += xiang_items[k]["data"]
            cells += [r_squared(own, observed), r_squared(flipped, observed)]
        print(f"      {img:<10}{len(sel):>6}{cells[0]:>15.3f}{cells[1]:>9.3f}"
              f"{cells[2]:>15.3f}{cells[3]:>9.3f}")


def prior_variation_report(fit):
    """F.14: the comparison repeated with every item on one base world prior, uniform over the five
    positions, in place of Experiment 1's. A control on the base world prior, which no other block
    varies. `fit` is each class's located (n, Lambda) under the elicited prior (F.3)."""
    print()
    print("=" * 100)
    print("F.14 THE PRIOR VARIED (a control): every item given the same base world prior, uniform over the five")
    print("     positions, in place of Experiment 1's elicited prior. Entries, partition, ladders and")
    print("     read-outs as above. Under this base world prior a class's items differ only in their data.")
    print("=" * 100)
    print("  (a) R^2 against Experiment 3 at the located fit of F.3 (n and Lambda as fitted on the")
    print("      elicited prior), under each base world prior, over item x position rows.")
    print(f"      {'class':<14}{'image':<10}{'rows':>6}   {'elicited:':<10}{'q_lit':>7}{'tempered':>10}{'model':>8}"
          f"   {'uniform:':<9}{'q_lit':>7}{'tempered':>10}{'model':>8}")
    for cls in CLASSES_F:
        n, lam = fit[cls]
        for img in ("shape", "artifact", None):
            rows = sum(sum(xiang_items[k]["observed"]) for k in by_class(cls, img))
            line = f"      {cls:<14}{img or 'both':<10}{rows:>6}"
            for prior in PRIORS_F:
                line += f"   {'':<10}" if prior == "elicited" else f"   {'':<9}"
                for level in ("q_lit", "tempered", "model"):
                    width = 10 if level == "tempered" else (7 if level == "q_lit" else 8)
                    line += f"{class_r_squared(cls, n, lam, level, img, prior)[0]:>{width}.3f}"
            print(line)
    print()
    print("  (b) The fit located again under the uniform base world prior, by F.2's plane and F.3's search.")
    print(f"      {'class':<14}{'read-out':<9}{'n':>5}{'Lambda':>10}{'R^2':>8}   {'n band':>12}"
          f"{'Lambda band':>20}   n: reading;  Lambda: reading")
    for cls in CLASSES_F:
        for level in LEVELS_F:
            best = (-1.0, None, None)
            for n in N_LADDER_F:
                for lam in LAMBDA_LADDER_F:
                    v = class_r_squared(cls, n, lam, level, prior="uniform")[0]
                    if v > best[0]:
                        best = (v, n, lam)
            f = locate_fit(cls, level, best[1], best[2], prior="uniform")
            print(f"      {cls:<14}{level:<9}{f['n']:>5}{f['lam']:>10.2f}{f['r2']:>8.3f}   "
                  f"{f['n_band'][0]:>5} to {f['n_band'][1]:<4}{f['lam_band'][0]:>9.2f} to "
                  f"{f['lam_band'][1]:<8.2f}   {f['n_reading']};  {f['lam_reading']}")
    print()
    print("  (c) Movement from the uniform base world prior, whose mean scale position is "
          f"{mean_position(UNIFORM_PRIOR_F):.2f}, at the located fit of F.3.")
    print(f"      {'class':<14}{'condition':<10}{'q_lit':>8}{'model':>8}{'data':>8}"
          f"{'q_lit-p_0':>13}{'model-p_0':>13}{'data-p_0':>12}")
    p = mean_position(UNIFORM_PRIOR_F)
    for cls in CLASSES_F:
        n, lam = fit[cls]
        for img in ("shape", "artifact"):
            keys = by_class(cls, img)
            d = sum(mean_position(xiang_items[k]["data"]) for k in keys) / len(keys)
            ql = sum(mean_position(profile_of(k, n, lam, "q_lit", "uniform")) for k in keys) / len(keys)
            m = sum(mean_position(profile_of(k, n, lam, "model", "uniform")) for k in keys) / len(keys)
            print(f"      {cls:<14}{img:<10}{ql:>8.2f}{m:>8.2f}{d:>8.2f}"
                  f"{ql - p:>+13.2f}{m - p:>+13.2f}{d - p:>+12.2f}")
    print("      the same, for the literal listener at every n of the ladder, each class at its located Lambda:")
    print(f"      {'class':<14}" + "".join(f"{'n=' + str(n):>9}" for n in N_LADDER_F))
    for cls in CLASSES_F:
        lam = fit[cls][1]
        key = by_class(cls)[0]
        print(f"      {cls:<14}" + "".join(
            f"{mean_position(profile_of(key, n, lam, 'q_lit', 'uniform')) - p:>+9.2f}"
            for n in N_LADDER_F))


# ---------------------------------------------------------------- F.15: individual responses
CLASS_OF_ADJ_TYPE = {"Maximum": "absolute_max", "Minimum": "absolute_min", "Relative": "relative"}


def read_individual_responses():
    """Experiment 2, one row per participant and image, keyed as `xiang_items` is. Returns
    {(subj, key): {position: [responses]}}; a cell with more than one response at a position
    comes from a participant id that recurs (data README)."""
    cells = {}
    with open(EXPT2_CSV, newline="") as handle:
        for row in csv.DictReader(handle):
            key = (row["adj"], row["img_set"].replace("-", "_"), row["img_type"],
                   CLASS_OF_ADJ_TYPE[row["adj_type"]])
            cells.setdefault((row["subj"], key), {}).setdefault(
                int(row["scale_updated"]), []).append(int(row["response2"]))
    return cells


def cut_of(responses):
    """A participant's five responses to an item, in the adjective's own orientation: the
    cut k if the sentence is judged true from position k up to 5 (6 if of no image), and
    None if the judgements are not of that form."""
    y = [responses[k][0] for k in range(1, 6)]
    if not any(y):
        return 6
    k = y.index(1) + 1
    return k if all(y[k - 1:]) else None


def nearest_unimodal(p):
    """The least total-variation distance from a distribution on the six cuts to one that
    rises to a single mode and falls after it, and that unimodal distribution: monotone fits
    by pooling adjacent violators on each side of every candidate mode."""
    def monotone(xs, rising):
        blocks = [[x, 1] for x in (xs if rising else xs[::-1])]
        i = 0
        while i < len(blocks) - 1:
            if blocks[i][0] / blocks[i][1] > blocks[i + 1][0] / blocks[i + 1][1]:
                blocks[i] = [blocks[i][0] + blocks[i + 1][0], blocks[i][1] + blocks[i + 1][1]]
                del blocks[i + 1]
                i = max(i - 1, 0)
            else:
                i += 1
        out = [b[0] / b[1] for b in blocks for _ in range(b[1])]
        return out if rising else out[::-1]
    best = None
    for m in range(len(p)):
        q = monotone(p[:m + 1], True) + monotone(p[m:], False)[1:]
        q[m] = max(q[m], p[m])
        total = sum(q)
        q = [x / total for x in q]
        d = 0.5 * sum(abs(a - b) for a, b in zip(p, q))
        if best is None or d < best[0]:
            best = (d, q)
    return best


def unimodality_test(counts, rng, draws=2000):
    """The distance of `nearest_unimodal` and its p: the share of `draws` samples of the same
    size from the nearest unimodal distribution that lie at least as far from unimodal."""
    n = sum(counts)
    d, q = nearest_unimodal([c / n for c in counts])
    at_least = 0
    for _ in range(draws):
        sample = rng.choices(range(6), weights=q, k=n)
        if nearest_unimodal([sample.count(k) / n for k in range(6)])[0] >= d - 1e-12:
            at_least += 1
    return d, (at_least + 1) / (draws + 1)


def individual_responses_report():
    """F.15: Experiment 2 read response by response. Data alone; no model enters the block."""
    import random
    print()
    print("=" * 100)
    print("F.15 INDIVIDUAL RESPONSES: Experiment 2 read participant by participant. Each participant")
    print("     judged all five images of an item; in the adjective's own orientation a response is an")
    print("     INTERVAL if it is true from a cut k up to position 5 (k = 6: true of no image).")
    print("=" * 100)
    cells = read_individual_responses()
    complete = {c: r for c, r in cells.items()
                if sorted(r) == [1, 2, 3, 4, 5] and all(len(v) == 1 for v in r.values())}
    print(f"  {len({s for s, _ in cells})} participant ids, {len(cells)} participant x item cells;"
          f" {len(cells) - len(complete)} cells with a recurring id are set aside below,")
    print("  and kept in the mean judgements.")

    def mean_tvj(exclude):
        sums = {}
        for (subj, key), r in cells.items():
            if subj in exclude:
                continue
            for k, v in r.items():
                acc = sums.setdefault((key, k), [0, 0])
                acc[0] += sum(v)
                acc[1] += len(v)
        return {kk: v[0] / v[1] for kk, v in sums.items()}

    every = mean_tvj(set())
    gap = max(abs(every[(key, k + 1)] - xiang_items[key]["tvj"][k])
              for key in xiang_items for k in range(5) if xiang_items[key]["tvj"][k] is not None)
    print(f"  check: these responses' means against the aggregate's tvj column, largest difference {gap:.1e}")

    cuts = {}
    for (subj, key), r in complete.items():
        cuts.setdefault(key, []).append((subj, cut_of(r)))
    groups = [(cls, img) for cls in ("absolute_max", "absolute_min", "relative")
              for img in ("shape", "artifact")]
    print()
    print("  (a) Responses that are intervals, by class and image type.")
    for cls, img in groups:
        got = [c for key, v in cuts.items() if key[3] == cls and key[2] == img for _, c in v]
        inside = sum(c is not None for c in got)
        print(f"      {cls:<14}{img:<10}{inside:>5} / {len(got):<5}= {inside / len(got):.3f}")

    def one_place(exclude, label):
        rng = random.Random(500)
        tested = {}
        for key in sorted(cuts):
            got = [c for subj, c in cuts[key] if c is not None and subj not in exclude]
            if not got:
                continue
            counts = [got.count(k) for k in range(1, 7)]
            tested[key] = (counts,) + unimodality_test(counts, rng)
        print(f"      {label}: items whose cuts are not unimodal at p < 0.05, and the pooled cuts")
        print(f"      {'':<24}{'items':>9}   " + "".join(f"{'k=' + str(k):>7}" for k in range(1, 7))
              + "   responses")
        for cls, img in groups:
            keys = [k for k in tested if k[3] == cls and k[2] == img]
            below = sum(tested[k][2] < 0.05 for k in keys)
            pooled = [sum(tested[k][0][j] for k in keys) for j in range(6)]
            total = sum(pooled)
            print(f"      {cls:<14}{img:<10}{below:>4} / {len(keys):<3}  "
                  + "".join(f"{x / total:>7.2f}" for x in pooled) + f"   {total:>6}")
        return tested

    print()
    print("  (b) One place or several. Per item, the total-variation distance from the distribution of")
    print("      its cuts (interval responses only) to the nearest unimodal one on k = 1..6, and its p by")
    print("      2,000 draws of the item's size from that unimodal distribution (seed 500).")
    tested = one_place(set(), "all participants")
    print("      minimum items with p < 0.05 (counts at k = 1..6, distance, p):")
    for key, (counts, d, pv) in tested.items():
        if key[3] == "absolute_min" and pv < 0.05:
            print(f"        {key[0]:<9}{key[1]:<24}{key[2]:<10}{counts}  {d:.3f}  {pv:.3f}")

    print()
    print("  (c) The participants who cut at k = 5 on at least half the minimum items they answered with")
    print("      an interval: only the top image marked.")
    share = {}
    for key, v in cuts.items():
        if key[3] != "absolute_min":
            continue
        for subj, c in v:
            if c is not None:
                share.setdefault(subj, []).append(c == 5)
    top = {s for s, v in share.items() if sum(v) / len(v) >= 0.5}
    image_of = {subj: key[2] for subj, key in complete}
    for img in ("shape", "artifact"):
        everyone = {s for s in share if image_of[s] == img}
        mine = top & everyone
        always = sum(all(share[s]) for s in mine)
        min5 = [(s, c) for key, v in cuts.items() if key[3] == "absolute_min" and key[2] == img
                for s, c in v if c == 5]
        theirs = sum(s in top for s, _ in min5)
        rel = {flag: [c == 5 for key, v in cuts.items() if key[3] == "relative" and key[2] == img
                      for s, c in v if c is not None and (s in top) == flag] for flag in (True, False)}
        print(f"      {img:<9}{len(mine):>3} of {len(everyone)} participants ({always} at k = 5 on every"
              f" minimum item); their share of the minimum class's cuts at 5: {theirs} / {len(min5)}"
              f" = {theirs / len(min5):.2f}")
        print(f"      {'':<9}relative items cut at 5: theirs {sum(rel[True])} / {len(rel[True])}"
              f" = {sum(rel[True]) / len(rel[True]):.2f}; the others' {sum(rel[False])} / {len(rel[False])}"
              f" = {sum(rel[False]) / len(rel[False]):.2f}")

    print()
    print("  (d) (b) without the participants of (c).")
    without = one_place(top, "without them")
    print("      minimum items, counts at k = 1..6 without them, and the item's elicited prior's mean position:")
    for key, (counts, d, pv) in without.items():
        if key[3] == "absolute_min":
            print(f"        {key[0]:<9}{key[1]:<24}{key[2]:<10}{counts}  p {pv:.3f}"
                  f"   elicited prior mean {mean_position(xiang_items[key]['prior']):.2f}")

    print()
    print("  (e) F.12 (c)'s sums, maximum adjective + minimum antonym, in the maximum adjective's")
    print("      orientation, with and without the participants of (c).")
    pairs = {}
    for key in xiang_items:
        if key[3] in CLASSES_F:
            pairs.setdefault((key[1], key[2]), []).append(key)
    print(f"      {'':<18}{'image':<10}{'sets':>5}  " + "".join(f"{k:>7}" for k in range(1, 6)))
    for label, exclude in (("all participants", set()), ("without them", top)):
        tvj = mean_tvj(exclude)
        for img in ("shape", "artifact"):
            sel = [(next(k for k in v if k[3] == "absolute_max"), next(k for k in v if k[3] == "absolute_min"))
                   for (s, i), v in pairs.items() if i == img and sorted(k[3] for k in v) == list(CLASSES_F)
                   and all(None not in xiang_items[k]["tvj"] for k in v)]
            sums = [sum(tvj[(a, k)] + tvj[(b, 6 - k)] for a, b in sel) / len(sel) for k in range(1, 6)]
            print(f"      {label:<18}{img:<10}{len(sel):>5}  " + "".join(f"{x:>7.2f}" for x in sums))
    print("      The image sets are F.12 (c)'s, and the rows for all participants are its sums, recomputed")
    print("      here as this block's check.")

    print()
    print("  (f) F.13 (b)'s L0_emp against Experiment 3, with and without the participants of (c).")
    print(f"      {'':<18}{'class':<14}{'image':<10}{'R2':>8}{'L0_emp':>9}{'data':>8}{'data - L0_emp':>16}")
    for label, exclude in (("all participants", set()), ("without them", top)):
        tvj = mean_tvj(exclude)
        for cls in CLASSES_F:
            for img in ("shape", "artifact"):
                keys = by_class(cls, img)
                own = [empirical_literal_listener(xiang_items[k]["prior"],
                                                  [tvj[(k, i)] for i in range(1, 6)]) for k in keys]
                r2 = r_squared([x for q in own for x in q],
                               [x for k in keys for x in xiang_items[k]["data"]])
                l0 = sum(mean_position(q) for q in own) / len(keys)
                data = sum(mean_position(xiang_items[k]["data"]) for k in keys) / len(keys)
                print(f"      {label:<18}{cls:<14}{img:<10}{r2:>8.3f}{l0:>9.2f}{data:>8.2f}{data - l0:>+16.2f}")
    print("      The rows for all participants are F.13 (b)'s, recomputed here as this block's check.")

    print()
    print("  (g) (e)'s pairs cut by cut, with and without the participants of (c). If a minimum adjective's")
    print("      truth region were the complement of its maximum antonym's on the same images, a maximum")
    print("      cut at k would be a minimum cut at 7 - k. Per image set: the total-variation distance")
    print("      between the minimum adjective's cuts and the maximum adjective's cuts so mapped; the share")
    print("      of each adjective's cuts at its modal cut (the smallest k, if several); and whether the")
    print("      two modal cuts differ ('displaced'). Means over the sets of (e).")
    print(f"      {'':<18}{'image':<10}{'sets':>5}{'distance':>10}{'modal share: min':>18}{'max':>6}"
          f"{'displaced':>12}")
    listed = []
    for label, exclude in (("all participants", set()), ("without them", top)):
        for img in ("shape", "artifact"):
            sel = [(next(k for k in v if k[3] == "absolute_max"), next(k for k in v if k[3] == "absolute_min"))
                   for (s, i), v in pairs.items() if i == img and sorted(k[3] for k in v) == list(CLASSES_F)
                   and all(None not in xiang_items[k]["tvj"] for k in v)]
            rows = []
            for a, b in sel:
                got = {key: [c for subj, c in cuts[key] if c is not None and subj not in exclude]
                       for key in (a, b)}
                low = [got[b].count(k) for k in range(1, 7)]
                high = [got[a].count(k) for k in range(1, 7)][::-1]          # k -> 7 - k
                p, q = [x / sum(low) for x in low], [x / sum(high) for x in high]
                distance = 0.5 * sum(abs(x - y) for x, y in zip(p, q))
                rows.append((b, low, high, distance, max(p), max(q), p.index(max(p)) != q.index(max(q))))
            n = len(rows)
            print(f"      {label:<18}{img:<10}{n:>5}{sum(r[3] for r in rows) / n:>10.2f}"
                  f"{sum(r[4] for r in rows) / n:>18.2f}{sum(r[5] for r in rows) / n:>6.2f}"
                  f"{sum(r[6] for r in rows):>8} / {n}")
            if exclude:
                listed += [r for r in rows if r[3] >= 0.40]
    print("      sets at a distance of at least 0.40 without them (minimum adjective, image set, its cuts")
    print("      at k = 1..6; the maximum adjective's cuts so mapped; distance):")
    for key, low, high, distance, *_ in listed:
        print(f"        {key[0]:<9}{key[1]:<24}{key[2]:<10}{low}  {high}  {distance:.2f}")


def degen_priors_report():
    """F.16: the two elicited priors of Degen, Tessler & Goodman (2015), item by item. Data alone; no
    model enters the block. Without a connection the block prints a notice and returns."""
    import io
    import statistics
    import urllib.request
    print()
    print("=" * 100)
    print("F.16 TWO ELICITATIONS OF ONE PRIOR: Degen, Tessler & Goodman (2015). The smoothed elicited priors of the")
    print("     give-a-number task (the elicited prior their paper reports) and of the four-step task (in their")
    print("     repository only), over the counts 0..15, read from the repository at a pinned commit.")
    print("=" * 100)
    priors = {}
    try:
        for task, name in DEGEN_PRIOR_FILES:
            with urllib.request.urlopen(DEGEN_PRIORS_URL + name, timeout=30) as handle:
                rows = csv.DictReader(io.StringIO(handle.read().decode()))
                priors[task] = {row["Item"]: [float(row[str(k)]) for k in range(16)] for row in rows}
    except Exception as error:
        print(f"  NOT RUN: the files could not be read ({type(error).__name__}). The block needs a connection;")
        print("  the files are not kept here because the repository carries no licence.")
        return
    first, second = (priors[task] for task, _ in DEGEN_PRIOR_FILES)
    items = sorted(set(first) & set(second))
    print(f"  items: {len(first)} and {len(second)}, {len(items)} in common")

    def expected_count(p):
        return sum(k * x for k, x in enumerate(p)) / sum(p)

    gaps = [abs(expected_count(first[i]) - expected_count(second[i])) for i in items]
    print(f"  {'':<16}{'mean expected count':>21}{'median P(15 of 15)':>21}{'largest':>10}{'items with P > 1/2':>21}")
    for (task, _), table in zip(DEGEN_PRIOR_FILES, (first, second)):
        top = [table[i][15] for i in items]
        print(f"  {task:<16}{statistics.mean(expected_count(table[i]) for i in items):>21.2f}"
              f"{statistics.median(top):>21.3f}{max(top):>10.3f}{sum(x > 0.5 for x in top):>21}")
    print(f"  expected count, item by item: |difference| median {statistics.median(gaps):.2f},"
          f" largest {max(gaps):.2f} (of 15)")


def appendix_f_report():
    """Every number Appendix F and Sec. 4.2 quote about these data, in sixteen blocks."""
    print("=" * 100)
    print("F.1  CONFIGURATION, AND THE CHECKS THAT IT IS THE SAME MODEL")
    print("=" * 100)
    print(f"  n swept over {list(N_LADDER_F)}; Lambda over {[int(x) for x in LAMBDA_LADDER_F]};"
          f" Z = {evaluation_network.zeta[-1].item():.1f}, K = {len(evaluation_network.zeta)}")
    for n in (256, 512, 1024):
        print(f"    n = {n}: theta_L = log(2n - 1) = {math.log(2 * n - 1):.4f} >= Z, outside the ladder")
    print("  partition R1 in s, each class in its own orientation (endpoint image's cell first named):")
    for n in N_LADDER_F:
        cells = {c: ", ".join(f"{x:.4f}" for x in partition_edges(c, n)[1:-1]) for c in CLASSES_F}
        print(f"    n = {n:>3}  theta_L = {appendix_f_network(n).theta_L:.4f}   "
              f"max edges [{cells['absolute_max']}]   min edges [{cells['absolute_min']}]")
    for cls in CLASSES_F:
        y = ENTRY_OF_CLASS[cls]
        print(f"  {cls:<13} -> the model's \"{y}\" (Eq. A1); its complement ker chi completes the ensemble")
    _probe = appendix_f_network(4).respawn(lexical_strength=8.0)
    _gap_fp = max(float((field_fixed_point(_probe, _probe.lexical_field(y),
                                           _probe.learned_theta_u())
                         - _probe.closed_form_fixed_point(
                             y, theta_u=_probe.learned_theta_u())[0]
                         ).abs().max()) for y in UTTERANCES)
    _gap_th = abs(field_stationary_maximizer(_probe, [_probe.lexical_field(y)
                                                      for y in UTTERANCES])
                  - _probe.learned_theta_u())
    print(f"  this cell's field-valued closed form against `closed_form_fixed_point`: "
          f"{_gap_fp:.1e}")
    print(f"  this cell's field-valued maximizer against `theta_u_stationary_points`: "
          f"{_gap_th:.1e}")
    print(f"  the default n = 10 inventory still learns theta_u* = "
          f"{evaluation_network.learned_theta_u():.4f}")
    _worst_part = 0.0
    _worst_chi = 0.0
    for n in N_LADDER_F:
        z = appendix_f_network(n).zeta
        w = appendix_f_network(n).weights
        for cls in CLASSES_F:
            masks, _ = partition_masks(cls, n, z)
            _worst_part = max(_worst_part, abs(sum(float(w[m].sum()) for m in masks)
                                               - float(w.sum())))
            chi = exclusion_indicator(ENTRY_OF_CLASS[cls], z, appendix_f_network(n).theta_L)
            end_cell = masks[4] if cls == "absolute_max" else masks[0]
            kept = (chi == 0) if cls == "absolute_max" else (chi == 1)
            _worst_chi = max(_worst_chi, float((end_cell != kept).sum()))
    print(f"  the five cells partition the grid at every n, to {_worst_part:.1e}")
    print(f"  the endpoint cell is the entry's own region (all: kept; some: excluded) at every n:"
          f" {int(_worst_chi)} nodes differ")
    print()
    print(f"  {len(xiang_items)} items in the file; {len(MODELLED)} in the two modelled "
          f"classes; {len(xiang_items) - len(MODELLED)} relative items carry no prediction")
    _absent = sorted({k[1] for k in MODELLED if not all(xiang_items[k]["observed"])})
    print(f"  {len(MODELLED) - len(SCORED)} modelled item(s) drew no Experiment 3 response "
          f"at all and are scored nowhere below: image set(s) {', '.join(_absent)}")
    print(f"  scored rows: " + ", ".join(
        f"{cls.split('_')[-1]} {sum(sum(xiang_items[k]['observed']) for k in by_class(cls))}"
        for cls in CLASSES_F))

    print()
    print("=" * 100)
    print("F.2  THE PLANE: R^2 AGAINST EXPERIMENT 3 over item x position rows, at every (n, Lambda).")
    print("     'q_lit' is the literal listener ell_0 - phi_L; 'model' this network at its own theta_u*.")
    print("=" * 100)
    plane_best = {}
    for cls in CLASSES_F:
        for level in LEVELS_F:
            print(f"\n  {cls}, {level}")
            print("    " + "n | Lambda".rjust(10) + "".join(f"{int(l):>8}" for l in LAMBDA_LADDER_F))
            best = (-1.0, None, None)
            for n in N_LADDER_F:
                line = f"    {n:>10}"
                for lam in LAMBDA_LADDER_F:
                    v = class_r_squared(cls, n, lam, level)[0]
                    if v > best[0]:
                        best = (v, n, lam)
                    line += f"{v:>8.3f}"
                print(line)
            plane_best[(cls, level)] = best
            print(f"    best cell: n = {best[1]}, Lambda = {int(best[2])}, R^2 = {best[0]:.3f}")

    print()
    print("=" * 100)
    print("F.3  THE FIT, LOCATED APART FROM THE PLANE. For each class and read-out, every integer n")
    print("     from half to twice the best cell's, and for each a golden-section search in log Lambda")
    print(f"     from half to twice its Lambda, both kept within the ladders. The bands are the points")
    print(f"     evaluated within {RESOLUTION_F} of the located maximum, the width the search resolves.")
    print("=" * 100)
    fits = {}
    print(f"  {'class':<14}{'read-out':<9}{'n':>5}{'Lambda':>10}{'R^2':>8}   {'n band':>12}"
          f"{'Lambda band':>20}   searched: n, Lambda")
    for cls in CLASSES_F:
        for level in LEVELS_F:
            _, nb_, lb_ = plane_best[(cls, level)]
            f = locate_fit(cls, level, nb_, lb_)
            fits[(cls, level)] = f
            print(f"  {cls:<14}{level:<9}{f['n']:>5}{f['lam']:>10.2f}{f['r2']:>8.3f}   "
                  f"{f['n_band'][0]:>5} to {f['n_band'][1]:<4}{f['lam_band'][0]:>9.2f} to "
                  f"{f['lam_band'][1]:<8.2f}   {f['n_bracket'][0]}-{f['n_bracket'][1]}, "
                  f"{f['lam_bracket'][0]:g}-{f['lam_bracket'][1]:g}")
    print("  What each located value is:")
    for cls in CLASSES_F:
        for level in LEVELS_F:
            f = fits[(cls, level)]
            print(f"    {cls:<14}{level:<9} n: {f['n_reading']};  Lambda: {f['lam_reading']}")
    FIT_F = {cls: (fits[(cls, "model")]["n"], fits[(cls, "model")]["lam"]) for cls in CLASSES_F}
    print("  The blocks below are read at each class's located fit for the full network:")
    for cls in CLASSES_F:
        print(f"    {cls:<14} n = {FIT_F[cls][0]}, Lambda = {FIT_F[cls][1]:.2f}")

    print()
    print("=" * 100)
    print("F.4  THE CLASS PROFILES, BY IMAGE TYPE, at the located fit. Mass in each of the five positions.")
    print("     'data' is the authors' Experiment 3; 'tempered' the tempered control, theta_u = 0 at the")
    print("     model's sigma, (1/2)(ell_0 - phi_L).")
    print("=" * 100)
    for cls in CLASSES_F:
        n, lam = FIT_F[cls]
        for img in ("shape", "artifact"):
            keys = by_class(cls, img)
            rows = {level: [0.0] * 5 for level in ("p_0", "q_lit", "tempered", "model", "data")}
            for k in keys:
                item = xiang_items[k]
                profiles = {"p_0": item["prior"], "data": item["data"]}
                for level in ("q_lit", "tempered", "model"):
                    profiles[level] = profile_of(k, n, lam, level)
                for level, p in profiles.items():
                    for i in range(5):
                        rows[level][i] += p[i] / len(keys)
            print(f"\n  {cls:<13}{img:<10}({len(keys)} items, n = {n}, Lambda = {lam:.2f})")
            for level in ("p_0", "q_lit", "tempered", "model", "data"):
                v = rows[level]
                print(f"    {level:<9}" + "".join(f"{x:>7.3f}" for x in v)
                      + f"   global maximum at {1 + max(range(5), key=lambda i: v[i])}"
                      + f"   mean position {mean_position(v):.2f}")

    print()
    print("=" * 100)
    print("F.5  R^2 AT THE LOCATED FIT, and the two steps between the literal listener and the model:")
    print("     'tempering' = tempered - q_lit, a finite sigma_S at theta_u = 0; 'utility' = model -")
    print("     tempered, theta_u* at the same sigma.")
    print("=" * 100)
    print(f"  {'class':<14}{'items':>7}{'rows':>7}{'n':>5}{'Lambda':>9}{'model':>9}{'tempered':>10}"
          f"{'q_lit':>9}{'tempering':>11}{'utility':>9}")
    pooled = {"model": [[], []], "tempered": [[], []], "q_lit": [[], []]}
    for cls in CLASSES_F:
        n, lam = FIT_F[cls]
        scores = {}
        for level in ("model", "tempered", "q_lit"):
            value, (p, o) = class_r_squared(cls, n, lam, level)
            scores[level] = value
            pooled[level][0].extend(p)
            pooled[level][1].extend(o)
        n_rows = sum(sum(xiang_items[k]["observed"]) for k in by_class(cls))
        print(f"  {cls:<14}{len(by_class(cls)):>7}{n_rows:>7}{n:>5}{lam:>9.2f}"
              f"{scores['model']:>9.3f}{scores['tempered']:>10.3f}{scores['q_lit']:>9.3f}"
              f"{scores['tempered'] - scores['q_lit']:>+11.3f}"
              f"{scores['model'] - scores['tempered']:>+9.3f}")
    _pooled = {level: r_squared(*pooled[level]) for level in pooled}
    print(f"  {'both classes':<14}{len(SCORED):>7}{len(pooled['model'][0]):>7}{'(each)':>14}"
          f"{_pooled['model']:>9.3f}{_pooled['tempered']:>10.3f}{_pooled['q_lit']:>9.3f}"
          f"{_pooled['tempered'] - _pooled['q_lit']:>+11.3f}"
          f"{_pooled['model'] - _pooled['tempered']:>+9.3f}")

    print()
    print("=" * 100)
    print("F.6  THE IMAGE-TYPE DIFFERENCE, class by class: shape minus artifact in the mean")
    print("     scale position, in the elicited prior, in the data, in the literal listener and in the model.")
    print("=" * 100)
    print(f"  {'class':<14}{'':<8}" + "".join(f"{s:>12}" for s in ("shape", "artifact", "difference")))
    for cls in CLASSES_F:
        n, lam = FIT_F[cls]
        for source in ("p_0", "data", "q_lit", "model"):
            means = {}
            for img in ("shape", "artifact"):
                keys = by_class(cls, img)
                means[img] = sum(mean_position(
                    xiang_items[k]["prior"] if source == "p_0" else
                    xiang_items[k]["data"] if source == "data" else
                    profile_of(k, n, lam, source)) for k in keys) / len(keys)
            print(f"  {cls:<14}{source:<8}{means['shape']:>12.2f}{means['artifact']:>12.2f}"
                  f"{means['shape'] - means['artifact']:>+12.2f}")

    print()
    print("=" * 100)
    print("F.7  THE GAP BETWEEN THE TWO CLASSES, by image condition: maximum minus minimum")
    print("     in the mean scale position.")
    print("=" * 100)
    print(f"  {'':<8}{'shape':>36}{'artifact':>36}")
    print(f"  {'source':<8}" + "".join(f"{s:>12}" for s in ("max", "min", "max - min") * 2))
    for source in ("p_0", "data", "q_lit", "model"):
        line = f"  {source:<8}"
        for img in ("shape", "artifact"):
            means = {}
            for cls in CLASSES_F:
                n, lam = FIT_F[cls]
                keys = by_class(cls, img)
                means[cls] = sum(mean_position(
                    xiang_items[k]["prior"] if source == "p_0" else
                    xiang_items[k]["data"] if source == "data" else
                    profile_of(k, n, lam, source)) for k in keys) / len(keys)
            line += (f"{means['absolute_max']:>12.2f}{means['absolute_min']:>12.2f}"
                     f"{means['absolute_max'] - means['absolute_min']:>+12.2f}")
        print(line)

    print()
    print("=" * 100)
    print("F.8  MOVEMENT FROM THE ELICITED PRIOR: the mean scale position of the literal listener,")
    print("     the model and the data, and each one's displacement from the elicited prior. Reported, not explained.")
    print("=" * 100)
    print(f"    {'class':<14}{'condition':<10}{'p_0':>8}{'q_lit':>8}{'model':>8}{'data':>8}"
          f"{'q_lit-p_0':>13}{'model-p_0':>13}{'data-p_0':>12}")
    for cls in CLASSES_F:
        n, lam = FIT_F[cls]
        for img in ("shape", "artifact"):
            keys = by_class(cls, img)
            p = sum(mean_position(xiang_items[k]["prior"]) for k in keys) / len(keys)
            d = sum(mean_position(xiang_items[k]["data"]) for k in keys) / len(keys)
            ql = sum(mean_position(profile_of(k, n, lam, "q_lit")) for k in keys) / len(keys)
            m = sum(mean_position(profile_of(k, n, lam, "model")) for k in keys) / len(keys)
            print(f"    {cls:<14}{img:<10}{p:>8.2f}{ql:>8.2f}{m:>8.2f}{d:>8.2f}"
                  f"{ql - p:>+13.2f}{m - p:>+13.2f}{d - p:>+12.2f}")
    print("  the same, for the literal listener at every n of the ladder, each class at its located Lambda:")
    print(f"    {'class':<14}{'condition':<10}" + "".join(f"{'n=' + str(n):>9}" for n in N_LADDER_F))
    for cls in CLASSES_F:
        lam = FIT_F[cls][1]
        for img in ("shape", "artifact"):
            keys = by_class(cls, img)
            p = sum(mean_position(xiang_items[k]["prior"]) for k in keys) / len(keys)
            print(f"    {cls:<14}{img:<10}" + "".join(
                f"{sum(mean_position(profile_of(k, n, lam, 'q_lit')) for k in keys) / len(keys) - p:>+9.2f}"
                for n in N_LADDER_F))

    print()
    print("=" * 100)
    print("F.9  THE PARITY OF THE TWO ENTRIES' LOADINGS at every n (Appendix C's argument, at")
    print("     this cell's configuration). kappa_y = B^T W chi_y: the tilt (odd) and width")
    print("     (even) coordinates of each entry's indicator.")
    print("=" * 100)
    print(f"  {'n':>5}{'tilt all':>12}{'width all':>12}{'tilt some':>12}{'width some':>12}"
          f"{'tilt diff':>12}{'width sum':>12}")
    for n in N_LADDER_F:
        net = appendix_f_network(n)
        k_all = net.project(exclusion_indicator("all", net.zeta, net.theta_L))
        k_some = net.project(exclusion_indicator("some", net.zeta, net.theta_L))
        print(f"  {n:>5}{float(k_all[0]):>12.6f}{float(k_all[1]):>12.6f}{float(k_some[0]):>12.6f}"
              f"{float(k_some[1]):>12.6f}{float(k_all[0] - k_some[0]):>12.1e}"
              f"{float(k_all[1] + k_some[1]):>12.1e}")
    print("  at every n the maximum and minimum entries differ in the even (width) coordinate alone.")
    _net4 = appendix_f_network(4)
    _zeta4 = _net4.zeta
    _mid = (_zeta4 < 0.0).to(_zeta4.dtype)
    _mid[_zeta4.abs() < 1e-12] = 0.5
    print(f"  a cut at the midpoint of the log-odds scale carries no even component at all: "
          f"width {float(_net4.project(_mid)[1]):.1e}")
    print("  (the node on the cut is half-weighted; left unweighted the grid reads it as "
          f"{float(_net4.project((_zeta4 < 0.0).to(_zeta4.dtype))[1]):.4f}.)")

    print()
    print("=" * 100)
    print("F.10 HOW FAR THE UTILITY LEVEL REACHES, AND WHAT FIXES theta_u*'s SIGN, at the located fit.")
    print("=" * 100)
    print("  (a) The ensemble is {chi, ker chi} and chi + ker chi = 1, so sum_y phi_L,y = Lambda * 1")
    print("      and the Lambda term cancels against B^T W 1 = 0 (Appendix C Sec. 2):")
    print("          sum_y c_y = B^T W (2 ell_0 - Lambda * 1) = 2 B^T W ell_0.")
    print("      Eq. (B2) gives theta_u* the sign of <mu_u, sum_y c_y>, so THAT SIGN IS A PROPERTY")
    print("      OF THE PRIOR ALONE, independent of Lambda and of which entry was uttered.")
    worst = 0.0
    for cls in CLASSES_F:
        for n in N_LADDER_F:
            for lam in (4.0, 32.0, 256.0, 1024.0):
                for k in by_class(cls):
                    net, chi, _ = item_network(k, n)
                    total = (net.project(net.base_log_prior - lam * chi)
                             + net.project(net.base_log_prior - lam * (1.0 - chi)))
                    worst = max(worst, float((total - 2.0 * net.project(net.base_log_prior))
                                             .abs().max()))
    print(f"      max |sum_y c_y - 2 B^T W ell_0| over {len(N_LADDER_F)} n x 4 Lambda x "
          f"{len(SCORED)} items: {worst:.1e}")
    print()
    print("  (b) theta_u* per condition. The SIGN VARIES ITEM BY ITEM and |theta_u*| diverges near")
    print("      Appendix B's degenerate ray, so a MEAN over items is not a usable statistic here;")
    print("      the median and the sign counts are reported instead.")
    print(f"    {'class':<14}{'image':<10}{'neg':>5}{'pos':>5}{'median':>12}"
          f"{'max |theta_u*|':>16}{'min |<mu,sum c>|':>18}")
    for cls in CLASSES_F:
        n, lam = FIT_F[cls]
        for img in ("shape", "artifact"):
            thetas, couplings_seen = [], []
            for k in by_class(cls, img):
                net, _, th, _ = settled(k, n, lam)
                thetas.append(th)
                couplings_seen.append(abs(2.0 * float(net.dot(net.mu_u,
                                                              net.project(net.base_log_prior)))))
            order = sorted(thetas)
            mid = (order[len(order) // 2] if len(order) % 2
                   else 0.5 * (order[len(order) // 2 - 1] + order[len(order) // 2]))
            print(f"    {cls:<14}{img:<10}{sum(1 for t in thetas if t < 0):>5}"
                  f"{sum(1 for t in thetas if t > 0):>5}{mid:>12.1f}"
                  f"{max(abs(t) for t in thetas):>16.1f}{min(couplings_seen):>18.4f}")
    print()
    print("  (c) Whether the sign reaches the belief. Eq. (24)'s limit is the same from either sign,")
    print("      and the gap to it falls as 1/|theta_u|.")
    print(f"    {'class':<14}{'image':<10}{'max |settled - Eq24|':>22}{'R2 settled':>13}"
          f"{'R2 at Eq24':>13}")
    for cls in CLASSES_F:
        n, lam = FIT_F[cls]
        for img in ("shape", "artifact"):
            gap, ps, pl, obs = 0.0, [], [], []
            for k in by_class(cls, img):
                item = xiang_items[k]
                net, phi_S, _, phi_L = settled(k, n, lam)
                _, _, masks = item_network(k, n)
                lim = amplification_limit(net, phi_L)
                gap = max(gap, float((phi_S - lim).abs().max()))
                qs = net.read_out(phi_S) * net.weights
                ql = net.read_out(lim) * net.weights
                for i in range(5):
                    if item["observed"][i]:
                        ps.append(float(qs[masks[i]].sum()))
                        pl.append(float(ql[masks[i]].sum()))
                        obs.append(item["data"][i])
            print(f"    {cls:<14}{img:<10}{gap:>22.1e}{r_squared(ps, obs):>13.3f}"
                  f"{r_squared(pl, obs):>13.3f}")
    print()
    print("  (d) What the amplification does, in mean scale position. Eq. (24) doubles the span(B)")
    print("      component of the tempered control, so the utility level's displacement is measured")
    print("      from that control; 'tempered - q_lit' is the tempering's own share.")
    print(f"    {'class':<14}{'image':<10}{'tempered - q_lit':>18}{'model - tempered':>18}"
          f"{'data - tempered':>17}{'model - data':>14}")
    for cls in CLASSES_F:
        n, lam = FIT_F[cls]
        for img in ("shape", "artifact"):
            keys = by_class(cls, img)
            q = sum(mean_position(profile_of(k, n, lam, "q_lit")) for k in keys) / len(keys)
            t = sum(mean_position(profile_of(k, n, lam, "tempered")) for k in keys) / len(keys)
            m = sum(mean_position(profile_of(k, n, lam, "model")) for k in keys) / len(keys)
            d = sum(mean_position(xiang_items[k]["data"]) for k in keys) / len(keys)
            print(f"    {cls:<14}{img:<10}{t - q:>+18.2f}{m - t:>+18.2f}{d - t:>+17.2f}"
                  f"{m - d:>+14.2f}")

    # F.11 is the control for Appendix A's half-width ladder: Appendix A varies Z and
    # reports the four criteria, which are masses above a fixed cut and the mode's own
    # place; this block varies Z and reports R^2, which scores AGREEMENT OF SHAPE across
    # items. The partition is fixed in s, so only its node masks are rebuilt per grid; the
    # Z = 6.0 row must reproduce F.5's model column exactly, which is this block's check.
    print()
    print("=" * 100)
    print("F.11 R^2 AT EVERY HALF-WIDTH Z, at the located fit, against Appendix A's ladder of the")
    print("     same Z. Nodes scaled with Z to hold the spacing, as in Appendix A. A class whose")
    print("     theta_L is not below Z has no row there (Eq. A5).")
    print("=" * 100)
    print(f"  {'Z':>6}{'K':>5}" + "".join(f"{c:>16}" for c in CLASSES_F))
    for half_width in HALF_WIDTHS:
        nodes = int(round(2 * half_width / NODE_SPACING)) + 1
        line = f"  {half_width:>6.1f}{nodes:>5}"
        for cls in CLASSES_F:
            n, lam = FIT_F[cls]
            if math.log(2 * n - 1) >= half_width:
                line += f"{'theta_L >= Z':>16}"
                continue
            probe = appendix_f_network(n).respawn(grid_half_width=half_width, num_nodes=nodes)
            masks, widths = partition_masks(cls, n, probe.zeta)
            predicted, observed = [], []
            for key in by_class(cls):
                item = xiang_items[key]
                grid = probe.respawn(base_prior=elicited_prior(item["prior"], masks, widths))
                chi = exclusion_indicator(ENTRY_OF_CLASS[cls], grid.zeta, grid.theta_L)
                theta_u = field_stationary_maximizer(grid, [lam * chi, lam * (1.0 - chi)])
                q = grid.read_out(field_fixed_point(grid, lam * chi, theta_u)) * grid.weights
                profile = [float(q[m].sum()) for m in masks]
                for i in range(5):
                    if item["observed"][i]:
                        predicted.append(profile[i])
                        observed.append(item["data"][i])
            line += f"{r_squared(predicted, observed):>16.3f}"
        print(line)
    print("     The Z = 6.0 row is F.5's model column, recomputed here as this block's check.")

    antonym_report()
    flip_report()
    prior_variation_report(FIT_F)
    individual_responses_report()
    degen_priors_report()


appendix_f_report()
````

## <a id="appg" name="appg"></a>Appendix G: the computational complexity of the simulation

This appendix counts what it costs to run **this simulation**: the arithmetic `code cell 1`
performs on a serial machine, as a function of the node count $K$, the number of utility directions
$m$, the number of utterances $\lvert Y\rvert$ an exposure ensemble holds, and $\theta_u$. The
simulation is one instantiation of the architecture, and the architecture is one representation of
the process it models, so these counts represent the process's cost, through this instantiation.
They are not the complexity of the architecture: another instantiation of the same equations, with
another integrator or on parallel hardware, would count differently. Each quantity below is
therefore attributed to its source, either the equations every instantiation of the architecture
shares or a choice this instantiation makes.

**Notation.** For costs $f$ and $g$ that depend on the sizes above, $f=O(g)$ means that $f$ is at
most a constant multiple of $g$ once the sizes are large, and $f=\Theta(g)$ that it is also at least
one, so that $g$ is its rate of growth exactly. $\Theta$ is written wherever both bounds are shown,
and $O$ only where the upper one alone is. Throughout $m<K$, since the basis orthonormalizes $m+1$
functions on $K$ nodes, so a term such as $m^3$ is absorbed into $Km^2$. The atom count $n$ appears in no count below, and enters two quantities they depend on: through
$\theta_L$ it sets $\theta_u^\ast$ and, with it, the condition number of Eq. (G1); and through the
lexical field it sets how much of the start lies in the slowest mode, which moves the step count of
Eq. (G2) at a fixed $\theta_u$ (§3).

---

### <a id="appg-1" name="appg-1"></a>1. The spectrum of $H$ in closed form

At the model's $\sigma=1$, and with $B^{\mathsf T}WB=I$, which the basis satisfies by construction,
$H$ in the quadrature metric of Text cell 3 §8 reads, in the symmetrized coordinates
$\operatorname{diag}(W^{1/2},I)\,H\,\operatorname{diag}(W^{-1/2},I)$, which have $H$'s eigenvalues,

$$H=\begin{pmatrix}2I_K&-\theta_u\hat B\\-\theta_u\hat B^{\mathsf T}&(1+\theta_u^2)I_m\end{pmatrix},
\qquad \hat B=W^{1/2}B,$$

and $\hat B$ has orthonormal columns. On the $K-m$ directions of $\mathbb R^K$ orthogonal to those
columns $H$ acts as $2$. Each column $\hat b_j$ pairs with the $j$th utility coordinate into the
block $\begin{pmatrix}2&-\theta_u\\-\theta_u&1+\theta_u^2\end{pmatrix}$, of trace $3+\theta_u^2$ and
determinant $2+\theta_u^2$, whose eigenvalues are therefore $1$ and $\theta_u^2+2$. Hence

$$\operatorname{spec}H=\{\,1\ (m\text{ times}),\ \ 2\ (K-m\text{ times}),\ \
\theta_u^2+2\ (m\text{ times})\,\},\qquad
\kappa(H)=\frac{\lambda_{\max}(H)}{\lambda_{\min}(H)}=\theta_u^2+2 .
\tag{G1}$$

**The condition number $\kappa(H)$ depends on $\theta_u$ alone**, and on neither $K$ nor $m$. Code
Cell G checks Eq. (G1) at three node counts, three basis sizes and five values of $\theta_u$, and
finds it exact to $2.1\times10^{-15}$ of $\lambda_{\max}$. $\kappa(H)$ is not the **stiffness
ratio** of *Integration cost and conditioning* (Text cell 4), which divides Eq. (28)'s $\varphi_u$
rate by the $\varphi_S$ rate and is a ratio of two diagonal entries of $H$. At the evaluation
network's $\theta_u^\ast$ the cell prints both: $\kappa(H)=810.69$ against a stiffness ratio of
$404.85$. The step count below is set by $\kappa(H)$.

**The same split holds at any $\sigma$ and any basis**, and it is how `stiffest_state_rate` finds
$\lambda_{\max}(H)$ without decomposing $H$. $H$ couples $\varphi_S$ to $\varphi_u$ only through
$\hat B=W^{1/2}B$, whose singular values are the square roots of the eigenvalues $\gamma_j$ of
$B^{\mathsf T}WB$. So $H$ acts as $a=1/\sigma_L+1/\sigma_S$ on the directions of $\varphi_S$ that
$B$ does not reach, and splits into one block per $\gamma_j$,

$$\begin{pmatrix}a&-\theta_u\sqrt{\gamma_j}/\sigma_S\\-\theta_u\sqrt{\gamma_j}/\sigma_S&
1/\sigma_u+\theta_u^2\gamma_j/\sigma_S\end{pmatrix},$$

each with a larger root of at least $a$. $\lambda_{\max}(H)$ is the largest of those $m$ roots,
found from the $m\times m$ matrix $B^{\mathsf T}WB$ alone, and Eq. (G1) is the case $\sigma=1$,
$B^{\mathsf T}WB=I$. Code Cell G compares it with the largest eigenvalue of the dense $H$: they agree
to $9.8\times10^{-16}$ of $\lambda_{\max}$ over the $45$ configurations above, and to
$9.6\times10^{-16}$ over $90$ more at three settings of $\sigma\ne1$ with an orthonormal, a random
and a rank-deficient basis.

### <a id="appg-2" name="appg-2"></a>2. The cost, operation by operation

| Operation | Where | Cost |
|---|---|---|
| Grid, base world prior and its normalization | `build_grid`, the constructor | $\Theta(K)$ |
| Basis: Gram–Schmidt on $m+1$ functions | the constructor | $\Theta(Km^2)$ |
| $\theta_u^\ast$ by Eq. (B2): the Gram check, $\lvert Y\rvert$ lexical fields and projections, one quadratic | `theta_u_stationary_points` | $\Theta(Km^2+\lvert Y\rvert Km)$ |
| **Construction, on every `respawn`** | | $\boldsymbol{\Theta(Km(m+\lvert Y\rvert))}$ |
| Settled state in closed form, Eqs. (15)–(16): the Gram matrix, one projection, an $m\times m$ solve | `closed_form_fixed_point` | $\Theta(Km^2)$ |
| Literal listener; read-out, Eq. (12) | `literal_fixed_point`, `read_out` | $\Theta(K)$ |
| $\lambda_{\max}(H)$ blockwise from $B^{\mathsf T}WB$ (§1), twice per inference (the tolerance and $\tau_\varepsilon$) | `stiffest_state_rate` | $\Theta(Km^2)$ |
| One Euler step of Eqs. (18)–(19): the prediction $\theta_uB\varphi_u$ and the projection $B^{\mathsf T}W\varepsilon_S$, and with the history recorded, $\mathcal F$ as well | `infer` | $\Theta(Km)$ |
| The number of Euler steps, $N$ | `infer` | Eq. (G2) |
| One update of Eq. (20): $\lvert Y\rvert$ inferences, each from the same start | `learn_theta_u` | $\lvert Y\rvert$ inferences |

Every $\Theta$ in the table holds because the operation performs a fixed amount of arithmetic for
given sizes, whatever the input. One row is a choice of **this instantiation** and not a
requirement of the equations it integrates: the matrix the closed form's solve inverts is a
multiple of the identity, since $B^{\mathsf T}WB=I$.

### <a id="appg-3" name="appg-3"></a>3. How many Euler steps

With the error units eliminated, each eigenvalue $\lambda$ of $H$ gives rates $s$ solving
$\tau_\varepsilon\tau_\varphi s^2+\tau_\varphi s+\lambda=0$ (`fast_time_constant`). By default
$\tau_\varepsilon=\tau_\varphi/(4\lambda_{\max}(H))$ and $\mathrm dt=\tau_\varepsilon/2$, so
$\mathrm dt=\tau_\varphi/(8\lambda_{\max}(H))$.

`infer` does not step Eqs. (18)–(19) by forward Euler on the combined system: each step updates the
error units first, and then the states from the **updated** errors. Per eigenvalue $\lambda$ of $H$,
with $\alpha=\mathrm dt/\tau_\varepsilon$ and $\beta=\mathrm dt/\tau_\varphi$, a step is therefore
the map, on the mode's error and state coordinates,

$$\begin{pmatrix}1-\alpha&\alpha\\-\beta\lambda(1-\alpha)&1-\alpha\beta\lambda\end{pmatrix},
\qquad\det=1-\alpha,\qquad\operatorname{tr}=2-\alpha-\alpha\beta\lambda .$$

Under `infer`'s guards, $\mathrm dt\le\tau_\varepsilon\le\tau_\varphi/(4\lambda_{\max}(H))$, so
$\alpha\le1$ and $\alpha\beta\lambda\le\alpha^2/4$ for every $\lambda\le\lambda_{\max}(H)$. The
discriminant $\operatorname{tr}^2-4\det$ is then at least $\alpha^3/2>0$, and
$1\mp\operatorname{tr}+\det>0$, so **both factors are real and lie in $[0,1)$**: the integration
is stable and does not oscillate, with no separate check. At the defaults $\alpha=\tfrac12$, and
the slow factor of the slowest mode, $\lambda_{\min}(H)=1$ of Eq. (G1), is

$$\mu_1=1-\frac1{8\kappa(H)}-\frac1{64\,\kappa(H)^2}+O\big(\kappa(H)^{-3}\big),
\qquad -8\kappa(H)\log\mu_1=1+\frac3{16\,\kappa(H)}+O\big(\kappa(H)^{-2}\big),$$

so each step shrinks that mode by $1-1/(8\kappa(H))$ to leading order. Integration stops once the largest derivative
has stayed below the tolerance for ten steps. Every other mode shrinks faster, so the count is set
by the slowest one. Let $a$ be the largest derivative at the start and $a_1$ the slowest mode's
share of it:

$$8\,\kappa(H)\,\log\frac{a_1}{\mathrm{tol}}\;\lesssim\;N\;\lesssim\;
8\,\kappa(H)\,\log\frac{C\,a}{\mathrm{tol}}+10,
\tag{G2}$$

to leading order in $1/\kappa(H)$, where $C$ depends only on how far the eigenvectors of the
linear system are from orthogonal. So $N=O\big(\kappa(H)\log(a/\mathrm{tol})\big)$ always, and
$N=\Theta\big(\kappa(H)\log(a/\mathrm{tol})\big)$ whenever $a_1$ is a fixed fraction of $a$. A start
with no component in the slowest mode would settle sooner.

Code Cell G integrates fourteen configurations with $\theta_u$ fixed, **a control**: the step
count is read here as a function of $\theta_u$, which each configuration's own $\theta_u^\ast$
does not vary. In each, the largest derivative's decay per step over the second half of the run,
times $8\kappa(H)$, is $1.0334$ at $\kappa(H)=6$ and falls to $1.0005$ at $\kappa(H)=402$, and the
cell prints $-8\kappa(H)\log\mu_1$ beside it: the two agree to within $0.0002$ in every row, which
the leading-order factor alone does not explain. The runs therefore decay at the slowest mode's
rate, which shows that $a_1>0$, so the lower bound applies to every run printed. $N/\kappa(H)$ runs from $176.2$ to $204.0$. At a fixed $\theta_u$ it agrees across
the three node counts to within $0.1$, and it moves little with $m$: $182.3$, $187.0$ and
$187.5$ at $m=1,2,3$ and $\kappa(H)=102$.

**The atom count moves the step count without moving the spectrum.** At $\theta_u=5$, a control,
and $K=101$, the atom counts $n=2$, $10$ and $100$ give $5{,}300$, $5{,}310$ and $5{,}100$ steps at the
same $\lambda_{\max}(H)=27$ and the same largest derivative $a$ at the start. What moves is $a_1$,
the slowest mode's share of the start, which the lexical field sets: it reads $7.47$, $8.12$ and
$3.04$, and $\log(a_1/\mathrm{tol})/(-\log\mu_1)$ plus the ten steps of patience reproduces each
count to within one step (block 2b).

**What in the step count the equations supply, and what this instantiation does.** $N\,\mathrm dt$
is the settling time in the time of Eqs. (18)–(19). In units of $\tau_\varphi$ it is $25.5$ at
$\kappa(H)=6$ and $22.0$ at $\kappa(H)=402$. It falls slightly because the tolerance is keyed to
$\lambda_{\max}(H)$, so $\log(a/\mathrm{tol})$ shrinks as $\theta_u$ grows. $N$ counts the same
time in steps, $N=(N\,\mathrm dt/\tau_\varphi)\cdot\tau_\varphi/\mathrm dt$ with
$\tau_\varphi/\mathrm dt=8\lambda_{\max}(H)$. Of that factor, the equations every instantiation
shares supply the separation: commitment 7 of Text cell 3 requires
$\tau_\varphi/\tau_\varepsilon\ge4\lambda_{\max}(H)$, which *Integration cost and conditioning*
reports. This instantiation supplies the rest: it sets $\tau_\varepsilon$ at that bound, integrates
by Euler at $\mathrm dt=\tau_\varepsilon/2$, and stops at a tolerance keyed to
$\lambda_{\max}(H)$. So the $\theta_u^2$ in $N$ enters through the separation the equations
require, and that it appears as a count of steps, with the constant $8$, is this instantiation's.

### <a id="appg-4" name="appg-4"></a>4. The total

One inference by integration, the two evaluations of $\lambda_{\max}(H)$ followed by $N$ steps:

$$T_{\text{infer}}=\Theta(Km^2)+\Theta(Km)\,N
=O\!\left(Km\left(m+(\theta_u^2+2)\log\frac{a}{\mathrm{tol}}\right)\right).
\tag{G3}$$

The total is an $O$ and not a $\Theta$ because $N$ is. One inference in closed form costs
$\Theta(Km^2)$, and building a network costs $\Theta(Km(m+\lvert Y\rvert))$.

Learning by Eq. (20) runs until its own update falls below its tolerance, and raises at
`num_updates`. No bound on the number of updates $U$ is derived here beyond that cap, so

$$T_{\text{learn}}=O\!\left(U_{\max}\,\lvert Y\rvert\,Km\left(m+\max_t(\theta_t^2+2)\,
\log\frac{a}{\mathrm{tol}}\right)\right),$$

with $U_{\max}$ the cap, and the maximum taken over the $\theta_u$ the flow passes through.

The memory is $\Theta(Km)$ for $B$, and $\Theta(mN)$ for the history of an inference when it is
recorded.

**The counts describe arithmetic and not seconds.** At the sizes this notebook runs, the time per Euler
step, printed on Code Cell G's `cost:` lines, grows far more slowly than $K$ from $K=101$ to
$K=3201$ on the machine that ran it, because a fixed overhead per operation outweighs the
arithmetic at these sizes. The notebook therefore quotes step
counts wherever a comparison has to survive a change of machine (Text cell 4).

<a id="codeg" name="codeg"></a>

````python
# === Code Cell G: the numbers Appendix G quotes. EVALUATION ===
#
# The cost of THIS SIMULATION, one instantiation of the architecture (Appendix G, opening). Three
# blocks: the spectrum of H against Eq. (G1); Euler step counts against Eq. (G2), and at fixed
# theta_u against the atom count n; and the wall clock, printed behind `cost:` because it moves with the machine. Every helper is
# local to this cell, so `code cell 1` is untouched.

import time

SPECTRUM_NODE_COUNTS = (51, 101, 201)
SPECTRUM_BASIS_DEGREES = (1, 2, 3)
STEP_COUNT_THETAS = (2.0, 5.0, 10.0, 20.0)     # theta_u fixed: a stated control (B4)
N_LADDER = (2, 10, 100)
N_LADDER_THETA = 5.0                           # theta_u fixed: a stated control (B4)
TIMING_THETA = 20.0
TIMING_NODE_COUNTS = (101, 401, 1601, 3201)
GENERAL_SIGMAS = ((2.0, 1.0, 1.0), (1.0, 3.0, 1.0), (2.5, 4.0, 1.5))


def state_hessian(net, theta_u):
    """H in the quadrature metric, whole and dense, in symmetrized coordinates: the
    matrix `stiffest_state_rate` (Code Cell 1) avoids building, kept here to check it."""
    root_w_basis = torch.sqrt(net.weights)[:, None] * net.basis
    gram = net.basis.T @ (net.weights[:, None] * net.basis)
    eye_K = torch.eye(net.num_nodes, dtype=net.dtype)
    eye_m = torch.eye(net.num_basis, dtype=net.dtype)
    a = 1.0 / net.sigma_lexical + 1.0 / net.sigma_state
    return torch.cat([
        torch.cat([a * eye_K, -theta_u * root_w_basis / net.sigma_state], dim=1),
        torch.cat([-theta_u * root_w_basis.T / net.sigma_state,
                   eye_m / net.sigma_utility + theta_u ** 2 * gram / net.sigma_state], dim=1),
    ], dim=0)


def recurrence_factors(lam, dt, tau_error, tau_state=1.0):
    """The two per-step factors of one mode of H, eigenvalue `lam`, under `infer`'s update
    order: the error units first, then the states from the updated errors (Sec. 3)."""
    alpha, beta = dt / tau_error, dt / tau_state
    trace = 2.0 - alpha - alpha * beta * lam
    det = 1.0 - alpha
    root = math.sqrt(trace * trace - 4.0 * det)
    return 0.5 * (trace + root), 0.5 * (trace - root)


def predicted_spectrum(K, m, theta_u):
    """Eq. (G1), ascending."""
    return sorted([1.0] * m + [2.0] * (K - m) + [theta_u ** 2 + 2.0] * m)


def complexity_report(net, patience=10):
    theta_star = net.theta_u
    thetas = (0.0, 1.0, 10.0, theta_star, 1.0e3)

    print("APPENDIX G: THE COMPUTATIONAL COMPLEXITY OF THE SIMULATION")
    print("  Eq. (G1): spec H = {1 (m times), 2 (K - m times), theta_u^2 + 2 (m times)} at sigma = 1,")
    print("  so the condition number kappa(H) = lambda_max / lambda_min = theta_u^2 + 2.")
    print()

    # 1. the spectrum
    print(f"  1. THE SPECTRUM OF H AGAINST EQ. (G1), at theta_u in "
          f"{{{', '.join(f'{t:g}' for t in thetas[:3])}, {theta_star:.4f} (theta*), {thetas[-1]:g}}}")
    print(f"    {'K':>5}{'m':>4}   largest |lambda_k - (G1)_k| / lambda_max    largest |`stiffest_state_rate`"
          f" - dense lambda_max| / lambda_max")
    print(f"    {'':>9}   over the five theta_u")
    worst, worst_rate = 0.0, 0.0
    for K in SPECTRUM_NODE_COUNTS:
        for m in SPECTRUM_BASIS_DEGREES:
            probe = net.respawn(num_nodes=K, basis_degree=m, mu_u=1.0)
            gap, rate_gap = 0.0, 0.0
            for t in thetas:
                eigenvalues = torch.linalg.eigvalsh(state_hessian(probe, t))
                predicted = torch.tensor(predicted_spectrum(K, m, t), dtype=probe.dtype)
                gap = max(gap, float((eigenvalues - predicted).abs().max()) / float(predicted.max()))
                top = float(eigenvalues.max())
                rate_gap = max(rate_gap, abs(probe.stiffest_state_rate(t) - top) / top)
            worst, worst_rate = max(worst, gap), max(worst_rate, rate_gap)
            print(f"    {K:>5}{m:>4}   {gap:.1e}{'':>36}{rate_gap:.1e}")
    count = len(SPECTRUM_NODE_COUNTS) * len(SPECTRUM_BASIS_DEGREES) * len(thetas)
    print(f"    largest over all {count} configurations: {worst:.1e} against Eq. (G1), "
          f"{worst_rate:.1e} for `stiffest_state_rate`")

    # the blockwise lambda_max away from Eq. (G1)'s premises: sigma != 1, and a basis that is
    # not orthonormal or not of full rank, set on probe networks for this check alone
    generator = torch.Generator().manual_seed(0)
    general_gap, general_count = 0.0, 0
    for sigmas in GENERAL_SIGMAS:
        for m in (2, 3):
            probe = net.respawn(num_nodes=51, basis_degree=m, mu_u=1.0, sigma_lexical=sigmas[0],
                                sigma_state=sigmas[1], sigma_utility=sigmas[2])
            random_basis = torch.randn(51, m, generator=generator, dtype=probe.dtype)
            deficient_basis = random_basis.clone()
            deficient_basis[:, -1] = 2.0 * deficient_basis[:, 0]
            for basis in (probe.basis, random_basis, deficient_basis):
                probe.basis = basis
                for t in thetas:
                    top = float(torch.linalg.eigvalsh(state_hessian(probe, t)).max())
                    general_gap = max(general_gap, abs(probe.stiffest_state_rate(t) - top) / top)
                    general_count += 1
    print(f"    the same for sigma (L, S, u) in {', '.join(str(s) for s in GENERAL_SIGMAS)}, m = 2, 3, and")
    print(f"    an orthonormal, a random and a rank-deficient basis: largest over {general_count} "
          f"configurations {general_gap:.1e}")
    lam_max = net.stiffest_state_rate()
    lam_min = float(torch.linalg.eigvalsh(state_hessian(net, theta_star)).min())
    phi_u_rate = 1.0 + theta_star ** 2              # Eq. (28) at sigma = 1, lambda_max(G) = 1
    print(f"    at the evaluation network's theta* = {theta_star:.4f}:")
    print(f"      lambda_max from `stiffest_state_rate` {lam_max:.4f}, theta*^2 + 2 = "
          f"{theta_star ** 2 + 2.0:.4f}, lambda_min {lam_min:.4f}")
    print(f"      kappa(H) = {lam_max / lam_min:.2f}; the stiffness ratio of Text cell 4, Eq. (28)'s "
          f"phi_u rate over the phi_S rate 2, is {phi_u_rate / 2.0:.2f}")
    print()

    # 2. the step count
    print("  2. EULER STEPS AGAINST EQ. (G2), under \"some\", theta_u FIXED as a control (the step")
    print("  count is read as a function of theta_u, which theta* does not vary), tau_epsilon and dt")
    print("  at their defaults. 'decay x 8 kappa' is the largest derivative's decay per step over the")
    print("  second half of the run, times 8 kappa(H); 'recurrence' is the same quantity predicted by")
    print("  the slow factor of `infer`'s update order at lambda_min(H) = 1 (Sec. 3).")
    print(f"    {'K':>5}{'m':>4}{'theta_u':>9}{'kappa(H)':>10}{'steps N':>10}{'N / kappa':>11}"
          f"{'decay x 8 kappa':>17}{'recurrence':>12}{'N dt / tau_phi':>16}")
    worst_factor, every_mode_real = 0.0, True
    configurations = ([(K, 2, t) for K in SPECTRUM_NODE_COUNTS for t in STEP_COUNT_THETAS]
                      + [(101, m, 10.0) for m in (1, 3)])
    for K, m, t in configurations:
        probe = net.respawn(num_nodes=K, basis_degree=m, mu_u=1.0, theta_u=t)
        kappa = t ** 2 + 2.0
        result = probe.infer("some", patience=patience)
        if not result["converged"]:
            raise RuntimeError(f"K = {K}, m = {m}, theta_u = {t:g} did not converge")
        steps = result["steps"]
        trace = [row["max_derivative"] for row in result["history"]]
        first, last = steps // 2, steps - 1
        decay = (math.log(trace[first]) - math.log(trace[last])) / (last - first)
        slow = recurrence_factors(1.0, result["dt"], result["tau_error"])[0]
        for lam in (1.0, 2.0, kappa):                                  # Eq. (G1)'s spectrum
            try:
                factors = recurrence_factors(lam, result["dt"], result["tau_error"])
            except ValueError:                                         # complex factors
                every_mode_real = False
                continue
            worst_factor = max(worst_factor, *(abs(f) for f in factors))
        print(f"    {K:>5}{m:>4}{t:>9g}{kappa:>10g}{steps:>10}{steps / kappa:>11.1f}"
              f"{decay * 8.0 * kappa:>17.4f}{-math.log(slow) * 8.0 * kappa:>12.4f}"
              f"{steps * result['dt']:>16.2f}")   # tau_phi = 1
    print(f"    every mode's two factors real: {every_mode_real}; the largest in modulus over every "
          f"eigenvalue of every row: {worst_factor:.6f}")
    print()

    # 2b. the atom count, at fixed theta_u and so fixed kappa(H)
    print(f"  2b. THE ATOM COUNT n AT FIXED theta_u = {N_LADDER_THETA:g} (a control) AND K = 101, under \"some\".")
    print("  The spectrum does not move, and neither does the largest derivative a at the start;")
    print("  the slowest mode's share a_1 does, because the lexical field sets the start's")
    print("  modal amplitudes. a_1 is read off the run: the last step's largest derivative over")
    print("  the slow factor to the power of its step. 'from a_1' is log(a_1 / tol) / -log(slow")
    print("  factor) + patience, Eq. (G2)'s count with the recurrence's own factor.")
    print(f"    {'n':>5}{'lambda_max(H)':>15}{'a':>12}{'a_1':>10}{'steps N':>10}{'from a_1':>10}")
    for n in N_LADDER:
        probe = net.respawn(num_atoms=n, theta_u=N_LADDER_THETA)
        result = probe.infer("some", patience=patience)
        if not result["converged"]:
            raise RuntimeError(f"n = {n} did not converge")
        trace = [row["max_derivative"] for row in result["history"]]
        slow = recurrence_factors(1.0, result["dt"], result["tau_error"])[0]
        last = result["steps"] - 1
        a_1 = trace[last] / slow ** last
        predicted = math.log(a_1 / result["derivative_tolerance"]) / -math.log(slow) + patience
        print(f"    {n:>5}{probe.stiffest_state_rate():>15.4f}{trace[0]:>12.4f}{a_1:>10.4f}"
              f"{result['steps']:>10}{predicted:>10.1f}")
    print()

    # 3. the wall clock
    print(f"  3. THE WALL CLOCK, which moves with the machine, at theta_u = {TIMING_THETA:g} (a control): one")
    print("  evaluation of lambda_max(H), and one Euler step, timed as the difference between a full")
    print(f"  run of `infer` and a run capped at one step, which pays the same set-up; theta_u = {TIMING_THETA:g}")
    print("  gives enough steps for the difference to stand clear of the set-up's own variation.")
    for K in TIMING_NODE_COUNTS:
        probe = net.respawn(num_nodes=K, theta_u=TIMING_THETA)
        start = time.perf_counter()
        probe.stiffest_state_rate()
        setup = time.perf_counter() - start
        start = time.perf_counter()
        probe.infer("some", record_history=False, max_steps=1)
        one_step = time.perf_counter() - start
        start = time.perf_counter()
        result = probe.infer("some", record_history=False)
        full = time.perf_counter() - start
        per_step = (full - one_step) / (result["steps"] - 1)
        print(f"cost: K = {K:>5}: lambda_max(H) {setup:.5f} s, "
              f"{result['steps']} steps at {per_step * 1e6:.1f} microseconds each")
    print()


complexity_report(evaluation_network)
````
